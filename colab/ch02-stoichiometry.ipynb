## 2.1 Relative masses of atoms and molecules

### A. Unified atomic mass unit

一个 atom 的质量太小（10<sup>−27</sup> – 10<sup>−25</sup> kg），直接用 kg 很不方便。于是选一个 standard：**carbon-12 atom**。

**unified atomic mass unit**（符号 u）= one twelfth of the mass of a carbon-12 atom ≈ 1.66 × 10<sup>−27</sup> kg。

所有 relative mass 都是「这个粒子的质量是 1 u 的多少倍」，是两个质量之比，所以**没有单位**。

> **LO 2.1.1** — define the unified atomic mass unit as one twelfth of the mass of a carbon-12 atom

### B. 四个 relative mass

| term | 比的是什么 | 用在哪 |
|---|---|---|
| relative isotopic mass | the mass of **one atom of a particular isotope** ÷ 1 u | 单个 isotope，例如 <sup>35</sup>Cl = 35.0（近似整数） |
| relative atomic mass, *A*ᵣ | the **weighted average mass** of the atoms of an element ÷ 1 u | element（天然 isotope 混合物），例如 Cl = 35.5 |
| relative molecular mass, *M*ᵣ | the weighted average mass of **one molecule** ÷ 1 u | simple molecular substance，例如 CH₄ = 16.0 |
| relative formula mass | the weighted average mass of **one formula unit** ÷ 1 u | ionic compound（没有 molecule），例如 Mg(OH)₂ = 58.3 |

relative isotopic mass 的判分（9701/21 Oct/Nov 2024 Q1(c)(i)，2 分）：

- M1 **mass of an (atom of an) isotope**
- M2 compared to the unified atomic mass unit **OR** on a scale on which a carbon-12 atom has a mass of exactly 12 units **OR** divided by 1/12 mass of a carbon-12 atom

讲义提醒：M2 写成 "compared to the unified atomic mass" 少了 *unit* 会被拒。*A*ᵣ 的定义比它多一个关键词 **weighted average**——因为大部分 element 是 isotopes 的混合物，所以 Periodic Table 上的 *A*ᵣ 不是整数（Cl = 35.5）。

*M*ᵣ / relative formula mass 就是把 formula 里所有 atom 的 *A*ᵣ 加起来：

- CH₄：12.0 + 4 × 1.0 = 16.0
- Mg(OH)₂：24.3 + 2 × (16.0 + 1.0) = 58.3
- CuSO₄·5H₂O：63.5 + 32.1 + 4 × 16.0 + 5 × 18.0 = 249.6（water of crystallisation 也要算进去）

> **LO 2.1.2** — define relative atomic mass, *A*ᵣ, relative isotopic mass, relative molecular mass, *M*ᵣ, and relative formula mass in terms of the unified atomic mass unit

### C. 从 mass spectrum 算 *A*ᵣ

mass spectrometer 把 atom 变成 +1 ion，按 *m*/*e* 分开；每个 peak 对应一个 isotope，peak 的高度是它的 relative abundance。

*A*ᵣ = Σ(relative isotopic mass × relative abundance) ÷ Σ(relative abundance)

- abundance 不一定是百分数（boron：<sup>10</sup>B 23、<sup>11</sup>B 100），所以分母要写 **Σ abundance**，不能直接 ÷ 100。
- boron：(10 × 23 + 11 × 100) ÷ (23 + 100) = 10.8
- chlorine：(35 × 75 + 37 × 25) ÷ 100 = 35.5

```python
# @title 🔬 mass spectrum → Aᵣ { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-03">
<h3>Mass spectrum → <i>A</i>ᵣ</h3>
<div class="row">element <select class="ars-pre"></select>
  <span class="faint">拖柱顶的圆点，或改下面的数字，调 relative abundance</span></div>
<div class="ars-note faint"></div>
<svg class="ars-svg" viewBox="0 0 760 330" aria-label="mass spectrum"></svg>
<div class="row ars-inputs"></div>
<div class="card ars-out"></div>

<h4>自测：这张 spectrum 的 <i>A</i>ᵣ 是多少？（1 d.p.，填对变绿）</h4>
<svg class="ars-qsvg" viewBox="0 0 760 250" aria-label="mass spectrum quiz"></svg>
<div class="row"><span><i>A</i>ᵣ =</span><input type="text" class="ars-qin" autocomplete="off" inputmode="decimal">
  <span class="ars-qres"></span>
  <button class="ars-qnew">换一题</button></div>
<button class="reveal">看算式</button><div class="ans ars-qans"></div>

<style>
.k .ars-svg,.k .ars-qsvg{background:var(--card);border:1px solid var(--line);border-radius:8px;margin:6px 0;touch-action:none}
.k .ars-svg text,.k .ars-qsvg text{font-size:14px}
.k .ars-svg text.ars-ax,.k .ars-qsvg text.ars-ax{fill:var(--mut);font-size:13px}
.k .ars-svg text.ars-ar{fill:var(--s2);font-weight:600}
.k .ars-svg text.ars-ab,.k .ars-qsvg text.ars-ab{font-weight:600}
.k .ars-bar{fill:var(--acc);fill-opacity:.82}
.k .ars-hit{fill:transparent;cursor:ns-resize}
.k .ars-knob{fill:var(--card);stroke:var(--acc);stroke-width:2.5;cursor:ns-resize}
.k .ars-svg.drag .ars-knob.on{fill:var(--acc)}
.k .ars-axis{stroke:var(--fg);stroke-width:1.2}
.k .ars-grid{stroke:var(--line);stroke-width:1}
.k .ars-arl{stroke:var(--s2);stroke-width:2;stroke-dasharray:6 4}
.k .ars-inputs label{display:inline-flex;gap:5px;align-items:center;margin-right:10px}
.k .ars-inputs input{width:6em}
.k .ars-out .ars-eq{font-size:16px;margin:2px 0}
.k .ars-out .ars-big{font-size:20px;font-weight:650}
.k .ars-out .err{color:var(--bad)}
.k input.ars-qin{width:6em;text-align:center}
.k input.ars-qin.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.ars-qin.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .ars-qres.good{color:var(--ok)} .k .ars-qres.badv{color:var(--bad)}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var PRE = [
  { id: 'boron', sym: 'B', iso: [[10, 23], [11, 100]], dp: 1, step: 0,
    note: 'abundance 不是百分数（¹¹B 记作 100）——分母是 23 + 100 = 123，不是 100。' },
  { id: 'neon', sym: 'Ne', iso: [[20, 90.9], [21, 0.3], [22, 8.8]], dp: 1, step: 1,
    note: '²¹Ne 只有 0.3%，柱子几乎看不见，但它照样进算式。' },
  { id: 'chlorine', sym: 'Cl', iso: [[35, 75], [37, 25]], dp: 1, step: 0,
    note: 'Periodic Table 上 Cl = 35.5 就是这么来的：³⁵Cl : ³⁷Cl = 3 : 1。' },
  { id: 'lead', sym: 'Pb', iso: [[204, 2], [206, 24], [207, 22], [208, 52]], dp: 1, step: 0,
    note: '四个 isotope：每个 peak 都要乘、都要加。' },
  { id: 'iridium', sym: 'Ir', iso: [[190.96, 15.18], [192.96, 34.82]], dp: 2, step: 2,
    note: '9701/21 Oct/Nov 2024：用 relative isotopic mass（不是整数），答案给 2 d.p. → 192.35。' },
  { id: 'magnesium', sym: 'Mg', iso: [[24, 79], [25, 10], [26, 11]], dp: 1, step: 0,
    note: '²⁴Mg 占 79%，所以 <i>A</i>ᵣ 被拉向 24。' }
];
var sel = $root.querySelector('.ars-pre'), svg = $root.querySelector('.ars-svg'),
    out = $root.querySelector('.ars-out'), inBox = $root.querySelector('.ars-inputs'), noteBox = $root.querySelector('.ars-note');
PRE.forEach(function (p, i) { sel.appendChild(K.el('option', { value: i }, p.id)); });

/* number → plain string without float noise */
function num(x, dp) {
  if (dp != null) return x.toFixed(dp);
  return String(+x.toFixed(6));
}
function isoName(m, sym) {
  var a = Math.round(m);
  return K.sup(a) + sym + (Math.abs(m - a) > 1e-9 ? ' (' + num(m) + ')' : '');
}
function arOf(iso) {
  var top = 0, bot = 0;
  iso.forEach(function (r) { top += r[0] * r[1]; bot += r[1]; });
  return { top: top, bot: bot, ar: bot > 0 ? top / bot : NaN };
}

/* geometry of a chart: plot box + scales */
function frame(svgEl, iso, yTop, H) {
  var ms = iso.map(function (r) { return r[0]; });
  var lo = Math.min.apply(null, ms) - 1, hi = Math.max.apply(null, ms) + 1;
  var W = +svgEl.getAttribute('viewBox').split(' ')[2];
  var g = { x0: 78, x1: W - 30, y0: H - 52, y1: 46, lo: lo, hi: hi, yTop: yTop };
  g.X = function (m) { return g.x0 + (m - lo) / (hi - lo) * (g.x1 - g.x0); };
  g.Y = function (v) { return g.y0 - v / yTop * (g.y0 - g.y1); };
  svgEl.setAttribute('data-plot', [g.x0, g.y1, g.x1 - g.x0, g.y0 - g.y1].join(','));
  svgEl.setAttribute('data-ytop', yTop);
  return g;
}
function topFor(max) { var t = K.ticks(Math.max(max, 1e-6) * 1.15, 4); return t[t.length - 1]; }

/* draw one spectrum; opts: {sym, dp, step, drag, arLine} */
function draw(svgEl, iso, yTop, opts) {
  while (svgEl.firstChild) svgEl.removeChild(svgEl.firstChild);
  /* viewBox width = rendered width, so 1 user unit = 1 CSS px and text never shrinks below its font-size */
  var H = +svgEl.getAttribute('viewBox').split(' ')[3];
  var W = Math.round(svgEl.getBoundingClientRect().width) || 760;
  svgEl.setAttribute('viewBox', '0 0 ' + Math.max(W, 420) + ' ' + H);
  var g = frame(svgEl, iso, yTop, H);
  var avoid = [], placed = [];
  var ticks = K.ticks(yTop, 4).filter(function (v) { return v <= yTop + 1e-9; });
  ticks.forEach(function (v) {
    if (v > 0) svgEl.appendChild(K.svg('line', { 'class': 'ars-grid', x1: g.x0, x2: g.x1, y1: g.Y(v), y2: g.Y(v) }));
  });
  svgEl.appendChild(K.svg('line', { 'class': 'ars-axis', x1: g.x0, x2: g.x1, y1: g.y0, y2: g.y0 }));
  svgEl.appendChild(K.svg('line', { 'class': 'ars-axis', x1: g.x0, x2: g.x0, y1: g.y0, y2: g.y1 - 6 }));
  avoid.push([g.x0, g.y0, g.x1, g.y0], [g.x0, g.y0, g.x0, g.y1 - 6]);
  var bw = Math.min(26, (g.x1 - g.x0) / (g.hi - g.lo) * 0.3);
  var bars = iso.map(function (r, i) {
    var x = g.X(r[0]), y = g.Y(r[1]);
    var rect = K.svg('rect', { 'class': 'ars-bar', x: x - bw / 2, y: y, width: bw, height: g.y0 - y, 'data-v': r[1], 'data-m': r[0], 'data-i': i });
    svgEl.appendChild(rect);
    avoid.push([x - bw / 2, y, x - bw / 2, g.y0], [x + bw / 2, y, x + bw / 2, g.y0], [x - bw / 2, y, x + bw / 2, y]);
    return { x: x, y: y, rect: rect };
  });
  var ar = arOf(iso).ar, ax = null;
  if (opts.arLine && isFinite(ar)) {
    ax = g.X(ar);
    var ln = K.svg('line', { 'class': 'ars-arl', x1: ax, x2: ax, y1: g.y0, y2: g.y1 - 20, 'data-ar': ar });
    svgEl.appendChild(ln);
    avoid.push([ax, g.y0, ax, g.y1 - 20]);
  }
  iso.forEach(function (r, i) {
    K.label(svgEl, bars[i].x, g.y0, num(r[0]), { cls: 'ars-ax ars-me', cands: [[0, 22, 'middle'], [0, 38, 'middle']], avoid: avoid, placed: placed });
  });
  if (opts.drag) bars.forEach(function (b, i) {
    svgEl.appendChild(K.svg('rect', { 'class': 'ars-hit', x: b.x - bw / 2 - 8, y: g.y1, width: bw + 16, height: g.y0 - g.y1 + 8, 'data-i': i }));
    var k = K.svg('circle', { 'class': 'ars-knob', cx: b.x, cy: b.y, r: 7, 'data-i': i });
    svgEl.appendChild(k);
    placed.push({ x: b.x - 9, y: b.y - 9, w: 18, h: 18 });
  });
  /* text: y ticks, m/e under the axis, abundance on top of each bar, then the A_r label */
  ticks.forEach(function (v) {
    K.label(svgEl, g.x0, g.Y(v), num(v), { cls: 'ars-ax', cands: [[-8, 5, 'end'], [-8, 9, 'end'], [-8, 1, 'end']], avoid: avoid, placed: placed, pad: 1 });
  });
  K.label(svgEl, g.x0, g.y1 - 6, 'relative abundance', { cls: 'ars-ax', cands: [[-6, -12, 'start'], [0, -12, 'start']], avoid: avoid, placed: placed });
  var me = K.label(svgEl, g.x1, g.y0, 'm/e', { cls: 'ars-ax', cands: [[0, 40, 'end'], [0, 22, 'end']], avoid: avoid, placed: placed });
  me.textContent = '';
  ['m', '/', 'e'].forEach(function (c) { var s = K.svg('tspan', c === '/' ? {} : { 'font-style': 'italic' }); s.textContent = c; me.appendChild(s); });
  iso.forEach(function (r, i) {
    var dy = opts.drag ? -18 : -9;   /* -7 lets the 14 px bbox (descent + pad) touch the bar top → falls to a far candidate */
    K.label(svgEl, bars[i].x, bars[i].y, num(r[1], opts.step), {
      cls: 'ars-ab', avoid: avoid, placed: placed,
      cands: [[0, dy, 'middle'], [12, dy, 'start'], [-12, dy, 'end'], [14, 5, 'start'], [-14, 5, 'end'], [0, dy - 16, 'middle'], [16, -18, 'start'], [-16, -18, 'end']]
    }).setAttribute('data-i', i);
  });
  if (ax != null) {
    var t = K.label(svgEl, ax, g.y1 - 20, 'Ar = ' + num(ar, opts.dp), {
      cls: 'ars-ar', avoid: avoid, placed: placed,
      cands: [[0, -6, 'middle'], [6, -6, 'start'], [-6, -6, 'end'], [8, 12, 'start'], [-8, 12, 'end'], [8, 30, 'start'], [-8, 30, 'end']]
    });
    /* italic A, subscript r — same width as the plain text K.label measured */
    t.textContent = '';
    var a = K.svg('tspan', { 'font-style': 'italic' }); a.textContent = 'A'; t.appendChild(a);
    var rest = K.svg('tspan'); rest.textContent = 'ᵣ = ' + num(ar, opts.dp); t.appendChild(rest);
  }
  return g;
}

/* ---------- main chart ---------- */
var cur = null, iso = [], yTop = 100, geo = null, inputs = [];
function load(i) {
  cur = PRE[i];
  iso = cur.iso.map(function (r) { return r.slice(); });
  yTop = topFor(Math.max.apply(null, iso.map(function (r) { return r[1]; })));
  noteBox.innerHTML = cur.note;
  inBox.innerHTML = '';
  inputs = iso.map(function (r, j) {
    var lab = K.el('label', {}, isoName(r[0], cur.sym) + ' ');
    var inp = K.el('input', { type: 'number', min: 0, step: Math.pow(10, -cur.step), 'data-i': j, 'class': 'ars-ab-in' });
    inp.value = num(r[1], cur.step);
    inp.addEventListener('input', function () {
      var v = parseFloat(inp.value);
      iso[j][1] = isFinite(v) && v >= 0 ? v : 0;
      if (iso[j][1] > yTop) yTop = topFor(iso[j][1]);
      render(false);
    });
    lab.appendChild(inp); inBox.appendChild(lab);
    return inp;
  });
  render(false);
}
function render(syncInputs) {
  geo = draw(svg, iso, yTop, { sym: cur.sym, dp: cur.dp, step: cur.step, drag: true, arLine: true });
  if (syncInputs) inputs.forEach(function (inp, j) { inp.value = num(iso[j][1], cur.step); });
  var s = arOf(iso);
  var terms = iso.map(function (r) { return num(r[0]) + ' × ' + num(r[1], cur.step); }).join(' + ');
  var bots = iso.map(function (r) { return num(r[1], cur.step); }).join(' + ');
  var A = '<i>A</i>ᵣ';
  if (!(s.bot > 0)) {
    out.innerHTML = '<p class="err">Σ abundance = 0：至少要有一个 peak。</p>';
  } else {
    out.innerHTML =
      '<p class="ars-eq">' + A + ' = Σ(<i>m</i>/<i>e</i> × abundance) ÷ Σ abundance</p>' +
      '<p class="ars-eq ars-work">' + A + ' = (' + terms + ') ÷ (' + bots + ')</p>' +
      '<p class="ars-eq">　 = ' + num(s.top) + ' ÷ ' + num(s.bot, cur.step) + ' = <span class="ars-big ars-res">' + num(s.ar, cur.dp) + '</span>' +
      ' <span class="faint">（' + cur.dp + ' d.p.）</span></p>' +
      '<p class="faint">红色虚线就是 ' + A + ' 在 <i>m</i>/<i>e</i> 轴上的位置：weighted average，所以它靠近最高的那根柱，而不是落在正中间。</p>';
  }
  K.fit();
}
sel.addEventListener('change', function () { load(+sel.value); });

/* drag a bar top (knob or the column under it) */
var dragI = -1;
function svgY(ev) {
  var pt = svg.createSVGPoint(); pt.x = ev.clientX; pt.y = ev.clientY;
  return pt.matrixTransform(svg.getScreenCTM().inverse()).y;
}
function setFromPointer(ev) {
  var v = (geo.y0 - svgY(ev)) / (geo.y0 - geo.y1) * yTop;
  v = Math.max(0, Math.min(yTop, v));
  var f = Math.pow(10, cur.step); v = Math.round(v * f) / f;
  iso[dragI][1] = v;
  render(true);
  var k = svg.querySelector('.ars-knob[data-i="' + dragI + '"]'); if (k) k.classList.add('on');
}
svg.addEventListener('pointerdown', function (ev) {
  var t = ev.target.closest('[data-i]');
  if (!t || !(t.classList.contains('ars-knob') || t.classList.contains('ars-hit') || t.classList.contains('ars-bar'))) return;
  dragI = +t.getAttribute('data-i');
  svg.classList.add('drag');
  try { svg.setPointerCapture(ev.pointerId); } catch (e) { /* synthetic events */ }
  ev.preventDefault();
  setFromPointer(ev);
});
svg.addEventListener('pointermove', function (ev) { if (dragI >= 0) setFromPointer(ev); });
function endDrag() { dragI = -1; svg.classList.remove('drag'); }
svg.addEventListener('pointerup', endDrag);
svg.addEventListener('pointercancel', endDrag);

load(0);

/* ---------- self-test ---------- */
var qsvg = $root.querySelector('.ars-qsvg'), qin = $root.querySelector('.ars-qin'),
    qres = $root.querySelector('.ars-qres'), qans = $root.querySelector('.ars-qans'), q = null;
function rnd(a, b) { return a + Math.floor(Math.random() * (b - a + 1)); }
function newQ() {
  var n = rnd(2, 3), m0 = rnd(6, 120), gap = n === 2 ? rnd(1, 2) : 1, list = [], used = {};
  for (var i = 0; i < n; i++) list.push([m0 + i * gap, 0]);
  list.forEach(function (r) { r[1] = rnd(1, 20) * 5; });
  list[rnd(0, n - 1)][1] = 100;              /* biggest peak scaled to 100, like many spectra */
  q = { iso: list, ar: arOf(list).ar };
  qsvg.setAttribute('data-ans', q.ar);
  draw(qsvg, list, topFor(100), { step: 0 });
  qin.value = ''; qin.classList.remove('good', 'badv'); qres.textContent = ''; qres.className = 'ars-qres';
  var s = arOf(list);
  qans.innerHTML = '<i>A</i>ᵣ = (' + list.map(function (r) { return r[0] + ' × ' + r[1]; }).join(' + ') + ') ÷ (' +
    list.map(function (r) { return r[1]; }).join(' + ') + ') = ' + num(s.top) + ' ÷ ' + num(s.bot) + ' = <b>' + num(s.ar, 1) + '</b>';
  var btn = $root.querySelector('button.reveal');
  if (qans.classList.contains('show')) btn.click();
  K.fit();
}
qin.addEventListener('input', function () {
  var v = parseFloat(qin.value.replace(/[−–]/g, '-'));
  qin.classList.remove('good', 'badv'); qres.className = 'ars-qres'; qres.textContent = '';
  if (!qin.value.trim()) return;
  var ok = /^\s*[\d.]+\s*$/.test(qin.value) && Math.abs(v - q.ar) <= 0.05 + 1e-9;
  qin.classList.add(ok ? 'good' : 'badv'); qres.classList.add(ok ? 'good' : 'badv');
  qres.textContent = ok ? '✓' : '✗　分母是 Σ abundance，不是 100';
});
$root.querySelector('.ars-qnew').addEventListener('click', newQ);
newQ();

/* redraw both charts when the output area changes width (Colab panel, narrow window) */
var lastW = svg.getBoundingClientRect().width;
if (window.ResizeObserver) new ResizeObserver(function () {
  var w = svg.getBoundingClientRect().width;
  if (Math.abs(w - lastW) < 1) return;
  lastW = w;
  render(false);
  draw(qsvg, q.iso, topFor(100), { step: 0 });
}).observe(svg);

K.init($root);
})(document.getElementById("k-ch02-03"));</script>"""))
```

## 2.2 The mole and the Avogadro constant

**mole**：the amount of substance that contains **6.02 × 10<sup>23</sup>** specified particles（atoms、molecules、ions 或 electrons）。这个数就是 **Avogadro constant, *L***（6.02 × 10<sup>23</sup> mol<sup>−1</sup>）。

1 mol 任何物质的质量，以 g 计，在数值上等于它的 *A*ᵣ / *M*ᵣ——这就是 **molar mass, *M***，单位 g mol<sup>−1</sup>。

| 关系 | 公式 |
|---|---|
| mass ↔ amount | *n* = *m* / *M* |
| amount ↔ particles | *N* = *n* × *L* |
| gas volume ↔ amount（room conditions） | *n* = *V* / 24.0（*V* 用 dm<sup>3</sup>） |
| solution ↔ amount | *n* = *c* × *V*（*V* 用 dm<sup>3</sup>；1 dm<sup>3</sup> = 1000 cm<sup>3</sup>） |

**一定要说清楚是什么 particle**：「1 mol of oxygen」有歧义——1 mol O atoms 是 16.0 g，1 mol O₂ molecules 是 32.0 g、含 2 × 6.02 × 10<sup>23</sup> 个 O atoms。

例：0.18 g ethanedioic acid, HO₂CCO₂H（*M*ᵣ = 90.0）里有多少个 carbon atom？

1. *n*(acid) = 0.18 / 90.0 = 0.0020 mol
2. 每个 molecule 有 2 个 C：*n*(C) = 2 × 0.0020 = 0.0040 mol
3. *N*(C) = 0.0040 × 6.02 × 10<sup>23</sup> = **2.4 × 10<sup>21</sup>**（2 s.f.，跟 0.18 一样）

```python
# @title 🔬 mole 换算图 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-05">
<h3>mole 换算中枢：一切都先变成 <i>n</i></h3>
<p class="mut">在<b>任意一个</b>盒子里填数，其余盒子自动算出；用到的箭头会变蓝，下面写出每一步的算式。结果按你输入数据的 s.f. 显示（至少 2 位）。</p>
<div class="row">formula <input type="text" class="mm-f" value="HO2CCO2H" spellcheck="false" size="16">
  <span class="mm-pre"></span></div>
<div class="mm-mr card"></div>
<svg class="mm-map" viewBox="0 0 808 500">
  <defs>
    <marker id="mm-a-on" viewBox="0 0 10 10" refX="9" refY="5" markerWidth="4.2" markerHeight="4.2" orient="auto"><path d="M0,0 L10,5 L0,10 z" style="fill:var(--acc)"/></marker>
    <marker id="mm-a-off" viewBox="0 0 10 10" refX="9" refY="5" markerWidth="5" markerHeight="5" orient="auto"><path d="M0,0 L10,5 L0,10 z" style="fill:var(--faint)"/></marker>
  </defs>
  <rect x="0" y="0" width="808" height="500" rx="10" class="mm-bgr"/>
  <g class="mm-arrows"></g>
  <circle cx="404" cy="250" r="70" class="mm-core"/>
  <g class="mm-labels"></g>
  <foreignObject x="334" y="196" width="140" height="108"><div xmlns="http://www.w3.org/1999/xhtml" class="mm-in mm-n" data-box="n">
    <div class="mm-t">amount, <i>n</i> / mol</div><input type="text" class="mm-v" data-k="n" inputmode="decimal" spellcheck="false"></div></foreignObject>
  <foreignObject x="284" y="8" width="240" height="112"><div xmlns="http://www.w3.org/1999/xhtml" class="mm-box" data-box="gas">
    <div class="mm-t">gas volume, <i>V</i> / dm³</div><div class="mm-s">at room conditions</div>
    <input type="text" class="mm-v" data-k="gas" inputmode="decimal" spellcheck="false"></div></foreignObject>
  <foreignObject x="14" y="194" width="230" height="112"><div xmlns="http://www.w3.org/1999/xhtml" class="mm-box" data-box="mass">
    <div class="mm-t">mass, <i>m</i> / g</div><div class="mm-s mm-msub">&#160;</div>
    <input type="text" class="mm-v" data-k="mass" inputmode="decimal" spellcheck="false"></div></foreignObject>
  <foreignObject x="564" y="194" width="230" height="112"><div xmlns="http://www.w3.org/1999/xhtml" class="mm-box" data-box="part">
    <div class="mm-t">number of particles, <i>N</i></div><select class="mm-kind"></select>
    <input type="text" class="mm-v" data-k="part" spellcheck="false"></div></foreignObject>
  <foreignObject x="254" y="380" width="300" height="112"><div xmlns="http://www.w3.org/1999/xhtml" class="mm-box" data-box="sol">
    <div class="mm-t">solution</div>
    <div class="mm-two"><label><span><i>c</i> / mol dm⁻³</span><input type="text" class="mm-v" data-k="c" inputmode="decimal" spellcheck="false"></label>
    <label><span><i>V</i> / cm³</span><input type="text" class="mm-v" data-k="vol" inputmode="decimal" spellcheck="false" value="250"></label></div></div></foreignObject>
</svg>
<div class="mm-work card"></div>
<p class="faint"><i>L</i> 用 6.02 × 10²³ mol⁻¹；gas 的 molar volume 用 24.0 dm³ mol⁻¹（room conditions；gas volume 这一格只对 gas 有意义）。<i>N</i> 可以写成 <code>2.4e21</code>、<code>2.4 x 10^21</code> 或直接写 2.4 × 10²¹。</p>
<style>
.k .mm-map{margin:8px 0;user-select:none}
.k .mm-bgr{fill:var(--card);stroke:var(--line)}
.k .mm-core{fill:var(--acc-soft);stroke:var(--acc);stroke-width:2}
.k .mm-arrows path{fill:none;stroke:var(--faint);stroke-width:2;stroke-dasharray:5 4}
.k .mm-arrows path.on{stroke:var(--acc);stroke-width:4;stroke-dasharray:none}
.k .mm-labels text{font-size:16px;fill:var(--mut)}
.k .mm-labels text.on{fill:var(--acc);font-weight:650}
.k .mm-box,.k .mm-in{height:100%;display:flex;flex-direction:column;justify-content:center;align-items:center;gap:3px;text-align:center;font-size:15px;line-height:1.25}
.k .mm-box{border:2px solid var(--line);border-radius:10px;background:var(--bg);padding:4px 8px}
.k .mm-box.src{border-color:var(--acc);background:var(--acc-soft)}
.k .mm-in.src .mm-v{border-color:var(--acc);border-width:2px}
.k .mm-t{font-weight:600}
.k .mm-s{font-size:14px;color:var(--mut)}
.k .mm-map input.mm-v{width:150px;text-align:center;font-size:16px;padding:2px 6px}
.k .mm-in input.mm-v{width:118px}
.k .mm-map input.mm-v.bad{border-color:var(--bad);background:var(--bad-soft)}
.k .mm-map select{font-size:14px;max-width:210px;padding:1px 4px}
.k .mm-two{display:flex;gap:10px}
.k .mm-two label{display:flex;flex-direction:column;align-items:center;font-size:14px;color:var(--mut);cursor:text}
.k .mm-two input.mm-v{width:120px;color:var(--fg)}
.k .mm-mr{margin:6px 0;font-size:15px}
.k .mm-mr .err,.k .mm-work .err{color:var(--bad)}
.k .mm-mr .big{font-size:19px}
.k .mm-pre button{padding:1px 8px;font-size:14px}
.k .mm-work ol{margin:4px 0}
.k .mm-work li{margin:3px 0}
.k .mm-work .h{font-weight:600}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var AR = {}, NOAR = 'Tc Po At Rn Fr Ra Ac Np Pu Am Cm Bk Cf Es Fm Md No Lr Rf Db Sg Bh Hs Mt Ds Rg Cn Nh Fl Mc Lv Ts Og Pm'.split(' ');
('H 1.0 He 4.0 Li 6.9 Be 9.0 B 10.8 C 12.0 N 14.0 O 16.0 F 19.0 Ne 20.2 Na 23.0 Mg 24.3 Al 27.0 Si 28.1 P 31.0 S 32.1 Cl 35.5 ' +
 'Ar 39.9 K 39.1 Ca 40.1 Sc 45.0 Ti 47.9 V 50.9 Cr 52.0 Mn 54.9 Fe 55.8 Co 58.9 Ni 58.7 Cu 63.5 Zn 65.4 Ga 69.7 Ge 72.6 As 74.9 ' +
 'Se 79.0 Br 79.9 Kr 83.8 Rb 85.5 Sr 87.6 Y 88.9 Zr 91.2 Nb 92.9 Mo 95.9 Ru 101.1 Rh 102.9 Pd 106.4 Ag 107.9 Cd 112.4 In 114.8 ' +
 'Sn 118.7 Sb 121.8 Te 127.6 I 126.9 Xe 131.3 Cs 132.9 Ba 137.3 Hf 178.5 Ta 180.9 W 183.8 Re 186.2 Os 190.2 Ir 192.2 Pt 195.1 ' +
 'Au 197.0 Hg 200.6 Tl 204.4 Pb 207.2 Bi 209.0 La 138.9 Ce 140.1 Pr 140.9 Nd 144.4 Sm 150.4 Eu 152.0 Gd 157.3 Tb 158.9 Dy 162.5 ' +
 'Ho 164.9 Er 167.3 Tm 168.9 Yb 173.1 Lu 175.0 Th 232.0 Pa 231.0 U 238.0').split(' ').forEach(function (v, i, a) {
  if (i % 2 === 0) AR[v] = Math.round(+a[i + 1] * 10);          /* tenths, so sums stay exact */
});
var L = 6.02e23, VM = 24.0;
function d1(t) { return (t / 10).toFixed(1); }                   /* tenths → '249.6' */
var SUBD = '₀₁₂₃₄₅₆₇₈₉';
function subs(s) { return String(s).replace(/\d/g, function (c) { return SUBD[+c]; }); }

/* ---------- formula ---------- */
function parseFormula(raw) {
  var s = String(raw).replace(/[₀-₉]/g, function (c) { return SUBD.indexOf(c); }).replace(/\s+/g, '');
  if (!s) return { err: '先输入一个 formula。' };
  var parts = s.split(/[.·•∙*]/), out = [], pretty = [];
  for (var p = 0; p < parts.length; p++) {
    var t = parts[p], coef = 1, m = t.match(/^\d+/);
    if (!t) return { err: '「·」两边都要有东西。' };
    if (m) {
      if (p === 0) return { err: 'formula 开头不要写系数（' + m[0] + '）。' };
      coef = +m[0]; t = t.slice(m[0].length);
      if (!coef) return { err: '系数不能是 0。' };
    }
    var stack = [{}], order = [], i = 0, pr = '';
    while (i < t.length) {
      var ch = t[i];
      if (ch === '(' || ch === '[') { stack.push({}); pr += ch; i++; continue; }
      if (ch === ')' || ch === ']') {
        if (stack.length < 2) return { err: '多了一个「' + ch + '」。' };
        var g = stack.pop(), n = t.slice(i + 1).match(/^\d+/), k = n ? +n[0] : 1;
        for (var e in g) stack[stack.length - 1][e] = (stack[stack.length - 1][e] || 0) + g[e] * k;
        pr += ch + (n ? subs(n[0]) : ''); i += 1 + (n ? n[0].length : 0); continue;
      }
      var sm = t.slice(i).match(/^([A-Z][a-z]?)(\d*)/);
      if (!sm) return { err: '不认识的符号「' + t.slice(i).match(/^[a-z]+|^./)[0] + '」——element symbol 要大写字母开头。' };
      var sym = sm[1];
      if (!(sym in AR)) {
        return { err: NOAR.indexOf(sym) >= 0 ? sym + ' 在 Periodic Table 上没有给 Aᵣ，算不了。' : '不认识的符号「' + sym + '」：Periodic Table 上没有这个 element。' };
      }
      var c = sm[2] ? +sm[2] : 1;
      if (!c) return { err: sym + ' 后面的数字不能是 0。' };
      stack[stack.length - 1][sym] = (stack[stack.length - 1][sym] || 0) + c;
      if (order.indexOf(sym) < 0) order.push(sym);
      pr += sym + subs(sm[2]); i += sm[0].length;
    }
    if (stack.length > 1) return { err: '括号没有配对：少了「)」。' };
    var cnt = stack[0], mr = 0;
    order.forEach(function (e) { mr += AR[e] * cnt[e]; });
    out.push({ coef: coef, cnt: cnt, order: order, mr: mr });
    pretty.push((coef > 1 ? coef : '') + pr);
  }
  var tot = {}, ord = [], mr = 0;
  out.forEach(function (q) {
    q.order.forEach(function (e) { tot[e] = (tot[e] || 0) + q.cnt[e] * q.coef; if (ord.indexOf(e) < 0) ord.push(e); });
    mr += q.mr * q.coef;
  });
  /* working: main part element by element, each water-of-crystallisation part as k × Mr(part) */
  var terms = out[0].order.map(function (e) { var k = out[0].cnt[e]; return (k > 1 ? k + ' × ' : '') + d1(AR[e]); });
  out.slice(1).forEach(function (q) { terms.push((q.coef > 1 ? q.coef + ' × ' : '') + d1(q.mr)); });
  return { parts: out, tot: tot, order: ord, mr: mr, pretty: pretty.join('·'), terms: terms };
}

/* ---------- numbers ---------- */
var SUPD = { '⁰': '0', '¹': '1', '²': '2', '³': '3', '⁴': '4', '⁵': '5', '⁶': '6', '⁷': '7', '⁸': '8', '⁹': '9', '⁻': '-', '⁺': '+' };
function parseNum(raw) {
  var s = String(raw).trim().replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹⁻⁺]+/g, function (m) { return '^' + m.split('').map(function (c) { return SUPD[c]; }).join(''); })
    .replace(/[−–]/g, '-').replace(/\s+/g, '').replace(/,/g, '');
  var m = s.match(/^(\d*\.?\d*)(?:[x×*X]10(?:\^|\*\*)?([-+]?\d+)|[eE]([-+]?\d+))?$/);
  if (!m || !/\d/.test(m[1])) return null;
  var man = m[1], ex = m[2] || m[3] || '0', v = parseFloat(man) * Math.pow(10, +ex);
  var dg = man.replace('.', '').replace(/^0+/, ''), sf = man.indexOf('.') >= 0 ? dg.length : dg.replace(/0+$/, '').length || 1;
  if (man.indexOf('.') < 0 && /0$/.test(man)) sf = Math.max(sf, Math.min(3, dg.length));   /* '250' → 3 s.f. */
  return { v: v, sf: Math.max(2, Math.min(sf, 5)) };
}
function fmt(x, sf) {
  if (!isFinite(x)) return '—';
  if (x === 0) return '0';
  var e = Math.floor(Math.log10(Math.abs(x)));
  var r = +x.toPrecision(sf); e = Math.floor(Math.log10(Math.abs(r)));
  if (e >= -3 && e < 5 && sf - 1 - e >= 0) return r.toFixed(sf - 1 - e);
  if (e >= -3 && e < 5) return String(Math.round(r));
  var man = (r / Math.pow(10, e)).toFixed(sf - 1);
  return man + ' × 10' + K.sup(e);
}

/* ---------- map geometry ---------- */
var CX = 404, CY = 250, R = 70, GAP = 9, svg = $root.querySelector('.mm-map'), ga = $root.querySelector('.mm-arrows'), gl = $root.querySelector('.mm-labels');
function cy(x, up) { var d = Math.sqrt(R * R - (x - CX) * (x - CX)); return up ? CY - d : CY + d; }
function cx(y, left) { var d = Math.sqrt(R * R - (y - CY) * (y - CY)); return left ? CX - d : CX + d; }
/* in = box → n, out = n → box; side says where its label goes */
var ARW = {
  gas:  { 'in': [[388, 124], [388, cy(388, 1) - GAP]], out: [[420, cy(420, 1) - 4], [420, 120 + GAP]], li: 'left', lo: 'right' },
  sol:  { 'in': [[388, 376], [388, cy(388, 0) + GAP]], out: [[420, cy(420, 0) + 4], [420, 380 - GAP]], li: 'left', lo: 'right' },
  mass: { 'in': [[248, 236], [cx(236, 1) - GAP, 236]], out: [[cx(264, 1) - 4, 264], [244 + GAP, 264]], li: 'above', lo: 'below' },
  part: { 'in': [[560, 236], [cx(236, 0) + GAP, 236]], out: [[cx(264, 0) + 4, 264], [564 - GAP, 264]], li: 'above', lo: 'below' }
};
var CANDS = {
  left:  [[-10, 5, 'end'], [-10, -8, 'end'], [-10, 18, 'end']],
  right: [[10, 5, 'start'], [10, -8, 'start'], [10, 18, 'start']],
  above: [[0, -9, 'middle'], [-14, -9, 'middle'], [14, -9, 'middle']],
  below: [[0, 21, 'middle'], [-14, 21, 'middle'], [14, 21, 'middle']]
};
var paths = {};
Object.keys(ARW).forEach(function (k) {
  ['in', 'out'].forEach(function (d) {
    var p = ARW[k][d], el = K.svg('path', { d: 'M' + p[0][0] + ',' + p[0][1] + ' L' + p[1][0] + ',' + p[1][1], 'data-arrow': k + '-' + d, 'marker-end': 'url(#mm-a-off)' });
    ga.appendChild(el); paths[k + '-' + d] = el;
  });
});
function boxRects() {   /* everything opaque on the map, for K.label */
  return [{ x: 284, y: 8, w: 240, h: 112 }, { x: 14, y: 194, w: 230, h: 112 }, { x: 564, y: 194, w: 230, h: 112 },
          { x: 254, y: 380, w: 300, h: 112 }, { x: CX - R, y: CY - R, w: 2 * R, h: 2 * R }];
}
function drawLabels(txt, on) {
  gl.innerHTML = '';
  var placed = boxRects(), avoid = [];
  Object.keys(ARW).forEach(function (k) { ['in', 'out'].forEach(function (d) { var p = ARW[k][d]; avoid.push([p[0][0], p[0][1], p[1][0], p[1][1]]); }); });
  Object.keys(ARW).forEach(function (k) {
    ['in', 'out'].forEach(function (d) {
      var p = ARW[k][d], mx = (p[0][0] + p[1][0]) / 2, my = (p[0][1] + p[1][1]) / 2;
      var t = K.label(gl, mx, my, txt[k + '-' + d], { cands: CANDS[ARW[k][d === 'in' ? 'li' : 'lo']], avoid: avoid, placed: placed, pad: 3 });
      t.setAttribute('data-for', k + '-' + d);
      /* quantity symbols italic, after placement (same glyph widths give the same box) */
      t.innerHTML = t.textContent.replace(/\b([MLVc])\b/g, '<tspan font-style="italic">$1</tspan>');
      if (on[k + '-' + d]) t.classList.add('on');
    });
  });
}

/* ---------- state ---------- */
var fIn = $root.querySelector('.mm-f'), kindSel = $root.querySelector('.mm-kind'), mrBox = $root.querySelector('.mm-mr'), work = $root.querySelector('.mm-work');
var inp = {}; K.$$($root, 'input.mm-v').forEach(function (e) { inp[e.getAttribute('data-k')] = e; });
var F = null, src = 'mass', kindWanted = 'el:C';
var PRE = [['HO2CCO2H', 'HO₂CCO₂H'], ['CuSO4.5H2O', 'CuSO₄·5H₂O'], ['Mg(OH)2', 'Mg(OH)₂'], ['CO2', 'CO₂'], ['NaCl', 'NaCl'], ['Ca3(PO4)2', 'Ca₃(PO₄)₂']];
var preBox = $root.querySelector('.mm-pre');
PRE.forEach(function (p) {
  var b = K.el('button', { type: 'button', 'data-f': p[0] }, p[1]);
  b.addEventListener('click', function () { fIn.value = p[0]; onFormula(); });
  preBox.appendChild(b);
});

function kindInfo() {
  var v = kindSel.value || 'mol';
  if (v === 'all') { var n = 0; for (var e in F.tot) n += F.tot[e]; return { k: n, name: 'atoms', short: 'atoms' }; }
  if (v.indexOf('el:') === 0) { var s = v.slice(3); return { k: F.tot[s], name: s + ' atoms', short: s, el: s }; }
  return { k: 1, name: 'molecules / formula units', short: '' };
}
function fillKinds() {
  var keep = kindSel.value || kindWanted;
  kindSel.innerHTML = '';
  kindSel.appendChild(K.el('option', { value: 'mol' }, 'molecules / formula units'));
  F.order.forEach(function (e) { kindSel.appendChild(K.el('option', { value: 'el:' + e }, 'atoms of ' + e + '（' + F.tot[e] + ' per formula）')); });
  var all = 0; for (var e in F.tot) all += F.tot[e];
  if (F.order.length > 1 || all > 1) kindSel.appendChild(K.el('option', { value: 'all' }, 'all atoms（' + all + ' per formula）'));
  kindSel.value = keep; if (kindSel.value !== keep) kindSel.value = 'mol';
}

function onFormula() {
  var r = parseFormula(fIn.value);
  if (r.err) {
    F = null; fIn.classList.add('bad');
    mrBox.innerHTML = '<span class="err">' + r.err + '</span>';
  } else {
    F = r; fIn.classList.remove('bad');
    fillKinds();
    mrBox.innerHTML = '<span class="big"><i>M</i><sub>r</sub>(' + F.pretty + ') = ' + F.terms.join(' + ') + ' = <b>' + d1(F.mr) + '</b></span>' +
      '　→　<i>M</i> = ' + d1(F.mr) + ' g mol⁻¹';
  }
  $root.querySelector('.mm-msub').innerHTML = F ? '<i>M</i> = ' + d1(F.mr) + ' g mol⁻¹' : '<span class="err">formula 有错</span>';
  K.$$($root, '.mm-pre button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-f') === fIn.value.trim()); });
  calc();
}

function calc() {
  K.$$($root, '[data-box]').forEach(function (b) { b.classList.toggle('src', b.getAttribute('data-box') === (src === 'c' ? 'sol' : src)); });
  Object.keys(inp).forEach(function (k) { inp[k].classList.remove('bad'); });
  var box = src === 'c' ? 'sol' : src, lines = [], n = null, sf = 3, err = null;
  var KI = F ? kindInfo() : { k: 1, name: 'molecules / formula units', short: '' };
  var M = F ? F.mr / 10 : null, Ms = F ? d1(F.mr) : '?';
  var volP = parseNum(inp.vol.value);
  var vol = volP && volP.v > 0 ? volP.v : null;

  function need(k) { var p = parseNum(inp[k].value); if (!p || !(p.v > 0)) { inp[k].classList.add('bad'); err = '「' + (inp[k].value || '空') + '」看不懂——要一个正数。'; return null; } return p; }
  if (box === 'mass') {
    var p = need('mass');
    if (p && !F) err = 'formula 有错，算不出 <i>M</i>。';
    if (p && F) { n = p.v / M; sf = p.sf; lines.push('<i>n</i> = <i>m</i> ÷ <i>M</i> = ' + inp.mass.value.trim() + ' ÷ ' + Ms + ' = <b>' + fmt(n, sf) + ' mol</b>'); }
  } else if (box === 'gas') {
    p = need('gas');
    if (p) { n = p.v / VM; sf = p.sf; lines.push('<i>n</i> = <i>V</i> ÷ 24.0 = ' + inp.gas.value.trim() + ' ÷ 24.0 = <b>' + fmt(n, sf) + ' mol</b>'); }
  } else if (box === 'part') {
    p = need('part');
    if (p) {
      sf = p.sf; var np = p.v / L;
      if (KI.k > 1 || KI.el) {
        lines.push('<i>n</i>(' + KI.short + ') = <i>N</i> ÷ <i>L</i> = ' + fmt(p.v, sf) + ' ÷ 6.02 × 10²³ = ' + fmt(np, sf) + ' mol');
        n = np / KI.k;
        lines.push('每个 ' + F.pretty + ' 有 ' + KI.k + ' 个 ' + (KI.el ? KI.el + ' atom' : 'atom') + '：<i>n</i>(' + F.pretty + ') = ' + fmt(np, sf) + ' ÷ ' + KI.k + ' = <b>' + fmt(n, sf) + ' mol</b>');
      } else {
        n = np; lines.push('<i>n</i> = <i>N</i> ÷ <i>L</i> = ' + fmt(p.v, sf) + ' ÷ 6.02 × 10²³ = <b>' + fmt(n, sf) + ' mol</b>');
      }
    }
  } else if (box === 'sol') {
    var pc = need('c'), pv = need('vol');
    if (pc && pv) {
      n = pc.v * pv.v / 1000; sf = Math.min(pc.sf, pv.sf);
      lines.push('<i>n</i> = <i>c</i> × <i>V</i> ÷ 1000 = ' + inp.c.value.trim() + ' × ' + inp.vol.value.trim() + ' ÷ 1000 = <b>' + fmt(n, sf) + ' mol</b>');
    }
  } else {
    p = need('n');
    if (p) { n = p.v; sf = p.sf; lines.push('<i>n</i> = ' + inp.n.value.trim() + ' mol（直接给出）'); }
  }

  var on = {};
  if (box !== 'n' && n != null) on[box + '-in'] = 1;
  if (n != null) {
    var out = {};
    out.n = fmt(n, sf);
    if (F) { out.mass = fmt(n * M, sf); }
    out.gas = fmt(n * VM, sf);
    var Npart = n * KI.k * L;
    out.part = fmt(Npart, sf);
    if (vol) out.c = fmt(n / (vol / 1000), sf);
    var tl = [];
    if (box !== 'mass') tl.push(F ? '<i>m</i> = <i>n</i> × <i>M</i> = ' + out.n + ' × ' + Ms + ' = <b>' + out.mass + ' g</b>' : '<i>m</i>：formula 有错，算不出 <i>M</i>');
    if (box !== 'part') {
      if (KI.k > 1 || KI.el) {
        var nx = n * KI.k;
        tl.push('<i>n</i>(' + KI.short + ') = ' + KI.k + ' × ' + out.n + ' = ' + fmt(nx, sf) + ' mol；<i>N</i>(' + KI.short + ') = ' + fmt(nx, sf) + ' × 6.02 × 10²³ = <b>' + out.part + '</b>');
      } else tl.push('<i>N</i> = <i>n</i> × <i>L</i> = ' + out.n + ' × 6.02 × 10²³ = <b>' + out.part + '</b>');
    }
    if (box !== 'gas') tl.push('<i>V</i> = <i>n</i> × 24.0 = ' + out.n + ' × 24.0 = <b>' + out.gas + ' dm³</b>');
    if (box !== 'sol') tl.push(vol ? '<i>c</i> = <i>n</i> ÷ (<i>V</i> ÷ 1000) = ' + out.n + ' ÷ (' + inp.vol.value.trim() + ' ÷ 1000) = <b>' + out.c + ' mol dm⁻³</b>'
                                    : '<i>c</i>：先在 solution 盒子里填 <i>V</i> / cm³');
    ['mass', 'gas', 'part', 'sol'].forEach(function (b) { if (b !== box) on[b + '-out'] = 1; });
    if (box !== 'n') inp.n.value = out.n;
    if (box !== 'mass') inp.mass.value = F ? out.mass : '';
    if (box !== 'gas') inp.gas.value = out.gas;
    if (box !== 'part') inp.part.value = out.part;
    if (box !== 'sol') inp.c.value = vol ? out.c : '';
    work.innerHTML = '<div class="h">① 先变成 <i>n</i>（' + { mass: 'mass', gas: 'gas volume', part: 'number of particles', sol: 'solution', n: 'amount' }[box] + ' → amount）</div><ol>' +
      lines.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>' +
      '<div class="h">② 再从 <i>n</i> 出发</div><ol>' + tl.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>' +
      '<p class="faint">结果取 ' + sf + ' s.f.（跟你填的数据一样）；中间步骤不要提前四舍五入。</p>';
  } else {
    Object.keys(inp).forEach(function (k) { if (k !== 'vol' && k !== (box === 'sol' ? 'c' : box)) inp[k].value = ''; });
    work.innerHTML = '<p class="err">' + (err || '填一个数') + '</p>';
  }
  Object.keys(paths).forEach(function (k) {
    paths[k].classList.toggle('on', !!on[k]);
    paths[k].setAttribute('marker-end', on[k] ? 'url(#mm-a-on)' : 'url(#mm-a-off)');
  });
  var kk = KI.k > 1 || KI.el ? KI.k + ' × ' : '';
  drawLabels({
    'mass-in': '÷ M', 'mass-out': '× M',
    'gas-in': '÷ 24.0', 'gas-out': '× 24.0',
    'part-in': kk ? '÷ (' + kk + 'L)' : '÷ L', 'part-out': '× ' + kk + 'L',
    'sol-in': 'c × V ÷ 1000', 'sol-out': '÷ (V ÷ 1000)'
  }, on);
  K.fit();
}

Object.keys(inp).forEach(function (k) {
  inp[k].addEventListener('input', function () {
    if (k !== 'vol') src = k === 'c' ? 'sol' : k;
    calc();
  });
});
kindSel.addEventListener('change', calc);
fIn.addEventListener('input', onFormula);
inp.mass.value = '0.18';
F = parseFormula(fIn.value); fillKinds(); kindSel.value = kindWanted;
onFormula();

K.init($root);
})(document.getElementById("k-ch02-05"));</script>"""))
```

## 2.3 Formulas

### A. Ionic compounds 的 formula

规则只有一条：**positive charges 的总数 = negative charges 的总数**（整体 neutral）。

ion 的 charge 可以从 Periodic Table 上的位置预测：

| 位置 | 怎样达到 noble gas configuration | ion charge | 例 |
|---|---|---|---|
| Group 1 | 失 1 个 electron | +1 | Na<sup>+</sup>、K<sup>+</sup> |
| Group 2 | 失 2 个 | +2 | Mg<sup>2+</sup>、Ca<sup>2+</sup> |
| Group 13 | 失 3 个 | +3 | Al<sup>3+</sup> |
| Group 15 | 得 3 个 | −3 | N<sup>3−</sup>（nitride） |
| Group 16 | 得 2 个 | −2 | O<sup>2−</sup>（oxide）、S<sup>2−</sup>（sulfide） |
| Group 17 | 得 1 个 | −1 | Cl<sup>−</sup>（chloride）、Br<sup>−</sup>、I<sup>−</sup> |
| transition element | charge 不固定 | 用 Roman numeral 写 oxidation number | iron(II) → Fe<sup>2+</sup>，iron(III) → Fe<sup>3+</sup> |

Group 1 / 2 / 13 的 compound 不写 Roman numeral：magnesium oxide，不是 magnesium(II) oxide。

**必须背的 ion**（syllabus 点名）：

| ion | name | ion | name |
|---|---|---|---|
| NO₃<sup>−</sup> | nitrate | NH₄<sup>+</sup> | ammonium |
| CO₃<sup>2−</sup> | carbonate | Zn<sup>2+</sup> | zinc |
| SO₄<sup>2−</sup> | sulfate | Ag<sup>+</sup> | silver |
| OH<sup>−</sup> | hydroxide | HCO₃<sup>−</sup> | hydrogencarbonate |
| PO₄<sup>3−</sup> | phosphate | | |

写 formula 时，compound ion 要乘 2 以上就加括号：Zn<sup>2+</sup> + 2 NO₃<sup>−</sup> → Zn(NO₃)₂；(NH₄)₂SO₄；Ca₃(PO₄)₂。

> **LO 2.3.1** — write formulas of ionic compounds from ionic charges and oxidation numbers (shown by a Roman numeral), including:

```python
# @title 🔬 ionic formula 拼装器 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-07">
<h3>Ionic formula 拼装器</h3>
<div class="tabbar">
  <button data-tab="build">拼 formula</button>
  <button data-tab="pt">从 Periodic Table 预测 charge</button>
  <button data-tab="quiz">必背 ion 测验</button>
</div>

<div data-pane="build">
  <p class="mut">选一个 cation 和一个 anion。上排每个 <b class="if-p">+</b> 方块是一个正 charge，下排每个 <b class="if-n">−</b> 方块是一个负 charge；
    方块<b>上下对齐、数目相等</b>，compound 才是 neutral。</p>
  <div class="if-pick">
    <div class="if-side"><div class="if-head">cation（positive ion）</div><div class="if-cats"></div></div>
    <div class="if-side"><div class="if-head">anion（negative ion）</div><div class="if-ans"></div></div>
  </div>
  <div class="card if-stage">
    <div class="if-board"><div class="if-row if-top"></div><div class="if-row if-bot"></div></div>
    <div class="row if-steps">
      <span>cation 个数</span><button class="if-cm" aria-label="fewer cations">−</button><b class="if-nc"></b><button class="if-cp" aria-label="more cations">+</button>
      <span class="if-gap">anion 个数</span><button class="if-am" aria-label="fewer anions">−</button><b class="if-na"></b><button class="if-ap" aria-label="more anions">+</button>
      <button class="if-auto">自动配平</button><button class="if-one">从 1 : 1 自己配</button>
    </div>
    <div class="if-tally"></div>
  </div>
  <div class="card if-out"></div>
</div>

<div data-pane="pt">
  <p class="mut">点一个元素：它在哪个 Group，就决定失去 / 得到几个 electron 才达到 noble gas configuration。</p>
  <div class="if-pt"></div>
  <div class="row faint"><span class="if-key if-key-c"></span> 形成 cation　<span class="if-key if-key-a"></span> 形成 anion　<span class="if-key if-key-x"></span> 一般不形成简单 ion</div>
  <div class="card if-ptout"></div>
</div>

<div data-pane="quiz">
  <p class="mut">syllabus 点名必须背的 9 个 ion。随机出题：给 name 选 formula，或给 formula 选 name。</p>
  <div class="card if-quiz"></div>
  <div class="row"><button class="if-next">下一题 →</button><span class="if-score faint"></span></div>
</div>

<style>
.k .if-p{color:var(--pos)} .k .if-n{color:var(--neg)}
.k .if-pick{display:grid;grid-template-columns:repeat(auto-fit,minmax(290px,1fr));gap:10px;margin:8px 0}
.k .if-head{font-size:13px;color:var(--mut);margin-bottom:4px;font-weight:600}
.k .if-sub{font-size:12px;color:var(--faint);margin:4px 0 2px}
.k .if-chips{display:flex;flex-wrap:wrap;gap:5px}
.k .if-chips button{min-width:3.4em;padding:3px 8px;font-size:15px}
.k .if-cats button.on{background:var(--pos);border-color:var(--pos)}
.k .if-ans button.on{background:var(--neg);border-color:var(--neg)}
.k .if-stage{margin:8px 0;overflow-x:auto}
.k .if-board{display:inline-block;padding:4px 2px 2px;min-width:100%}
.k .if-row{display:flex;align-items:flex-end}
.k .if-bot{align-items:flex-start}
.k .if-grp{display:flex;flex-direction:column;align-items:center}
.k .if-lab{font-size:14px;line-height:20px;height:20px;white-space:nowrap;color:var(--fg)}
.k .if-sqs{display:flex;border-radius:5px;outline:2px solid;outline-offset:-2px}
.k .if-top .if-sqs{outline-color:var(--pos)} .k .if-bot .if-sqs{outline-color:var(--neg)}
.k .if-sq{width:52px;height:34px;display:flex;align-items:center;justify-content:center;font-size:20px;font-weight:700}
.k .if-sq+.if-sq{border-left:1px dashed var(--line)}
.k .if-top .if-sq{color:var(--pos);background:color-mix(in srgb,var(--pos) 16%,var(--card))}
.k .if-bot .if-sq{color:var(--neg);background:color-mix(in srgb,var(--neg) 16%,var(--card))}
.k .if-sq.if-un{background:repeating-linear-gradient(45deg,var(--bad-soft) 0 5px,var(--card) 5px 10px)}
.k .if-gap{margin-left:14px}
.k .if-steps b{min-width:1.2em;text-align:center;font-size:17px}
.k .if-steps button{padding:2px 9px}
.k .if-tally{font-size:14px;margin-top:4px}
.k .if-tally .if-ok{color:var(--ok);font-weight:650} .k .if-tally .if-no{color:var(--bad);font-weight:650}
.k .if-big{font-size:30px;line-height:1.25;margin:0 0 2px}
.k .if-big .if-name{font-size:19px;color:var(--mut);margin-left:14px}
.k .if-out ol{margin:6px 0 4px}
.k .if-out li{margin:2px 0}
.k .if-warn{color:var(--bad)}
.k .if-note{border-left:3px solid var(--line);padding:2px 9px;margin-top:6px;color:var(--mut);font-size:14px}
/* mini periodic table */
.k .if-pt{display:grid;grid-template-columns:32px repeat(8,minmax(46px,60px));gap:4px;margin:8px 0;max-width:540px}
.k .if-pt .if-gh,.k .if-pt .if-ph{font-size:12px;color:var(--faint);text-align:center;align-self:end}
.k .if-pt .if-ph{align-self:center}
.k .if-pt button{padding:2px 0 3px;display:flex;flex-direction:column;align-items:center;line-height:1.15;border-width:2px}
.k .if-pt button small{font-size:12px;color:var(--faint)}
.k .if-pt button b{font-size:17px}
.k .if-pt .c{border-color:color-mix(in srgb,var(--pos) 55%,var(--line))}
.k .if-pt .a{border-color:color-mix(in srgb,var(--neg) 55%,var(--line))}
.k .if-pt .x{border-style:dashed;opacity:.75}
.k .if-pt button.on{background:var(--acc-soft);color:var(--fg);border-color:var(--acc)}
.k .if-pt button.on small{color:var(--mut)}
.k .if-key{display:inline-block;width:14px;height:14px;border:2px solid;border-radius:3px;vertical-align:-2px}
.k .if-key-c{border-color:var(--pos)} .k .if-key-a{border-color:var(--neg)} .k .if-key-x{border-color:var(--faint);border-style:dashed}
/* quiz */
.k .if-qq{font-size:17px;margin:2px 0 8px}
.k .if-qq b{font-size:21px}
.k .if-opts{display:grid;grid-template-columns:repeat(auto-fit,minmax(150px,1fr));gap:8px}
.k .if-opts button{font-size:18px;padding:6px 10px}
.k .if-opts button:disabled{cursor:default;opacity:1;color:var(--fg)}
.k .if-opts button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .if-opts button.badv{background:var(--bad-soft);border-color:var(--bad)}
.k .if-why{margin-top:8px;min-height:1.5em}
.k .if-why .if-ok{color:var(--ok);font-weight:650} .k .if-why .if-no{color:var(--bad);font-weight:650}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var $ = function (s) { return $root.querySelector(s); };
var SUB = '₀₁₂₃₄₅₆₇₈₉';
function sub(n) { return String(n).split('').map(function (d) { return SUB[+d]; }).join(''); }
function chg(q) { var a = Math.abs(q); return (a === 1 ? '' : K.sup(a)) + (q > 0 ? '⁺' : '⁻'); }
function gcd(a, b) { return b ? gcd(b, a % b) : a; }

/* [key, formula part, compound ion?, charge, name, kind, Group] */
var CAT = [
  ['Li', 'Li', 0, 1, 'lithium', 'g', 1], ['Na', 'Na', 0, 1, 'sodium', 'g', 1], ['K', 'K', 0, 1, 'potassium', 'g', 1],
  ['Mg', 'Mg', 0, 2, 'magnesium', 'g', 2], ['Ca', 'Ca', 0, 2, 'calcium', 'g', 2], ['Ba', 'Ba', 0, 2, 'barium', 'g', 2],
  ['Al', 'Al', 0, 3, 'aluminium', 'g', 13],
  ['Zn', 'Zn', 0, 2, 'zinc', 'd'], ['Ag', 'Ag', 0, 1, 'silver', 'd'],
  ['Cu2', 'Cu', 0, 2, 'copper(II)', 't'], ['Fe2', 'Fe', 0, 2, 'iron(II)', 't'], ['Fe3', 'Fe', 0, 3, 'iron(III)', 't'], ['Cr3', 'Cr', 0, 3, 'chromium(III)', 't'],
  ['NH4', 'NH₄', 1, 1, 'ammonium', 'x']
].map(function (r) { return { key: r[0], f: r[1], cx: !!r[2], q: r[3], name: r[4], kind: r[5], grp: r[6] }; });
var AN = [
  ['Cl', 'Cl', 0, 1, 'chloride'], ['Br', 'Br', 0, 1, 'bromide'], ['I', 'I', 0, 1, 'iodide'],
  ['O', 'O', 0, 2, 'oxide'], ['S', 'S', 0, 2, 'sulfide'], ['N', 'N', 0, 3, 'nitride'],
  ['OH', 'OH', 1, 1, 'hydroxide'], ['NO3', 'NO₃', 1, 1, 'nitrate'], ['HCO3', 'HCO₃', 1, 1, 'hydrogencarbonate'],
  ['CO3', 'CO₃', 1, 2, 'carbonate'], ['SO4', 'SO₄', 1, 2, 'sulfate'], ['PO4', 'PO₄', 1, 3, 'phosphate']
].map(function (r) { return { key: r[0], f: r[1], cx: !!r[2], q: -r[3], name: r[4] }; });
function ion(x) { return x.f + chg(x.q); }
function part(x, n) { return n === 1 ? x.f : (x.cx ? '(' + x.f + ')' : x.f) + sub(n); }
/* lowest whole-number ratio of cation : anion */
function ratio(c, a) { var p = c.q, m = -a.q, g = gcd(p, m); return [m / g, p / g]; }
function formula(c, a) { var r = ratio(c, a); return part(c, r[0]) + part(a, r[1]); }
function cname(c, a) { return c.name + ' ' + a.name; }
$root.__ionf = { CAT: CAT, AN: AN, formula: formula, name: cname };   /* exposed for the test */

/* ---------------- tab 1: build ---------------- */
var cat = CAT[6], an = AN[10], nc = 2, na = 3;
function chips(box, list, groups, onpick) {
  groups.forEach(function (g) {
    box.appendChild(K.el('div', { 'class': 'if-sub' }, g[0]));
    var row = K.el('div', { 'class': 'if-chips' });
    list.filter(g[1]).forEach(function (x) {
      var b = K.el('button', { 'data-ion': x.key, title: x.name }, ion(x));
      b.addEventListener('click', function () { onpick(x); });
      row.appendChild(b);
    });
    box.appendChild(row);
  });
}
chips($('.if-cats'), CAT, [['Group 1 / 2 / 13 metal', function (x) { return x.kind === 'g'; }],
  ['d-block metal', function (x) { return x.kind === 'd' || x.kind === 't'; }],
  ['compound ion', function (x) { return x.kind === 'x'; }]], function (x) { cat = x; auto(); });
chips($('.if-ans'), AN, [['simple ion（单原子）', function (x) { return !x.cx; }],
  ['compound ion', function (x) { return x.cx; }]], function (x) { an = x; auto(); });

function auto() { var r = ratio(cat, an); nc = r[0]; na = r[1]; draw(); }
function row(box, x, n, other, top) {
  box.innerHTML = '';
  var per = Math.abs(x.q), mine = n * per, k = 0;
  for (var i = 0; i < n; i++) {
    var g = K.el('div', { 'class': 'if-grp' }), sq = K.el('div', { 'class': 'if-sqs' });
    for (var j = 0; j < per; j++, k++)
      sq.appendChild(K.el('div', { 'class': 'if-sq' + (k >= other ? ' if-un' : '') }, top ? '+' : '−'));
    var lab = K.el('div', { 'class': 'if-lab' }, ion(x));
    if (top) { g.appendChild(lab); g.appendChild(sq); } else { g.appendChild(sq); g.appendChild(lab); }
    box.appendChild(g);
  }
  return mine;
}
function draw() {
  K.$$($root, '.if-cats button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-ion') === cat.key); });
  K.$$($root, '.if-ans button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-ion') === an.key); });
  var pos = nc * cat.q, neg = na * -an.q;
  row($('.if-top'), cat, nc, neg, true);
  row($('.if-bot'), an, na, pos, false);
  $('.if-nc').textContent = nc; $('.if-na').textContent = na;
  $('.if-cm').disabled = nc <= 1; $('.if-cp').disabled = nc >= 4;
  $('.if-am').disabled = na <= 1; $('.if-ap').disabled = na >= 4;
  var bal = pos === neg;
  $('.if-tally').innerHTML = '<span class="if-p">+ charges：' + nc + ' × (+' + cat.q + ') = +' + pos + '</span>　' +
    '<span class="if-n">− charges：' + na + ' × (−' + (-an.q) + ') = −' + neg + '</span>　' +
    (bal ? '<span class="if-ok">→ total charge 0 ✓</span>'
         : '<span class="if-no">→ total charge ' + (pos > neg ? '+' : '−') + Math.abs(pos - neg) + '，不平衡</span>');
  $('.if-tally').setAttribute('data-pos', pos); $('.if-tally').setAttribute('data-neg', neg);
  var out = $('.if-out'), r = ratio(cat, an), f = formula(cat, an), nm = cname(cat, an);
  if (!bal) {
    out.innerHTML = '<p class="if-warn">还没平衡：+ 方块 ' + pos + ' 个、− 方块 ' + neg + ' 个（斜线方块没有对上）。' +
      (pos < neg ? (nc < 4 ? '加 cation' : '减 anion') : (na < 4 ? '加 anion' : '减 cation')) + '，或按「自动配平」。</p>';
    out.setAttribute('data-state', 'unbalanced'); K.fit(); return;
  }
  if (nc !== r[0]) {
    var g = nc / r[0];
    out.innerHTML = '<p>charge 平衡了，但 ' + nc + ' : ' + na + ' 不是最简比——ionic compound 的 formula 写<b>最简整数比</b>：÷ ' + g +
      ' → ' + r[0] + ' : ' + r[1] + '，即 <b>' + f + '</b>（不是 ' + part(cat, nc) + part(an, na) + '）。</p>';
    out.setAttribute('data-state', 'reducible'); K.fit(); return;
  }
  out.setAttribute('data-state', 'ok');
  var p = cat.q, m = -an.q, L = p * m / gcd(p, m), steps = [];
  steps.push('charges：' + ion(cat) + ' 是 +' + p + '，' + ion(an) + ' 是 −' + m + '。');
  steps.push(p === m ? '两者 charge 大小相等 → 1 : 1。'
    : 'lowest common multiple of ' + p + ' 和 ' + m + ' = ' + L + ' → 要 ' + L + ' ÷ ' + p + ' = <b>' + r[0] + '</b> 个 ' + ion(cat) +
      '、' + L + ' ÷ ' + m + ' = <b>' + r[1] + '</b> 个 ' + ion(an) + '。');
  var br = [cat, an].filter(function (x, i) { return x.cx && r[i] > 1; });
  if (br.length) steps.push(br.map(function (x) { return x.f; }).join('、') + ' 是 compound ion，要不止一个 → 整个 ion 加括号再写 subscript：' +
    br.map(function (x) { return part(x, r[x === cat ? 0 : 1]); }).join('、') + '。');
  else if ([cat, an].some(function (x) { return x.cx; })) steps.push('compound ion 只要 1 个 → 不用加括号。');
  steps.push('formula 里不写 charge：<b>' + f + '</b>。');
  var why = cat.kind === 'g' ? cat.f + ' 在 Group ' + cat.grp + '，只有一种 charge → name 不写 Roman numeral（' + cat.name + '，不是 ' + cat.name + '(' + ['', 'I', 'II', 'III'][cat.q] + ')）。'
    : cat.kind === 'd' ? cat.f + ' 是 d-block，但只形成 ' + ion(cat) + ' 一种 ion → syllabus 直接叫 ' + cat.name + '，不写 Roman numeral。'
    : cat.kind === 't' ? cat.f + ' 是 transition element，有不止一种 oxidation number' + (cat.f === 'Fe' ? '（Fe²⁺ / Fe³⁺）' : cat.f === 'Cu' ? '（Cu⁺ / Cu²⁺）' : '（+3、+6 …）') +
      ' → 用 Roman numeral 写出这里的 +' + p + '：' + cat.name + '。'
    : 'NH₄⁺ 是 compound ion（ammonium），不是 metal ion。';
  var odd = (cat.q === 3 && /CO3$/.test(an.key)) || (cat.key === 'NH4' && (an.key === 'O' || an.key === 'N')) ? '<div class="if-note">⚠ 这里只练按 charge 拼 formula；' + f + ' 这个 compound 实际上不稳定 / 不存在。</div>' : '';
  out.innerHTML = '<p class="if-big"><span class="if-f">' + f + '</span><span class="if-name">' + nm + '</span></p>' +
    '<ol>' + steps.map(function (s) { return '<li>' + s + '</li>'; }).join('') + '</ol>' +
    '<p>name：先 cation 后 anion → <b>' + nm + '</b>。' + why + '</p>' +
    '<div class="if-note">criss-cross 速记：把对方 charge 的数字交叉写成自己的 subscript，再约到最简（Mg²⁺ + O²⁻ → Mg₂O₂ → MgO）。</div>' + odd;
  K.fit();
}
$('.if-cm').addEventListener('click', function () { if (nc > 1) { nc--; draw(); } });
$('.if-cp').addEventListener('click', function () { if (nc < 4) { nc++; draw(); } });
$('.if-am').addEventListener('click', function () { if (na > 1) { na--; draw(); } });
$('.if-ap').addEventListener('click', function () { if (na < 4) { na++; draw(); } });
$('.if-auto').addEventListener('click', auto);
$('.if-one').addEventListener('click', function () { nc = 1; na = 1; draw(); });
draw();

/* ---------------- tab 2: predict from the Periodic Table ---------------- */
var GROUPS = [1, 2, 13, 14, 15, 16, 17, 18];
var PT = {   /* period → symbols in GROUPS order */
  2: ['Li', 'Be', 'B', 'C', 'N', 'O', 'F', 'Ne'],
  3: ['Na', 'Mg', 'Al', 'Si', 'P', 'S', 'Cl', 'Ar'],
  4: ['K', 'Ca', 'Ga', 'Ge', 'As', 'Se', 'Br', 'Kr']
};
var ENAME = { Li: 'lithium', Be: 'beryllium', B: 'boron', C: 'carbon', N: 'nitrogen', O: 'oxygen', F: 'fluorine', Ne: 'neon',
  Na: 'sodium', Mg: 'magnesium', Al: 'aluminium', Si: 'silicon', P: 'phosphorus', S: 'sulfur', Cl: 'chlorine', Ar: 'argon',
  K: 'potassium', Ca: 'calcium', Ga: 'gallium', Ge: 'germanium', As: 'arsenic', Se: 'selenium', Br: 'bromine', Kr: 'krypton' };
var IDE = { N: 'nitride', P: 'phosphide', As: 'arsenide', O: 'oxide', S: 'sulfide', Se: 'selenide', F: 'fluoride', Cl: 'chloride', Br: 'bromide' };
var QG = { 1: 1, 2: 2, 13: 3, 15: -3, 16: -2, 17: -1 };
var NOBLE = { 2: 'He', 10: 'Ne', 18: 'Ar', 36: 'Kr' };
var pt = $('.if-pt');
pt.appendChild(K.el('div'));
GROUPS.forEach(function (g) { pt.appendChild(K.el('div', { 'class': 'if-gh' }, 'Group ' + g)); });
[2, 3, 4].forEach(function (per) {
  pt.appendChild(K.el('div', { 'class': 'if-ph' }, 'P' + per));
  PT[per].forEach(function (s, i) {
    var g = GROUPS[i], q = QG[g], Z = K.chem.SYM.indexOf(s) + 1;
    var kind = s === 'B' || !q ? 'x' : q > 0 ? 'c' : 'a';
    var b = K.el('button', { 'class': kind, 'data-el': s }, '<small>' + Z + '</small><b>' + s + '</b>');
    b.addEventListener('click', function () { pick(s, g, per); });
    pt.appendChild(b);
  });
});
function cfg(Z, q) { return K.chem.config(Z, q).full; }
function pick(s, g, per) {
  K.$$(pt, 'button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-el') === s); });
  var Z = K.chem.SYM.indexOf(s) + 1, q = QG[g], out = $('.if-ptout'), nm = ENAME[s], h;
  out.setAttribute('data-el', s);
  if (g === 18) {
    h = '<p class="if-big">' + s + '<span class="if-name">' + nm + '</span></p><p>Group 18 noble gas：' + s + ' ' + cfg(Z, 0) +
      ' 已经是 noble gas configuration → 不形成 ion。</p>';
  } else if (g === 14) {
    h = '<p class="if-big">' + s + '<span class="if-name">' + nm + '</span></p><p>Group 14：要失去或得到 <b>4</b> 个 electron 才达到 noble gas configuration，能量代价太大 → ' +
      nm + ' 一般<b>不形成简单 ion</b>，而是形成 covalent bonds（例如 ' + { C: 'CH₄、CO₂', Si: 'SiO₂、SiCl₄', Ge: 'GeCl₄' }[s] + '）。</p>';
  } else if (s === 'B') {
    h = '<p class="if-big">B<span class="if-name">boron</span></p><p>按 Group 13 会预测 B³⁺，但 boron atom 很小，失去 3 个 electron 的能量太高 → boron <b>不形成 B³⁺</b>，' +
      '它的 compounds 是 covalent（例如 BCl₃）。同 Group 往下的 Al 才形成 Al³⁺。</p>';
  } else {
    var ionf = s + chg(q), ionZ = Z - q, ng = NOBLE[ionZ], c1 = cfg(Z, 0), c2 = cfg(Z, q);
    var same = ng && cfg(K.chem.SYM.indexOf(ng) + 1, 0) === c2;
    var X = q > 0 ? { f: s, cx: false, q: q, name: nm } : { f: s, cx: false, q: q, name: IDE[s] };
    var ex = q > 0 ? [[X, AN[0]], [X, AN[3]]] : [[CAT[1], X], [CAT[3], X]];
    h = '<p class="if-big"><span class="if-f">' + ionf + '</span><span class="if-name">' + (q > 0 ? nm + ' ion' : IDE[s] + ' ion') + '</span></p>' +
      '<p>Group ' + g + ' → ' + (q > 0 ? '<b>失去 ' + q + '</b> 个 electron → cation，charge +' + q : '<b>得到 ' + (-q) + '</b> 个 electron → anion，charge −' + (-q)) + '。</p>' +
      '<p>' + s + '　' + c1 + '　→　' + ionf + '　' + c2 + (same ? '（和 ' + ng + ' 一样）' : '（3d¹⁰ 全满，同样稳定）') + '</p>' +
      (q > 0 ? '<p>name：cation 就用 element 名 → <b>' + nm + '</b>（Group ' + g + ' 只有一种 charge，不写 Roman numeral）。</p>'
             : '<p>name：anion 把 element 名改成 <b>-ide</b> 结尾：' + nm + ' → <b>' + IDE[s] + '</b>。</p>') +
      '<p>例：' + ex.map(function (e) { return formula(e[0], e[1]) + ' ' + cname(e[0], e[1]); }).join('；') + '</p>' +
      (s === 'Be' ? '<div class="if-note">Be²⁺ 很小、charge density 很高，beryllium compounds 有明显的 covalent character。</div>' : '');
  }
  out.innerHTML = h; K.fit();
}
pick('Mg', 2, 3);

/* ---------------- tab 3: the 9 ions to learn ---------------- */
var MUST = [['NO₃⁻', 'nitrate', ['NO₃²⁻', 'NO₂⁻']], ['CO₃²⁻', 'carbonate', ['CO₃⁻', 'CO₂²⁻']], ['SO₄²⁻', 'sulfate', ['SO₄⁻', 'SO₃²⁻']],
  ['OH⁻', 'hydroxide', ['OH²⁻', 'OH⁺']], ['NH₄⁺', 'ammonium', ['NH₃⁺', 'NH₄⁻']], ['Zn²⁺', 'zinc', ['Zn⁺', 'Zn³⁺']],
  ['Ag⁺', 'silver', ['Ag²⁺', 'Ag⁻']], ['HCO₃⁻', 'hydrogencarbonate', ['HCO₃²⁻', 'H₂CO₃⁻']], ['PO₄³⁻', 'phosphate', ['PO₄²⁻', 'PO₃³⁻']]];
var FAKE_NAMES = { 'NO₃⁻': 'nitrite', 'SO₄²⁻': 'sulfite' };
function shuffle(a) { for (var i = a.length - 1; i > 0; i--) { var j = Math.floor(Math.random() * (i + 1)), t = a[i]; a[i] = a[j]; a[j] = t; } return a; }
var deck = [], right = 0, done = 0;
function nextQ() {
  if (!deck.length) deck = shuffle(MUST.slice());
  var it = deck.pop(), dir = Math.random() < 0.5 ? 'f' : 'n', opts, ans, qtxt;
  var others = shuffle(MUST.filter(function (m) { return m !== it; }));
  if (dir === 'f') {   /* name → formula */
    qtxt = '<b>' + it[1] + '</b> 的 formula 是？'; ans = it[0];
    opts = [it[0], it[2][0], it[2][1], others[0][0]];
  } else {             /* formula → name */
    qtxt = '<b>' + it[0] + '</b> 叫什么？'; ans = it[1];
    opts = [it[1]].concat(FAKE_NAMES[it[0]] ? [FAKE_NAMES[it[0]]] : []).concat(others.map(function (m) { return m[1]; })).slice(0, 4);
  }
  opts = shuffle(opts);
  var box = $('.if-quiz');
  box.innerHTML = '<p class="if-qq">' + qtxt + '</p><div class="if-opts"></div><div class="if-why"></div>';
  box.setAttribute('data-ion', it[0]); box.setAttribute('data-dir', dir); box.setAttribute('data-ans', ans);
  var wrap = K.$(box, '.if-opts');
  opts.forEach(function (o) {
    var b = K.el('button', { 'data-v': o }, o);
    b.addEventListener('click', function () {
      var ok = o === ans; done++; if (ok) right++;
      K.$$(wrap, 'button').forEach(function (x) {
        x.disabled = true;
        if (x.getAttribute('data-v') === ans) x.classList.add('good');
      });
      if (!ok) b.classList.add('badv');
      K.$(box, '.if-why').innerHTML = (ok ? '<span class="if-ok">✓ 对</span>' : '<span class="if-no">✗ 错</span>') +
        '　' + it[0] + ' = <b>' + it[1] + '</b>' + (ok ? '' : '（你选的是 ' + o + '）');
      score(); K.fit();
    });
    wrap.appendChild(b);
  });
  score(); K.fit();
}
function score() { $('.if-score').textContent = done ? '本轮答对 ' + right + ' / ' + done : '共 9 个 ion，答完一轮会重新洗牌'; }
$('.if-next').addEventListener('click', nextQ);
nextQ();

K.init($root);
})(document.getElementById("k-ch02-07"));</script>"""))
```

### B. Equations：配平、state symbols、ionic equations

**配平**：atom 不会凭空产生或消失，左右两边每种 atom 的数目必须相等。

1. 写出 reactants 和 products 的正确 formula——**之后只改前面的系数，绝不改 formula**；
2. 数两边每种 atom；
3. 一次配一种 atom，最后配出现在多个物质里的那一种（通常是 O 或 H）。

例：Fe₂O₃ + CO → Fe + CO₂ → 先配 Fe：Fe₂O₃ + CO → **2**Fe + CO₂ → 设 CO 和 CO₂ 的系数都是 *x*（C 守恒），O 两边分别是 3 + *x* 和 2*x* → *x* = 3：
Fe₂O₃ + 3CO → 2Fe + 3CO₂。

真题里「Construct an equation」常给产物的描述，要自己写 formula：

- bismuth 把 water 还原成 Bi₂O₃，同时放出一种 squeaky pop 的气体（H₂）：2Bi + 3H₂O → Bi₂O₃ + 3H₂（9701/22 Feb/Mar 2024 Q1(b)(i)）
- Mg₃N₂ 和 excess water 只生成 ammonia 和 magnesium hydroxide：Mg₃N₂ + 6H₂O → 3Mg(OH)₂ + 2NH₃（9701/21 May/June 2024 Q2(c)(ii)）

**state symbols**：(s) solid、(l) liquid、(g) gas、(aq) aqueous（溶在水里）。注意 water 本身是 (l)，不是 (aq)；precipitate 是 (s)。

**ionic equation**：只写真正参与反应的 ions / atoms / molecules；溶液里前后没有变化的 ion 叫 **spectator ions**，要删掉。做法：把所有 (aq) 的 ionic compound 拆成 ions → 两边相同的划掉 → 检查 atoms 和 **charges** 都平衡。

| full equation | ionic equation | spectator ions |
|---|---|---|
| Zn(s) + CuSO₄(aq) → ZnSO₄(aq) + Cu(s) | Zn(s) + Cu<sup>2+</sup>(aq) → Zn<sup>2+</sup>(aq) + Cu(s) | SO₄<sup>2−</sup> |
| H₂SO₄(aq) + 2NaOH(aq) → Na₂SO₄(aq) + 2H₂O(l) | H<sup>+</sup>(aq) + OH<sup>−</sup>(aq) → H₂O(l) | Na<sup>+</sup>、SO₄<sup>2−</sup> |
| Br₂(aq) + 2KI(aq) → 2KBr(aq) + I₂(aq) | Br₂(aq) + 2I<sup>−</sup>(aq) → 2Br<sup>−</sup>(aq) + I₂(aq) | K<sup>+</sup> |
| CuSO₄(aq) + 2NaOH(aq) → Cu(OH)₂(s) + Na₂SO₄(aq) | Cu<sup>2+</sup>(aq) + 2OH<sup>−</sup>(aq) → Cu(OH)₂(s) | Na<sup>+</sup>、SO₄<sup>2−</sup> |
| Pb(NO₃)₂(aq) + 2KI(aq) → PbI₂(s) + 2KNO₃(aq) | Pb<sup>2+</sup>(aq) + 2I<sup>−</sup>(aq) → PbI₂(s) | K<sup>+</sup>、NO₃<sup>−</sup> |

> **LO 2.3.2** — (a) write and construct equations (which should be balanced), including ionic equations (which should not include spectator ions)
>
> (b) use appropriate state symbols in equations

```python
# @title 🔬 配平 + ionic equation { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-09">
<h3>配平 equation + ionic equation</h3>
<div class="tabbar"><button data-tab="bal">① 配平 equation</button><button data-tab="ion">② ionic equation</button></div>

<div data-pane="bal">
  <p class="mut">用每个物质下面的 <b>−</b> / <b>+</b> 调系数。formula 是灰色锁定的：改 subscript 就变成另一种物质了，只能改前面的系数。</p>
  <div class="row bl-pick"></div>
  <p class="faint bl-ctx"></p>
  <div class="card bl-eqcard"><div class="bl-eq"></div></div>
  <table class="bl-tab"></table>
  <div class="bl-st"></div>
  <div class="row"><button class="bl-reset">系数全部回到 1</button><button class="bl-hint">提示：配平顺序</button></div>
  <div class="bl-hintbox"></div>
</div>

<div data-pane="ion">
  <p class="mut">做法：把 (aq) 的 ionic compound 拆成 ions → 两边一模一样的 ion（spectator ions）划掉 → 检查 atoms 和 charges 都平衡。</p>
  <div class="row bl-ipick"></div>
  <div class="card"><div class="faint">full equation</div><div class="bl-full"></div></div>
  <div class="row bl-steps">
    <button class="bl-s1">① 拆成 ions</button>
    <button class="bl-s2" disabled>② 划掉 spectator ions</button>
    <button class="bl-s3" disabled>③ 检查</button>
    <button class="bl-ire">重来</button>
  </div>
  <div class="bl-work"></div>
  <div class="bl-msg"></div>
  <div class="bl-final"></div>
</div>

<style>
.k .bl-pick button,.k .bl-ipick button{font-size:14px}
.k .bl-eqcard{margin:8px 0}
.k .bl-eq{display:flex;flex-wrap:wrap;align-items:flex-start;justify-content:center;gap:6px 4px;font-size:24px}
.k .bl-term{display:inline-flex;flex-direction:column;align-items:center;gap:4px}
.k .bl-top{display:inline-flex;align-items:baseline;gap:3px;white-space:nowrap}
.k .bl-co{color:var(--acc);font-weight:700;min-width:.7em;text-align:right}
.k .bl-co.one{color:var(--faint);font-weight:400}
.k .bl-f{background:var(--line);color:var(--fg);border-radius:6px;padding:0 7px;cursor:not-allowed;user-select:none}
.k .bl-pm{display:inline-flex;gap:4px}
.k .bl-pm button{font-size:15px;line-height:1;padding:3px 10px;min-width:30px}
.k .bl-op{padding:0 6px;color:var(--mut);align-self:flex-start;line-height:1.55}
.k .bl-tab td,.k .bl-tab th{text-align:center;min-width:64px}
.k .bl-tab td.w{text-align:left;color:var(--mut);font-size:13.5px}
.k .bl-tab tr.eq td{background:var(--ok-soft)}
.k .bl-tab tr.eq td.n{color:var(--ok);font-weight:700}
.k .bl-tab tr.ne td.n{color:var(--bad);font-weight:700}
.k .bl-st,.k .bl-msg{margin:6px 0;padding:7px 11px;border-radius:6px;border-left:3px solid var(--line);background:var(--card)}
.k .bl-st:empty,.k .bl-msg:empty{display:none}
.k .bl-st.ok,.k .bl-msg.ok{border-left-color:var(--ok);background:var(--ok-soft)}
.k .bl-st.warn,.k .bl-msg.warn{border-left-color:var(--bad);background:var(--bad-soft)}
.k .bl-big{font-size:20px}
.k .bl-hintbox{display:none;margin-top:6px;padding:7px 11px;border-left:3px solid var(--acc);background:var(--acc-soft);border-radius:0 6px 6px 0}
.k .bl-hintbox.show{display:block}
.k .bl-full{font-size:19px;margin-top:2px}
.k .bl-st-s{color:var(--mut);font-size:.85em}
.k .bl-why .bl-st-s{font-size:1em}
.k .bl-work{margin:8px 0}
.k .bl-work:empty{display:none}
.k .bl-line{display:flex;flex-wrap:wrap;align-items:center;gap:6px 4px;font-size:19px;padding:8px 10px;background:var(--card);border:1px solid var(--line);border-radius:8px}
.k .bl-chip{display:inline-block;border:1px solid var(--line);border-radius:6px;padding:0 6px;background:var(--bg)}
.k .bl-chip.ion{border-color:var(--acc)}
.k .bl-live .bl-chip{cursor:pointer}
.k .bl-live .bl-chip:hover{background:var(--acc-soft)}
.k .bl-chip.gone{text-decoration:line-through;text-decoration-color:var(--bad);text-decoration-thickness:2px;opacity:.5}
.k .bl-chip.flash{background:var(--bad-soft);border-color:var(--bad)}
.k .bl-q{font-family:"DejaVu Sans","Segoe UI","Helvetica Neue",Arial,sans-serif}
.k .bl-why{font-size:13.5px;color:var(--mut);margin:6px 2px 0}
.k .bl-why li{margin:1px 0}
.k .bl-final table td,.k .bl-final table th{text-align:center}
.k .bl-final table td.w{text-align:left;color:var(--mut)}
.k .bl-final .good{color:var(--ok);font-weight:700}
.k button:disabled{opacity:.45;cursor:default}
.k button:disabled:hover{border-color:var(--line)}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var Q = function (s) { return $root.querySelector(s); };
var SUBD = '₀₁₂₃₄₅₆₇₈₉';
function sub(f) { return f.replace(/\d/g, function (d) { return SUBD[d]; }); }
/* charge in its own Latin font: Noto Sans draws a lone ⁺/⁻ as a near-invisible speck (I⁻ looked like I·) */
function chg(q) { return q === 0 ? '' : '<span class="bl-q">' + K.sup(Math.abs(q) === 1 ? '' : Math.abs(q)) + (q > 0 ? '⁺' : '⁻') + '</span>'; }
function signed(q) { return q === 0 ? '0' : (q > 0 ? '+' : '−') + Math.abs(q); }
function gcd(a, b) { return b ? gcd(b, a % b) : a; }
function gcdAll(a) { return a.reduce(function (g, x) { return gcd(g, x); }, 0); }
/* 'Mg(OH)2' → {Mg:1, O:2, H:2}, keys in order of first appearance */
function atoms(f) {
  var st = [{}], order = [], i = 0, m;
  function add(o, e, n) { if (!(e in o)) o[e] = 0; o[e] += n; if (order.indexOf(e) < 0) order.push(e); }
  while (i < f.length) {
    if (f[i] === '(') { st.push({}); i++; }
    else if (f[i] === ')') {
      m = f.slice(i + 1).match(/^\d*/)[0]; i += 1 + m.length;
      var g = st.pop(), k = +m || 1;
      for (var e in g) add(st[st.length - 1], e, g[e] * k);
    } else {
      m = f.slice(i).match(/^([A-Z][a-z]?)(\d*)/);
      add(st[st.length - 1], m[1], +m[2] || 1); i += m[0].length;
    }
  }
  st[0].__order = order;
  return st[0];
}

/* ------------------------------------------------------------ tab 1: balancing */
var EQ = [
  { L: ['Fe2O3', 'CO'], R: ['Fe', 'CO2'], ans: [1, 3, 2, 3], ctx: 'blast furnace：iron(III) oxide 被 carbon monoxide 还原',
    hint: '先配 Fe：右边要 <b>2</b>Fe。再配 O：设 CO 和 CO₂ 系数都是 <i>x</i>，左边 3 + <i>x</i>，右边 2<i>x</i> → <i>x</i> = 3。' },
  { L: ['Al(OH)3'], R: ['Al2O3', 'H2O'], ans: [2, 1, 3], ctx: '加热 aluminium hydroxide（thermal decomposition）',
    hint: '先配 Al：Al₂O₃ 有 2 个 Al → <b>2</b>Al(OH)₃。左边现在 6 个 H → <b>3</b>H₂O。最后核对 O：左 6，右 3 + 3 = 6。' },
  { L: ['Fe', 'HCl'], R: ['FeCl2', 'H2'], ans: [1, 2, 1, 1], ctx: 'iron + dilute hydrochloric acid',
    hint: 'FeCl₂ 有 2 个 Cl → <b>2</b>HCl。H 自动平衡：左 2，右 H₂ 也是 2。' },
  { L: ['C3H8', 'O2'], R: ['CO2', 'H2O'], ans: [1, 5, 3, 4], ctx: 'propane 的 complete combustion',
    hint: '先配 C → <b>3</b>CO₂；再配 H：8 个 H → <b>4</b>H₂O；O 最后配：右边 3×2 + 4×1 = 10 → <b>5</b>O₂。' },
  { L: ['Bi', 'H2O'], R: ['Bi2O3', 'H2'], ans: [2, 3, 1, 3], ctx: 'bismuth 把 water 还原成 Bi₂O₃，放出 H₂（9701/22 Feb/Mar 2024 Q1(b)(i)）',
    hint: 'Bi₂O₃ 需要 <b>2</b>Bi；它有 3 个 O → <b>3</b>H₂O；左边 6 个 H → <b>3</b>H₂。' },
  { L: ['Mg3N2', 'H2O'], R: ['Mg(OH)2', 'NH3'], ans: [1, 6, 3, 2], ctx: 'Mg₃N₂ + excess water → ammonia + magnesium hydroxide（9701/21 May/June 2024 Q2(c)(ii)）',
    hint: '先配 Mg → <b>3</b>Mg(OH)₂；再配 N → <b>2</b>NH₃；H：右边 3×2 + 2×3 = 12 → <b>6</b>H₂O；最后核对 O：6 = 3×2。' }
];
var cur = 0, co = [];
EQ.forEach(function (e, i) {
  var b = K.el('button', { 'data-eq': i }, e.L.map(sub).join(' + ') + ' → ' + e.R.map(sub).join(' + '));
  b.addEventListener('click', function () { pick(i); });
  Q('.bl-pick').appendChild(b);
});
function species() { var e = EQ[cur]; return e.L.concat(e.R); }
function eqText(c) {
  var e = EQ[cur], s = species();
  function side(a, off) { return a.map(function (f, j) { var n = c[off + j]; return (n > 1 ? n : '') + sub(f); }).join(' + '); }
  return side(e.L, 0) + ' → ' + side(e.R, e.L.length);
}
function pick(i) {
  cur = i; co = species().map(function () { return 1; });
  $root.querySelectorAll('.bl-pick button').forEach(function (b) { b.classList.toggle('on', +b.getAttribute('data-eq') === i); });
  Q('.bl-ctx').textContent = EQ[i].ctx;
  Q('.bl-hintbox').classList.remove('show'); Q('.bl-hintbox').innerHTML = EQ[i].hint;
  Q('.bl-hint').textContent = '提示：配平顺序';
  buildEq(); update();
}
function buildEq() {
  var e = EQ[cur], box = Q('.bl-eq'); box.innerHTML = '';
  species().forEach(function (f, k) {
    if (k === e.L.length) box.appendChild(K.el('span', { 'class': 'bl-op' }, '→'));
    else if (k) box.appendChild(K.el('span', { 'class': 'bl-op' }, '+'));
    var t = K.el('span', { 'class': 'bl-term', 'data-i': k },
      '<span class="bl-top"><span class="bl-co"></span><span class="bl-f" data-tip="formula 不能改：改 subscript 就是另一种物质。只能改前面的系数。">' + sub(f) + '</span></span>' +
      '<span class="bl-pm"><button class="bl-minus" aria-label="减少系数">−</button><button class="bl-plus" aria-label="增加系数">+</button></span>');
    t.querySelector('.bl-minus').addEventListener('click', function () { if (co[k] > 1) { co[k]--; update(); } });
    t.querySelector('.bl-plus').addEventListener('click', function () { if (co[k] < 12) { co[k]++; update(); } });
    box.appendChild(t);
  });
}
function update() {
  var e = EQ[cur], s = species(), A = s.map(atoms), els = [];
  A.forEach(function (a) { a.__order.forEach(function (x) { if (els.indexOf(x) < 0) els.push(x); }); });
  $root.querySelectorAll('.bl-term').forEach(function (t) {
    var k = +t.getAttribute('data-i'), c = t.querySelector('.bl-co');
    c.textContent = co[k]; c.classList.toggle('one', co[k] === 1);
    t.querySelector('.bl-minus').disabled = co[k] <= 1;
    t.querySelector('.bl-plus').disabled = co[k] >= 12;
  });
  var rows = '<tr><th>element</th><th colspan="2">左边 reactants</th><th colspan="2">右边 products</th></tr>', bad = [];
  els.forEach(function (x) {
    function side(from, to) {
      var parts = [], tot = 0;
      for (var k = from; k < to; k++) if (A[k][x]) { parts.push(co[k] + '×' + A[k][x]); tot += co[k] * A[k][x]; }
      return { w: parts.join(' + '), n: tot };
    }
    var l = side(0, e.L.length), r = side(e.L.length, s.length), ok = l.n === r.n;
    if (!ok) bad.push(x);
    rows += '<tr class="' + (ok ? 'eq' : 'ne') + '" data-el="' + x + '"><th>' + x + '</th><td class="w">' + l.w + '</td><td class="n">' + l.n +
      '</td><td class="w">' + r.w + '</td><td class="n">' + r.n + '</td></tr>';
  });
  Q('.bl-tab').innerHTML = rows;
  var st = Q('.bl-st'), g = gcdAll(co);
  st.className = 'bl-st';
  if (bad.length) {
    st.innerHTML = '还没平衡：<b>' + bad.join('、') + '</b> 两边数目不相等。';
    $root.setAttribute('data-bal', 'no');
  } else if (g > 1) {
    st.classList.add('warn');
    st.innerHTML = 'atoms 都平衡了，但所有系数都能 ÷ ' + g + ' —— equation 要用<b>最简整数比</b>：' + eqText(co.map(function (n) { return n / g; }));
    $root.setAttribute('data-bal', 'reduce');
  } else {
    st.classList.add('ok');
    st.innerHTML = '<span class="bl-big">✓ ' + eqText(co) + '</span><br>每种 element 两边数目相等，系数是最简整数比。';
    $root.setAttribute('data-bal', 'ok');
  }
  K.fit();
}
Q('.bl-reset').addEventListener('click', function () { co = co.map(function () { return 1; }); update(); });
Q('.bl-hint').addEventListener('click', function () {
  var on = Q('.bl-hintbox').classList.toggle('show');
  Q('.bl-hint').textContent = on ? '收起提示' : '提示：配平顺序'; K.fit();
});
pick(0);

/* ------------------------------------------------------------ tab 2: ionic equations */
/* species: n, f (ASCII), st, and either ions [[f, per formula unit, charge], …] or why (stays whole) */
var IR = [
  { name: 'Zn + CuSO₄',
    L: [{ n: 1, f: 'Zn', st: 's', why: 'metal，(s)' }, { n: 1, f: 'CuSO4', st: 'aq', ions: [['Cu', 1, 2], ['SO4', 1, -2]] }],
    R: [{ n: 1, f: 'ZnSO4', st: 'aq', ions: [['Zn', 1, 2], ['SO4', 1, -2]] }, { n: 1, f: 'Cu', st: 's', why: 'metal，(s)' }] },
  { name: 'H₂SO₄ + NaOH',
    L: [{ n: 1, f: 'H2SO4', st: 'aq', ions: [['H', 2, 1], ['SO4', 1, -2]], note: 'strong acid，在 water 里完全 ionise，所以也拆' },
        { n: 2, f: 'NaOH', st: 'aq', ions: [['Na', 1, 1], ['OH', 1, -1]] }],
    R: [{ n: 1, f: 'Na2SO4', st: 'aq', ions: [['Na', 2, 1], ['SO4', 1, -2]] }, { n: 2, f: 'H2O', st: 'l', why: 'covalent molecule，(l)' }] },
  { name: 'Br₂ + KI',
    L: [{ n: 1, f: 'Br2', st: 'aq', why: 'covalent molecule：溶在 water 里也不 ionise' }, { n: 2, f: 'KI', st: 'aq', ions: [['K', 1, 1], ['I', 1, -1]] }],
    R: [{ n: 2, f: 'KBr', st: 'aq', ions: [['K', 1, 1], ['Br', 1, -1]] }, { n: 1, f: 'I2', st: 'aq', why: 'covalent molecule：溶在 water 里也不 ionise' }] },
  { name: 'CuSO₄ + NaOH',
    L: [{ n: 1, f: 'CuSO4', st: 'aq', ions: [['Cu', 1, 2], ['SO4', 1, -2]] }, { n: 2, f: 'NaOH', st: 'aq', ions: [['Na', 1, 1], ['OH', 1, -1]] }],
    R: [{ n: 1, f: 'Cu(OH)2', st: 's', why: 'precipitate，(s)' }, { n: 1, f: 'Na2SO4', st: 'aq', ions: [['Na', 2, 1], ['SO4', 1, -2]] }] },
  { name: 'Pb(NO₃)₂ + KI',
    L: [{ n: 1, f: 'Pb(NO3)2', st: 'aq', ions: [['Pb', 1, 2], ['NO3', 2, -1]] }, { n: 2, f: 'KI', st: 'aq', ions: [['K', 1, 1], ['I', 1, -1]] }],
    R: [{ n: 1, f: 'PbI2', st: 's', why: 'precipitate，(s)' }, { n: 2, f: 'KNO3', st: 'aq', ions: [['K', 1, 1], ['NO3', 1, -1]] }] }
];
var ir = 0, step = 0, parts = null, struck = {};
IR.forEach(function (r, i) {
  var b = K.el('button', { 'data-r': i }, r.name);
  b.addEventListener('click', function () { ipick(i); });
  Q('.bl-ipick').appendChild(b);
});
function spTxt(f, q, st, n) { return (n > 1 ? n : '') + sub(f) + chg(q || 0) + '<span class="bl-st-s">(' + st + ')</span>'; }
function fullTxt(r) {
  function side(a) { return a.map(function (s) { return spTxt(s.f, 0, s.st, s.n); }).join(' + '); }
  return side(r.L) + ' → ' + side(r.R);
}
/* split one side into particles, merging identical ones */
function split(a) {
  var out = [], idx = {};
  function push(f, q, st, n, whole) {
    var key = f + '|' + q + '|' + st;
    if (key in idx) { out[idx[key]].n += n; return; }
    idx[key] = out.length; out.push({ f: f, q: q, st: st, n: n, key: key, ion: !whole });
  }
  a.forEach(function (s) {
    if (s.ions) s.ions.forEach(function (io) { push(io[0], io[2], s.st, s.n * io[1], false); });
    else push(s.f, 0, s.st, s.n, true);
  });
  return out;
}
function isSpec(key) {
  var l = parts.L.filter(function (p) { return p.key === key; })[0], r = parts.R.filter(function (p) { return p.key === key; })[0];
  return !!(l && r && l.ion && l.n === r.n);
}
function specKeys() { return parts.L.filter(function (p) { return isSpec(p.key); }).map(function (p) { return p.key; }); }
function msg(t, cls) { var m = Q('.bl-msg'); m.className = 'bl-msg' + (cls ? ' ' + cls : ''); m.innerHTML = t; K.fit(); }
function ipick(i) {
  ir = i; step = 0; parts = null; struck = {};
  $root.querySelectorAll('.bl-ipick button').forEach(function (b) { b.classList.toggle('on', +b.getAttribute('data-r') === i); });
  Q('.bl-full').innerHTML = fullTxt(IR[i]);
  Q('.bl-work').innerHTML = ''; Q('.bl-final').innerHTML = ''; msg('');
  Q('.bl-s1').disabled = false; Q('.bl-s2').disabled = true; Q('.bl-s3').disabled = true;
  Q('.bl-s2').classList.remove('on');
  $root.setAttribute('data-ionic', 'start');
  K.fit();
}
function drawWork() {
  var r = IR[ir], h = '<div class="faint">拆开后（' + (step >= 2 ? '点 ion 划掉；两边同一个 ion 会一起划掉' : '先看哪些拆了、哪些没拆') + '）</div>';
  function side(a) {
    return a.map(function (p) {
      return '<span class="bl-chip' + (p.ion ? ' ion' : '') + (struck[p.key] ? ' gone' : '') + '" data-key="' + p.key + '">' +
        spTxt(p.f, p.q, p.st, p.n) + '</span>';
    }).join('<span class="bl-op">+</span>');
  }
  h += '<div class="bl-line' + (step >= 2 ? ' bl-live' : '') + '">' + side(parts.L) + '<span class="bl-op">→</span>' + side(parts.R) + '</div>';
  var why = [];
  r.L.concat(r.R).forEach(function (s) {
    var t = spTxt(s.f, 0, s.st, 1);
    if (s.why) why.push('<li>' + t + ' 不拆：' + s.why + '</li>');
    else if (s.note) why.push('<li>' + t + '：' + s.note + '</li>');
  });
  var aq = r.L.concat(r.R).filter(function (s) { return s.ions && !s.note; }).map(function (s) { return spTxt(s.f, 0, s.st, 1); });
  if (aq.length) why.unshift('<li>' + aq.filter(function (t, j) { return aq.indexOf(t) === j; }).join('、') + '：(aq) ionic compound → 拆成 ions（系数要乘进去）</li>');
  h += '<ul class="bl-why">' + why.join('') + '</ul>';
  Q('.bl-work').innerHTML = h;
  $root.querySelectorAll('.bl-work .bl-chip').forEach(function (c) { c.addEventListener('click', function () { strike(c.getAttribute('data-key')); }); });
  K.fit();
}
function strike(key) {
  if (step < 2) return;
  var p = parts.L.concat(parts.R).filter(function (x) { return x.key === key; })[0], t = spTxt(p.f, p.q, p.st, 1);
  if (struck[key]) { delete struck[key]; msg(''); drawWork(); return; }
  if (!isSpec(key)) {
    var why = !p.ion ? t + ' 不是 ion，是 reactant / product 本身 —— 它参与了反应，不能划掉。'
      : t + ' 只在' + (parts.L.indexOf(p) >= 0 ? '左' : '右') + '边出现，另一边找不到一模一样的 ' + t + ' —— 它参与了反应（变成了别的 species），<b>不是 spectator</b>，不能划掉。';
    msg('✗ ' + why, 'warn');
    drawWork();
    $root.querySelectorAll('.bl-work .bl-chip[data-key="' + key + '"]').forEach(function (c) { c.classList.add('flash'); });
    $root.setAttribute('data-ionic', 'wrong');
    return;
  }
  struck[key] = 1; msg(''); drawWork();
  $root.setAttribute('data-ionic', 'striking');
}
Q('.bl-s1').addEventListener('click', function () {
  var r = IR[ir]; parts = { L: split(r.L), R: split(r.R) }; step = 1;
  Q('.bl-s1').disabled = true; Q('.bl-s2').disabled = false;
  drawWork(); msg('只拆 (aq) 的 ionic compound（和 strong acid）；(s)、(l)、(g) 和 covalent molecules 保持原样。下一步：② 找出 spectator ions。');
  $root.setAttribute('data-ionic', 'split');
});
Q('.bl-s2').addEventListener('click', function () {
  if (step >= 2) return;
  step = 2; Q('.bl-s2').classList.add('on'); Q('.bl-s3').disabled = false;
  drawWork(); msg('点击你认为是 spectator 的 ion：反应前后完全没变（同一个 ion、同样的 state、同样的数目）。划好了按 ③ 检查。');
  $root.setAttribute('data-ionic', 'striking');
});
Q('.bl-s3').addEventListener('click', function () {
  var sk = specKeys(), left = sk.filter(function (k) { return !struck[k]; });
  if (left.length) {
    msg('还有 ' + left.length + ' 种 spectator ion 没划掉 —— 找两边一模一样的 ion。', 'warn');
    $root.setAttribute('data-ionic', 'incomplete');
    return;
  }
  var L = parts.L.filter(function (p) { return !struck[p.key]; }), R = parts.R.filter(function (p) { return !struck[p.key]; });
  var g = gcdAll(L.concat(R).map(function (p) { return p.n; }));
  function line(a) { return a.map(function (p) { return spTxt(p.f, p.q, p.st, p.n / g); }).join(' + '); }
  var eq = line(L) + ' → ' + line(R);
  var els = [];
  L.concat(R).forEach(function (p) { atoms(p.f).__order.forEach(function (x) { if (els.indexOf(x) < 0) els.push(x); }); });
  function cnt(a, x) { return a.reduce(function (t, p) { return t + (atoms(p.f)[x] || 0) * p.n / g; }, 0); }
  function charge(a) {
    var io = a.filter(function (p) { return p.q; }), tot = io.reduce(function (t, p) { return t + p.q * p.n / g; }, 0);
    if (!io.length) return { w: '', n: 0 };
    return { w: io.map(function (p) { var n = p.n / g; return (n > 1 ? n + '×' : '') + '(' + signed(p.q) + ')'; }).join(' + '), n: tot };
  }
  var rows = '<tr><th></th><th>左边</th><th>右边</th><th></th></tr>', allOk = true;
  els.forEach(function (x) {
    var a = cnt(L, x), b = cnt(R, x); if (a !== b) allOk = false;
    rows += '<tr><th>' + x + '</th><td>' + a + '</td><td>' + b + '</td><td class="' + (a === b ? 'good' : '') + '">' + (a === b ? '✓' : '✗') + '</td></tr>';
  });
  var cl = charge(L), cr = charge(R); if (cl.n !== cr.n) allOk = false;
  rows += '<tr><th>charge</th><td class="w">' + (cl.w ? cl.w + ' = ' + signed(cl.n) : '0（没有 ion）') + '</td><td class="w">' + (cr.w ? cr.w + ' = ' + signed(cr.n) : '0（没有 ion）') + '</td><td class="' + (cl.n === cr.n ? 'good' : '') + '">' + (cl.n === cr.n ? '✓' : '✗') + '</td></tr>';
  var specs = sk.slice().sort(function (a, b) { return (+b.split('|')[1] > 0) - (+a.split('|')[1] > 0); }).map(function (k) { var p = parts.L.filter(function (x) { return x.key === k; })[0]; return sub(p.f) + chg(p.q); });
  Q('.bl-final').innerHTML = '<div class="bl-st ok"><div class="faint">ionic equation</div><span class="bl-big bl-ionic">' + eq + '</span>' +
    (g > 1 ? '<br><span class="mut">（剩下的系数同除以 ' + g + '，写成最简整数比）</span>' : '') +
    '<br>spectator ions：<b class="bl-specs">' + specs.join('、') + '</b></div>' +
    '<table class="bl-chk">' + rows + '</table>' +
    '<p class="' + (allOk ? 'good' : '') + '">' + (allOk ? '✓ atoms 和 charge 两边都平衡。' : '✗ 不平衡') + '</p>';
  msg('');
  Q('.bl-s3').disabled = true; Q('.bl-s2').disabled = true; Q('.bl-s2').classList.remove('on');
  $root.setAttribute('data-ionic', allOk ? 'done' : 'unbalanced');
  K.fit();
});
Q('.bl-ire').addEventListener('click', function () { ipick(ir); });
ipick(0);

K.init($root);
})(document.getElementById("k-ch02-09"));</script>"""))
```

### C. Empirical formula 和 molecular formula

- **empirical formula**：the simplest whole-number ratio of the atoms of each element present in a compound。
- **molecular formula**：the actual number of atoms of each element present in one molecule。
- molecular formula 一定是 empirical formula 的整数倍：C₆H₁₂O₆ → CH₂O，C₂H₄Br₂ → CH₂Br。ionic compound 没有 molecule，它的 formula 就是 empirical formula。

> **LO 2.3.3** — define and use the terms empirical and molecular formula

**anhydrous / hydrated / water of crystallisation**

- **water of crystallisation**：以固定的 mol 数结合在 crystal structure 里的 water，formula 里用「·」隔开，系数是整数：CuSO₄·5H₂O。
- **hydrated**：含 water of crystallisation 的 compound（blue CuSO₄·5H₂O）。
- **anhydrous**：不含 water of crystallisation（white CuSO₄）。
- 加热：CuSO₄·5H₂O → CuSO₄ + 5H₂O；加水则反过来。同一个 salt 可以有不同 hydrate：CoCl₂·6H₂O、CoCl₂·2H₂O。

> **LO 2.3.4** — understand and use the terms anhydrous, hydrated and water of crystallisation

**怎么算**（三列表格法）：

1. 每个 element 的 mass（或 %）——没给的那个用减法：O = 总 mass − 其余；
2. ÷ *A*ᵣ → amount (mol)；
3. ÷ 最小的那个 → ratio；接近 .5 就 ×2，接近 .33 / .67 就 ×3，**不要把 1.5 四舍五入成 2**；
4. molecular formula：*n* = *M*ᵣ ÷ (empirical formula mass)，再把 empirical formula 乘 *n*。

例：100.0 g 的 X（只含 S 和 Cl）含 47.49 g S，*M*ᵣ = 135.2（9701/21 May/June 2026 Q1(e)）：

| | S | Cl |
|---|---|---|
| mass / g | 47.49 | 100.0 − 47.49 = 52.51 |
| ÷ *A*ᵣ | 47.49 / 32.1 = 1.48 | 52.51 / 35.5 = 1.48 |
| ratio | 1 | 1 |

empirical formula SCl，mass = 67.6；135.2 / 67.6 = 2 → **S₂Cl₂**。

hydrated salt 求 *x* 时，把 H₂O 当成一个「element」（*M*ᵣ = 18.0）放进表格。

> **LO 2.3.5** — calculate empirical and molecular formulas, using given data

```python
# @title 🔬 empirical → molecular formula 分步解 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-11">
<h3>empirical formula → molecular formula：分步解</h3>
<p class="mut">选一个预设或者自己改数值，再按「下一步」一行一行展开：mass → ÷ <i>A</i>ᵣ → ÷ smallest → × <i>k</i> 成整数 → formula。每一步都写出算式，跟 mark scheme 的 working 一样。</p>
<div class="tabbar">
  <button data-tab="mass">① by mass</button>
  <button data-tab="pct">② by %</button>
  <button data-tab="mf">③ molecular formula</button>
  <button data-tab="hyd">④ hydrated salt</button>
</div>
<div data-pane="mass" class="em-pane"></div>
<div data-pane="pct" class="em-pane"></div>
<div data-pane="mf" class="em-pane"></div>
<div data-pane="hyd" class="em-pane"></div>

<style>
.k .em-top{display:flex;flex-wrap:wrap;gap:8px 14px;align-items:center;margin:4px 0}
.k .em-top select{font-size:14px;max-width:100%}
.k .em-ctx{margin:2px 0 6px}
.k .em-in{display:flex;flex-wrap:wrap;gap:8px 18px;align-items:center}
.k .em-in label{display:inline-flex;align-items:center;gap:6px;white-space:nowrap}
.k .em-in input[type=number]{width:6.2em}
.k .em-in input[type=text]{width:7em}
.k .em-in .em-auto{color:var(--mut);font-size:14px}
.k .em-btns{margin:8px 0 2px}
.k .em-btns button:disabled{opacity:.45;cursor:default}
.k .em-err{margin:6px 0;padding:7px 11px;border-left:3px solid var(--bad);background:var(--bad-soft);border-radius:0 6px 6px 0}
.k .em-err:empty{display:none}
.k .em-wrap{overflow-x:auto}
.k .em-tab{margin:8px 0 4px;font-size:14.5px}
.k .em-tab th,.k .em-tab td{text-align:center;vertical-align:middle;min-width:92px}
.k .em-tab th.em-rl,.k .em-tab td.em-rl{text-align:left;background:var(--acc-soft);font-weight:600;white-space:nowrap}
.k .em-tab th.em-sym{font-size:17px}
.k .em-tab tr.hid{display:none}
.k .em-tab tr.new td{animation:em-in .5s ease-out}
@keyframes em-in{from{background:var(--acc-soft)}to{background:transparent}}
.k .em-tab td.em-sub{background:var(--hl)}
.k .em-tab .em-tag{display:block;font-size:12px;color:var(--mut);font-weight:400}
.k .em-tab td.em-min{outline:2px solid var(--acc);outline-offset:-3px;font-weight:700}
.k .em-tab .em-v{font-weight:700}
.k .em-tab .em-w{color:var(--mut)}
.k .em-tab .em-int{color:var(--ok);font-weight:700;font-size:17px}
.k .em-steps{margin:6px 0;padding-left:24px}
.k .em-steps li{margin:3px 0}
.k .em-steps li.hid{display:none}
.k .em-steps .em-warn{color:var(--bad);font-weight:600}
.k .em-res{margin:8px 0 2px;padding:8px 12px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0;font-size:16px}
.k .em-res.hid,.k .em-res:empty{display:none}
.k .em-res .em-big{font-size:22px;font-weight:700;margin:0 4px}
.k .em-res.warn{border-left-color:var(--bad);background:var(--bad-soft)}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var AR = { H: 1.0, B: 10.8, C: 12.0, N: 14.0, O: 16.0, F: 19.0, Ne: 20.2, Na: 23.0, Mg: 24.3, Al: 27.0, Si: 28.1, P: 31.0,
  S: 32.1, Cl: 35.5, K: 39.1, Ca: 40.1, Cr: 52.0, Mn: 54.9, Fe: 55.8, Co: 58.9, Cu: 63.5, Zn: 65.4, Br: 79.9, Sr: 87.6,
  Rh: 102.9, Ag: 107.9, I: 126.9, Ba: 137.3, Ir: 192.2, Pb: 207.2, Bi: 209.0 };
var SUBD = '₀₁₂₃₄₅₆₇₈₉';
function sub(f) { return String(f).replace(/\d/g, function (d) { return SUBD[d]; }); }
function unsub(f) { return String(f).replace(/[₀-₉]/g, function (d) { return SUBD.indexOf(d); }); }
function minus(s) { return String(s).replace(/-/g, '−'); }
/* 3 s.f., × 10ⁿ instead of e-notation */
function sf(x, n) {
  n = n || 3;
  if (!isFinite(x)) return '—';
  if (x === 0) return '0';
  var a = Math.abs(x);
  if (a >= 1e-3 && a < 1e6) {
    var s = x.toPrecision(n);
    if (s.indexOf('e') < 0) return s;
    return String(Math.round(x));               // e.g. 1.23e+3 → 1230
  }
  var e = Math.floor(Math.log10(a)), m = x / Math.pow(10, e);
  if (Math.abs(+m.toPrecision(n)) >= 10) { e++; m /= 10; }
  return m.toPrecision(n) + ' × 10' + K.sup(String(e));
}
function dp(s) { s = String(s).trim(); var i = s.indexOf('.'); return i < 0 ? 0 : s.length - i - 1; }
function fx(x, d) { return x.toFixed(d); }
/* 'CH2Br' / 'CH₂Br' → [['C',1],['H',2],['Br',1]] or null */
function parse(f) {
  f = unsub(String(f).replace(/\s+/g, ''));
  if (!f) return null;
  var out = [], re = /([A-Z][a-z]?)(\d*)/g, m, used = 0;
  while ((m = re.exec(f))) {
    if (m.index !== used) return null;
    used += m[0].length;
    var n = m[2] ? +m[2] : 1;
    if (!n) return null;
    var hit = out.filter(function (p) { return p[0] === m[1]; })[0];
    if (hit) hit[1] += n; else out.push([m[1], n]);
  }
  return used === f.length ? out : null;
}
function unknown(parts) { return parts.filter(function (p) { return !(p[0] in AR); }).map(function (p) { return p[0]; }); }
function fmass(parts) { return parts.reduce(function (s, p) { return s + p[1] * AR[p[0]]; }, 0); }
/* '12.0 + 2 × 1.0 + 79.9' */
function fmassW(parts) {
  return parts.map(function (p) { return (p[1] > 1 ? p[1] + ' × ' : '') + fx(AR[p[0]], 1); }).join(' + ');
}
function fstr(parts) { return parts.map(function (p) { return p[0] + (p[1] > 1 ? sub(p[1]) : ''); }).join(''); }
/* smallest k (1–6) that makes every ratio a whole number (± 0.1) */
function findK(r) {
  for (var k = 1; k <= 6; k++) {
    if (r.every(function (x) { return Math.abs(x * k - Math.round(x * k)) <= 0.1; })) return k;
  }
  return 0;
}
function fracName(x) {
  var f = x - Math.floor(x);
  if (Math.abs(f - 0.5) < 0.1) return '.5';
  if (Math.abs(f - 1 / 3) < 0.1) return '⅓';
  if (Math.abs(f - 2 / 3) < 0.1) return '⅔';
  if (Math.abs(f - 0.25) < 0.1) return '¼';
  if (Math.abs(f - 0.75) < 0.1) return '¾';
  return '';
}

/* ------------------------------------------------------------ presets
   cols: element (or salt / H2O) columns; one column may be found by subtraction (sub: true).
   total: the whole (compound mass, or 100 for %); inputs: everything that is not 'sub'. */
var MODES = {
  mass: {
    q: 'mass / g', presets: [
      { name: 'Mg 燃烧：0.486 g Mg → 0.806 g MgO', total: '0.806', totalLab: 'mass of oxide / g',
        cols: [{ s: 'Mg', v: '0.486', lab: 'mass of Mg / g' }, { s: 'O', sub: true }],
        ctx: '0.486 g magnesium 在 air 里加热到 constant mass，得到 0.806 g oxide。O 的 mass 没直接给 → 用减法。' },
      { name: 'phosphorus oxide：1.55 g P → 3.55 g oxide', total: '3.55', totalLab: 'mass of oxide / g',
        cols: [{ s: 'P', v: '1.55', lab: 'mass of P / g' }, { s: 'O', sub: true }],
        ctx: '1.55 g phosphorus 在 excess oxygen 中燃烧，得到 3.55 g oxide。注意 ratio 会出现 2.5。' },
      { name: 'X（S + Cl）：100.0 g X 含 47.49 g S，Mᵣ = 135.2', total: '100.0', totalLab: 'mass of X / g', Mr: '135.2',
        cols: [{ s: 'S', v: '47.49', lab: 'mass of S / g' }, { s: 'Cl', sub: true }],
        ctx: '9701/21 May/June 2026 Q1(e)：X 只含 S 和 Cl；Cl 的 mass 用减法。' }
    ] },
  pct: {
    q: '% by mass', presets: [
      { name: 'hydrocarbon：85.7% C，14.3% H', total: '100',
        cols: [{ s: 'C', v: '85.7', lab: '% C' }, { s: 'H', v: '14.3', lab: '% H' }],
        ctx: '两个 % 都给了（加起来 100）。可以把 % 直接当成 100 g 样品里的 mass。' },
      { name: 'vitamin C：40.9% C，4.5% H，其余是 O', total: '100', Mr: '176',
        cols: [{ s: 'C', v: '40.9', lab: '% C' }, { s: 'H', v: '4.5', lab: '% H' }, { s: 'O', sub: true }],
        ctx: 'vitamin C 只含 C、H、O；O 的 % 没给 → 100 − 其余。ratio 会出现 1.32 ≈ 1⅓。' }
    ] },
  hyd: {
    q: null, presets: [
      { name: 'CuSO₄·xH₂O：2.50 g → 加热后 1.60 g', total: '2.50', totalLab: 'mass of hydrated salt / g', q: 'mass / g',
        cols: [{ s: 'CuSO4', salt: true, v: '1.60', lab: 'mass of anhydrous CuSO₄ / g' }, { s: 'H2O', sub: true }],
        ctx: '2.50 g blue CuSO₄·xH₂O 加热到 constant mass，剩 1.60 g white anhydrous CuSO₄。失去的 mass 就是 water。' },
      { name: 'RhCl₃·xH₂O：20.5% water of crystallisation', total: '100', q: '% by mass',
        cols: [{ s: 'RhCl3', salt: true, sub: true }, { s: 'H2O', v: '20.5', lab: '% water' }],
        ctx: '9701/21 Oct/Nov 2024 Q1(d)(i)：hydrated rhodium(III) chloride 含 20.5% by mass water of crystallisation。RhCl₃ 的 % 用减法。' },
      { name: 'CoCl₂·xH₂O：2.38 g → 加热后 1.30 g', total: '2.38', totalLab: 'mass of hydrated salt / g', q: 'mass / g',
        cols: [{ s: 'CoCl2', salt: true, v: '1.30', lab: 'mass of anhydrous CoCl₂ / g' }, { s: 'H2O', sub: true }],
        ctx: '2.38 g CoCl₂·xH₂O 加热到 constant mass，剩 1.30 g anhydrous CoCl₂。' }
    ] },
  mf: {
    presets: [
      { name: 'CH₂Br，Mᵣ = 187.8', ef: 'CH2Br', Mr: '187.8' },
      { name: 'C₃H₅，Mᵣ = 82', ef: 'C3H5', Mr: '82' },
      { name: 'CCl₃，Mᵣ = 237', ef: 'CCl3', Mr: '237' },
      { name: 'CH₂，Mᵣ = 112', ef: 'CH2', Mr: '112' },
      { name: 'SCl，Mᵣ = 135.2（9701/21 May/June 2026 Q1(e)）', ef: 'SCl', Mr: '135.2' }
    ] }
};

/* ------------------------------------------------------------ pane skeleton */
function build(mode) {
  var pane = $root.querySelector('[data-pane="' + mode + '"]');
  pane.innerHTML =
    '<div class="em-top"><label>预设 <select class="em-preset"></select></label></div>' +
    '<p class="faint em-ctx"></p>' +
    '<div class="card em-in"></div>' +
    '<div class="row em-btns"><button class="em-next">下一步 ▸</button><button class="em-all">全部显示</button>' +
    '<button class="em-reset">从头开始</button><span class="faint em-prog"></span></div>' +
    '<div class="em-err"></div>' +
    '<div class="em-wrap"><table class="em-tab"></table></div>' +
    '<ol class="em-steps"></ol>' +
    '<div class="em-res hid"></div>';
  var st = { mode: mode, pane: pane, preset: 0, vals: {}, shown: 1, nSteps: 1 };
  var sel = pane.querySelector('.em-preset');
  MODES[mode].presets.forEach(function (p, i) { sel.appendChild(K.el('option', { value: i }, p.name)); });
  sel.addEventListener('change', function () { load(st, +sel.value); });
  pane.querySelector('.em-next').addEventListener('click', function () { st.shown = Math.min(st.nSteps, st.shown + 1); render(st, true); });
  pane.querySelector('.em-all').addEventListener('click', function () { st.shown = st.nSteps; render(st, false); });
  pane.querySelector('.em-reset').addEventListener('click', function () { st.shown = 1; render(st, false); });
  load(st, 0);
}

function load(st, i) {
  st.preset = i; st.shown = 1;
  var p = MODES[st.mode].presets[i], box = st.pane.querySelector('.em-in');
  st.pane.querySelector('.em-preset').value = i;
  st.pane.querySelector('.em-ctx').textContent = p.ctx || '';
  box.innerHTML = '';
  function field(key, lab, val, type) {
    var l = K.el('label', {}, '<span>' + lab + '</span>');
    var inp = K.el('input', { type: type || 'number', step: 'any', 'data-k': key, value: val, spellcheck: 'false' });
    inp.addEventListener('input', function () { render(st, false); });
    l.appendChild(inp); box.appendChild(l);
  }
  if (st.mode === 'mf') {
    field('ef', 'empirical formula', p.ef, 'text');
    field('Mr', '<i>M</i>ᵣ', p.Mr);
  } else {
    if (st.mode !== 'pct' && p.total !== '100') field('total', p.totalLab, p.total);
    p.cols.forEach(function (c, j) { if (!c.sub) field('c' + j, c.lab, c.v); });
    var subc = p.cols.filter(function (c) { return c.sub; })[0];
    if (subc) box.appendChild(K.el('span', { 'class': 'em-auto' },
      (subc.salt || subc.s === 'H2O' ? sub(subc.s) : subc.s) + '：用减法（见下表）'));
    if (st.mode !== 'hyd') field('Mr', '<i>M</i>ᵣ（可选，留空只求 empirical formula）', p.Mr || '');
  }
  render(st, false);
}

function val(st, k) { var e = st.pane.querySelector('input[data-k="' + k + '"]'); return e ? e.value.trim() : ''; }

/* ------------------------------------------------------------ calculations */
function solveTable(st) {
  var p = MODES[st.mode].presets[st.preset], hyd = st.mode === 'hyd';
  var totS = (st.mode === 'pct' || p.total === '100') ? '100' : val(st, 'total');
  var tot = parseFloat(totS);
  var cols = p.cols.map(function (c, j) {
    var o = { s: c.s, salt: !!c.salt, sub: !!c.sub, disp: hyd ? sub(c.s) : c.s };
    if (hyd || c.salt) {
      var parts = parse(c.s); o.M = fmass(parts); o.Mw = fmassW(parts);
    } else o.M = AR[c.s];
    if (!c.sub) { o.vs = val(st, 'c' + j); o.v = parseFloat(o.vs); }
    return o;
  });
  var err = '';
  cols.forEach(function (c) { if (!c.sub && !(c.v > 0)) err = c.disp + ' 的数值要是正数。'; });
  if (!(tot > 0)) err = '总量要是正数。';
  var sc = cols.filter(function (c) { return c.sub; })[0];
  if (!err && sc) {
    var given = cols.filter(function (c) { return !c.sub; });
    var d = Math.max(dp(totS), Math.max.apply(null, given.map(function (c) { return dp(c.vs); })));
    sc.v = +(tot - given.reduce(function (s, c) { return s + c.v; }, 0)).toFixed(d);
    sc.vs = fx(sc.v, d);
    sc.w = totS + ' − ' + given.map(function (c) { return c.vs; }).join(' − ');
    if (!(sc.v > 0)) err = '减法得到 ' + sc.disp + ' = ' + minus(sc.vs) + '：总量必须比其余加起来大。';
  }
  if (!err && !sc && st.mode === 'pct') {
    var sum = cols.reduce(function (s, c) { return s + c.v; }, 0);
    if (Math.abs(sum - 100) > 0.5) err = '各 % 加起来是 ' + +sum.toFixed(2) + '，不是 100——要么改数值，要么有一个 element 要用减法。';
  }
  if (err) return { err: err };
  cols.forEach(function (c) { c.n = c.v / c.M; });
  var nmin = Math.min.apply(null, cols.map(function (c) { return c.n; }));
  cols.forEach(function (c) { c.r = c.n / nmin; c.isMin = c.n === nmin; });
  var k = findK(cols.map(function (c) { return c.r; }));
  cols.forEach(function (c) { c.int = k ? Math.round(c.r * k) : null; });
  return { cols: cols, nmin: nmin, k: k, p: p, hyd: hyd, q: p.q || MODES[st.mode].q };
}

/* the Aᵣ-not-Mᵣ trap, named with an element actually in the table */
var DIAT = { O: 1, Cl: 1, H: 1, N: 1, Br: 1, F: 1, I: 1 };
function diatomicNote(C) {
  var d = C.filter(function (c) { return DIAT[c.s]; })[0];
  if (!d) return '用 <i>A</i>ᵣ（单个 atom），不是 <i>M</i>ᵣ。';
  return '用 <i>A</i>ᵣ，不是 <i>M</i>ᵣ（' + d.s + ' 是 ' + fx(AR[d.s], 1) + '，不是 ' + d.s + '₂ 的 ' + fx(2 * AR[d.s], 1) + '）。';
}

/* ------------------------------------------------------------ render: table modes */
function render(st, animate) {
  if (st.mode === 'mf') return renderMf(st, animate);
  var R = solveTable(st), pane = st.pane, tab = pane.querySelector('.em-tab'), ol = pane.querySelector('.em-steps'),
    res = pane.querySelector('.em-res');
  pane.querySelector('.em-err').textContent = R.err || '';
  tab.innerHTML = ''; ol.innerHTML = ''; res.innerHTML = ''; res.className = 'em-res hid';
  pane.removeAttribute('data-ef'); pane.removeAttribute('data-mf'); pane.removeAttribute('data-x');
  if (R.err) { st.nSteps = 1; progress(st); K.fit(); return; }
  var C = R.cols, hyd = R.hyd, mrS = hyd ? '' : val(st, 'Mr'), Mr = parseFloat(mrS), hasMr = !hyd && Mr > 0;
  st.nSteps = hasMr ? 6 : 5;

  function row(cls, lab, cells) {
    var tr = K.el('tr', { 'class': cls }, '<td class="em-rl">' + lab + '</td>');
    cells.forEach(function (c) { tr.appendChild(K.el('td', c[1] ? { 'class': c[1] } : {}, c[0])); });
    tab.appendChild(tr); return tr;
  }
  tab.appendChild(K.el('tr', {}, '<th class="em-rl"></th>' + C.map(function (c) { return '<th class="em-sym">' + c.disp + '</th>'; }).join('')));
  var rows = [];
  rows.push(row('em-r1', R.q, C.map(function (c) {
    return c.sub ? ['<span class="em-tag">用减法</span><span class="em-w">' + c.w + ' =</span> <span class="em-v">' + c.vs + '</span>', 'em-sub']
                 : ['<span class="em-v">' + c.vs + '</span>'];
  })));
  rows.push(row('em-r2', '÷ ' + (hyd ? '<i>M</i>ᵣ' : '<i>A</i>ᵣ') + ' → mol', C.map(function (c) {
    return ['<span class="em-w">' + c.vs + ' ÷ ' + fx(c.M, 1) + ' =</span><br><span class="em-v">' + sf(c.n) + '</span>', c.isMin ? 'em-min' : ''];
  })));
  rows.push(row('em-r3', '÷ smallest', C.map(function (c) {
    return ['<span class="em-w">' + sf(c.n) + ' ÷ ' + sf(R.nmin) + ' =</span><br><span class="em-v">' + fx(c.r, 2) + '</span>'];
  })));
  rows.push(row('em-r4', R.k > 1 ? '× ' + R.k + ' → 整数' : '取整数', C.map(function (c) {
    if (!R.k) return ['<span class="em-warn">?</span>'];
    var x = c.r * R.k;
    return ['<span class="em-w">' + (R.k > 1 ? fx(c.r, 2) + ' × ' + R.k + ' = ' + fx(x, 2) : fx(c.r, 2)) + ' →</span> <span class="em-int">' + c.int + '</span>'];
  })));

  /* ---- step notes (one per row + result) */
  var notes = [];
  var sc = C.filter(function (c) { return c.sub; })[0], given = C.filter(function (c) { return !c.sub; });
  var n1;
  if (R.q === '% by mass') {
    n1 = '把 % 当成 100 g 样品里每种 ' + (hyd ? 'component' : 'element') + ' 的 mass。';
  } else {
    n1 = '写出每种 ' + (hyd ? 'component（salt 和 H₂O）' : 'element') + ' 的 mass。';
  }
  if (sc) n1 += ' <b>' + sc.disp + ' 没有直接给，用减法</b>：' + sc.disp + ' = ' + sc.w + ' = <b>' + sc.vs + '</b>' + (R.q === '% by mass' ? '%' : ' g') + '。';
  notes.push(n1);
  var n2 = hyd ? '把 H₂O 当成一个「element」：每列 ÷ 它的 <i>M</i>ᵣ（' + C.map(function (c) { return '<i>M</i>ᵣ(' + c.disp + ') = ' + c.Mw + ' = ' + fx(c.M, 1); }).join('；') + '）。'
               : '每列 ÷ 该 element 的 <i>A</i>ᵣ，得到 amount (mol)。' + diatomicNote(C);
  notes.push(n2);
  var mn = C.filter(function (c) { return c.isMin; })[0];
  notes.push('最小的是 <i>n</i>(' + mn.disp + ') = ' + sf(R.nmin) + ' mol（表里框起来的那格）→ 每列都 ÷ ' + sf(R.nmin) + '。中间步骤不要提前四舍五入。');
  var n4;
  if (!R.k) {
    n4 = '<span class="em-warn">× 1 到 × 6 都化不成整数比——检查一下数据。</span>';
  } else if (R.k > 1) {
    var odd = C.filter(function (c) { return Math.abs(c.r - Math.round(c.r)) > 0.1; })[0];
    var fr = fracName(odd.r);
    n4 = fx(odd.r, 2) + ' 不是整数' + (fr ? '（小数部分 ≈ ' + fr + '）' : '') + '，<b>不能四舍五入成 ' + Math.round(odd.r) + '</b>（那就是另一个 compound 了）。' +
      '全部 × ' + R.k + '：' + C.map(function (c) { return fx(c.r, 2); }).join(' : ') + ' → ' + C.map(function (c) { return c.int; }).join(' : ') + '。';
  } else {
    var far = C.some(function (c) { return Math.abs(c.r - c.int) >= 0.005; });
    n4 = far ? '都很接近整数（小偏差来自数据取整），直接取整：' + C.map(function (c) { return c.int; }).join(' : ') + '。'
             : '已经是整数比：' + C.map(function (c) { return c.int; }).join(' : ') + '。';
  }
  notes.push(n4);

  /* ---- result */
  var resHtml = '', ef = '', ok = !!R.k, warn = false;
  if (ok && hyd) {
    var salt = C.filter(function (c) { return c.salt; })[0], w = C.filter(function (c) { return !c.salt; })[0];
    if (salt.int === 1) {
      ef = sub(salt.s) + '·' + (w.int > 1 ? w.int : '') + 'H₂O';
      notes.push('<i>x</i> = <i>n</i>(H₂O) ÷ <i>n</i>(' + salt.disp + ') = ' + fx(w.r, 2) + (Math.abs(w.r - w.int) > 0.005 ? ' ≈ ' : ' → ') + '<b>' + w.int + '</b>。');
      resHtml = '<i>x</i> = <span class="em-big">' + w.int + '</span> → <span class="em-big">' + ef + '</span>';
      pane.setAttribute('data-x', w.int);
    } else {
      warn = true;
      notes.push('<span class="em-warn">salt : H₂O = ' + salt.int + ' : ' + w.int + '，<i>x</i> 不是整数——检查数据。</span>');
      resHtml = 'salt : H₂O = ' + salt.int + ' : ' + w.int + '，<i>x</i> 不是整数';
    }
  } else if (ok) {
    var parts = C.map(function (c) { return [c.s, c.int]; });
    ef = fstr(parts);
    notes.push('按整数比写 formula（1 不写出来）：empirical formula = <b>' + ef + '</b>。');
    resHtml = 'empirical formula = <span class="em-big">' + ef + '</span>';
    if (hasMr) {
      var efm = fmass(parts), n = Mr / efm, ni = Math.round(n);
      if (ni >= 1 && Math.abs(n - ni) <= 0.1) {
        var mf = fstr(parts.map(function (q) { return [q[0], q[1] * ni]; }));
        notes.push('molecular formula：empirical formula mass = ' + fmassW(parts) + ' = ' + fx(efm, 1) + '；<i>n</i> = <i>M</i>ᵣ ÷ ' + fx(efm, 1) + ' = ' + mrS + ' ÷ ' + fx(efm, 1) +
          ' = ' + fx(n, 2) + (Math.abs(n - ni) > 0.005 ? ' ≈ ' + ni : '') + ' → ' + ef + (ni > 1 ? ' × ' + ni : '') + ' = <b>' + mf + '</b>。');
        resHtml += '<br>molecular formula = (' + ef + ')' + (ni > 1 ? sub(ni) : '') + ' = <span class="em-big">' + mf + '</span>';
        pane.setAttribute('data-mf', mf);
      } else {
        warn = true;
        notes.push('<span class="em-warn">empirical formula mass = ' + fx(efm, 1) + '；' + mrS + ' ÷ ' + fx(efm, 1) + ' = ' + fx(n, 2) + '，不是整数——<i>M</i>ᵣ 和 empirical formula 对不上。</span>');
        resHtml += '<br><span class="em-warn">' + mrS + ' ÷ ' + fx(efm, 1) + ' = ' + fx(n, 2) + ' 不是整数</span>';
      }
    }
  } else {
    warn = true; resHtml = '化不成整数比';
  }
  if (ef) pane.setAttribute('data-ef', ef);
  notes.forEach(function (h, i) { ol.appendChild(K.el('li', { 'class': i < st.shown ? '' : 'hid' }, h)); });
  rows.forEach(function (tr, i) {
    if (i >= st.shown) tr.classList.add('hid');
    else if (animate && i === st.shown - 1) tr.classList.add('new');
  });
  res.innerHTML = resHtml;
  res.className = 'em-res' + (st.shown >= 5 ? '' : ' hid') + (warn ? ' warn' : '');
  progress(st);
  K.fit();
}

/* ------------------------------------------------------------ render: molecular formula mode */
function renderMf(st, animate) {
  var pane = st.pane, tab = pane.querySelector('.em-tab'), ol = pane.querySelector('.em-steps'), res = pane.querySelector('.em-res');
  tab.innerHTML = ''; ol.innerHTML = ''; res.innerHTML = ''; res.className = 'em-res hid';
  pane.removeAttribute('data-ef'); pane.removeAttribute('data-mf');
  st.nSteps = 3;
  var efS = val(st, 'ef'), mrS = val(st, 'Mr'), Mr = parseFloat(mrS), parts = parse(efS), err = '';
  if (!parts) err = '看不懂「' + efS + '」：element symbol 首字母大写，数字跟在后面，例如 CH2Br。';
  else if (unknown(parts).length) err = unknown(parts).join('、') + ' 不在本 cell 的 <i>A</i>ᵣ 表里（用 syllabus Periodic Table 的常见 element）。';
  else if (!(Mr > 0)) err = '<i>M</i>ᵣ 要是正数。';
  pane.querySelector('.em-err').innerHTML = err;
  if (err) { st.nSteps = 1; progress(st); K.fit(); return; }
  var ef = fstr(parts), efm = fmass(parts), n = Mr / efm, ni = Math.round(n), good = ni >= 1 && Math.abs(n - ni) <= 0.1;
  pane.setAttribute('data-ef', ef);
  tab.appendChild(K.el('tr', {}, '<th class="em-rl"></th>' + parts.map(function (p) { return '<th class="em-sym">' + p[0] + '</th>'; }).join('') + '<th>' + ef + '</th>'));
  var r1 = K.el('tr', { 'class': 'em-r1' }, '<td class="em-rl">atoms × <i>A</i>ᵣ</td>' + parts.map(function (p) {
    return '<td><span class="em-w">' + p[1] + ' × ' + fx(AR[p[0]], 1) + ' =</span><br><span class="em-v">' + fx(p[1] * AR[p[0]], 1) + '</span></td>';
  }).join('') + '<td><span class="em-w">合计</span><br><span class="em-v">' + fx(efm, 1) + '</span></td>');
  var r2 = K.el('tr', { 'class': 'em-r2' }, '<td class="em-rl">× <i>n</i></td>' + parts.map(function (p) {
    return '<td><span class="em-w">' + p[1] + ' × ' + (good ? ni : '?') + ' →</span> <span class="em-int">' + (good ? p[1] * ni : '?') + '</span></td>';
  }).join('') + '<td><span class="em-w">' + fx(efm, 1) + ' × ' + (good ? ni : '?') + ' =</span><br><span class="em-v">' + (good ? fx(efm * ni, 1) : '?') + '</span></td>');
  tab.appendChild(r1); tab.appendChild(r2);
  var notes = [
    'empirical formula mass（' + ef + '）= ' + fmassW(parts) + ' = <b>' + fx(efm, 1) + '</b>。',
    '<i>n</i> = <i>M</i>ᵣ ÷ empirical formula mass = ' + mrS + ' ÷ ' + fx(efm, 1) + ' = ' + fx(n, 2) +
      (good ? (Math.abs(n - ni) > 0.005 ? ' ≈ <b>' + ni + '</b>' : ' → <b>' + ni + '</b>') + '。' : '。<span class="em-warn">不是整数——<i>M</i>ᵣ 和 empirical formula 对不上。</span>'),
    good ? '每个 subscript 都 × ' + ni + '：molecular formula = <b>' + fstr(parts.map(function (p) { return [p[0], p[1] * ni]; })) + '</b>' +
      (ni === 1 ? '（<i>n</i> = 1：molecular formula 和 empirical formula 一样）' : '') + '。'
         : '<span class="em-warn">没法写出 molecular formula。</span>'
  ];
  var mf = good ? fstr(parts.map(function (p) { return [p[0], p[1] * ni]; })) : '';
  if (good) pane.setAttribute('data-mf', mf);
  notes.forEach(function (h, i) { ol.appendChild(K.el('li', { 'class': i < st.shown ? '' : 'hid' }, h)); });
  [r1, r2].forEach(function (tr, i) {
    var vis = i === 0 ? st.shown >= 1 : st.shown >= 2;
    if (!vis) tr.classList.add('hid'); else if (animate && i === st.shown - 1) tr.classList.add('new');
  });
  res.innerHTML = good ? 'molecular formula = (' + ef + ')' + (ni > 1 ? sub(ni) : '') + ' = <span class="em-big">' + mf + '</span>'
                       : '<span class="em-warn">' + mrS + ' ÷ ' + fx(efm, 1) + ' = ' + fx(n, 2) + '，不是整数</span>';
  res.className = 'em-res' + (st.shown >= 3 ? '' : ' hid') + (good ? '' : ' warn');
  progress(st);
  K.fit();
}

function progress(st) {
  var p = st.pane;
  p.setAttribute('data-step', st.shown);
  p.querySelector('.em-prog').textContent = '第 ' + Math.min(st.shown, st.nSteps) + ' / ' + st.nSteps + ' 步';
  p.querySelector('.em-next').disabled = st.shown >= st.nSteps;
  p.querySelector('.em-all').disabled = st.shown >= st.nSteps;
}

['mass', 'pct', 'mf', 'hyd'].forEach(build);

K.init($root);
})(document.getElementById("k-ch02-11"));</script>"""))
```

## 2.4 Reacting masses and volumes (of solutions and gases)

### A. Reacting masses、limiting reagent、percentage yield

balanced equation 的系数就是 **mole ratio**（stoichiometry）。所有题都是同一个套路：

> 已知量 → **mol** → 用系数换成目标物的 **mol** → 目标量

例：79.8 g Fe₂O₃ 与 9.36 g C 加热：2Fe₂O₃ + 3C → 4Fe + 3CO₂，求 CO₂ 的 mass。

1. *n*(Fe₂O₃) = 79.8 / 159.6 = 0.500 mol；*n*(C) = 9.36 / 12.0 = 0.780 mol
2. 0.500 mol Fe₂O₃ 需要 0.500 × 3/2 = 0.750 mol C < 0.780 → **C in excess**（剩 0.030 mol），**Fe₂O₃ is the limiting reagent**
3. 产物永远按 limiting reagent 算：*n*(CO₂) = 0.500 × 3/2 = 0.750 mol → 0.750 × 44.0 = **33.0 g**

**percentage yield** = actual yield ÷ theoretical (predicted) yield × 100%。theoretical yield 是 limiting reagent 全部转化、没有 side reaction 时算出来的量。yield 低于 100% 的原因：side reactions、reaction 没有进行完全（reversible）、操作中损失产物。

例：18.0 g Al 与 excess Cl₂：2Al + 3Cl₂ → 2AlCl₃，实得 71.0 g AlCl₃。theoretical = 18.0 / 27.0 × 133.5 = 89.0 g → yield = 71.0 / 89.0 × 100 = **79.8%**。

**significant figures**：答案的 s.f. 跟题目数据里最少的那个一致；**中间步骤不要先四舍五入**，最后一步才 round。0.004 是 1 s.f.，0.0040 是 2 s.f.。mark scheme 对 s.f. 是认真的：9701/22 May/June 2026 Q4(b)(iii) 的 1.875 × 10<sup>−3</sup> mol，接受 1.88 × 10<sup>−3</sup> / 1.9 × 10<sup>−3</sup>，但**拒绝** 1.880 × 10<sup>−3</sup>、1.90 × 10<sup>−3</sup>、2.0 × 10<sup>−3</sup>（凭空多出来的 0 = 不合理的精度），也拒绝截断成 1.87 × 10<sup>−3</sup>。

```python
# @title 🔬 limiting reagent + percentage yield { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-13">
<h3>limiting reagent 与 percentage yield</h3>
<p class="mut">拖动 slider（或直接填数）改两个 reactant 的 mass。每条横条的长度 = <i>n</i> ÷ equation 里的系数，也就是这份 reactant「够反应几份 equation」。
<b>短的那条先用完 → 它就是 limiting reagent</b>；反应在虚线处停止，长的那条多出来的浅色段就是 excess。products 全部按 limiting reagent 算。</p>
<div class="row lr-pick"></div>
<div class="card lr-eqcard"><div class="lr-eq"></div><div class="lr-mr faint"></div></div>
<div class="lr-ctl"></div>
<svg class="lr-fig" viewBox="0 0 860 300"></svg>
<div class="card lr-work"></div>
<h4>percentage yield</h4>
<div class="row lr-yrow">actual yield of <select class="lr-yp"></select> = <input type="text" class="lr-act" inputmode="decimal" spellcheck="false"> g</div>
<div class="card lr-yield"></div>
<p class="faint">theoretical yield = limiting reagent 全部转化、没有 side reaction 时算出的量。yield &lt; 100% 的原因：side reactions、reaction 没有进行完全（reversible）、操作中损失产物。<i>A</i>ᵣ 取 Periodic Table 的 1 位小数；结果取 3 s.f.，中间步骤不提前四舍五入。</p>
<style>
.k .lr-pick button{font-size:14px}
.k .lr-eqcard{margin:8px 0;text-align:center}
.k .lr-eq{font-size:22px;letter-spacing:.2px}
.k .lr-eq .co{color:var(--acc);font-weight:700;margin-right:2px}
.k .lr-eq .ra{color:var(--s1);font-weight:600}
.k .lr-eq .rb{color:var(--s4);font-weight:600}
.k .lr-eq .pr{color:var(--s3);font-weight:600}
.k .lr-mr{margin-top:2px;font-size:13.5px}
.k .lr-ctl{display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:6px 18px;margin:6px 0}
.k .lr-c{display:flex;align-items:center;gap:8px;flex-wrap:nowrap}
.k .lr-c .nm{font-weight:650;min-width:5.6em}
.k .lr-c input[type=range]{flex:1;min-width:90px}
.k .lr-c input[type=text]{width:5.6em;text-align:right}
.k .lr-c input.bad,.k .lr-act.bad{border-color:var(--bad);background:var(--bad-soft)}
.k .lr-act{width:5.6em;text-align:right}
.k .lr-fig{margin:6px 0;background:var(--card);border:1px solid var(--line);border-radius:8px}
.k .lr-fig text{font-size:16px}
.k .lr-fig text.nm{font-weight:700}
.k .lr-fig text.lim{fill:var(--acc);font-weight:700}
.k .lr-fig text.exc{fill:var(--mut)}
.k .lr-fig text.cut{fill:var(--mut)}
.k .lr-fig .used{stroke:none}
.k .lr-fig .left{stroke-width:1.5;stroke-dasharray:4 3;fill-opacity:.22}
.k .lr-fig .xi{stroke:var(--fg);stroke-width:1.6;stroke-dasharray:6 4;opacity:.65}
.k .lr-fig .bk{fill:var(--card)}
.k .lr-fig .sep{stroke:var(--line);stroke-width:1}
.k .lr-work ol{margin:4px 0}
.k .lr-work li{margin:3px 0}
.k .lr-work .h{font-weight:600}
.k .lr-work .err,.k .lr-yield .err{color:var(--bad)}
.k .lr-tag{display:inline-block;padding:0 7px;border-radius:5px;background:var(--acc-soft);color:var(--acc);font-weight:700}
.k .lr-yield .big{font-size:19px}
.k .lr-yield .warn{color:var(--bad)}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var AR = { H: 10, C: 120, O: 160, Mg: 243, Al: 270, Cl: 355, Fe: 558, Sr: 876 };   /* tenths */
var SUBD = '₀₁₂₃₄₅₆₇₈₉';
/* species: pretty formula, composition */
var SP = {
  'Fe₂O₃': [['Fe', 2], ['O', 3]], 'C': [['C', 1]], 'Fe': [['Fe', 1]], 'CO₂': [['C', 1], ['O', 2]],
  'Al': [['Al', 1]], 'Cl₂': [['Cl', 2]], 'AlCl₃': [['Al', 1], ['Cl', 3]],
  'SrO': [['Sr', 1], ['O', 1]], 'Sr': [['Sr', 1]], 'Sr₃Al₂O₆': [['Sr', 3], ['Al', 2], ['O', 6]],
  'Mg': [['Mg', 1]], 'O₂': [['O', 2]], 'MgO': [['Mg', 1], ['O', 1]]
};
function mrT(f) { return SP[f].reduce(function (t, p) { return t + AR[p[0]] * p[1]; }, 0); }   /* tenths */
function d1(t) { return (t / 10).toFixed(1); }
function mrWork(f) {
  var c = SP[f];
  if (c.length === 1 && c[0][1] === 1) return f + ' ' + d1(mrT(f));
  return f + ' = ' + c.map(function (p) { return (p[1] > 1 ? p[1] + ' × ' : '') + d1(AR[p[0]]); }).join(' + ') + ' = ' + d1(mrT(f));
}
/* r: reactants [coef, formula], p: products, m: default masses, max / step: sliders, y: product index for yield, act: actual yield */
var PRE = [
  { id: 'fe', r: [[2, 'Fe₂O₃'], [3, 'C']], p: [[4, 'Fe'], [3, 'CO₂']], m: ['79.8', '9.36'], max: [200, 25], step: [0.1, 0.01], y: 1, act: '30.0',
    ctx: '正文的例子：79.8 g Fe₂O₃ 与 9.36 g C 一起加热。' },
  { id: 'al', r: [[2, 'Al'], [3, 'Cl₂']], p: [[2, 'AlCl₃']], m: ['18.0', '100.0'], max: [50, 200], step: [0.1, 0.5], y: 0, act: '71.0',
    ctx: '18.0 g Al 与 excess Cl₂，实得 71.0 g AlCl₃。' },
  { id: 'sr', r: [[6, 'SrO'], [2, 'Al']], p: [[3, 'Sr'], [1, 'Sr₃Al₂O₆']], m: ['100', '20.0'], max: [250, 40], step: [0.5, 0.1], y: 0, act: '38.0',
    ctx: '100 g SrO 与足量的 Al 加热，制取 strontium。' },
  { id: 'mg', r: [[2, 'Mg'], [1, 'O₂']], p: [[2, 'MgO']], m: ['2.43', '2.00'], max: [6, 4], step: [0.01, 0.01], y: 0, act: '3.80',
    ctx: 'Mg ribbon 在 O₂ 中燃烧。试着把 O₂ 拖到 1.60 g：恰好按 mole ratio 反应，没有 excess。' }
];
var COL = ['var(--s1)', 'var(--s4)'], PCOL = 'var(--s3)';

function fmt(x, sf) {
  sf = sf || 3;
  if (!isFinite(x)) return '—';
  if (x === 0) return '0';
  var r = +x.toPrecision(sf), e = Math.floor(Math.log10(Math.abs(r)));
  if (e >= -3 && e < 5) return sf - 1 - e >= 0 ? r.toFixed(sf - 1 - e) : String(Math.round(r));
  return (r / Math.pow(10, e)).toFixed(sf - 1) + ' × 10' + K.sup(e);
}
function dps(s) { var i = s.indexOf('.'); return i < 0 || /×/.test(s) ? 0 : s.length - i - 1; }
function sfOf(s) { var d = s.replace('.', '').replace(/^0+/, ''); return Math.max(1, d.length); }
function ratio(a, b) { return a === b ? '' : ' × ' + a + '/' + b; }
function co(c) { return c > 1 ? c : ''; }

var pick = $root.querySelector('.lr-pick'), eqBox = $root.querySelector('.lr-eq'), mrBox = $root.querySelector('.lr-mr');
var ctl = $root.querySelector('.lr-ctl'), svg = $root.querySelector('.lr-fig'), work = $root.querySelector('.lr-work');
var ySel = $root.querySelector('.lr-yp'), act = $root.querySelector('.lr-act'), yBox = $root.querySelector('.lr-yield');
var P = PRE[0], rng = [], txt = [], last = null;

PRE.forEach(function (q) {
  var b = K.el('button', { type: 'button', 'data-pre': q.id },
    q.r.map(function (x) { return co(x[0]) + x[1]; }).join(' + ') + ' → ' + q.p.map(function (x) { return co(x[0]) + x[1]; }).join(' + '));
  b.addEventListener('click', function () { load(q); });
  pick.appendChild(b);
});

function load(q) {
  P = q;
  K.$$(pick, 'button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-pre') === q.id); });
  eqBox.innerHTML = q.r.map(function (x, i) { return '<span class="co">' + co(x[0]) + '</span><span class="' + (i ? 'rb' : 'ra') + '">' + x[1] + '</span>'; }).join(' + ') +
    ' → ' + q.p.map(function (x) { return '<span class="co">' + co(x[0]) + '</span><span class="pr">' + x[1] + '</span>'; }).join(' + ');
  mrBox.innerHTML = q.ctx + '　<i>M</i><sub>r</sub>：' + q.r.concat(q.p).map(function (x) { return mrWork(x[1]); }).join('；');
  ctl.innerHTML = ''; rng = []; txt = [];
  q.r.forEach(function (x, i) {
    var row = K.el('label', { 'class': 'lr-c', 'data-r': i });
    row.innerHTML = '<span class="nm" style="color:' + COL[i] + '">mass ' + x[1] + '</span>';
    var s = K.el('input', { type: 'range', min: 0, max: q.max[i], step: q.step[i], value: q.m[i], 'data-r': i });
    var t = K.el('input', { type: 'text', value: q.m[i], inputmode: 'decimal', spellcheck: 'false', 'data-r': i });
    s.addEventListener('input', function () { t.value = (+s.value).toFixed(dps(String(q.step[i]))); calc(); });
    t.addEventListener('input', function () { var v = parseFloat(t.value); if (isFinite(v)) s.value = v; calc(); });
    row.appendChild(s); row.appendChild(t); row.appendChild(K.el('span', null, 'g'));
    ctl.appendChild(row); rng.push(s); txt.push(t);
  });
  ySel.innerHTML = '';
  q.p.forEach(function (x, i) { ySel.appendChild(K.el('option', { value: i }, x[1])); });
  ySel.value = q.y; act.value = q.act;
  calc();
}

function num(s) { s = String(s).trim(); return /^\d*\.?\d+$|^\d+\.$/.test(s) ? parseFloat(s) : NaN; }

function calc() {
  var R = P.r.map(function (x, i) {
    var s = txt[i].value.trim(), m = num(s), M = mrT(x[1]) / 10;
    txt[i].classList.toggle('bad', !(m >= 0));
    return { c: x[0], f: x[1], s: s, m: m, M: M, n: m / M, x: m / M / x[0] };
  });
  last = null;
  if (R.some(function (r) { return !(r.m >= 0); })) {
    work.innerHTML = '<p class="err">mass 要填一个 ≥ 0 的数。</p>'; svg.innerHTML = ''; yBox.innerHTML = ''; K.fit(); return;
  }
  var A = R[0], B = R[1], tie = Math.abs(A.x - B.x) <= 1e-9 * Math.max(A.x, B.x);
  var li = tie ? 0 : (A.x < B.x ? 0 : 1), L = R[li], E = R[1 - li], xi = L.x;
  var lines = [];
  lines.push('<i>n</i>(' + A.f + ') = ' + A.s + ' ÷ ' + d1(mrT(A.f)) + ' = ' + fmt(A.n) + ' mol；<i>n</i>(' + B.f + ') = ' + B.s + ' ÷ ' + d1(mrT(B.f)) + ' = ' + fmt(B.n) + ' mol');
  var need = A.n * B.c / A.c, needS = fmt(need), haveS = fmt(B.n), exS = '', exM = '';
  if (A.m === 0 || B.m === 0) {
    lines.push('没有 ' + (A.m === 0 ? A.f : B.f) + '，reaction 不发生，products 都是 0。');
  } else {
    var cmp = tie ? '= ' : (B.x > A.x ? '> ' : '< ');
    lines.push(fmt(A.n) + ' mol ' + A.f + ' 需要 ' + fmt(A.n) + ratio(B.c, A.c) + ' = ' + needS + ' mol ' + B.f + '；实际有 ' + haveS + ' mol（' + cmp + needS + '）→ ' +
      (tie ? '恰好按 mole ratio 反应，<span class="lr-tag">没有 limiting / excess</span>，两者都用完'
           : '<b>' + E.f + ' in excess</b>，<span class="lr-tag" data-lim="' + L.f + '">' + L.f + ' is the limiting reagent</span>'));
    if (!tie) {
      /* excess = what is there − what the limiting reagent uses up; a difference keeps the fewest decimal places of the numbers subtracted */
      var used = L.n * E.c / L.c, usedS = fmt(used), hS = fmt(E.n), dp = Math.min(dps(usedS), dps(hS)), ex = E.n - used;
      exS = /×/.test(usedS + hS) ? fmt(ex) : ex.toFixed(dp);
      if (+exS === 0) exS = fmt(ex);
      exM = fmt(ex * E.M, Math.max(2, Math.min(3, sfOf(exS))));
      lines.push('excess ' + E.f + '：' + hS + ' − ' + (li === 1 ? usedS + '（= ' + fmt(L.n) + ratio(E.c, L.c) + '）' : usedS) + ' = <b>' + exS + ' mol</b> 剩下 = ' + exS + ' × ' + d1(mrT(E.f)) + ' = <b>' + exM + ' g</b>');
    }
  }
  var prods = P.p.map(function (x) {
    var n = A.m === 0 || B.m === 0 ? 0 : L.n * x[0] / L.c, M = mrT(x[1]) / 10;
    return { c: x[0], f: x[1], n: n, M: M, g: n * M };
  });
  if (A.m > 0 && B.m > 0) {
    prods.forEach(function (p) {
      lines.push('<i>n</i>(' + p.f + ') = ' + (p.c === L.c ? '<i>n</i>(' + L.f + ')' : fmt(L.n) + ratio(p.c, L.c)) + ' = ' + fmt(p.n) + ' mol → <i>m</i> = ' + fmt(p.n) + ' × ' + d1(mrT(p.f)) + ' = <b data-prod="' + p.f + '">' + fmt(p.g) + ' g</b>');
    });
  }
  work.innerHTML = '<div class="h">working（products 永远按 limiting reagent 算）</div><ol>' + lines.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>';
  last = { R: R, li: li, tie: tie, xi: xi, prods: prods, exS: exS, exM: exM };
  draw(last);
  yieldCalc();
}

/* ---------- figure ---------- */
var X0 = 150, WB = 560, BH = 22;
function draw(s) {
  var R = s.R, rows = [], y = 40;
  R.forEach(function (r, i) { rows.push({ kind: 'r', i: i, top: y }); y += 82; });
  y += 12; var sepY = y - 10;
  s.prods.forEach(function (p, i) { rows.push({ kind: 'p', i: i, top: y }); y += 58; });
  var H = y + 6;
  svg.setAttribute('viewBox', '0 0 860 ' + H);
  svg.innerHTML = '';
  var mx = Math.max(R[0].x, R[1].x) || 1, sc = WB / mx, xX = X0 + s.xi * sc;
  var gB = K.svg('g'), gK = K.svg('g'), gL = K.svg('g'); svg.appendChild(gB); svg.appendChild(gK); svg.appendChild(gL);
  var placed = [], avoid = [];
  function rect(x, y, w, h, cls, col, extra) {
    var e = K.svg('rect', Object.assign({ x: x, y: y, width: Math.max(0, w), height: h, rx: 3, 'class': cls, style: 'fill:' + col + ';stroke:' + col }, extra || {}));
    gB.appendChild(e); if (w > 0) placed.push({ x: x, y: y, w: w, h: h }); return e;
  }
  gB.appendChild(K.svg('line', { x1: 10, y1: sepY, x2: 850, y2: sepY, 'class': 'sep' }));
  rows.forEach(function (row) {
    var by = row.top + 22, mid = by + BH / 2;
    if (row.kind === 'r') {
      var r = R[row.i], col = COL[row.i], w = r.x * sc, isL = !s.tie && row.i === s.li;
      rect(X0, by, Math.min(w, s.xi * sc), BH, 'used', col, { 'data-bar': 'used-' + row.i });
      if (w - s.xi * sc > 0.5) rect(xX, by, w - s.xi * sc, BH, 'left', col, { 'data-bar': 'left-' + row.i });
      lab(X0 - 12, mid + 6, r.f, 'nm', [[0, 0, 'end']]).setAttribute('style', 'fill:' + col);
      lab(X0, by - 7, 'n = ' + fmt(r.n) + ' mol　÷ ' + r.c + ' = ' + fmt(r.x), '', [[0, 0, 'start'], [0, -4, 'start']]);
      if (r.m === 0) return;
      if (s.tie) lab(X0, by + BH + 20, '全部用完（恰好按 mole ratio）', 'lim', [[0, 0, 'start']]);
      else if (isL) lab(X0, by + BH + 20, 'limiting reagent：全部用完', 'lim', [[0, 0, 'start']]).setAttribute('data-role', 'lim-' + row.i);
      else lab(xX, by + BH + 20, 'excess：剩 ' + s.exS + ' mol = ' + s.exM + ' g', 'exc', [[6, 0, 'start'], [w - s.xi * sc, 0, 'end'], [-6, 0, 'end']]).setAttribute('data-role', 'exc-' + row.i);
    } else {
      var p = s.prods[row.i];
      rect(X0, by, s.xi * sc, BH, 'used', PCOL, { 'data-bar': 'prod-' + row.i });
      lab(X0 - 12, mid + 6, p.f, 'nm', [[0, 0, 'end']]).setAttribute('style', 'fill:' + PCOL);
      lab(X0, by - 7, 'n = ' + fmt(p.n) + ' mol　÷ ' + p.c + ' = ' + fmt(p.n / p.c) + '　→ ' + fmt(p.g) + ' g', '', [[0, 0, 'start'], [0, -4, 'start']]).setAttribute('data-role', 'prod-' + row.i);
    }
  });
  if (s.xi > 0) {
    /* the cut-off line is decorative (not in avoid): every label gets an opaque card-coloured backing,
       a glyph halo alone leaves the line visible between glyphs and it reads as a strike-through */
    gB.appendChild(K.svg('line', { x1: xX, y1: 30, x2: xX, y2: H - 6, 'class': 'xi' }));
    lab(xX, 22, (s.tie ? '两个 reactant 同时用完' : 'limiting reagent 用完') + ' → 反应在这里停止', 'cut', [[0, 0, 'middle'], [-6, 0, 'end'], [6, 0, 'start']]);
  }
  function lab(x, y, t, cls, cands) {
    /* K.label reads the viewBox from the element it is given, so pass the <svg> itself (a <g> has none →
       labels running off the right edge go unnoticed), then move the text into the label layer */
    var e = K.label(svg, x, y, t, { cls: cls, cands: cands, placed: placed, avoid: avoid, pad: 2 });
    gL.appendChild(e);
    var b = e.getBBox();
    gK.appendChild(K.svg('rect', { x: b.x - 3, y: b.y, width: b.width + 6, height: b.height, 'class': 'bk' }));
    e.innerHTML = e.textContent.replace(/\b([nm])\b(?= =)/g, '<tspan font-style="italic">$1</tspan>');
    return e;
  }
}

/* ---------- percentage yield ---------- */
function yieldCalc() {
  if (!last) { yBox.innerHTML = ''; return; }
  var p = last.prods[+ySel.value], a = num(act.value);
  act.classList.toggle('bad', !(a >= 0));
  if (!(a >= 0)) { yBox.innerHTML = '<p class="err">actual yield 要填一个 ≥ 0 的数（单位 g）。</p>'; K.fit(); return; }
  if (!(p.g > 0)) { yBox.innerHTML = '<p class="err">theoretical yield 是 0，算不了 percentage yield。</p>'; K.fit(); return; }
  var pct = a / p.g * 100;
  yBox.innerHTML = '<div>theoretical yield of ' + p.f + ' = ' + fmt(p.g) + ' g（上面按 limiting reagent 算出来的）</div>' +
    '<div class="big">percentage yield = actual ÷ theoretical × 100% = ' + act.value.trim() + ' ÷ ' + fmt(p.g) + ' × 100% = <b class="lr-pct">' + fmt(pct) + '%</b></div>' +
    (pct > 100 ? '<div class="warn">超过 100% 不可能是纯的 product——通常是没干燥（含 water / solvent）或混有杂质。</div>' : '');
  K.fit();
}
ySel.addEventListener('change', yieldCalc);
act.addEventListener('input', yieldCalc);
load(PRE[0]);

K.init($root);
})(document.getElementById("k-ch02-13"));</script>"""))
```

### B. Volumes of gases

同温同压下，相同体积的任何 gas 含相同数目的 molecules，所以 **gas 的体积比 = mole 比**。room conditions 下 1 mol gas 占 **24.0 dm<sup>3</sup>**（= 24 000 cm<sup>3</sup>）。

- 0.40 mol N₂ 的体积：0.40 × 24.0 = 9.6 dm<sup>3</sup>
- 120 cm<sup>3</sup> CH₄ 的 mass：0.120 / 24.0 = 0.00500 mol → × 16.0 = 0.0800 g
- 14 g C₂H₄ 完全燃烧：C₂H₄ + 3O₂ → 2CO₂ + 2H₂O，0.50 mol C₂H₄ → 1.0 mol CO₂ → 24 dm<sup>3</sup>

**hydrocarbon 燃烧**：C*ₓ*H*ᵧ* + (*x* + *y*/4) O₂ → *x* CO₂ + (*y*/2) H₂O

例：50 cm<sup>3</sup> hydrocarbon 恰好与 250 cm<sup>3</sup> O₂ 反应，生成 150 cm<sup>3</sup> CO₂（room temperature，H₂O 是 liquid、体积忽略）。体积比 1 : 5 : 3 → *x* = 3；O 守恒：5 × 2 = 3 × 2 + *y*/2 → *y* = 8 → **C₃H₈**。

不是 room conditions 时（例如 400 °C、101 kPa），就要用 *pV* = *nRT*：*p* 用 Pa、*V* 用 m<sup>3</sup>、*T* 用 K（chapter 4）。

```python
# @title 🔬 gas volumes：hydrocarbon 燃烧 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-15">
<h3>gas volumes：hydrocarbon 燃烧</h3>
<div class="tabbar"><button data-tab="fwd">① 正向：算体积</button><button data-tab="rev">② 反向：从体积推 formula</button></div>

<div data-pane="fwd">
  <p class="mut">选一个 hydrocarbon、填体积，下面画出燃烧<b>前</b>和冷却回 room temperature <b>后</b>的 gas 体积。H₂O 冷却后是 liquid，体积 ≈ 0，所以总体积会变小（contraction）。</p>
  <div class="row cb-pre"></div>
  <div class="row">
    <label class="cb-lab"><span>C<sub><i>x</i></sub>：<i>x</i> =</span> <input type="number" class="cb-x" min="1" max="8" step="1" value="3"></label>
    <label class="cb-lab"><span>H<sub><i>y</i></sub>：<i>y</i> =</span> <select class="cb-y"></select></label>
    <label class="cb-lab"><span><i>V</i>(hydrocarbon) / cm³</span> <input type="text" class="cb-v" value="50" inputmode="decimal" size="6"></label>
  </div>
  <div class="row">
    <label class="cb-lab"><input type="checkbox" class="cb-ex"> <span>excess O₂：一共加入</span></label>
    <label class="cb-lab"><input type="text" class="cb-o2" value="300" inputmode="decimal" size="6" disabled> <span>cm³ O₂</span></label>
  </div>
  <p class="faint cb-note"></p>
  <svg class="cb-bars" viewBox="0 0 820 200"></svg>
  <div class="card cb-eq"></div>
  <div class="card cb-work"></div>

  <h4><i>n</i> ↔ <i>V</i>（room conditions：1 mol gas = 24.0 dm³）</h4>
  <div class="row">
    <label class="cb-lab"><span><i>n</i>(<span class="cb-fn"></span>) / mol</span> <input type="text" class="cb-n" value="0.50" inputmode="decimal" size="7"></label>
    <span class="cb-lab">⇄</span>
    <label class="cb-lab"><span><i>V</i>(<span class="cb-fn"></span>) / dm³</span> <input type="text" class="cb-vd" inputmode="decimal" size="7"></label>
  </div>
  <div class="card cb-nv"></div>
  <p class="faint">不是 room conditions 时（例如 400 °C、101 kPa）不能用 24.0 dm³ mol⁻¹，要用 <i>pV</i> = <i>nRT</i>（<i>p</i> / Pa、<i>V</i> / m³、<i>T</i> / K，chapter 4）。</p>
</div>

<div data-pane="rev">
  <p class="mut">同温同压下体积比 = mole 比，所以三个体积就能定出 C<sub><i>x</i></sub>H<sub><i>y</i></sub>。改下面三个数看推导怎么变。</p>
  <div class="row">
    <label class="cb-lab"><span><i>V</i>(C<sub><i>x</i></sub>H<sub><i>y</i></sub>) / cm³</span> <input type="text" class="cb-rh" value="50" inputmode="decimal" size="6"></label>
    <label class="cb-lab"><span><i>V</i>(O₂ reacted) / cm³</span> <input type="text" class="cb-ro" value="250" inputmode="decimal" size="6"></label>
    <label class="cb-lab"><span><i>V</i>(CO₂ formed) / cm³</span> <input type="text" class="cb-rc" value="150" inputmode="decimal" size="6"></label>
  </div>
  <div class="card cb-rwork"></div>

  <h4>练一题</h4>
  <div class="cb-q"></div>
  <div class="row"><button class="cb-new">换一题</button></div>
</div>

<style>
.k .cb-lab{display:inline-flex;align-items:center;gap:6px}
.k .cb-pre button{padding:2px 9px}
.k .cb-bars{margin:8px 0;user-select:none}
.k .cb-bars text{font-size:16px}
.k .cb-bars text.cb-row{font-weight:650}
.k .cb-bars text.cb-tot{fill:var(--mut)}
.k .cb-bars .cb-frame{fill:var(--card);stroke:var(--line)}
.k .cb-bars rect.seg{stroke-width:2}
.k .cb-bars .hc{fill:color-mix(in srgb,var(--s4) 28%,transparent);stroke:var(--s4)}
.k .cb-bars .o2{fill:color-mix(in srgb,var(--s1) 26%,transparent);stroke:var(--s1)}
.k .cb-bars .co2{fill:color-mix(in srgb,var(--s5) 28%,transparent);stroke:var(--s5)}
.k .cb-bars .gone{fill:none;stroke:var(--faint);stroke-dasharray:6 4}
.k .cb-bars .h2o{stroke:var(--s3);stroke-width:4}
.k .cb-bars .lead{stroke:var(--faint);stroke-width:1.2}
.k .cb-eq{font-size:17px;margin:6px 0}
.k .cb-eq .sm{font-size:14px;color:var(--mut)}
.k .cb-work,.k .cb-nv,.k .cb-rwork{margin:6px 0;font-size:15px}
.k .cb-work ol,.k .cb-nv ol,.k .cb-rwork ol{margin:2px 0}
.k .cb-work li,.k .cb-nv li,.k .cb-rwork li{margin:3px 0}
.k .cb-err{color:var(--bad)}
.k input.bad{border-color:var(--bad);background:var(--bad-soft)}
.k .cb-q .qt{margin:4px 0 8px;line-height:1.55}
.k .cb-q .ans ol{margin:2px 0}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);   /* show the first tab now: getBBox() of a hidden pane is all zeros */
var SUBD = '₀₁₂₃₄₅₆₇₈₉';
function subs(n) { return String(n).replace(/\d/g, function (c) { return SUBD[+c]; }); }
function fml(x, y) { return 'C' + (x > 1 ? subs(x) : '') + 'H' + subs(y); }
function fmlH(x, y) { return fml(x, y); }   /* Unicode digits: <sub> digits come out bigger / lower than the ₂ in O₂ next to them */
function ymin(x) { return x === 1 ? 4 : Math.max(2, x + (x % 2)); }
function ylist(x) { var o = []; for (var y = ymin(x); y <= 2 * x + 2; y += 2) o.push(y); return o; }
/* coefficient: 3.5 → 3½ */
function coef(v) { var w = Math.floor(v); return Math.abs(v - w - 0.5) < 1e-9 ? (w ? w : '') + '½' : String(v); }
/* exact-ish volume: 87.5, 250 */
function vol(v) { return String(Math.round(v * 100) / 100); }
function parsePos(s) { var t = String(s).trim().replace(/[−–]/g, '-'); if (!/^\d*\.?\d+$|^\d+\.$/.test(t)) return null; var v = parseFloat(t); return v > 0 ? v : null; }
function sfOf(s) {
  var t = String(s).trim(), dg = t.replace('.', '').replace(/^0+/, '');
  var sf = t.indexOf('.') >= 0 ? dg.length : (dg.replace(/0+$/, '').length || 1);
  return Math.max(2, Math.min(sf, 5));
}
function fmt(x, sf) {
  if (!isFinite(x)) return '—';
  var r = +x.toPrecision(sf), e = Math.floor(Math.log10(Math.abs(r)));
  if (e >= -3 && e < 5) return sf - 1 - e >= 0 ? r.toFixed(sf - 1 - e) : String(Math.round(r));
  return (r / Math.pow(10, e)).toFixed(sf - 1) + ' × 10' + K.sup(e);
}
function q(s) { return $root.querySelector(s); }

/* ================= forward ================= */
var xIn = q('.cb-x'), ySel = q('.cb-y'), vIn = q('.cb-v'), exIn = q('.cb-ex'), o2In = q('.cb-o2');
var svg = q('.cb-bars'), eqBox = q('.cb-eq'), work = q('.cb-work'), note = q('.cb-note');
var PRE = [[1, 4], [2, 6], [2, 4], [2, 2], [3, 8], [3, 6], [4, 10]];
PRE.forEach(function (p) {
  var b = K.el('button', { type: 'button', 'data-f': fml(p[0], p[1]) }, fml(p[0], p[1]));
  b.addEventListener('click', function () { xIn.value = p[0]; fillY(p[1]); fwd(); });
  q('.cb-pre').appendChild(b);
});
function getX() { var x = Math.round(+xIn.value); return x >= 1 && x <= 8 ? x : null; }
function fillY(want) {
  var x = getX(); if (!x) return;
  var keep = want || +ySel.value, L = ylist(x);
  ySel.innerHTML = L.map(function (y) { return '<option value="' + y + '">' + y + '</option>'; }).join('');
  ySel.value = L.indexOf(keep) >= 0 ? keep : L[L.length - 1];
}

function seg(x, y, w, h, cls) { var r = K.svg('rect', { x: x, y: y, width: Math.max(w, 0), height: h, 'class': 'seg ' + cls, rx: 3 }); svg.appendChild(r); return r; }

var CURRENT = null;
function fwd() {
  var x = getX();
  xIn.classList.toggle('bad', !x);
  o2In.disabled = !exIn.checked;
  if (!x) { work.innerHTML = '<span class="cb-err"><i>x</i> 要是 1–8 的整数。</span>'; return; }
  if (!ySel.options.length || ylist(x).indexOf(+ySel.value) < 0) fillY();
  var y = +ySel.value, F = fml(x, y), FH = fmlH(x, y), a = x + y / 4;
  CURRENT = { x: x, y: y, a: a };
  K.$$($root, '.cb-pre button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-f') === F); });
  K.$$($root, '.cb-fn').forEach(function (s) { s.innerHTML = FH; });
  note.textContent = x >= 5 ? '注意：C₅ 起的 hydrocarbon 在 room temperature 多半是 liquid；真题的体积题用的是 C₁–C₄ 这类 gas。这里只当比例练习。' : '';

  /* equation */
  var dbl = Math.abs(a - Math.round(a)) > 1e-9;
  eqBox.innerHTML = '<div>' + FH + '(g) + ' + coef(a) + 'O₂(g) → ' + (x > 1 ? x : '') + 'CO₂(g) + ' + (y / 2 > 1 ? y / 2 : '') + 'H₂O(l)</div>' +
    (dbl ? '<div class="sm">× 2 变成整数：2' + FH + ' + ' + 2 * a + 'O₂ → ' + 2 * x + 'CO₂ + ' + y + 'H₂O</div>' : '') +
    '<div class="sm">体积比 ' + FH + ' : O₂ : CO₂ = <b class="cb-ratio">1 : ' + vol(a) + ' : ' + x + '</b>（= mole ratio；H₂O 是 liquid，不算 gas 体积）</div>';

  var V = parsePos(vIn.value);
  vIn.classList.toggle('bad', !V);
  if (!V) { work.innerHTML = '<span class="cb-err">「' + (vIn.value || '空') + '」看不懂——<i>V</i> 要是正数。</span>'; svg.innerHTML = ''; nv(); K.fit(); return; }
  var need = V * a, co2 = V * x;
  var sup = exIn.checked ? parsePos(o2In.value) : need;
  o2In.classList.toggle('bad', exIn.checked && !sup);
  var lines = ['<i>V</i>(O₂) needed = ' + vol(V) + ' × (' + x + ' + ' + y + '/4) = ' + vol(V) + ' × ' + vol(a) + ' = <b class="cb-need">' + vol(need) + ' cm³</b>',
               '<i>V</i>(CO₂) = ' + vol(V) + ' × ' + x + ' = <b class="cb-co2">' + vol(co2) + ' cm³</b>'];
  if (exIn.checked && (!sup || sup < need - 1e-9)) {
    lines.push('<span class="cb-err">' + (sup ? '加入的 O₂（' + vol(sup) + ' cm³）少于需要的 ' + vol(need) + ' cm³ → 会 incomplete combustion（生成 CO / C），这里不算。'
                                            : '「' + (o2In.value || '空') + '」看不懂——O₂ 体积要是正数。') + '</span>');
    work.innerHTML = '<ol>' + lines.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>';
    svg.innerHTML = ''; nv(); K.fit(); return;
  }
  var left = sup - need, before = V + sup, after = co2 + left, contr = before - after;
  if (exIn.checked) lines.push('O₂ left over = ' + vol(sup) + ' − ' + vol(need) + ' = <b class="cb-left">' + vol(left) + ' cm³</b>');
  lines.push('before = ' + vol(V) + ' + ' + vol(sup) + ' = ' + vol(before) + ' cm³；after（冷却回 room temperature）' + (exIn.checked ? '= ' + vol(co2) + ' + ' + vol(left) + ' ' : '') + '= <b class="cb-after">' + vol(after) + ' cm³</b>');
  lines.push('contraction = ' + vol(before) + ' − ' + vol(after) + ' = <b class="cb-con">' + vol(contr) + ' cm³</b>（= ' + vol(V) + ' × (1 + ' + y + '/4)，与 O₂ 加多少无关）');
  work.innerHTML = '<ol>' + lines.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>';
  draw(F, V, sup, need, co2, left, before, after, contr);
  nv(); K.fit();
}

function draw(F, V, sup, need, co2, left, before, after, contr) {
  svg.innerHTML = '';
  svg.setAttribute('viewBox', '0 0 820 262');   /* full room while placing labels; cropped to the content at the end */
  var frame = K.svg('rect', { x: 1, y: 1, width: 818, height: 260, rx: 10, 'class': 'cb-frame' });
  svg.appendChild(frame);
  var X0 = 132, W = 660, H = 38, Y1 = 60, Y2 = 150, s = W / before;
  var bars = [{ x: X0, y: Y1, w: before * s, h: H }, { x: X0, y: Y2, w: W, h: H }];
  var placed = bars.map(function (b) { return { x: b.x, y: b.y, w: b.w, h: b.h }; }), avoid = [], labs = [];
  /* before row */
  var segs1 = [[V, 'hc', F + ' ' + vol(V)], [sup, 'o2', 'O₂ ' + vol(sup)]];
  var segs2 = [[co2, 'co2', 'CO₂ ' + vol(co2)]];
  if (left > 1e-9) segs2.push([left, 'o2', 'O₂ left ' + vol(left)]);
  var rowLab = function (yy, t1, t2) {
    K.label(svg, 14, yy + H / 2, t1, { cands: [[0, -5, 'start']], placed: placed, cls: 'cb-row', pad: 1 });
    K.label(svg, 14, yy + H / 2, t2, { cands: [[0, 19, 'start']], placed: placed, cls: 'cb-tot', pad: 1 });
  };
  rowLab(Y1, 'before', vol(before) + ' cm³');
  rowLab(Y2, 'after', vol(after) + ' cm³');
  function lay(list, yy, up) {
    var cx = X0, out = [];
    list.forEach(function (sg) {
      var w = sg[0] * s, r = seg(cx, yy, w, H, sg[1]);
      r.setAttribute('data-seg', sg[2]);
      out.push({ x0: cx, w: w, t: sg[2] + ' cm³' });
      avoid.push([cx, yy, cx, yy + H]);
      cx += w;
    });
    avoid.push([cx, yy, cx, yy + H]);
    return out;
  }
  var L1 = lay(segs1, Y1, true), L2 = lay(segs2, Y2, false);
  var endAfter = X0 + after * s;
  /* contraction box */
  if (contr > 1e-9) {
    svg.appendChild(K.svg('rect', { x: endAfter, y: Y2, width: contr * s, height: H, 'class': 'seg gone', rx: 3 }));
    L2.push({ x0: endAfter, w: contr * s, t: 'contraction ' + vol(contr) + ' cm³', gone: 1 });
  }
  /* labels: inside the segment if it fits, else outside (above the before row, below the after row) */
  var probe = K.svg('text', {}); svg.appendChild(probe);
  function place(list, yy, up) {
    list.forEach(function (it) {
      probe.textContent = it.t;
      var tw = probe.getBBox().width, mx = it.x0 + it.w / 2, my = yy + H / 2;
      var t;
      if (tw + 14 <= it.w) t = K.label(svg, mx, my, it.t, { cands: [[0, 6, 'middle']], placed: placed.slice(2).concat(labs), pad: 2 });
      if (tw + 14 <= it.w) t.classList.remove('halo');   /* a card-coloured halo looks like a white outline on the tinted bar */
      else {
        var d = up ? -(H / 2 + 9) : H / 2 + 22, d2 = up ? d - 27 : d + 27;
        var c = [[0, d, 'middle'], [0, d2, 'middle'], [-6, d, 'end'], [6, d, 'start'], [-6, d2, 'end'], [6, d2, 'start']];
        t = K.label(svg, mx, my, it.t, { cands: c, placed: placed.concat(labs), avoid: avoid, pad: 2 });
        var bb = t.getBBox(), ly = up ? bb.y + bb.height : bb.y, ty = up ? yy : yy + H;
        if (Math.abs(ly - ty) > 3) {
          var ln = K.svg('line', { x1: mx, y1: ty, x2: Math.min(Math.max(mx, bb.x + 4), bb.x + bb.width - 4), y2: up ? ly + 1 : ly - 1, 'class': 'lead' });
          svg.insertBefore(ln, svg.firstChild.nextSibling);
          avoid.push([+ln.getAttribute('x1'), +ln.getAttribute('y1'), +ln.getAttribute('x2'), +ln.getAttribute('y2')]);
        }
      }
      t.setAttribute('data-lab', it.t);
      var b = t.getBBox(); labs.push({ x: b.x - 2, y: b.y - 2, w: b.width + 4, h: b.height + 4 });
    });
  }
  place(L1, Y1, true); place(L2, Y2, false);
  /* H₂O: zero-volume liquid = a tick at the end of the CO₂ segment */
  var hx = X0 + co2 * s;
  svg.appendChild(K.svg('line', { x1: hx, y1: Y2 - 7, x2: hx, y2: Y2 + H + 4, 'class': 'h2o' }));
  /* label in the empty band between the two rows, right above the tick: never competes with the after-row labels below */
  var ht = K.label(svg, hx, Y2, 'H₂O(l)：体积 ≈ 0', { cands: [[0, -11, 'middle'], [-4, -11, 'end'], [4, -11, 'start']], placed: placed.concat(labs), avoid: avoid, pad: 2 });
  ht.setAttribute('data-lab', 'H2O');
  probe.remove();
  /* crop: no empty band above / below when every label fits inside its bar */
  var y0 = 1e9, y1 = -1e9;
  K.$$(svg, ':scope > *').forEach(function (e) { if (e === frame) return; var b = e.getBBox(); y0 = Math.min(y0, b.y); y1 = Math.max(y1, b.y + b.height); });
  y0 = Math.floor(y0 - 14); y1 = Math.ceil(y1 + 14);
  svg.setAttribute('viewBox', '0 ' + y0 + ' 820 ' + (y1 - y0));
  frame.setAttribute('y', y0 + 1); frame.setAttribute('height', y1 - y0 - 2);
}

xIn.addEventListener('input', function () { fillY(); fwd(); });
ySel.addEventListener('change', fwd);
vIn.addEventListener('input', fwd);
o2In.addEventListener('input', fwd);
exIn.addEventListener('change', function () {
  if (exIn.checked && CURRENT) { var V = parsePos(vIn.value) || 0; o2In.value = vol(V * CURRENT.a + Math.max(50, Math.round(V * CURRENT.a / 5 / 10) * 10)); }
  fwd();
});

/* ---- n ↔ V ---- */
var nIn = q('.cb-n'), vdIn = q('.cb-vd'), nvBox = q('.cb-nv'), nvSrc = 'n';
function nv() {
  if (!CURRENT) return;
  var x = CURRENT.x, y = CURRENT.y, a = CURRENT.a, FH = fmlH(x, y);
  var srcIn = nvSrc === 'n' ? nIn : vdIn, p = parsePos(srcIn.value);
  nIn.classList.remove('bad'); vdIn.classList.remove('bad');
  if (!p) { srcIn.classList.add('bad'); nvBox.innerHTML = '<span class="cb-err">「' + (srcIn.value || '空') + '」看不懂——要一个正数。</span>'; return; }
  var sf = sfOf(srcIn.value), n, L = [];
  if (nvSrc === 'n') { n = p; vdIn.value = fmt(n * 24.0, sf); L.push('<i>V</i>(' + FH + ') = <i>n</i> × 24.0 = ' + nIn.value.trim() + ' × 24.0 = <b>' + fmt(n * 24, sf) + ' dm³</b>'); }
  else { n = p / 24.0; nIn.value = fmt(n, sf); L.push('<i>n</i>(' + FH + ') = <i>V</i> ÷ 24.0 = ' + vdIn.value.trim() + ' ÷ 24.0 = <b>' + fmt(n, sf) + ' mol</b>'); }
  var ns = fmt(n, sf);
  L.push('<i>n</i>(CO₂) = ' + x + ' × ' + ns + ' = ' + fmt(n * x, sf) + ' mol → <i>V</i>(CO₂) = ' + fmt(n * x, sf) + ' × 24.0 = <b class="cb-vco2">' + fmt(n * x * 24, sf) + ' dm³</b>');
  L.push('<i>n</i>(O₂) = ' + vol(a) + ' × ' + ns + ' = ' + fmt(n * a, sf) + ' mol → <i>V</i>(O₂) = ' + fmt(n * a, sf) + ' × 24.0 = <b class="cb-vo2">' + fmt(n * a * 24, sf) + ' dm³</b>');
  nvBox.innerHTML = '<ol>' + L.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol><span class="faint">取 ' + sf + ' s.f.（跟你填的数据一样）；中间步骤用未四舍五入的值。</span>';
}
nIn.addEventListener('input', function () { nvSrc = 'n'; nv(); K.fit(); });
vdIn.addEventListener('input', function () { nvSrc = 'v'; nv(); K.fit(); });

/* ================= reverse ================= */
var rh = q('.cb-rh'), ro = q('.cb-ro'), rc = q('.cb-rc'), rwork = q('.cb-rwork');
function solveSteps(V, O, C, tag) {
  var x = C / V, o = O / V, nw = 2 * o - 2 * x, y = 2 * nw, L = [];
  L.push('体积比 = mole 比：C<sub><i>x</i></sub>H<sub><i>y</i></sub> : O₂ : CO₂ = ' + vol(V) + ' : ' + vol(O) + ' : ' + vol(C) + ' = <b>1 : ' + vol(o) + ' : ' + vol(x) + '</b>');
  L.push('C 守恒：1 个 C<sub><i>x</i></sub>H<sub><i>y</i></sub> 生成 <i>x</i> 个 CO₂ → <i>x</i> = ' + vol(C) + ' ÷ ' + vol(V) + ' = <b' + tag + 'x">' + vol(x) + '</b>');
  L.push('O 守恒：' + vol(o) + ' O₂ 提供 ' + vol(2 * o) + ' 个 O；' + vol(x) + ' CO₂ 用掉 ' + vol(2 * x) + ' 个 → 剩 ' + vol(nw) + ' 个 O 进 ' + vol(nw) + ' H₂O → <i>y</i> = 2 × ' + vol(nw) + ' = <b' + tag + 'y">' + vol(y) + '</b>');
  L.push('（同一件事用公式：<i>x</i> + <i>y</i>/4 = ' + vol(o) + ' → <i>y</i> = 4 × (' + vol(o) + ' − ' + vol(x) + ') = ' + vol(y) + '）');
  var ok = Math.abs(x - Math.round(x)) < 1e-6 && Math.abs(y - Math.round(y)) < 1e-6 && x >= 1 && x <= 8 && Math.round(y) % 2 === 0 && y >= ymin(Math.round(x)) && y <= 2 * x + 2;
  return { x: Math.round(x), y: Math.round(y), ok: ok, lines: L };
}
function rev() {
  var V = parsePos(rh.value), O = parsePos(ro.value), C = parsePos(rc.value);
  [[rh, V], [ro, O], [rc, C]].forEach(function (p) { p[0].classList.toggle('bad', !p[1]); });
  if (!V || !O || !C) { rwork.innerHTML = '<span class="cb-err">三个体积都要是正数。</span>'; K.fit(); return; }
  var r = solveSteps(V, O, C, ' class="cb-r');
  r.lines.push(r.ok ? '→ formula <b class="cb-rf">' + fmlH(r.x, r.y) + '</b>　检查：' + fmlH(r.x, r.y) + ' + ' + coef(r.x + r.y / 4) + 'O₂ → ' + (r.x > 1 ? r.x : '') + 'CO₂ + ' + (r.y / 2 > 1 ? r.y / 2 : '') + 'H₂O ✓'
                    : '<span class="cb-err cb-rf">得不到合理的 C<sub><i>x</i></sub>H<sub><i>y</i></sub>（<i>x</i> 要是 1–8 的整数，<i>y</i> 是偶数且 ≤ 2<i>x</i> + 2）——检查数据。</span>');
  rwork.innerHTML = '<ol>' + r.lines.map(function (l) { return '<li>' + l + '</li>'; }).join('') + '</ol>';
  K.fit();
}
[rh, ro, rc].forEach(function (e) { e.addEventListener('input', rev); });

/* ---- practice ---- */
var POOL = [[1, 4], [2, 2], [2, 4], [2, 6], [3, 4], [3, 6], [3, 8], [4, 6], [4, 8], [4, 10]];
var qBox = q('.cb-q'), lastQ = -1;
function pick(a) { return a[Math.floor(Math.random() * a.length)]; }
function newQ() {
  var i; do { i = Math.floor(Math.random() * POOL.length); } while (i === lastQ); lastQ = i;
  var x = POOL[i][0], y = POOL[i][1], a = x + y / 4, V = pick([10, 20, 30, 40, 50]);
  var need = V * a, co2 = V * x, html, ans;
  if (Math.random() < 0.5) {
    html = vol(V) + ' cm³ 某 gaseous hydrocarbon C<sub><i>x</i></sub>H<sub><i>y</i></sub> 恰好与 ' + vol(need) + ' cm³ O₂ 完全燃烧，生成 ' + vol(co2) +
      ' cm³ CO₂（体积都在 room temperature and pressure 下测，H₂O 是 liquid）。求 <i>x</i>、<i>y</i>。';
    ans = solveSteps(V, need, co2, ' class="cb-a').lines;
  } else {
    var T = need + pick([20, 30, 40, 50, 60]), left = T - need, P = co2 + left;
    html = vol(V) + ' cm³ 某 gaseous hydrocarbon C<sub><i>x</i></sub>H<sub><i>y</i></sub> 与 ' + vol(T) + ' cm³ O₂（excess）混合点燃，冷却回 room temperature 后 gas 剩 ' + vol(P) +
      ' cm³；再通过 NaOH(aq)（吸收 CO₂），剩 ' + vol(left) + ' cm³。求 <i>x</i>、<i>y</i>。';
    ans = ['NaOH(aq) 吸收的是 CO₂：<i>V</i>(CO₂) = ' + vol(P) + ' − ' + vol(left) + ' = ' + vol(co2) + ' cm³',
           '剩下的 ' + vol(left) + ' cm³ 是没用完的 O₂：<i>V</i>(O₂ reacted) = ' + vol(T) + ' − ' + vol(left) + ' = ' + vol(need) + ' cm³'].concat(solveSteps(V, need, co2, ' class="cb-a').lines);
  }
  qBox.innerHTML = '<div class="qt">' + html + '</div>' +
    '<div class="row"><label class="cb-lab"><span><i>x</i> =</span> <input type="text" class="fill cb-fx" data-ans="' + x + '"></label>' +
    '<label class="cb-lab"><span><i>y</i> =</span> <input type="text" class="fill cb-fy" data-ans="' + y + '"></label>' +
    '<span class="mut">（formula：C<sub><i>x</i></sub>H<sub><i>y</i></sub>）</span></div>' +
    '<button class="reveal">看解法</button><div class="ans"><ol>' + ans.map(function (l) { return '<li>' + l + '</li>'; }).join('') +
    '<li>→ formula <b>' + fmlH(x, y) + '</b></li></ol></div>';
  qBox.setAttribute('data-f', fml(x, y));
  K.init($root);
  K.fit();
}
q('.cb-new').addEventListener('click', newQ);

fillY(8); fwd(); rev(); newQ();

K.init($root);
})(document.getElementById("k-ch02-15"));</script>"""))
```

### C. Volumes and concentrations of solutions

**concentration** *c* = *n* / *V*，单位 mol dm<sup>−3</sup>；*V* 必须先从 cm<sup>3</sup> 换成 dm<sup>3</sup>（÷ 1000）。

- 250 cm<sup>3</sup> 溶液含 2.0 g NaOH：*n* = 2.0 / 40.0 = 0.050 mol，*c* = 0.050 / 0.250 = 0.20 mol dm<sup>−3</sup>
- 55 cm<sup>3</sup> 0.20 mol dm<sup>−3</sup> CuSO₄ 里 CuSO₄ 的 mass：0.20 × 55/1000 = 0.011 mol → × 159.6 = 1.8 g（2 s.f.）

**titration** 三步：

1. 已知 *c* 和 *V* 的那个 → mol；
2. 用 equation 的 mole ratio 换成另一个的 mol；
3. ÷ 它的 *V*（dm<sup>3</sup>）→ *c*。

例：25.0 cm<sup>3</sup> NaOH(aq) 恰好被 15.10 cm<sup>3</sup> 0.200 mol dm<sup>−3</sup> H₂SO₄ 中和：2NaOH + H₂SO₄ → Na₂SO₄ + 2H₂O。

*n*(H₂SO₄) = 0.200 × 15.10/1000 = 0.00302 mol → *n*(NaOH) = 2 × 0.00302 = 0.00604 mol → *c* = 0.00604 / 0.0250 = **0.242 mol dm<sup>−3</sup>**

**dilution factor**：真题最爱「先把样品配成 *V*₁、取 *V*₂ 去滴定」。滴定只测了取出来那一份，算回原样品要 × *V*₁ / *V*₂（9701/21 May/June 2025 Q4 的 × 200/5、9701/22 May/June 2025 Q5 的 × 100.0/25.0 = × 4）。

```python
# @title 🔬 titration 计算（含稀释） { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-17">
<h3>titration 计算：从 burette 往回推到原样品</h3>
<p class="mut">选一道题，或者直接改下面任何一个数——流程图上的 mol 数和下面每一步的算式会即时更新。蓝色箭头就是往回推的方向：burette → conical flask → (× dilution factor) → 原样品。</p>
<div class="row tt-pre"></div>
<div class="tt-eq card"></div>
<div class="tt-form">
  <div class="tt-grp"><span class="tt-h">burette</span>
    <label><i>c</i> = <input type="text" class="tt-v" data-k="ct" spellcheck="false"> mol dm⁻³</label>
    <label>titre = <input type="text" class="tt-v" data-k="vt" spellcheck="false"> cm³</label></div>
  <div class="tt-grp"><span class="tt-h">mole ratio</span>
    <label><span class="tt-an"></span> : <span class="tt-ti"></span> = <input type="text" class="tt-v tt-int" data-k="ra" spellcheck="false"> : <input type="text" class="tt-v tt-int" data-k="rb" spellcheck="false"></label></div>
  <div class="tt-grp"><span class="tt-h">dilution</span>
    <label class="tt-chk"><input type="checkbox" class="tt-dil"> 先定容</label>
    <label class="tt-d1"><span>make up to <i>V</i>₁ =</span> <input type="text" class="tt-v" data-k="v1" spellcheck="false"> cm³</label>
    <label>pipette <i>V</i>₂ = <input type="text" class="tt-v" data-k="v2" spellcheck="false"> cm³</label></div>
  <div class="tt-grp"><span class="tt-h">要求</span>
    <select class="tt-tg">
      <option value="conc">concentration of original solution</option>
      <option value="mass">mass in original sample</option>
      <option value="pct">percentage by mass</option>
    </select>
    <label class="tt-f-v0"><span>original volume <i>V</i>₀ =</span> <input type="text" class="tt-v" data-k="v0" spellcheck="false"> cm³</label>
    <label class="tt-f-mr"><span><i>M</i><sub>r</sub> =</span> <input type="text" class="tt-v" data-k="mr" spellcheck="false"></label>
    <label class="tt-f-ms"><span>sample mass =</span> <input type="text" class="tt-v" data-k="ms" spellcheck="false"> g</label></div>
</div>
<svg class="tt-map" viewBox="0 0 780 548">
  <defs>
    <marker id="tt-a-back" viewBox="0 0 10 10" refX="9" refY="5" markerWidth="4.4" markerHeight="4.4" orient="auto"><path d="M0,0 L10,5 L0,10 z" style="fill:var(--acc)"/></marker>
    <marker id="tt-a-fwd" viewBox="0 0 10 10" refX="9" refY="5" markerWidth="5.5" markerHeight="5.5" orient="auto"><path d="M0,0 L10,5 L0,10 z" style="fill:var(--faint)"/></marker>
  </defs>
  <rect x="0" y="0" width="780" height="548" rx="10" class="tt-bgr"/>
  <g class="tt-art"></g>
  <g class="tt-wires"></g>
  <g class="tt-labels"></g>
</svg>
<div class="row tt-ctl">
  <button type="button" class="tt-next">下一步 ▸</button>
  <button type="button" class="tt-all">全部展开</button>
  <button type="button" class="tt-reset">从第 1 步重来</button>
  <span class="faint tt-prog"></span>
</div>
<div class="tt-work card"></div>
<style>
.k .tt-pre button{font-size:14px;padding:3px 10px}
.k .tt-eq{margin:6px 0;font-size:15px}
.k .tt-eq .src{color:var(--faint);font-size:13px}
.k .tt-form{display:grid;grid-template-columns:repeat(auto-fit,minmax(300px,1fr));gap:8px;margin:8px 0}
.k .tt-grp{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:6px 10px;display:flex;flex-wrap:wrap;gap:4px 12px;align-items:center;font-size:14px}
.k .tt-h{font-weight:650;width:100%;color:var(--mut);font-size:13px}
.k .tt-grp label{white-space:nowrap}
.k .tt-v{width:7.4em;text-align:center}
.k .tt-v.tt-int{width:2.8em}
.k .tt-v.bad{border-color:var(--bad);background:var(--bad-soft)}
.k .tt-grp select{max-width:100%;font-size:14px}
.k .tt-off{opacity:.45}
.k .tt-map{margin:8px 0;user-select:none}
.k .tt-bgr{fill:var(--card);stroke:var(--line)}
.k .tt-art .gl{fill:none;stroke:var(--mut);stroke-width:2;stroke-linejoin:round}
.k .tt-art .tick{stroke:var(--mut);stroke-width:1.2}
.k .tt-art .dim{opacity:.3}
.k .tt-wires path{fill:none}
.k .tt-wires path.fwd{stroke:var(--faint);stroke-width:2;stroke-dasharray:6 4}
.k .tt-wires path.back{stroke:var(--acc);stroke-width:3}
.k .tt-labels text{font-size:15px;fill:var(--fg)}
.k .tt-labels text.nm{fill:var(--mut)}
.k .tt-labels text.mol{fill:var(--acc);font-weight:650}
.k .tt-labels text.op{fill:var(--acc);font-weight:650}
.k .tt-labels text.fw{fill:var(--faint)}
.k .tt-labels text.dim{opacity:.4}
.k .tt-map .hid{visibility:hidden}
.k .tt-work ol{margin:4px 0;padding-left:24px}
.k .tt-work li{margin:5px 0}
.k .tt-work li.hid{display:none}
.k .tt-work .dil{background:var(--hl);border-radius:4px;padding:0 4px}
.k .tt-work .err{color:var(--bad)}
.k .tt-work .fin{font-size:16px}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

/* ---------- presets（数值是字符串：显示时保留题目原样的位数） ---------- */
var PRE = [
  { id: 'naoh', btn: '① NaOH vs H₂SO₄', eq: '2NaOH + H₂SO₄ → Na₂SO₄ + 2H₂O', src: '正文例子',
    an: 'NaOH', ti: 'H₂SO₄', tiAq: 'H₂SO₄(aq)', icon: 'beaker', sample: 'NaOH(aq)', hue: 'rgba(90,150,255,.28)',
    v: { ct: '0.200', vt: '15.10', ra: '2', rb: '1', v1: '250.0', v2: '25.0', v0: '25.0', mr: '40.0', ms: '1.00' }, dil: false, tg: 'conc' },
  { id: 'tab', btn: '② vitamin C tablet', eq: 'C₆H₈O₆ + I₂ → C₆H₆O₆ + 2HI（vitamin C : I₂ = 1 : 1）', src: '9701/21 May/June 2025 Q4(b)',
    an: 'vitamin C', ti: 'I₂', tiAq: 'I₂(aq)', icon: 'tablet', sample: 'vitamin C tablet', hue: 'rgba(180,110,30,.45)',
    v: { ct: '5.00 × 10⁻⁴', vt: '28.40', ra: '1', rb: '1', v1: '200.0', v2: '5.00', v0: '200.0', mr: '176', ms: '1.00' }, dil: true, tg: 'mass',
    note: 'flask 里另加的 ~150 cm³ 水是干扰项：它不改变 flask 里 vitamin C 的 mol 数。mark scheme 的答案写 0.099968（3 s.f. 即 0.100 g），也接受先 × 176 再 × 200/5 的顺序。' },
  { id: 'lemon', btn: '③ lemon', eq: 'C₆H₈O₆ + I₂ → C₆H₆O₆ + 2HI（vitamin C : I₂ = 1 : 1）', src: '9701/22 May/June 2025 Q5(b)',
    an: 'vitamin C', ti: 'I₂', tiAq: 'I₂(aq)', icon: 'lemon', sample: 'lemon', hue: 'rgba(180,110,30,.45)',
    v: { ct: '5.00 × 10⁻³', vt: '36.65', ra: '1', rb: '1', v1: '100.0', v2: '25.0', v0: '100.0', mr: '176', ms: '150.0' }, dil: true, tg: 'pct',
    note: 'mark scheme：0.086 % 或 0.087 % 都给分（0.129 g 取整后再除会得到 0.086 %）。' },
  { id: 'mn', btn: '④ Mn²⁺ with NaBiO₃', eq: '2Mn²⁺ + 5BiO₃⁻ + 14H⁺ → 2MnO₄⁻ + 5Bi³⁺ + 7H₂O', src: '9701/22 Feb/Mar 2024 Q1(e)',
    an: 'Mn²⁺', ti: 'BiO₃⁻', tiAq: 'NaBiO₃(aq)', icon: 'beaker', sample: 'Mn²⁺(aq)', hue: 'rgba(200,120,40,.35)',
    v: { ct: '0.100', vt: '21.50', ra: '2', rb: '5', v1: '1000', v2: '25.00', v0: '100.0', mr: '54.9', ms: '1.00' }, dil: true, tg: 'conc',
    note: '1.00 dm³ 写成 1000 cm³；dilution factor = 1000 / 25.00 = 40。' }
];

/* ---------- numbers ---------- */
var SUPD = { '⁰': '0', '¹': '1', '²': '2', '³': '3', '⁴': '4', '⁵': '5', '⁶': '6', '⁷': '7', '⁸': '8', '⁹': '9', '⁻': '-', '⁺': '+' };
function parseNum(raw) {
  var s = String(raw).trim().replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹⁻⁺]+/g, function (m) { return '^' + m.split('').map(function (c) { return SUPD[c]; }).join(''); })
    .replace(/[−–]/g, '-').replace(/\s+/g, '').replace(/,/g, '');
  var m = s.match(/^(\d*\.?\d*)(?:[x×*X]10(?:\^|\*\*)?([-+]?\d+)|[eE]([-+]?\d+))?$/);
  if (!m || !/\d/.test(m[1])) return null;
  var man = m[1], ex = m[2] || m[3] || '0', v = parseFloat(man) * Math.pow(10, +ex);
  var dg = man.replace('.', '').replace(/^0+/, ''), sf = man.indexOf('.') >= 0 ? dg.length : dg.replace(/0+$/, '').length || 1;
  if (man.indexOf('.') < 0 && /0$/.test(man)) sf = Math.max(sf, Math.min(3, dg.length));   /* '250' → 3 s.f. */
  return { v: v, sf: sf, plain: /^\d*\.?\d*$/.test(s) };
}
function fmt(x, sf) {
  if (!isFinite(x)) return '—';
  if (x === 0) return '0';
  var r = +x.toPrecision(sf), e = Math.floor(Math.log10(Math.abs(r)));
  if (e >= -3 && e < 5 && sf - 1 - e >= 0) return r.toFixed(sf - 1 - e);
  if (e >= -3 && e < 5) return String(Math.round(r));
  return (r / Math.pow(10, e)).toFixed(sf - 1) + ' × 10' + K.sup(e);
}
/* unrounded answer as a mark scheme prints it (5 s.f., trailing zeros dropped) */
function full(x) { var e = Math.floor(Math.log10(Math.abs(x))); return e >= -3 && e < 5 ? String(+x.toPrecision(5)) : fmt(x, 5); }
function fin(x, sf, unit) { var r = fmt(x, sf), f = full(x); return (f !== r ? f + ' → ' : '') + '<b class="fin">' + r + unit + '</b>（' + sf + ' s.f.）'; }
function trim(x) { return String(+x.toPrecision(6)); }
/* '100.0' cm³ → '0.1000' dm³: move the point, keep every written digit */
function toDm3(raw, p) {
  var s = raw.trim();
  if (!p.plain) return fmt(p.v / 1000, Math.max(p.sf, 2));
  if (s.indexOf('.') < 0) s += '.';
  var i = s.indexOf('.'), d = s.replace('.', ''), j = i - 3, out;
  out = j <= 0 ? '0.' + new Array(1 - j).join('0') + d : d.slice(0, j) + '.' + d.slice(j);
  out = out.replace(/^0+(?=\d)/, '');
  return out.replace(/\.$/, '');
}

/* ---------- DOM ---------- */
var svg = $root.querySelector('.tt-map'), gArt = $root.querySelector('.tt-art'), gW = $root.querySelector('.tt-wires'), gL = $root.querySelector('.tt-labels');
var work = $root.querySelector('.tt-work'), eqBox = $root.querySelector('.tt-eq'), dilBox = $root.querySelector('.tt-dil'), tgSel = $root.querySelector('.tt-tg');
var inp = {}; K.$$($root, 'input.tt-v').forEach(function (e) { inp[e.getAttribute('data-k')] = e; });
var P = PRE[0], shown = Infinity, nSteps = 0;

var preBox = $root.querySelector('.tt-pre');
PRE.forEach(function (p) {
  var b = K.el('button', { type: 'button', 'data-pre': p.id }, p.btn);
  b.addEventListener('click', function () { load(p); });
  preBox.appendChild(b);
});

/* ---------- geometry ---------- */
var W = 780, XO = 85, XV = 270, XP = 430, XC = 600, YF = 322;
var RA = 400, RB = 427, RC1 = 478, RC2 = 505, RC3 = 532;

function art(tag, a, cls) { var e = K.svg(tag, a); if (cls) e.setAttribute('class', cls); gArt.appendChild(e); return e; }
function drawArt(vt, dil) {
  gArt.innerHTML = '';
  /* original sample */
  var o = [];
  if (P.icon === 'beaker') {
    art('path', { d: 'M' + (XO - 38) + ',' + (YF - 12) + ' L' + (XO - 38) + ',' + (YF + 48) + ' L' + (XO + 38) + ',' + (YF + 48) + ' L' + (XO + 38) + ',' + (YF - 12) + ' L' + (XO + 38) + ',' + (YF - 30), style: 'fill:none' }, 'gl');
    art('rect', { x: XO - 37, y: YF + 6, width: 74, height: 41, style: 'fill:' + (P.id === 'mn' ? 'rgba(230,140,170,.35)' : 'rgba(120,170,255,.22)') });
    art('path', { d: 'M' + (XO - 38) + ',' + (YF - 30) + ' L' + (XO - 38) + ',' + (YF + 48) + ' L' + (XO + 38) + ',' + (YF + 48) + ' L' + (XO + 38) + ',' + (YF - 30) }, 'gl');
    o = [XO - 38, YF - 30, 76, 78];
  } else if (P.icon === 'tablet') {
    art('ellipse', { cx: XO, cy: YF + 4, rx: 36, ry: 24, style: 'fill:rgba(240,240,235,.9)' }, 'gl');
    art('line', { x1: XO - 30, y1: YF + 4, x2: XO + 30, y2: YF + 4 }, 'tick');
    o = [XO - 36, YF - 20, 72, 48];
  } else {
    art('ellipse', { cx: XO, cy: YF + 4, rx: 42, ry: 30, style: 'fill:rgba(245,215,40,.85)' }, 'gl');
    art('path', { d: 'M' + (XO - 42) + ',' + (YF + 4) + ' l-8,-3 l1,6 z M' + (XO + 42) + ',' + (YF + 4) + ' l8,-3 l-1,6 z', style: 'fill:rgba(245,215,40,.85)' }, 'gl');
    o = [XO - 50, YF - 26, 100, 60];
  }
  /* volumetric flask */
  /* no dilution → no volumetric flask in the picture (its name stays, dimmed) */
  var g = dil ? art('g', { 'class': 'vf' }) : K.svg('g');
  function sub(tag, a, cls) { var e = K.svg(tag, a); if (cls) e.setAttribute('class', cls); g.appendChild(e); return e; }
  var NJ = YF + 20 - 37;   /* neck meets the bulb (centre YF+20, r 38) */
  sub('path', { d: 'M' + (XV - 9) + ',' + NJ + ' L' + (XV - 9) + ',220 M' + (XV + 9) + ',' + NJ + ' L' + (XV + 9) + ',220', style: 'fill:none' }, 'gl');
  sub('path', { d: 'M' + (XV - 9) + ',' + NJ + ' A38,38 0 1,0 ' + (XV + 9) + ',' + NJ, style: 'fill:rgba(120,170,255,.2)' }, 'gl');
  sub('rect', { x: XV - 8, y: 240, width: 16, height: NJ - 240, style: 'fill:rgba(120,170,255,.2)' });
  sub('line', { x1: XV - 14, y1: 240, x2: XV + 14, y2: 240 }, 'tick');
  /* pipette */
  art('path', { d: 'M' + (XP - 3) + ',206 L' + (XP - 3) + ',' + (YF - 26) + ' M' + (XP + 3) + ',206 L' + (XP + 3) + ',' + (YF - 26) +
    ' M' + (XP - 3) + ',' + (YF + 26) + ' L' + (XP - 3) + ',' + (YF + 40) + ' L' + XP + ',' + (YF + 48) + ' L' + (XP + 3) + ',' + (YF + 40) + ' L' + (XP + 3) + ',' + (YF + 26) }, 'gl');
  art('ellipse', { cx: XP, cy: YF, rx: 11, ry: 26, style: 'fill:rgba(120,170,255,.2)' }, 'gl');
  art('line', { x1: XP - 10, y1: 228, x2: XP + 10, y2: 228 }, 'tick');
  /* conical flask */
  var cf = 'M' + (XC - 16) + ',266 L' + (XC - 16) + ',300 L' + (XC - 62) + ',372 L' + (XC + 62) + ',372 L' + (XC + 16) + ',300 L' + (XC + 16) + ',266';
  art('path', { d: 'M' + (XC - 39) + ',336 L' + (XC - 62) + ',372 L' + (XC + 62) + ',372 L' + (XC + 39) + ',336 z', style: 'fill:rgba(225,90,150,.25)' });
  art('path', { d: cf, style: 'fill:none' }, 'gl');
  /* burette: 0 cm³ at the top, liquid level at the titre reading */
  var top = 30, bot = 210, lev = top + Math.max(0, Math.min(50, vt || 0)) / 50 * (bot - top);
  art('rect', { x: XC - 10, y: lev, width: 20, height: bot - lev, style: 'fill:' + P.hue });
  art('rect', { x: XC - 11, y: 22, width: 22, height: bot - 22 + 2, style: 'fill:none' }, 'gl');
  for (var t = 0; t <= 50; t += 5) art('line', { x1: XC - 11, y1: top + t / 50 * (bot - top), x2: XC - (t % 10 ? 5 : 1), y2: top + t / 50 * (bot - top) }, 'tick');
  art('path', { d: 'M' + (XC - 20) + ',216 L' + (XC + 20) + ',216', style: 'stroke-width:4' }, 'gl');
  art('path', { d: 'M' + (XC - 5) + ',212 L' + (XC - 2) + ',236 L' + (XC + 2) + ',236 L' + (XC + 5) + ',212' }, 'gl');
  /* opaque-ish things for label placement */
  return [
    { x: o[0], y: o[1], w: o[2], h: o[3] },
    dil ? { x: XV - 39, y: 218, w: 78, h: YF + 59 - 218 } : { x: -99, y: -99, w: 1, h: 1 },
    { x: XP - 12, y: 204, w: 24, h: YF + 50 - 204 },
    { x: XC - 63, y: 264, w: 126, h: 110 },
    { x: XC - 21, y: 20, w: 42, h: 218 }
  ];
}

/* label text → italic quantity symbols (n V c m M) after placement; same glyphs, so the box only shrinks */
function ital(t) {
  var s = t.textContent.replace(/[&<>]/g, function (c) { return { '&': '&amp;', '<': '&lt;', '>': '&gt;' }[c]; });
  t.innerHTML = s.replace(/(^|[^A-Za-z0-9])([nVcmM])(?=[^A-Za-z0-9]|$)/g, '$1<tspan font-style="italic">$2</tspan>');
}
var placed, avoid;
function lab(x, y, text, cls, cands, step) {
  var t = K.label(gL, x, y, text, { cls: cls, cands: cands, placed: placed, avoid: avoid, pad: 3 });
  ital(t);
  if (step) t.setAttribute('data-step', step);
  return t;
}
var MID = [[0, 0, 'middle'], [0, 10, 'middle'], [0, -10, 'middle']];
function bb(t) { var b = t.getBBox(); return { x: b.x, y: b.y, w: b.width, h: b.height }; }
function wire(pts, cls, step, id) {
  var d = pts.map(function (p, i) { return (i ? 'L' : 'M') + p[0].toFixed(1) + ',' + p[1].toFixed(1); }).join(' ');
  var e = K.svg('path', { d: d, 'class': cls, 'marker-end': cls === 'back' ? 'url(#tt-a-back)' : 'url(#tt-a-fwd)' });
  if (step) e.setAttribute('data-step', step);
  if (id) e.setAttribute('data-w', id);
  gW.appendChild(e);
  K.segs(pts).forEach(function (s) { avoid.push(s); });
  return e;
}

/* ---------- compute + draw ---------- */
function val(k) { var p = parseNum(inp[k].value); if (!p || !(p.v > 0)) return null; return p; }
/* echo a datum as written ('5.00 × 10⁻³', '50.00'); only e-notation / '^' input is re-typeset */
function raw(k) { var t = inp[k].value.trim(), p = parseNum(t); return p && !p.plain && !/×\s*10[⁻⁰-⁹]/.test(t) ? fmt(p.v, Math.max(2, p.sf)) : t; }

function calc() {
  var dil = dilBox.checked, tg = tgSel.value;
  $root.querySelector('.tt-d1').classList.toggle('tt-off', !dil);
  inp.v1.disabled = !dil;
  $root.querySelector('.tt-f-v0').style.display = tg === 'conc' && dil ? '' : 'none';
  $root.querySelector('.tt-f-mr').style.display = tg !== 'conc' ? '' : 'none';
  $root.querySelector('.tt-f-ms').style.display = tg === 'pct' ? '' : 'none';
  $root.querySelector('.tt-an').textContent = P.an; $root.querySelector('.tt-ti').textContent = P.ti;
  K.$$($root, '.tt-pre button').forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-pre') === P.id); });
  eqBox.innerHTML = P.eq + '　<span class="src">' + P.src + '</span>' + (P.note ? '<div class="faint">' + P.note + '</div>' : '');

  var need = ['ct', 'vt', 'ra', 'rb', 'v2'];
  if (dil) need.push('v1');
  if (tg === 'conc' && dil) need.push('v0');
  if (tg !== 'conc') need.push('mr');
  if (tg === 'pct') need.push('ms');
  var p = {}, bad = [];
  Object.keys(inp).forEach(function (k) { inp[k].classList.remove('bad'); });
  need.forEach(function (k) { p[k] = val(k); if (!p[k]) { bad.push(k); inp[k].classList.add('bad'); } });
  if (!bad.length && dil && p.v1.v < p.v2.v) { bad.push('v1'); inp.v1.classList.add('bad'); }

  /* s.f. of the answer = fewest s.f. among measured data (ratio is exact) */
  var sf = 3;
  if (!bad.length) {
    sf = Math.min.apply(null, need.filter(function (k) { return k !== 'ra' && k !== 'rb'; }).map(function (k) { return p[k].sf; }));
    sf = Math.max(2, Math.min(4, sf));
  }
  var R = {}, steps = [];
  if (!bad.length) {
    var nt = p.ct.v * p.vt.v / 1000, a = p.ra.v, b = p.rb.v, na = nt * a / b, fac = dil ? p.v1.v / p.v2.v : 1, nT = na * fac;
    R.nt = fmt(nt, sf); R.na = fmt(na, sf); R.nT = fmt(nT, sf);
    var rat = raw('ra') + '/' + raw('rb');
    steps.push({ k: 'nt', h: '<i>n</i>(' + P.ti + ') = <i>c</i> × <i>V</i> ÷ 1000 = ' + raw('ct') + ' × ' + raw('vt') + ' ÷ 1000 = <b>' + R.nt + ' mol</b>' });
    steps.push({ k: 'na', h: 'mole ratio ' + P.an + ' : ' + P.ti + ' = ' + raw('ra') + ' : ' + raw('rb') + ' → <i>n</i>(' + P.an + ') in conical flask = ' + R.nt + ' × ' + rat + ' = <b>' + R.na + ' mol</b>' });
    if (dil) {
      R.fac = trim(fac);
      steps.push({ k: 'nT', h: '<span class="dil">dilution factor = <i>V</i>₁ / <i>V</i>₂ = ' + raw('v1') + ' / ' + raw('v2') + ' = ' + R.fac + '</span>：pipette 只取了 ' + raw('v2') + ' cm³，整个 ' + raw('v1') + ' cm³ 里有 <i>n</i>(' + P.an + ') = ' + R.na + ' × ' + raw('v1') + '/' + raw('v2') + ' = <b>' + R.nT + ' mol</b>' });
    }
    var nn = dil ? R.nT : R.na;
    if (tg === 'conc') {
      var vk = dil ? 'v0' : 'v2', vd = toDm3(inp[vk].value, p[vk]), c = nT / (p[vk].v / 1000);
      R.fin = fmt(c, sf) + ' mol dm⁻³'; R.op = '÷ ' + vd + ' dm³';
      steps.push({ k: 'fin', h: (dil ? '这些 ' + P.an + ' 全部来自 ' + raw('v0') + ' cm³ original solution：' : '') + '<i>c</i> = <i>n</i> ÷ <i>V</i> = ' + nn + ' ÷ ' + vd + ' = ' + fin(c, sf, ' mol dm⁻³') });
    } else {
      var m = nT * p.mr.v;
      R.fin = fmt(m, sf) + ' g'; R.op = '× ' + raw('mr');
      steps.push({ k: 'fin', h: '<i>m</i> = <i>n</i> × <i>M</i> = ' + nn + ' × ' + raw('mr') + ' = ' + fin(m, sf, ' g') });
      if (tg === 'pct') {
        R.pct = fmt(m / p.ms.v * 100, sf);
        steps.push({ k: 'pct', h: 'percentage by mass = ' + fmt(m, sf) + ' ÷ ' + raw('ms') + ' × 100 = ' + fin(m / p.ms.v * 100, sf, ' %') });
      }
    }
  }
  nSteps = steps.length;
  var kOf = {}; steps.forEach(function (s, i) { kOf[s.k] = i + 1; });

  /* ---------- diagram ---------- */
  gW.innerHTML = ''; gL.innerHTML = '';
  placed = drawArt(p.vt ? p.vt.v : parseFloat(inp.vt.value), dil).slice(); avoid = [];
  var tiTxt = P.tiAq + ', ' + raw('ct') + ' mol dm⁻³';
  var bur = [[-26, 0, 'end']];
  lab(XC, 46, 'burette', 'nm', bur);
  lab(XC, 73, tiTxt, '', bur);
  lab(XC, 100, 'titre = ' + raw('vt') + ' cm³', '', bur);
  var sampleTxt = tg === 'pct' ? raw('ms') + ' g ' + P.sample : tg === 'conc' && dil ? raw('v0') + ' cm³ ' + P.sample : P.sample;
  lab(XO, RA, 'original sample', 'nm', MID); lab(XO, RB, sampleTxt, '', MID);
  lab(XV, RA, 'volumetric flask', 'nm' + (dil ? '' : ' dim'), MID); lab(XV, RB, dil ? 'V₁ = ' + raw('v1') + ' cm³' : '（不稀释）', dil ? '' : 'dim', MID);
  lab(XP, RA, 'pipette', 'nm', MID); lab(XP, RB, 'V₂ = ' + raw('v2') + ' cm³', '', MID);
  lab(XC, RA, 'conical flask', 'nm', MID); lab(XC, RB, '+ indicator', 'nm', MID);

  /* legend in the empty top-left corner */
  wire([[22, 44], [70, 44]], 'fwd', 0, 'lg1'); lab(78, 44, '实验步骤（往前做）', 'fw', [[0, 5, 'start']]);
  wire([[22, 72], [70, 72]], 'back', 0, 'lg2'); lab(78, 72, '往回推 mol（计算方向）', 'op', [[0, 5, 'start']]);
  /* forward (experiment) arrows */
  var fw = [];
  var oR = placed[0].x + placed[0].w;
  if (dil) {
    fw.push(wire([[oR + 8, YF], [XV - 33 - 9, YF]], 'fwd', 0, 'f1'));
    fw.push(wire([[XV + 33 + 8, YF], [XP - 3 - 9, YF]], 'fwd', 0, 'f2'));
  } else fw.push(wire([[oR + 8, YF], [XP - 3 - 9, YF]], 'fwd', 0, 'f2'));
  fw.push(wire([[XP + 3 + 8, YF], [XC - 29 - 9, YF]], 'fwd', 0, 'f3'));
  var UP = [[0, -10, 'middle'], [0, -22, 'middle'], [0, 20, 'middle']];
  if (dil) lab((oR + XV - 33) / 2, YF, 'make up', 'fw', UP);
  if (dil) lab((XV + 33 + XP) / 2, YF, 'withdraw', 'fw', UP);
  else lab((oR + XP) / 2, YF, 'pipette', 'fw', UP);
  lab((XP + XC - 29) / 2, YF, 'run in', 'fw', UP);

  if (!bad.length) {
    /* burette n */
    var ntA = lab(XC, 112, 'n(' + P.ti + ')', 'mol', [[24, 0, 'start']], kOf.nt);
    var ntB = lab(XC, 139, '= ' + R.nt + ' mol', 'mol', [[24, 0, 'start']], kOf.nt);
    var cA = lab(XC, RC1, 'n(' + P.an + ')', 'mol', MID, kOf.na), cB = lab(XC, RC2, '= ' + R.na + ' mol', 'mol', MID, kOf.na);
    var nb = bb(ntB), xr = Math.min(766, nb.x + nb.w - 8), ca = bb(cA);
    var ry = ca.y + ca.h / 2;
    wire([[xr, nb.y + nb.h + 8], [xr, ry], [ca.x + ca.w + 9, ry]], 'back', kOf.na, 'ratio');
    lab(xr, (nb.y + nb.h + ry) / 2, '× ' + raw('ra') + '/' + raw('rb'), 'op', [[-8, 5, 'end'], [-8, 30, 'end'], [-8, -20, 'end']], kOf.na);
    var leftOf = ca, fromStep;
    if (dil) {
      var vA = lab(XV, RC1, 'n(' + P.an + ')', 'mol', MID, kOf.nT); lab(XV, RC2, '= ' + R.nT + ' mol', 'mol', MID, kOf.nT);
      var va = bb(vA);
      wire([[ca.x - 8, ry], [va.x + va.w + 9, ry]], 'back', kOf.nT, 'dil');
      lab((ca.x + va.x + va.w) / 2, ry, '× ' + raw('v1') + '/' + raw('v2') + ' (= ' + R.fac + ')', 'op', [[0, -14, 'middle'], [0, -26, 'middle']], kOf.nT);
      leftOf = va;
    }
    var finName = tg === 'conc' ? 'c(' + P.an + ')' : 'm(' + P.an + ')';
    var oA = lab(XO, RC1, finName, 'mol', MID, kOf.fin); lab(XO, RC2, '= ' + R.fin, 'mol', MID, kOf.fin);
    var oa = bb(oA);
    wire([[leftOf.x - 8, ry], [oa.x + oa.w + 9, ry]], 'back', kOf.fin, 'fin');
    lab((leftOf.x + oa.x + oa.w) / 2, ry, R.op, 'op', [[0, -14, 'middle'], [0, -26, 'middle'], [0, 46, 'middle']], kOf.fin);
    if (tg === 'pct') lab(XO, RC3, '= ' + R.pct + ' % by mass', 'mol', [[0, 0, 'middle'], [6, 0, 'start']], kOf.pct);
  }

  /* ---------- working ---------- */
  if (bad.length) {
    work.innerHTML = '<p class="err">' + (dil && p.v1 && p.v2 && p.v1.v < p.v2.v ? '<i>V</i>₁ 要比 <i>V</i>₂ 大：pipette 是从定容后的溶液里取一部分。' :
      '标红的格子看不懂——要一个正数（可以写 5.00 × 10⁻⁴、5.00e-4 或 5.00 x 10^-4）。') + '</p>';
  } else {
    work.innerHTML = '<ol>' + steps.map(function (s, i) { return '<li data-step="' + (i + 1) + '" data-k="' + s.k + '">' + s.h + '</li>'; }).join('') + '</ol>' +
      '<p class="faint">答案取 ' + sf + ' s.f.（数据里最少的 s.f.；mole ratio 是精确数）。中间步骤显示的是四舍五入后的值，计算时用的是完整数值。</p>';
  }
  applySteps();
  K.fit();
}

function applySteps() {
  var s = Math.min(shown, nSteps);
  K.$$($root, '[data-step]').forEach(function (e) { e.classList.toggle('hid', +e.getAttribute('data-step') > s); });
  $root.querySelector('.tt-prog').textContent = '已展开 ' + s + ' / ' + nSteps + ' 步';
  $root.querySelector('.tt-next').disabled = s >= nSteps;
}

function load(p) {
  P = p;
  Object.keys(p.v).forEach(function (k) { inp[k].value = p.v[k]; });
  dilBox.checked = p.dil; tgSel.value = p.tg;
  if (shown !== Infinity) shown = 0;
  calc();
}

Object.keys(inp).forEach(function (k) { inp[k].addEventListener('input', calc); });
dilBox.addEventListener('change', calc);
tgSel.addEventListener('change', calc);
$root.querySelector('.tt-next').addEventListener('click', function () { shown = Math.min(shown, nSteps) + 1; applySteps(); K.fit(); });
$root.querySelector('.tt-all').addEventListener('click', function () { shown = Infinity; applySteps(); K.fit(); });
$root.querySelector('.tt-reset').addEventListener('click', function () { shown = 0; applySteps(); K.fit(); });
load(PRE[1]);

K.init($root);
})(document.getElementById("k-ch02-17"));</script>"""))
```

### D. Deduce stoichiometric relationships

反过来用：实验测出两种物质反应的 mol 数 → 化成最简整数比 → 就是 equation 里的系数。

例：0.0012 mol W（C₅H₁₂O₄）恰好被 0.0020 mol acidified K₂Cr₂O₇ 完全氧化（9701/22 May/June 2026 Q4(c)(i)）。

0.0012 : 0.0020 = 3 : 5 → 3C₅H₁₂O₄ + **5**Cr₂O₇<sup>2−</sup> + 40H<sup>+</sup> → 3C₅H₆O₆ + 10Cr<sup>3+</sup> + 29H₂O

（M1 就是写对 5Cr₂O₇<sup>2−</sup>；之后 Cr、O、H 依次配平，最后用 charge 检查：左边 −10 + 40 = +30，右边 10 × (+3) = +30。）

> - (b) volumes of gases (e.g. in the burning of hydrocarbons)
> - (c) volumes and concentrations of solutions
> - (d) limiting reagent and excess reagent
>
> > When performing calculations, candidates' answers should reflect the number of significant figures given or asked for in the question. When rounding up or down, candidates should ensure that significant figures are neither lost unnecessarily nor used beyond what is justified (see also Mathematical requirements section).
>
> - (e) deduce stoichiometric relationships from calculations such as those in 2.4.1(a)–(d)

```python
# @title 📝 9701/22 Feb/Mar 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-20">
<div class="k-ptable"></div>
<h3>9701/22 Feb/Mar 2024</h3>
<p>Bismuth is an element in Group 15 of the Periodic Table.</p>
<div class="q"><h4>Q1(b)(i) <span class="mark">[1]</span></h4>
<p>Bismuth reduces water to form bismuth oxide, Bi₂O₃. A colourless gas that ignites with a squeaky pop also forms. Construct an equation for the reduction of water by bismuth.</p>
<button class="reveal">答案</button><div class="ans">
<p>2Bi + 3H₂O → Bi₂O₃ + 3H₂</p>
</div>
</div>
<div class="q"><h4>Q1(e) <span class="mark">[3]</span></h4>
<p>NaBiO₃ can be used to determine the concentration of Mn²⁺(aq):</p>
<p>equation 2　2Mn²⁺ + 5BiO₃⁻ + 14H⁺ → 2MnO₄⁻ + 5Bi³⁺ + 7H₂O</p>
<p>A student dilutes 100.0 cm³ of saturated Mn²⁺(aq) to 1.00 dm³. 25.00 cm³ of the diluted solution is titrated with 0.100 mol dm⁻³ NaBiO₃(aq); 21.50 cm³ is needed. Calculate the concentration, in mol dm⁻³, of Mn²⁺(aq) in the saturated solution. Show your working.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li><i>n</i>(NaBiO₃) = 21.50 / 1000 × 0.100 = 2.15 × 10⁻³ mol</li><li><i>n</i>(Mn²⁺) in 1.00 dm³ of diluted solution = 2.15 × 10⁻³ × 2/5 × 40 = 0.0344 mol（2 : 5 来自 equation 2，× 40 = 1000 / 25.00）</li><li>这 0.0344 mol 全部来自 100.0 cm³ saturated solution：<i>c</i> = 0.0344 × 1000 / 100 = <b>0.344 mol dm⁻³</b></li></ul>
<p>判分：三步各 1 分。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-20"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-21">
<div class="k-ptable"></div>
<h3>9701/21 May/June 2024</h3>
<div class="q"><h4>Q2(c)(ii) <span class="mark">[1]</span></h4>
<p>Magnesium nitride reacts with an excess of water to produce ammonia and magnesium hydroxide only. Construct an equation to describe this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p>Mg₃N₂ + 6H₂O → 3Mg(OH)₂ + 2NH₃</p>
</div>
</div>
<div class="q"><h4>Q2(d)(i) <span class="mark">[1]</span></h4>
<p>Figure 2.2 shows part of the lattice structure of a crystal of boron nitride. Use Figure 2.2 to deduce the empirical formula of boron nitride.</p>
<img src="" width="420" alt="">
<button class="reveal">答案</button><div class="ans">
<p>BN（B 和 N 交替排列，1 : 1）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-21"));</script>"""))
```

```python
# @title 📝 9701/21 Oct/Nov 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-22">
<div class="k-ptable"></div>
<h3>9701/21 Oct/Nov 2024</h3>
<p>Table 1.2 shows the relative abundances of isotopes in a sample of an alloy containing rhodium and iridium only.</p>
<table><tr><th>isotope</th><th>relative isotopic mass</th><th>relative abundance in alloy</th></tr><tr><td>¹⁰³Rh</td><td>102.91</td><td>50.00</td></tr><tr><td>¹⁹¹Ir</td><td>190.96</td><td>15.18</td></tr><tr><td>¹⁹³Ir</td><td>192.96</td><td>34.82</td></tr></table>
<div class="q"><h4>Q1(c)(i) <span class="mark">[2]</span></h4>
<p>Define relative isotopic mass.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 mass of an (atom of an) isotope</li><li>M2 compared to the mass of the unified atomic mass unit <b>OR</b> on a scale on which a carbon-12 atom has a mass of exactly 12 units <b>OR</b> divided by 1/12 mass of a carbon-12 atom</li></ul>
</div>
</div>
<div class="q"><h4>Q1(c)(ii) <span class="mark">[2]</span></h4>
<p>Use Table 1.2 to calculate the relative atomic mass, <i>A</i>ᵣ, of iridium in the alloy. Give your answer to two decimal places.</p>
<p><i>A</i>ᵣ(Ir) = <input class="fill" data-ans="192.35"></p>
<button class="reveal">答案</button><div class="ans">
<p>(190.96 × 15.18 + 192.96 × 34.82) ÷ (15.18 + 34.82) = <b>192.35</b></p>
<p>注意分母是 Ir 两个 isotope 的 abundance 之和 50.00，不是 100——Rh 不是 iridium。（本小问在分类表里归 chapter 22，因为数据来自 mass spectrum。）</p>
</div>
</div>
<div class="q"><h4>Q1(d)(i) <span class="mark">[2]</span></h4>
<p>Hydrated rhodium(III) chloride, RhCl₃·xH₂O, contains 20.5% by mass of water of crystallisation. Deduce the integer value of <i>x</i> in RhCl₃·xH₂O. Show your working.</p>
<p><i>x</i> = <input class="fill" data-ans="3"></p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 RhCl₃ : H₂O = 79.5 / 209.4 : 20.5 / 18.0 = 0.380 : 1.14 = 1 : 3 （或 18<i>x</i> ÷ (102.9 + 3 × 35.5 + 18<i>x</i>) = 0.205）</li><li>M2 <i>x</i> = <b>3</b></li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-22"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-23">
<div class="k-ptable"></div>
<h3>9701/21 May/June 2025</h3>
<img src="" width="520" alt="">
<div class="q"><h4>Q4(a) <span class="mark">[1]</span></h4>
<p>The structure of vitamin C is shown in Figure 4.1. Deduce the empirical formula of vitamin C.</p>
<button class="reveal">答案</button><div class="ans">
<p>C₃H₄O₃（molecular formula C₆H₈O₆，÷ 2）</p>
</div>
</div>
<div class="q"><h4>Q4(b)(i) <span class="mark">[1]</span></h4>
<p>A vitamin C tablet is dissolved in water to produce 200.0 cm³ of vitamin C solution. 5.00 cm³ of this solution is added to a flask with approximately 150 cm³ of water and an indicator. Exactly 28.40 cm³ of 5.00 × 10⁻⁴ mol dm⁻³ I₂(aq) reacts with the sample of vitamin C solution in the flask. [<i>M</i>ᵣ: vitamin C, 176]</p>
<img src="" width="520" alt="">
<p>Calculate the amount, in mol, of I₂(aq) added to the flask in this titration.</p>
<button class="reveal">答案</button><div class="ans">
<p>28.40 / 1000 × 5.00 × 10⁻⁴ = <b>1.42 × 10⁻⁵ mol</b></p>
</div>
</div>
<div class="q"><h4>Q4(b)(ii) <span class="mark">[2]</span></h4>
<p>Use your answer to (b)(i) to calculate the mass, in g, of vitamin C in the tablet. Show your working.</p>
<button class="reveal">答案</button><div class="ans">
<p>vitamin C : I₂ = 1 : 1（Figure 4.2）。</p>
<ul><li>M1 mass in the 5 cm³ sample = 1.42 × 10⁻⁵ × 176 = 2.4992 × 10⁻³ g（写出来可以是 2.50 × 10⁻³，但计算器里保留全部位数）</li><li>M2 mass in the tablet = 2.4992 × 10⁻³ × 200 / 5 = 0.099968 → <b>0.100 g</b>（3 s.f.）</li></ul>
<p>两个常见错：用已经 round 过的 2.49 × 10⁻³ 继续乘，得到 0.0996（中间步骤提前四舍五入）；把 0.099968 截断成 0.0999（应该四舍五入）。</p>
<p>也可以先 × 200/5 再 × 176，两种顺序都给分。那 150 cm³ 水是干扰项：它不改变 flask 里 vitamin C 的 mol 数。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-23"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-24">
<div class="k-ptable"></div>
<h3>9701/22 May/June 2025</h3>
<img src="" width="520" alt="">
<div class="q"><h4>Q5(a) <span class="mark">[1]</span></h4>
<p>Deduce the empirical formula of vitamin C.</p>
<button class="reveal">答案</button><div class="ans">
<p>C₃H₄O₃</p>
</div>
</div>
<div class="q"><h4>Q5(b)(i) <span class="mark">[1]</span></h4>
<p>All the vitamin C in 150.0 g of lemon is extracted and dissolved in water to make 100.0 cm³ of solution <b>L</b>. A 25.0 cm³ sample of <b>L</b> is titrated with 5.00 × 10⁻³ mol dm⁻³ I₂(aq); exactly 36.65 cm³ of I₂(aq) reacts. [<i>M</i>ᵣ: vitamin C, 176]</p>
<img src="" width="520" alt="">
<p>Calculate the amount, in mol, of I₂(aq) added in the titration.</p>
<button class="reveal">答案</button><div class="ans">
<p>5.00 × 10⁻³ × 36.65 / 1000 = <b>1.83 × 10⁻⁴ mol</b>（1.8325 × 10⁻⁴）</p>
</div>
</div>
<div class="q"><h4>Q5(b)(ii) <span class="mark">[2]</span></h4>
<p>Use your answer to (b)(i) to calculate the percentage by mass of vitamin C present in 150.0 g of lemon.</p>
<p>percentage = <input class="fill" data-ans="0.086|0.087"> %</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 mass of vitamin C in 150 g lemon = 1.8325 × 10⁻⁴ × 176 × 4 = 0.129 g（× 4 = 100.0 / 25.0）</li><li>M2 0.129 / 150 × 100 = <b>0.086%</b>（0.087 也接受）</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-24"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-25">
<div class="k-ptable"></div>
<h3>9701/22 Feb/Mar 2026</h3>
<div class="q"><h4>Q2(b) <span class="mark">[4]</span></h4>
<p>In the laboratory, NO is made in the reaction of copper metal with dilute nitric acid:</p>
<p>8HNO₃(aq) + 3Cu(s) → 3Cu(NO₃)₂(aq) + 4H₂O(l) + 2NO(g)</p>
<p>A student reacts 0.00393 mol Cu with excess dilute HNO₃(aq) at 5 °C and 101 kPa. Calculate the maximum volume, in cm³, of NO(g) that is produced under these conditions.</p>
<p>volume = <input class="fill" data-ans="59.9|60"> cm³</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <i>n</i>(NO) = 0.00393 × 2/3 = 0.00262 mol</li><li>M2 <i>p</i> = 101 000 Pa，<i>R</i> = 8.31，<i>T</i> = 278 K，<i>n</i> = 0.00262</li><li>M3 use of <i>pV</i> = <i>nRT</i>（<i>V</i> = <i>nRT</i> / <i>p</i>）</li><li>M4 <i>V</i> = 5.99 × 10⁻⁵ m³ × 10⁶ = <b>59.9 cm³</b></li></ul>
<p>5 °C 不是 room conditions，不能用 24.0 dm³ mol⁻¹。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-25"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-26">
<div class="k-ptable"></div>
<h3>9701/21 May/June 2026</h3>
<div class="q"><h4>Q1(e) <span class="mark">[3]</span></h4>
<p>Compound X contains sulfur and chlorine only. 100.0 g of X contains 47.49 g of sulfur. The relative molecular mass, <i>M</i>ᵣ, of X is 135.2. Determine the molecular formula of X. Show your working.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 mass Cl = 100.0 − 47.49 = 52.51 g</li><li>M2 47.49 / 32.1 : 52.51 / 35.5 = 1.48 : 1.48 → empirical formula SCl</li><li>M3 <i>M</i>ᵣ(SCl) = 67.6，135.2 / 67.6 = 2 → <b>S₂Cl₂</b></li></ul>
<p>判分：M2 允许 ecf；只写 S₂Cl₂ 不给过程，M3 也给。</p>
</div>
</div>
<div class="q"><h4>Q2(c)(ii) <span class="mark">[4]</span></h4>
<p>equation 1　2CaCO₃ + 2SO₂ + O₂ → 2CaSO₄ + 2CO₂</p>
<p>Use the ideal gas equation to calculate the volume of O₂(g) that reacts with 0.500 mol of SO₂ at 400 °C and 101 kPa in equation 1. Include units in your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <i>pV</i> = <i>nRT</i></li><li>M2 <i>n</i>(O₂) = 0.500 ÷ 2 = 0.250 mol</li><li>M3 <i>T</i> = 400 + 273 = 673 K</li><li>M4 <i>V</i> = 0.250 × 8.31 × 673 ÷ 101 000 = <b>0.0138 m³</b>（= 13.8 dm³ = 13 800 cm³）</li></ul>
<p>判分：单位跟数值配套即可（0.014 m³ / 14 dm³ / 14 000 cm³ 都给满分）；M4 允许 ecf。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-26"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-27">
<div class="k-ptable"></div>
<h3>9701/22 May/June 2026</h3>
<img src="" width="520" alt="">
<p>The molecular formula of X is C₅H₁₂O₄. Complete oxidation of 0.255 g of X occurs when it reacts with exactly 10.0 cm³ of 0.500 mol dm⁻³ acidified K₂Cr₂O₇(aq).</p>
<div class="q"><h4>Q4(b)(ii) <span class="mark">[1]</span></h4>
<p>Calculate the amount, in mol, of K₂Cr₂O₇(aq) used in the reaction.</p>
<p>amount = <input class="fill" data-ans="0.005|0.0050|5e-3"> mol</p>
<button class="reveal">答案</button><div class="ans">
<p>10.0 / 1000 × 0.500 = <b>0.005 mol</b>（5 × 10⁻³）</p>
</div>
</div>
<div class="q"><h4>Q4(b)(iii) <span class="mark">[1]</span></h4>
<p>Calculate the amount, in mol, of X present in the 0.255 g sample.</p>
<button class="reveal">答案</button><div class="ans">
<p><i>M</i>ᵣ(C₅H₁₂O₄) = 5 × 12.0 + 12 × 1.0 + 4 × 16.0 = 136 → 0.255 / 136 = <b>1.875 × 10⁻³ mol</b></p>
<p>判分：接受 1.88 × 10⁻³ / 1.9 × 10⁻³（2 × 10⁻³ 需有过程）；<b>拒绝</b> 1.880 × 10⁻³、1.90 × 10⁻³、2.0 × 10⁻³、1.87 × 10⁻³。</p>
</div>
</div>
<div class="q"><h4>Q4(c)(i) <span class="mark">[2]</span></h4>
<p>Compound X and W are a pair of structural isomers. 0.0012 mol of W is completely oxidised when it is heated under reflux with exactly 0.0020 mol of acidified K₂Cr₂O₇(aq) to produce compound Z, C₅H₆O₆. Complete the equation for this reaction.</p>
<p>…C₅H₁₂O₄ + …Cr₂O₇²⁻ + …H⁺ → …C₅H₆O₆ + …Cr³⁺ + …H₂O</p>
<button class="reveal">答案</button><div class="ans">
<p>3C₅H₁₂O₄ + 5Cr₂O₇²⁻ + 40H⁺ → 3C₅H₆O₆ + 10Cr³⁺ + 29H₂O</p>
<ul><li>M1 系数 5Cr₂O₇²⁻（0.0012 : 0.0020 = 3 : 5）</li><li>M2 全式配平正确</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-27"));</script>"""))
```

```python
# @title ✅ Learning outcomes 自查清单 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch02-28">
<h3>Learning outcomes 自查 <span class="faint count"></span></h3><p class="faint">勾选会存在这个浏览器里（localStorage）；Colab 不允许时只在当前页面有效。</p><div class="lolist" data-key="ch02"><div class="bar"><i></i></div><h4>2.1</h4>
<label class="chk"><input type="checkbox" data-id="2.1.1"><span><b>2.1.1</b> define the unified atomic mass unit as one twelfth of the mass of a carbon-12 atom</span></label>
<label class="chk"><input type="checkbox" data-id="2.1.2"><span><b>2.1.2</b> define relative atomic mass, <i>A</i>ᵣ, relative isotopic mass, relative molecular mass, <i>M</i>ᵣ, and relative formula mass in terms of the unified atomic mass unit</span></label>
<h4>2.2</h4>
<label class="chk"><input type="checkbox" data-id="2.2.1"><span><b>2.2.1</b> define and use the term mole in terms of the Avogadro constant</span></label>
<h4>2.3</h4>
<label class="chk"><input type="checkbox" data-id="2.3.1"><span><b>2.3.1</b> write formulas of ionic compounds from ionic charges and oxidation numbers (shown by a Roman numeral), including:<ul><li>(a) the prediction of ionic charge from the position of an element in the Periodic Table</li><li>(b) recall of the names and formulas for the following ions: NO₃⁻, CO₃²⁻, SO₄²⁻, OH⁻, NH₄⁺, Zn²⁺, Ag⁺, HCO₃⁻, PO₄³⁻</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="2.3.2"><span><b>2.3.2</b> (a) write and construct equations (which should be balanced), including ionic equations (which should not include spectator ions)<br>(b) use appropriate state symbols in equations</span></label>
<label class="chk"><input type="checkbox" data-id="2.3.3"><span><b>2.3.3</b> define and use the terms empirical and molecular formula</span></label>
<label class="chk"><input type="checkbox" data-id="2.3.4"><span><b>2.3.4</b> understand and use the terms anhydrous, hydrated and water of crystallisation</span></label>
<label class="chk"><input type="checkbox" data-id="2.3.5"><span><b>2.3.5</b> calculate empirical and molecular formulas, using given data</span></label>
<h4>2.4</h4>
<label class="chk"><input type="checkbox" data-id="2.4.1"><span><b>2.4.1</b> perform calculations including use of the mole concept, involving:<ul><li>(a) reacting masses (from formulas and equations) including percentage yield calculations</li><li>(b) volumes of gases (e.g. in the burning of hydrocarbons)</li><li>(c) volumes and concentrations of solutions</li><li>(d) limiting reagent and excess reagent</li></ul><br><span class="faint"><i>When performing calculations, candidates' answers should reflect the number of significant figures given or asked for in the question. When rounding up or down, candidates should ensure that significant figures are neither lost unnecessarily nor used beyond what is justified (see also Mathematical requirements section).</i></span><ul><li>(e) deduce stoichiometric relationships from calculations such as those in 2.4.1(a)–(d)</li></ul></span></label></div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch02-28"));</script>"""))
```
