## 背景：solid、liquid、gas

（讲义开头的复习内容，syllabus 4.1 / 4.2 都建立在它上面。）

| | solid | liquid | gas |
|---|---|---|---|
| arrangement | regular，particles 紧密接触 | 大多 random，particles 仍然靠得很近 | random，particles 相距很远 |
| movement | 只能在固定位置 vibrate | 可以互相 slide past | 向各个方向快速、random 地运动 |
| compress | 几乎不能 | 只能稍微压缩（fixed volume） | 很容易压缩 |

**change of state** 的本质：加热时 particles 的 kinetic energy 增大，足以 overcome particles 之间的 forces of attraction。

- **melting**：solid → liquid，发生在 melting point（1 atm 下）。
- **vaporisation**：liquid → gas。**evaporation** 在 boiling point 以下就会发生——表面能量最高的 molecules 先逃逸；到 boiling point 时整个 liquid 里都形成 vapour。
- **freezing / condensation** 反过来：particles 失去 kinetic energy，attraction 把它们拉近。
- **sublimation**：solid 直接变 gas（dry ice、iodine、naphthalene、aluminium chloride）；反过来叫 **deposition**。
- 在 closed system 里 liquid ⇌ gas 会达到 equilibrium：vaporisation 和 condensation 的 rate 相等。此时 vapour 对器壁的压强叫 **vapour pressure**，温度越高越大。

要 overcome 的是什么力，决定了 melting point / boiling point 的高低——这就是 4.2 的主题：

| type of structure | 熔化 / 沸腾时要 overcome 的力 | melting point / boiling point |
|---|---|---|
| giant ionic | ionic bonds（electrostatic attraction between oppositely charged ions） | high |
| giant metallic | metallic bonds | 一般 high |
| giant molecular | covalent bonds | very high |
| simple molecular | intermolecular forces（id-id、pd-pd、hydrogen bond），**不是** covalent bond | low |

## 4.1 The gaseous state: ideal and real gases and *pV* = *nRT*

### A. Kinetic theory：pressure 从哪里来

kinetic theory of gases 的假设：

1. gas molecules 不停地做 rapid、random 的运动；
2. molecules 之间的距离远大于 molecule 本身的直径，所以 molecules 自身的 volume 可以忽略（negligible）；
3. molecules 之间没有 intermolecular forces of attraction（或 repulsion）；
4. 所有 collision 都是 elastic 的：collision 中没有 kinetic energy 损失；
5. gas 的 temperature 与 molecules 的 average kinetic energy 成正比。

**pressure 的来源**：gas molecules 不断撞击容器壁，每次撞击对壁施加一个力；单位面积上受到的总力就是 pressure（1 Pa = 1 N m<sup>−2</sup>）。

- volume 变小（*T* 不变）：molecules 挤得更近，**单位时间内撞壁的次数更多** → pressure 增大；
- temperature 升高（*V* 不变）：molecules 运动更快，撞壁**更频繁、每次更用力** → pressure 增大；
- molecules 更多（*n* 增大）：撞壁更频繁 → pressure 增大。

答题时 "collisions of gas molecules with the walls of the container" 这一句是核心，只写 "molecules move around" 不得分。

```python
# @title 🔬 gas pressure：molecules 撞壁 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-04">
<h3>gas pressure：molecules 撞壁</h3><p class="faint">待合并：<code>colab/src/ch04/gas-box.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-04"));</script>"""))
```

### B. Ideal gas 与三条 gas law

完全符合上面 kinetic theory 的 gas 叫 **ideal gas**。syllabus 要的两点：

- ideal gas 的 particles **volume 为零**（zero particle volume）；
- particles 之间**没有 intermolecular forces of attraction**。

真实的 gas（**real gas**）只是接近，不完全符合。

> **LO 4.1.2** — understand that ideal gases have zero particle volume and no intermolecular forces of attraction

三条 gas law 都是 ideal gas 的行为，合起来就是 *pV* = *nRT*：

| law | 保持不变 | 关系 | 例（讲义） |
|---|---|---|---|
| Boyle's law | *n*、*T* | *pV* = constant → *p*₁*V*₁ = *p*₂*V*₂ | 240 cm<sup>3</sup> air at 100 kPa 压到 150 cm<sup>3</sup>：*p*₂ = 100 × 240 / 150 = **160 kPa** |
| Charles's law | *n*、*p* | *V* / *T* = constant → *V*₁ / *T*₁ = *V*₂ / *T*₂（*T* 必须用 K） | 1.13 dm<sup>3</sup> at 47 °C → 29 °C：*V*₂ = 1.13 × 302 / 320 = **1.07 dm<sup>3</sup>** |
| Avogadro's law | *p*、*T* | *V* / *n* = constant：同温同压下，相同 volume 的 gas 含相同数目的 molecules | room conditions 下 1 mol 任何 gas ≈ 24.0 dm<sup>3</sup> |

- Charles's law 的直线外推到 *V* = 0 时，温度是 −273 °C = 0 K（**absolute zero**）。real gas 在到达之前早就 liquefy / solidify 了。
- 两种 gas 混合（*T* 不变）：各自按 Boyle's law 算出在总 volume 里的 partial pressure，再相加。例：5 dm<sup>3</sup> He at 12 kPa 与 10 dm<sup>3</sup> Ne at 6 kPa 相连 → 15 dm<sup>3</sup> 里 He 4 kPa、Ne 4 kPa → 总 pressure **8 kPa**。

### C. The ideal gas equation *pV* = *nRT*

| 符号 | 量 | SI 单位 | 常见换算 |
|---|---|---|---|
| *p* | pressure | Pa | 1 kPa = 1000 Pa；1 atm = 101 325 Pa（题目常写 101 kPa） |
| *V* | volume | **m<sup>3</sup>** | 1 m<sup>3</sup> = 1000 dm<sup>3</sup> = 10<sup>6</sup> cm<sup>3</sup>；1 dm<sup>3</sup> = 10<sup>−3</sup> m<sup>3</sup>；1 cm<sup>3</sup> = 10<sup>−6</sup> m<sup>3</sup> |
| *n* | amount | mol | *n* = *m* / *M* |
| *R* | molar gas constant | J K<sup>−1</sup> mol<sup>−1</sup> | 8.31（data booklet；= 8.31 m<sup>3</sup> Pa K<sup>−1</sup> mol<sup>−1</sup>） |
| *T* | temperature | **K** | *T* / K = *θ* / °C + 273 |

**单位是这一节失分最多的地方**：*p* 用 Pa、*V* 用 m<sup>3</sup>、*T* 用 K。也可以 *p* 用 kPa、*V* 用 dm<sup>3</sup>——两边都放大 1000 倍，*pV* 不变（9701/23 May/June 2026 Q2(c) 的 mark scheme 明确接受「101 × 617.0 × 10<sup>−3</sup>」这种一致换算）。但 Pa 配 dm<sup>3</sup>、kPa 配 cm<sup>3</sup> 都是错的。

例 1：0.500 mol CO₂ at 150 kPa、19 °C 的 volume。

1. *p* = 150 000 Pa，*T* = 19 + 273 = 292 K
2. *V* = *nRT* / *p* = 0.500 × 8.31 × 292 / 150 000 = **8.09 × 10<sup>−3</sup> m<sup>3</sup>**（= 8.09 dm<sup>3</sup>）

例 2：5.00 dm<sup>3</sup> flask 里有 4.00 g O₂（*M*ᵣ = 32.0），127 °C 时的 pressure。

1. *n* = 4.00 / 32.0 = 0.125 mol；*V* = 5.00 × 10<sup>−3</sup> m<sup>3</sup>；*T* = 400 K
2. *p* = *nRT* / *V* = 0.125 × 8.31 × 400 / (5.00 × 10<sup>−3</sup>) = **8.31 × 10<sup>4</sup> Pa**（83.1 kPa）

**determination of *M*ᵣ**：把 *n* = *m* / *M* 代进去，

*pV* = (*m* / *M*) *RT*　→　*M* = *mRT* / (*pV*)

例 3：ethane 在 20 °C、1 atm 下的 density 是 1.264 g dm<sup>−3</sup>。1 dm<sup>3</sup>（= 1 × 10<sup>−3</sup> m<sup>3</sup>）里有 1.264 g：

*M* = 1.264 × 8.31 × 293 / (101 325 × 1 × 10<sup>−3</sup>) = **30.4 g mol<sup>−1</sup>**（C₂H₆ = 30.0，real gas 略有偏差）

讲义练习（答案在交互件里能直接验）：

- 2.00 dm<sup>3</sup> flask 含 5.28 g gas，200 kPa、20 °C → *M*ᵣ = 5.28 × 8.31 × 293 / (200 000 × 2.00 × 10<sup>−3</sup>) = **32.1**
- 0.25 mol CO 在 10 dm<sup>3</sup> flask 里 pressure 120 kPa → *T* = 120 000 × 0.010 / (0.25 × 8.31) = **578 K**
- 1.00 cm<sup>3</sup> ice（density 1.00 g cm<sup>−3</sup>）在 596 K、101 kPa 变成 steam 的 volume：*n* = 1.00 / 18.0 = 0.0556 mol，*V* = 0.0556 × 24.0 × 596 / 298 = **2.67 dm<sup>3</sup>**（用 *V* ∝ *T* 从 24.0 dm<sup>3</sup> at 298 K 换算；MCQ 选 C）

> **LO 4.1.3** — state and use the ideal gas equation *pV* = *nRT* in calculations, including in the determination of *M*ᵣ

### D. Real gases：什么时候偏离 ideal

real gas 不符合 kinetic theory 的正是 4.1.2 那两条：

1. molecules 之间**有** forces of attraction（id-id，polar molecule 还有 pd-pd，NH₃ 之类还有 hydrogen bond）；
2. molecules 自身的 volume **不能**忽略。

衡量偏离的量：**compression factor** *Z* = *pV* / *nRT*。ideal gas 永远 *Z* = 1。

| 条件 | 发生了什么 | 结果 |
|---|---|---|
| **low pressure、high temperature** | molecules 相距很远、运动很快：自身 volume 相对容器可以忽略，attraction 来不及起作用 | 最接近 ideal（*Z* ≈ 1） |
| low temperature（中等 pressure） | molecules 运动慢，attraction 把它们拉向彼此、拉离器壁 → 撞壁的力变小 | pressure 比 ideal 预测的**低**，*Z* < 1 |
| **very high pressure** | molecules 被挤得很近，自身 volume 占了容器的一大块，能自由运动的空间比 *V* 小；electron clouds 靠近时互相 repel | *pV* 比 ideal 预测的**大**，*Z* > 1 |
| 接近 boiling point | 快要 condense 成 liquid，attraction 已经很重要 | 偏离最大 |

结论（mark scheme 的句式）：**real gases behave most like an ideal gas at low pressure and high temperature**；偏离最大在 high pressure、low temperature。

**哪种 gas 更 ideal**：比的就是那两条——molecule 越小、intermolecular forces 越弱，越 ideal。

- noble gases（He、Ne）最接近：monatomic、non-polar、electrons 很少 → id-id forces 特别弱，atom 也小。
- polar molecule（NH₃、SO₂）、大 molecule（CO₂、C₂H₄）偏离较大。讲义的 molar volume at s.t.p.（3 s.f.）：He、N₂、O₂ 都是 22.4 dm<sup>3</sup>；CH₄、CO₂ 22.3；C₂H₄ 22.2；NH₃ 22.1；SO₂ 21.9——attraction 越强，molar volume 越被「拉小」。

```python
# @title 🔬 ideal vs real gas：Z = pV / nRT { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-09">
<h3>ideal vs real gas：Z = pV / nRT</h3><p class="faint">待合并：<code>colab/src/ch04/real-gas.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-09"));</script>"""))
```

## 4.2 Bonding and structure

很多 ionic、metallic、covalent substance 是 **crystalline** 的：particles（ions、atoms 或 molecules）在三维空间里规则地重复排列，这种排列叫 **crystal lattice**。syllabus 要会描述四类：

| type of structure | lattice 里的 particle | 把它们连起来的力 | 例（syllabus 点名） |
|---|---|---|---|
| **giant ionic** | positive ions 和 negative ions 交替 | ionic bonding（strong electrostatic attraction between oppositely charged ions，**all directions**） | sodium chloride、magnesium oxide |
| **simple molecular** | 小 molecules | molecule **内** strong covalent bonds；molecule **之间** weak intermolecular forces | iodine、buckminsterfullerene C₆₀、ice |
| **giant molecular**（giant covalent） | atoms | strong covalent bonds 贯穿整个 structure | silicon(IV) oxide、graphite、diamond |
| **giant metallic** | positive metal ions | metallic bonding：positive ions 与 sea of delocalised electrons 之间的 attraction | copper |

> **LO 4.2.1** — describe, in simple terms, the lattice structure of a crystalline solid which is:
> - (a) giant ionic, including sodium chloride and magnesium oxide
> - (b) simple molecular, including iodine, buckminsterfullerene C₆₀ and ice
> - (c) giant molecular, including silicon(IV) oxide, graphite and diamond
> - (d) giant metallic, including copper

### A. Giant ionic：NaCl、MgO

- 三维的 alternating positive and negative ions；lattice 的形状由 ions 的相对大小决定。
- NaCl 和 MgO 都是 **cubic**（同一种 lattice）：每个 Na<sup>+</sup> 被 6 个 Cl<sup>−</sup> 包围，每个 Cl<sup>−</sup> 也被 6 个 Na<sup>+</sup> 包围（**6 : 6 coordination**）。Cl<sup>−</sup>（0.181 nm）比 Na<sup>+</sup>（0.095 nm）大得多，Na<sup>+</sup> 嵌在 Cl<sup>−</sup> 之间的空隙里，离它们尽可能近。
- MgO 的 ions 是 Mg<sup>2+</sup> 和 O<sup>2−</sup>：charge 是 NaCl 的两倍、ions 又更小 → attraction 强得多 → melting point 2852 °C，远高于 NaCl 的 801 °C。

### B. Giant metallic：copper

- positive metal ions 排成规则的层（copper 是 cubic close packing），被 **sea of delocalised electrons** 包围；attraction 作用于所有方向。
- 受力时 layers 可以滑过彼此，新位置上 ions 与 delocalised electrons 之间的 metallic bonds 立即重新形成 → **malleable**（能锤成各种形状）、**ductile**（能拉成丝）。
- **alloy**：加入大小不同的另一种 atom（brass = copper + zinc；bronze = copper + tin）。Zn atoms 比 Cu 大，打乱层的规则排列，layers 不容易滑动 → 比纯 copper **更硬、更强**，但仍然 malleable。aluminium alloy（加 Cu、Mg、Si、Mn）轻、强、耐腐蚀，用于飞机机身、自行车架。

### C. Simple molecular：iodine、C₆₀、ice

- **iodine**：I₂ molecules 规则排列成 lattice。相邻两个 I₂ molecule 的 nuclei 距离比**同一个** I₂ 里两个 nuclei 的距离大——molecule 内是 strong covalent bond，molecule 之间只有 weak id-id forces。加热时只需要 overcome id-id forces → melting point 低，还会 sublime。
- **buckminsterfullerene C₆₀**：60 个 C atoms 组成足球形的笼子（12 个 pentagons + 20 个 hexagons），每个 C 连 3 个 C。它是 **simple molecule**（不是 giant structure）：molecules 之间只有 weak intermolecular forces → 约 600 °C sublime、比较软、能少量溶于 methylbenzene 之类的 solvent；有一部分 electrons delocalised，但比 graphite 少 → conductivity 远不如 graphite。
- **ice**：H₂O molecules 通过 **hydrogen bonds** 排成 open lattice，每个 O 周围 4 个 H（2 个 covalent bond + 2 个 hydrogen bond），tetrahedral 排列。这个空旷的结构让 ice 的 density 比 liquid water 小（ice 浮在水面上）；hydrogen bonding 的细节在 chapter 3。

### D. Giant molecular：diamond、graphite、SiO₂

diamond 和 graphite 是 carbon 的 **allotropes**（同一 element 的不同 structural forms）。

| | diamond | graphite | silicon(IV) oxide, SiO₂ |
|---|---|---|---|
| 每个 atom 连几个 | C 连 4 个 C，tetrahedral | C 连 3 个 C，hexagon 组成平面 layers | 每个 Si 连 4 个 O（tetrahedral）、每个 O 连 2 个 Si → formula SiO₂ |
| layer 之间 | — | weak id-id forces | — |
| delocalised electrons | 无：4 个 outer electrons 全部用于 bonding | 每个 C 的第 4 个 electron 在 p orbital 里，sideways overlap 形成 layer 上下的 delocalised electrons | 无 |
| melting point | very high（strong covalent bonds throughout） | very high（layer 内 strong covalent bonds） | very high（> 1600 °C） |
| hardness | very hard（3D network 很难破坏）→ drill tip | soft、slippery：layers 容易 slide → pencil lead、lubricant | hard |
| electrical conductivity | does not conduct | **conducts**：delocalised electrons move along / between the layers | does not conduct |

- SiO₂（quartz、sand 的主要成分）的结构和 diamond 很像：把 diamond 里每个 C 换成 Si，每两个 Si 之间插一个 O。
- **graphene** = 单独一层 graphite；**carbon nanotube** = 一层 graphene 卷成的圆筒（沿长轴导电、tensile strength 很高、melting point 约 3500 °C）。二者都不在 syllabus 点名的例子里，知道是 giant molecular 即可。

```python
# @title 🧊 lattice 3D：九种晶格 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-11">
<h3>lattice 3D：九种晶格</h3><p class="faint">待合并：<code>colab/src/ch04/lattice.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>/* che_note Colab kit — 3D helpers, inlined only into html3d cells by tools/build_colab.py.
   Loads three.js core (pinned) from jsDelivr with a dynamic import; no import map and no
   examples/jsm addons, so there is no bare 'three' specifier to resolve. Drag-rotate and
   wheel-zoom are implemented here. If the CDN or WebGL is unavailable, the cell's
   .fallback block (a static SVG) is shown instead. */
(function () {
  if (window.K3) return;
  var K3 = {};
  var URL3 = 'https://cdn.jsdelivr.net/npm/three@0.160.0/build/three.module.min.js';
  var loading = null;
  K3.load = function () {
    if (!loading) loading = import(URL3);
    return loading;
  };
  function webgl() {
    try { var c = document.createElement('canvas'); return !!(c.getContext('webgl2') || c.getContext('webgl')); }
    catch (e) { return false; }
  }
  function cssVar(el, name) { return getComputedStyle(el).getPropertyValue(name).trim() || '#ffffff'; }

  /* K3.start(root, build): build(T, api) fills api.group; returns nothing.
     root must contain .stage, and may contain .fallback plus buttons [data-3d=reset|spin]. */
  K3.start = function (root, build) {
    var stage = root.querySelector('.stage');
    var fb = root.querySelector('.fallback');
    function fail(why) {
      if (stage) stage.style.display = 'none';
      if (fb) { fb.classList.add('show'); var w = fb.querySelector('.why'); if (w) w.textContent = why; }
      root.setAttribute('data-3d', 'fallback');
      if (window.K) K.fit();
    }
    if (!webgl()) { fail('WebGL 不可用'); return; }
    K3.load().then(function (T) {
      var W = stage.clientWidth || 600, H = stage.clientHeight || 380;
      var renderer = new T.WebGLRenderer({ antialias: true, preserveDrawingBuffer: true });
      renderer.setPixelRatio(Math.min(window.devicePixelRatio || 1, 2));
      renderer.setSize(W, H);
      renderer.setClearColor(new T.Color(cssVar(root, '--card')));
      stage.appendChild(renderer.domElement);
      var scene = new T.Scene();
      var camera = new T.PerspectiveCamera(40, W / H, 0.05, 200);
      var DIST = 6;
      camera.position.set(0, 0, DIST);
      scene.add(new T.AmbientLight(0xffffff, 0.55));
      var key = new T.DirectionalLight(0xffffff, 0.9); key.position.set(3, 4, 5); scene.add(key);
      var rim = new T.DirectionalLight(0xffffff, 0.35); rim.position.set(-4, -2, -3); scene.add(rim);
      var group = new T.Group();
      scene.add(group);
      var api = { T: T, scene: scene, group: group, camera: camera, root: root, renderer: renderer,
                  dist: DIST, rx: 0.35, ry: -0.6, spin: false };
      api.color = function (name) { return new T.Color(cssVar(root, name)); };
      api.render = function () { renderer.render(scene, camera); };
      api.view = function () {
        group.rotation.set(api.rx, api.ry, 0);
        camera.position.set(0, 0, api.dist);
        camera.lookAt(0, 0, 0);
        api.render();
      };
      api.reset = function () { api.rx = api.rx0; api.ry = api.ry0; api.dist = api.dist0; api.view(); };
      build(T, api);
      api.rx0 = api.rx; api.ry0 = api.ry; api.dist0 = api.dist;

      /* drag to rotate, wheel / pinch to zoom */
      var c = renderer.domElement, drag = null;
      c.addEventListener('pointerdown', function (e) { drag = { x: e.clientX, y: e.clientY }; c.setPointerCapture(e.pointerId); c.style.cursor = 'grabbing'; });
      c.addEventListener('pointermove', function (e) {
        if (!drag) return;
        api.ry += (e.clientX - drag.x) * 0.01; api.rx += (e.clientY - drag.y) * 0.01;
        var lim = api.limit || 1.55;   /* a flat slice turned edge-on is just a line: cells may cap the tilt */
        api.rx = Math.max(-lim, Math.min(lim, api.rx));
        if (api.limit) api.ry = Math.max(-lim, Math.min(lim, api.ry));
        drag = { x: e.clientX, y: e.clientY }; api.view();
      });
      c.addEventListener('pointerup', function () { drag = null; c.style.cursor = 'grab'; });
      c.addEventListener('wheel', function (e) {
        e.preventDefault();
        api.dist = Math.max(1.2, Math.min(40, api.dist * (e.deltaY > 0 ? 1.1 : 0.9)));
        api.view();
      }, { passive: false });

      root.querySelectorAll('[data-3d=reset]').forEach(function (b) { b.addEventListener('click', api.reset); });
      root.querySelectorAll('[data-3d=spin]').forEach(function (b) {
        b.addEventListener('click', function () { api.spin = !api.spin; b.classList.toggle('on', api.spin); });
      });
      (function loop() {
        if (api.spin && document.visibilityState !== 'hidden') { api.ry += 0.008; api.view(); }
        requestAnimationFrame(loop);
      })();
      if (window.ResizeObserver) new ResizeObserver(function () {
        var w = stage.clientWidth, h = stage.clientHeight;
        if (!w || !h) return;
        renderer.setSize(w, h); camera.aspect = w / h; camera.updateProjectionMatrix(); api.render();
      }).observe(stage);
      api.view();
      root.setAttribute('data-3d', 'ready');
      root.__k3 = api;
    }).catch(function (e) { fail('three.js 载入失败（' + (e && e.message ? e.message : e) + '）'); });
  };

  /* ---------- builders ---------- */

  /* text label sprite */
  /* text sprite; the canvas is sized to the text so long labels are never cut off.
     size = world height of the text line */
  K3.label = function (T, text, color, size) {
    var cv = document.createElement('canvas'), g = cv.getContext('2d'), font = 'italic 600 40px "Noto Sans","DejaVu Sans",sans-serif';
    g.font = font;
    var w = Math.ceil(g.measureText(text).width) + 16;
    cv.width = w; cv.height = 64;
    g = cv.getContext('2d'); g.font = font; g.fillStyle = color; g.textAlign = 'center'; g.textBaseline = 'middle';
    g.fillText(text, w / 2, 34);
    var sp = new T.Sprite(new T.SpriteMaterial({ map: new T.CanvasTexture(cv), depthTest: false, transparent: true }));
    var h = size || 0.3;
    sp.scale.set(h * w / 64, h, 1);
    sp.renderOrder = 10;
    return sp;
  };

  /* x / y / z axes through the nucleus */
  K3.axes = function (T, api, len) {
    var g = new T.Group(), col = api.color('--mut').getStyle();
    var mat = new T.LineBasicMaterial({ color: api.color('--faint') });
    [['x', [1, 0, 0]], ['y', [0, 1, 0]], ['z', [0, 0, 1]]].forEach(function (a) {
      var v = new T.Vector3().fromArray(a[1]).multiplyScalar(len);
      g.add(new T.Line(new T.BufferGeometry().setFromPoints([v.clone().negate(), v]), mat));
      var s = K3.label(T, a[0], col, 0.3); s.position.copy(v.clone().multiplyScalar(1.1)); g.add(s);
    });
    /* nucleus: drawn on top, otherwise the p lobes (which meet at the origin) hide it. It must be
       'transparent' too — three.js draws all opaque objects before transparent ones, so an opaque
       dot would still be painted over by semi-transparent lobes whatever its renderOrder */
    var nuc = new T.Mesh(new T.SphereGeometry(0.06, 16, 12), new T.MeshBasicMaterial({ color: api.color('--fg'), depthTest: false, transparent: true }));
    nuc.renderOrder = 11;
    g.add(nuc);
    return g;
  };

  /* angular part of real hydrogen-like orbitals, as functions of a unit vector */
  var ANG = {
    s: function () { return 1; },
    px: function (x) { return x; }, py: function (x, y) { return y; }, pz: function (x, y, z) { return z; },
    dxy: function (x, y) { return 2 * x * y; }, dxz: function (x, y, z) { return 2 * x * z; }, dyz: function (x, y, z) { return 2 * y * z; },
    dx2y2: function (x, y) { return x * x - y * y; }, dz2: function (x, y, z) { return (3 * z * z - 1) / 2; }
  };
  /* boundary-surface style lobe mesh: radius ∝ angular part squared, two colours for the two phases */
  K3.orbital = function (T, api, type, scale, opacity) {
    var f = ANG[type], geo = new T.SphereGeometry(1, 120, 80), pos = geo.attributes.position;
    var cols = new Float32Array(pos.count * 3), cA = api.color('--s1'), cB = api.color('--s2'), v = new T.Vector3();
    var vals = [], mx = 0;
    for (var i = 0; i < pos.count; i++) {
      v.fromBufferAttribute(pos, i).normalize();
      var a = f(v.x, v.y, v.z); vals.push(a); mx = Math.max(mx, Math.abs(a));
    }
    for (i = 0; i < pos.count; i++) {
      v.fromBufferAttribute(pos, i).normalize();
      var r = type === 's' ? 1 : Math.pow(Math.abs(vals[i]) / mx, 2);
      pos.setXYZ(i, v.x * r * scale, v.y * r * scale, v.z * r * scale);
      var c = vals[i] >= 0 ? cA : cB; cols[3 * i] = c.r; cols[3 * i + 1] = c.g; cols[3 * i + 2] = c.b;
    }
    geo.setAttribute('color', new T.BufferAttribute(cols, 3));
    geo.computeVertexNormals();
    var mat = new T.MeshPhongMaterial({ vertexColors: true, shininess: 40, transparent: opacity < 1, opacity: opacity || 1,
                                        side: T.DoubleSide, depthWrite: !(opacity < 1) });
    return new T.Mesh(geo, mat);
  };

  /* hydrogen-like ns radial functions (a0 = 1) */
  var RAD = {
    '1s': function (r) { return Math.exp(-r); },
    '2s': function (r) { return (2 - r) * Math.exp(-r / 2) / 5.657; },
    '3s': function (r) { return (27 - 18 * r + 2 * r * r) * Math.exp(-r / 3) / 140.3; }
  };
  /* radius (a0) inside which the electron is found with probability p: integrates r²ψ² */
  K3.rBound = function (kind, p, R) {
    var f = RAD[kind], dr = R / 4000, tot = 0, acc = [], r, i;
    for (i = 0; i < 4000; i++) { r = (i + 0.5) * dr; tot += r * r * f(r) * f(r); acc.push(tot); }
    for (i = 0; i < 4000; i++) if (acc[i] >= p * tot) return (i + 1) * dr;
    return R;
  };
  /* 3D dot-density picture of an ns orbital: n points drawn with probability ∝ psi² per volume
     (r from r²psi², direction uniform), so the dots are the electron density. Colour = sign of psi.
     A dot cloud is a projection, so in a still view the nodes are hidden behind dots; they are
     drawn as translucent grey shells with a leader line to a "radial node" label outside the
     cloud, and seen by rotating / zooming / cutting. (Tried a strong dot layer on a cut face: the
     cloud then looks like it all lies in one plane.) With cut = true the front half (z > 0) is
     left out. userData.mat (caller sets size ∝ camera distance), userData.labels (sprites, size
     set by caller), userData.rMax = outermost dot radius (a0). */
  K3.dots = function (T, api, kind, n, scale, nodes, cut) {
    var f = RAD[kind], Rmax = K3.rBound(kind, 0.995, 80), g = new T.Group();
    var P = function (r) { return r * r * f(r) * f(r); }, mx = 0, i;
    for (i = 1; i <= 2000; i++) mx = Math.max(mx, P(i / 2000 * Rmax));
    var cA = api.color('--s1'), cB = api.color('--s2'), pts = [], cols = [], tries = 0;
    while (pts.length < n * 3 && tries < 5e6) {
      tries++;
      var r = Math.random() * Rmax;
      if (Math.random() * mx * 1.02 > P(r)) continue;
      var z = Math.random() * 2 - 1, ph = Math.random() * 2 * Math.PI, q = Math.sqrt(1 - z * z);
      var x = q * Math.cos(ph), y = q * Math.sin(ph);
      if (cut && z > 0) continue;
      pts.push(x * r * scale, y * r * scale, z * r * scale);
      var c = f(r) >= 0 ? cA : cB; cols.push(c.r, c.g, c.b);
    }
    var geo = new T.BufferGeometry();
    geo.setAttribute('position', new T.Float32BufferAttribute(pts, 3));
    geo.setAttribute('color', new T.Float32BufferAttribute(cols, 3));
    var mat = new T.PointsMaterial({ vertexColors: true, size: 0.02, sizeAttenuation: true, transparent: true, opacity: 0.7, depthWrite: false });
    g.add(new T.Points(geo, mat));
    /* leaders use the text colour: a grey line vanishes over the dense dots at the centre (3s node 1 is only ~7 % of the radius) */
    var grey = api.color('--mut'), ink = api.color('--fg'), labels = [], Rw = Rmax * scale;
    (nodes || []).forEach(function (rn, k) {
      var sh = new T.Mesh(new T.SphereGeometry(rn * scale, 48, 32, cut ? Math.PI : 0, cut ? Math.PI : 2 * Math.PI),
        new T.MeshPhongMaterial({ color: grey, transparent: true, opacity: 0.3, side: T.DoubleSide, depthWrite: false }));
      sh.renderOrder = 3; g.add(sh);
      /* leader from the shell to a label just outside the cloud; one direction per node so labels never stack */
      var a = [0.75, 2.39][k % 2], d = new T.Vector3(Math.cos(a), Math.sin(a), cut ? -0.25 : 0.25).normalize();
      var p0 = d.clone().multiplyScalar(rn * scale), p1 = d.clone().multiplyScalar(Rw * 1.04);
      var ln = new T.Line(new T.BufferGeometry().setFromPoints([p0, p1]), new T.LineBasicMaterial({ color: ink, depthTest: false, transparent: true }));
      ln.renderOrder = 12; g.add(ln);
      var dot = new T.Mesh(new T.SphereGeometry(Rw * 0.018, 10, 8), new T.MeshBasicMaterial({ color: ink, depthTest: false, transparent: true }));
      dot.position.copy(p0); dot.renderOrder = 12; g.add(dot);
      var lb = K3.label(T, nodes.length > 1 ? 'radial node ' + (k + 1) : 'radial node', ink.getStyle(), 0.3);
      lb.position.copy(d.clone().multiplyScalar(Rw * 1.04 + 0.01)); lb.center.set(Math.cos(a) > 0 ? 0 : 1, 0.5);
      lb.renderOrder = 13; g.add(lb); labels.push(lb);
    });
    g.userData = { mat: mat, labels: labels, rBound: K3.rBound(kind, 0.9, 80), rMax: Rmax };
    return g;
  };

  window.K3 = K3;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-11"));</script>"""))
```

### E. Structure 怎样决定 physical properties

| property | giant ionic | giant metallic | giant molecular | simple molecular |
|---|---|---|---|---|
| melting / boiling point | high：要破坏大量 strong ionic bonds | 一般 high | very high：要破坏大量 strong covalent bonds | low：只需 overcome weak intermolecular forces |
| electrical conductivity | solid **不**导电（ions 固定在 lattice 里）；molten 或 aqueous 导电（**ions move**） | solid、liquid 都导电（**delocalised electrons move**） | 不导电；**graphite 例外**（delocalised electrons） | 不导电（没有 mobile charged particles） |
| solubility | 很多溶于 water（ions 被 polar water molecules 包围），不溶于 non-polar solvent | 不溶（有的与 water 反应） | 不溶于任何 solvent | 大多不溶于 water（能形成 hydrogen bond 的除外），溶于 non-polar solvent |
| 受力 | hard 但 **brittle**：一层移动半个 ion，同号 ions 对齐、互相 repel → 裂开 | malleable、ductile | diamond / SiO₂ hard；graphite soft | soft |

答 conductivity 题的标准句式（mark scheme 逐字）：

- metal（sodium、tin）：**delocalised electrons move (through the structure)** / delocalised electrons are free to move；
- ionic compound molten：**ions move (through the liquid)**——不能说 electrons move；
- graphite：**delocalised electrons move through / between the layers**；
- diamond 不导电：**no delocalised electrons** / all (4) outer electrons are used in bonding。

比较 melting point 时必须说清**破坏的是哪一种力**：「SiO₂ is giant covalent AND SiCl₄ is simple molecular, so less energy is needed to overcome the intermolecular forces in SiCl₄ than the covalent bonds in SiO₂」（9701/21 Oct/Nov 2024 Q3(e)(iii)）。只说 "SiCl₄ has weak bonds" 是错的——SiCl₄ 熔化时 Si–Cl covalent bond 根本没断。

### F. 从数据推断 structure and bonding

题目给 melting point、conductivity（solid / molten / aqueous）、solubility、state at room temperature 或与 water 的反应，要你判断 structure 和 bonding，并**逐条说理由**。思路：

1. **melting point**：低（低于约 200 °C，或 room temperature 下是 liquid / gas）→ simple molecular；高 → giant（ionic、metallic 或 molecular 再看导电）。
2. **conductivity**：
   - solid 就导电（且 malleable、shiny）→ giant metallic（或 graphite）；
   - solid 不导电、molten / aqueous 导电 → giant ionic；
   - 都不导电 + melting point very high → giant molecular；
   - 都不导电 + melting point low → simple molecular。
3. 与 water 反应得 acidic solution、misty fumes（hydrolysis）→ 往往是 covalent chloride，simple molecular（chapter 9）。

判分时 structure 和 bonding 各占一分，**每一分都要「结论 AND 证据」**，例如 Bi₂O₃（9701/22 Feb/Mar 2024 Q1(b)(ii)）：

- M1 **giant AND** has a high melting point
- M2 **ionic AND** conducts electricity (only) when molten

> **LO 4.2.3** — deduce the type of structure and bonding present in a substance from given information

```python
# @title 🔬 推断 structure and bonding { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-15">
<h3>推断 structure and bonding</h3><p class="faint">待合并：<code>colab/src/ch04/deduce.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-15"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-17">
<h3>9701/22 Feb/Mar 2024</h3>
<p>Bismuth is an element in Group 15 of the Periodic Table. Bismuth reduces water to form bismuth oxide, Bi₂O₃.</p>
<div class="q"><h4>Q1(b)(ii) <span class="mark">[2]</span></h4>
<p>Bi₂O₃ is a yellow insoluble solid that melts at 1090 K. The molten compound conducts electricity. Deduce the structure and bonding of Bi₂O₃. Explain your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <b>giant AND</b> has a high melting point</li><li>M2 <b>ionic AND</b> conducts electricity (only) when molten</li></ul>
<p>每一分都是「结论 + 证据」：只写 "giant ionic" 没有理由得 0 分。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-17"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-18">
<h3>9701/21 May/June 2024</h3>
<p>Boron nitride is a white solid that melts above 2900 °C. Figure 2.2 shows part of the lattice structure of a crystal of boron nitride.</p>
<img src="" width="420" alt="">
<div class="q"><h4>Q2(d)(ii) <span class="mark">[1]</span></h4>
<p>Suggest the identity of another crystalline solid that has atoms arranged in layers similar to that of solid boron nitride.</p>
<button class="reveal">答案</button><div class="ans">
<p>graphite</p>
<p>（B 和 N 交替排成 hexagon 组成的平面 layers，layers 之间是 weak intermolecular forces——和 graphite 一样。）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-18"));</script>"""))
```

```python
# @title 📝 9701/21 Oct/Nov 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-19">
<h3>9701/21 Oct/Nov 2024</h3>
<p>The Group 14 elements show a change from non-metallic to metallic character down the group. Table 3.1 shows some properties of two Group 14 elements, C and Sn, in their standard states. The table is incomplete.</p>
<div class="q"><h4>Q3(a)(i) <span class="mark">[3]</span></h4>
<p>Complete Table 3.1.</p>
<table><tr><th>element</th><th>C (graphite)</th><th>Sn</th></tr><tr><td>state and appearance in standard state</td><td>grey shiny solid</td><td>silvery solid</td></tr><tr><td>electrical conductivity</td><td><input class="fill" data-ans="good|conductor|good conductor"></td><td>good</td></tr><tr><td>type of bonding</td><td><input class="fill" data-ans="covalent"></td><td>metallic</td></tr><tr><td>type of structure</td><td>giant</td><td><input class="fill" data-ans="giant"></td></tr></table>
<button class="reveal">答案</button><div class="ans">
<ul><li>C (graphite) electrical conductivity：good / conductor</li><li>C (graphite) type of bonding：covalent</li><li>Sn type of structure：giant</li></ul>
</div>
</div>
<div class="q"><h4>Q3(a)(ii) <span class="mark">[1]</span></h4>
<p>Identify the lattice structure shown by graphite.</p>
<button class="reveal">答案</button><div class="ans">
<p>giant molecular</p>
</div>
</div>
<div class="q"><h4>Q3(a)(iii) <span class="mark">[1]</span></h4>
<p>Explain why Sn has good electrical conductivity.</p>
<button class="reveal">答案</button><div class="ans">
<p>its delocalised electrons are free to move</p>
</div>
</div>
<div class="q"><h4>Q3(e)(iii) <span class="mark">[2]</span></h4>
<p>SiO₂ is a white solid that melts above 1700 °C. SiCl₄ is a colourless liquid at room temperature. Explain the difference in the melting points of these two compounds with reference to their structure and bonding.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 SiO₂ is <b>giant covalent AND</b> SiCl₄ is <b>simple covalent</b> / simple molecular / molecular</li><li>M2 so <b>less energy</b> is needed to overcome the <b>intermolecular forces</b> in SiCl₄ compared to the (covalent) <b>bonds</b> in SiO₂（or reverse argument）</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-19"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-20">
<h3>9701/22 Feb/Mar 2025</h3>
<p>The third compound of phosphorus and chlorine, <b>W</b>, has the molecular formula P₂Cl₄.</p>
<div class="q"><h4>Q1(a)(iv) <span class="mark">[2]</span></h4>
<p><b>W</b> is a liquid at room temperature and pressure. It reacts vigorously with water to form an acidic solution. Suggest the structure and bonding in <b>W</b>. Explain your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>(structure =) <b>simple / molecular</b>, because (it has a) low melting point / weak intermolecular forces</li><li>(bonding =) <b>covalent</b>, because (it is) hydrolysed</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-20"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-21">
<h3>9701/21 May/June 2025</h3>
<p>Solid sodium conducts electricity. Sodium oxide conducts electricity when molten but not when solid.</p>
<div class="q"><h4>Q1(a)(ii) <span class="mark">[1]</span></h4>
<p>Explain how solid sodium conducts electricity.</p>
<button class="reveal">答案</button><div class="ans">
<p>delocalised electrons move (through the structure)</p>
</div>
</div>
<div class="q"><h4>Q1(a)(iii) <span class="mark">[1]</span></h4>
<p>Explain why sodium oxide conducts electricity when molten.</p>
<button class="reveal">答案</button><div class="ans">
<p>(sodium and oxide) <b>ions</b> move (through the liquid)</p>
<p>写 "electrons move" 不给分——Na₂O 是 ionic，没有 delocalised electrons。solid 不导电是因为 ions 固定在 lattice 里。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-21"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-22">
<h3>9701/22 May/June 2025</h3>
<p>Diamond and graphite are both crystalline solids made from carbon atoms. Graphite conducts electricity. Diamond does not conduct electricity.</p>
<div class="q"><h4>Q1(a)(i) <span class="mark">[1]</span></h4>
<p>Name the type of lattice structure present in the crystalline solids diamond and graphite.</p>
<button class="reveal">答案</button><div class="ans">
<p>(both) giant molecular</p>
</div>
</div>
<div class="q"><h4>Q1(a)(ii) <span class="mark">[1]</span></h4>
<p>Explain how graphite conducts electricity.</p>
<button class="reveal">答案</button><div class="ans">
<p>delocalised electrons move through the structure <b>OR</b> delocalised electrons move through / between the layers</p>
</div>
</div>
<div class="q"><h4>Q1(a)(iii) <span class="mark">[1]</span></h4>
<p>Explain why diamond does not conduct electricity.</p>
<button class="reveal">答案</button><div class="ans">
<p>no delocalised electrons (to carry charge through the structure) <b>OR</b> all (4) outer electrons are used in bonding</p>
</div>
</div>
<div class="q"><h4>Q1(d)(i) <span class="mark">[3]</span></h4>
<p>The melting points of different oxides are shown in Table 1.1. Complete Table 1.1 by identifying the strongest force of attraction in each oxide that is broken during melting. Use the abbreviations: i.d. = instantaneous dipole–induced dipole; p.d. = permanent dipole–permanent dipole; H = hydrogen bond; C = covalent bond; I = ionic bond.</p>
<table><tr><th>oxide</th><th>melting point / °C</th><th>force of attraction broken during melting</th></tr><tr><td>SO₂</td><td>−73</td><td><input class="fill" data-ans="p.d.|pd"></td></tr><tr><td>H₂O</td><td>0</td><td><input class="fill" data-ans="H"></td></tr><tr><td>SiO₂</td><td>1610</td><td><input class="fill" data-ans="C"></td></tr><tr><td>Na₂O</td><td>1132</td><td><input class="fill" data-ans="I"></td></tr><tr><td>MgO</td><td>2852</td><td><input class="fill" data-ans="I"></td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p>SO₂ = p.d.，H₂O = H，SiO₂ = C，Na₂O = I，MgO = I</p>
<p>SO₂ 是 polar 的 bent molecule，最强的 intermolecular force 是 p.d.（id-id 也有，但题目问 strongest）。H₂O 熔化时破坏的是 hydrogen bonds，不是 O–H covalent bonds。</p>
</div>
</div>
<div class="q"><h4>Q1(d)(ii) <span class="mark">[2]</span></h4>
<p>A student suggests the following hypothesis: "The stronger the covalent bond between atoms in non-metal oxides, the higher the melting point." Use Table 1.1 to deduce if this hypothesis is true or false or if there is not enough information to make a conclusion. Explain your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <b>not enough information</b></li><li>M2 SO₂ and / or H₂O break intermolecular forces <b>OR</b> SO₂ and / or H₂O does not break covalent bonds <b>OR</b> only SiO₂ breaks covalent bonds</li></ul>
<p>（三个 non-metal oxide 里只有 SiO₂ 熔化时断 covalent bond，没法比较「covalent bond 强弱」对 melting point 的影响。）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-22"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-24">
<h3>9701/21 May/June 2026</h3>
<div class="q"><h4>Q2(c)(ii) <span class="mark">[4]</span></h4>
<p>equation 1　2CaCO₃ + 2SO₂ + O₂ → 2CaSO₄ + 2CO₂</p>
<p>Use the ideal gas equation to calculate the volume of O₂(g) that reacts with 0.500 mol of SO₂ at 400 °C and 101 kPa in equation 1. Include units in your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <i>pV</i> = <i>nRT</i></li><li>M2 <i>n</i>(O₂) = 0.500 ÷ 2 = 0.250 mol</li><li>M3 <i>T</i> = 400 + 273 = 673 K</li><li>M4 <i>V</i> = 0.250 × 8.31 × 673 ÷ 101 000 = <b>0.0138 m³</b>（= 13.8 dm³ = 13 800 cm³）</li></ul>
<p>判分：单位跟数值配套即可（0.014 m³ / 14 dm³ / 14 000 cm³ 都给满分）；273.15 也接受；M4 允许 ecf。</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-24"));</script>"""))
```

```python
# @title 📝 9701/23 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-25">
<h3>9701/23 May/June 2026</h3>
<p>Group 18 elements exist as colourless gases at room temperature and pressure.</p>
<table><tr><th>atom</th><th>relative volume of atom</th></tr><tr><td>neon</td><td>172</td></tr><tr><td>argon</td><td>287</td></tr><tr><td>krypton</td><td>355</td></tr><tr><td>xenon</td><td>446</td></tr></table>
<div class="q"><h4>Q2(a) <span class="mark">[2]</span></h4>
<p>Use Table 2.1 to describe and explain the variation in atomic radius down Group 18.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 (down the group) atomic radius increases <b>OR</b> atom volume increases <b>OR</b> Ne < Ar < Kr < Xe</li><li>M2 (down the group) the number of shells (of electrons) increases</li></ul>
<p>判分：shells = energy levels；只写 "sub-shells increase"、"shielding outweighs nuclear charge" 都不给 M2（这是 chapter 1 的内容）。</p>
</div>
</div>
<div class="q"><h4>Q2(b) <span class="mark">[3]</span></h4>
<p>Identify <b>one</b> Group 18 element from Table 2.1 that behaves most like an ideal gas under standard conditions. Explain your answer.</p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <b>neon</b> / Ne</li><li>M2 it has the <b>smallest</b> volume</li><li>M3 <b>weakest</b> van der Waals' / instantaneous dipole – induced dipole attractions / London (dispersion) forces</li></ul>
<p>判分：M2 和 M3 合起来要出现一次比较词（smaller / smallest、weaker / weakest / least），例如 "Ne has smaller volume and weak vdw"。用 mass 作 M2、只写 "smallest number of electrons" 都不给分。</p>
</div>
</div>
<div class="q"><h4>Q2(c) <span class="mark">[4]</span></h4>
<p>A sample of 1.775 g of gas X occupies 617.0 cm³ at 300 K and 101 kPa. Calculate the relative molecular mass, <i>M</i>ᵣ, of X assuming that X behaves as an ideal gas.</p>
<p><i>M</i>ᵣ of X = <input class="fill" data-ans="71.0|71"></p>
<button class="reveal">答案</button><div class="ans">
<ul><li>M1 <i>pV</i> = <i>nRT</i></li><li>M2 use of appropriate pressure value with volume：617.0 cm³ = 617.0 × 10⁻⁶ m³ 配 101 000 Pa（OR 101 kPa 配 617.0 × 10⁻³ dm³）</li><li>M3 <i>n</i> = 101 000 × 617.0 × 10⁻⁶ ÷ (8.31 × 300) = 0.024997 ≈ 0.0250 mol</li><li>M4 <i>M</i>ᵣ = 1.775 ÷ 0.024997 = <b>71.0</b></li></ul>
<p>判分：最后答案 71(.0) 四分全给；M3、M4 允许 ecf（单位换错了但后面算对，仍给后面的分）。（X 是 Cl₂。）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-25"));</script>"""))
```

```python
# @title ✅ Learning outcomes 自查清单 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
</style>
<div class="k" id="k-ch04-26">
<h3>Learning outcomes 自查 <span class="faint count"></span></h3><p class="faint">勾选会存在这个浏览器里（localStorage）；Colab 不允许时只在当前页面有效。</p><div class="lolist" data-key="ch04"><div class="bar"><i></i></div><h4>4.1</h4>
<label class="chk"><input type="checkbox" data-id="4.1.1"><span><b>4.1.1</b> explain the origin of pressure in a gas in terms of collisions between gas molecules and the wall of the container</span></label>
<label class="chk"><input type="checkbox" data-id="4.1.2"><span><b>4.1.2</b> understand that ideal gases have zero particle volume and no intermolecular forces of attraction</span></label>
<label class="chk"><input type="checkbox" data-id="4.1.3"><span><b>4.1.3</b> state and use the ideal gas equation <i>pV</i> = <i>nRT</i> in calculations, including in the determination of <i>M</i>ᵣ</span></label>
<h4>4.2</h4>
<label class="chk"><input type="checkbox" data-id="4.2.1"><span><b>4.2.1</b> describe, in simple terms, the lattice structure of a crystalline solid which is:<ul><li>(a) giant ionic, including sodium chloride and magnesium oxide</li><li>(b) simple molecular, including iodine, buckminsterfullerene C₆₀ and ice</li><li>(c) giant molecular, including silicon(IV) oxide, graphite and diamond</li><li>(d) giant metallic, including copper</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="4.2.2"><span><b>4.2.2</b> describe, interpret and predict the effect of different types of structure and bonding on the physical properties of substances, including melting point, boiling point, electrical conductivity and solubility</span></label>
<label class="chk"><input type="checkbox" data-id="4.2.3"><span><b>4.2.3</b> deduce the type of structure and bonding present in a substance from given information</span></label></div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch04-26"));</script>"""))
```
