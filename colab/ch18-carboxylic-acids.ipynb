carboxylic acid（–COOH，carboxyl group）有两面：作为 **acid**，它和 metal、alkali、carbonate 反应（18.1 B）；作为 **organic compound**，它和 alcohol 缩合成 ester、被 LiAlH₄ 还原成 alcohol（18.1 B、18.2）。ester 又能 hydrolyse 回 acid + alcohol。

## 18.1 Carboxylic acids

### A. 怎样制得

> **LO 18.1.1** — recall the reactions by which carboxylic acids can be produced:
> - (a) oxidation of primary alcohols and aldehydes with acidified K₂Cr₂O₇ or acidified KMnO₄ and refluxing
> - (b) hydrolysis of nitriles with dilute acid or dilute alkali followed by acidification
> - (c) hydrolysis of esters with dilute acid or dilute alkali and heat followed by acidification

| 起点 | reaction | reagents and conditions | LO | 详见 |
|---|---|---|---|---|
| alcohol | oxidation of 1° alcohol (full)<br>oxidation | acidified K₂Cr₂O₇ or acidified KMnO₄ and refluxing | 16.1.2(d)(ii) · 18.1.1(a) | [ch16 16.1 E →](#xref-missing-ch16-rx-oxidation) |
| aldehyde | oxidation of aldehyde<br>oxidation | acidified K₂Cr₂O₇ or acidified KMnO₄ and refluxing | 17.1.5 · 18.1.1(a) | [ch17 17.1 D →](#xref-missing-ch17-rx-tests) |
| ester | hydrolysis of esters<br>hydrolysis | dilute acid or dilute alkali and heat（alkali 之后要 acidification 才得 acid） | 16.1.1(f) · 18.1.1(c) · 18.2.2 | [ch18 18.2 B →](#scrollTo=ch18-rx-ester-hydrolysis) |
| nitrile | hydrolysis of nitriles<br>hydrolysis | dilute acid or dilute alkali followed by acidification | 18.1.1(b) · 19.2.3 | [ch19 19.2 B →](#xref-missing-ch19-rx-nitrile-hydrolysis) |

- (a) 1° alcohol 要 **reflux**（distillation 只到 aldehyde，[16.1 F](#xref-missing-ch16-rx-oxidation)）；
- (b)(c) 用 **dilute alkali** hydrolyse 时，得到的是 **carboxylate salt**（RCOO<sup>−</sup>Na<sup>+</sup>），必须再加 dilute acid **acidification** 才得到 RCOOH——「followed by acidification」是得分词；
- (b) nitrile 的 hydrolysis 详见 [19.2 B](#xref-missing-ch19-rx-nitrile-hydrolysis)；(c) ester hydrolysis 见本章 18.2 B。

命名：-oic acid，–COOH 的 C 是 C1（不写数字）：methanoic acid HCOOH、ethanoic acid CH₃COOH、propanoic acid CH₃CH₂COOH；salt 叫 -oate（sodium ethanoate CH₃COONa）。

carboxylic acid 是 **weak acid**：在水里只部分 dissociate，CH₃COOH ⇌ CH₃COO<sup>−</sup> + H<sup>+</sup>，pH < 7，但比同浓度 HCl 高。

### B. 反应

> **LO 18.1.2** — describe:
> - (a) the redox reaction with reactive metals to produce a salt and H₂(g)
> - (b) the neutralisation reaction with alkalis to produce a salt and H₂O(l)
> - (c) the acid–base reaction with carbonates to produce a salt and H₂O(l) and CO₂(g)
> - (d) esterification with alcohols with concentrated H₂SO₄ as catalyst
> - (e) reduction by LiAlH₄ to form a primary alcohol

| (x) | reagent | equation（ethanoic acid） | type | 观察 |
|---|---|---|---|---|
| (a) | reactive metal（Mg、Na） | 2CH₃COOH + Mg → (CH₃COO)₂Mg + H₂ | redox | effervescence（H₂），metal 溶解 |
| (b) | alkali（NaOH） | CH₃COOH + NaOH → CH₃COONa + H₂O | neutralisation | 无明显现象（可用 indicator 看） |
| (c) | carbonate（Na₂CO₃、NaHCO₃） | 2CH₃COOH + Na₂CO₃ → 2CH₃COONa + H₂O + CO₂ | acid–base | **effervescence（CO₂）** |
| (d) | alcohol + **concentrated H₂SO₄ as catalyst** | CH₃COOH + C₂H₅OH ⇌ CH₃COOC₂H₅ + H₂O | condensation（esterification） | 有水果香味 |
| (e) | **LiAlH₄**（dry ether） | CH₃COOH + 4[H] → CH₃CH₂OH + H₂O | reduction → **primary** alcohol | — |

- (c) 是**区分 acid 与 alcohol** 的经典 test：carbonate 只和 –COOH 反应放 CO₂，alcohol 不反应（Na(s) 两者都放 H₂）。判分（9701/22 May/June 2026 类题）：M1 reagent **Na₂CO₃ / NaHCO₃**；M2 **effervescence / fizzing in the acid（not the other）**；写「limewater turns milky」不给 M2；现象必须对应所写的 reagent。
- (e) 配平：COOH → CH₂OH 用 **4[H]**、放 1 H₂O；**NaBH₄ 不还原 COOH**。
- (d) 的细节（命名、可逆、条件）在 18.2 A。

```python
# @title 🔬 carboxylic acid + metal / alkali / carbonate / alcohol / LiAlH₄ { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-acid-reactions">
<h3>carboxylic acid + metal / alkali / carbonate / alcohol / LiAlH₄</h3><p class="faint">待合并：<code>colab/src/ch18/acid-reactions.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-acid-reactions"));</script>"""))
```

| ester | esterification<br>condensation | concentrated H₂SO₄ as catalyst | 16.1.2(f) · 18.1.2(d) · 18.2.1(a) | [ch18 18.2 A →](#scrollTo=ch18-rx-esterification) |

## 18.2 Esters

### A. esterification

> **LO 18.2.1** — recall the reaction (reagents and conditions) by which esters can be produced:
> - (a) the condensation reaction between an alcohol and a carboxylic acid with concentrated H₂SO₄ as catalyst

| 起点 | reaction | reagents and conditions | LO | 详见 |
|---|---|---|---|---|
| carboxylic acid, alcohol | esterification<br>condensation | concentrated H₂SO₄ as catalyst | 16.1.2(f) · 18.1.2(d) · 18.2.1(a) | [ch18 18.2 A →](#scrollTo=ch18-rx-esterification) |

carboxylic acid + alcohol，**concentrated H₂SO₄ as catalyst**，加热（reflux）→ ester + H₂O。是 **condensation**（两个 molecule 连在一起、放出 H₂O），而且**可逆**；conc. H₂SO₄ 既是 catalyst，也吸水使平衡右移。

**ester 的命名**：先写 **alcohol 那一半**的 alkyl（-yl），再写 **acid 那一半**的 -oate。

| acid | alcohol | ester | structural formula |
|---|---|---|---|
| ethanoic acid | ethanol | **ethyl ethanoate** | CH₃COOCH₂CH₃ |
| propanoic acid | methanol | **methyl propanoate** | CH₃CH₂COOCH₃ |
| methanoic acid | propan-1-ol | **propyl methanoate** | HCOOCH₂CH₂CH₃ |
| butanoic acid | ethanol | **ethyl butanoate** | CH₃CH₂CH₂COOCH₂CH₃ |

- structural formula 里 **acid 那一半写在前面**（RCOO–），名字里 **alcohol 那一半在前**——两者顺序相反，最容易写反。
- syllabus 范围：最多 6 + 6 个 C，**直链**。
- 判分（区分 acid / alcohol 的「named reagent」题）：写 named alcohol + (conc) H₂SO₄、或 named acid + (conc) H₂SO₄ 都可以。

### B. ester 的 hydrolysis

> **LO 18.2.2** — describe the hydrolysis of esters by dilute acid and by dilute alkali and heat

| | **dilute acid**（如 dilute HCl），heat | **dilute alkali**（如 NaOH(aq)），heat |
|---|---|---|
| equation | CH₃COOC₂H₅ + H₂O ⇌ CH₃COOH + C₂H₅OH | CH₃COOC₂H₅ + NaOH → CH₃COONa + C₂H₅OH |
| 可逆？ | **可逆**，得到平衡混合物（esterification 的逆反应） | **不可逆**，ester 完全反应（carboxylate 不再和 alcohol 反应） |
| product | carboxylic acid + alcohol | **carboxylate salt** + alcohol；要 acid 就再 acidify |

拆 ester：在 **C(=O)–O 的单键**处断开——C=O 那一侧变成 acid（或 carboxylate），O 那一侧变成 alcohol。例：methyl propanoate → propanoic acid + methanol。

```python
# @title 📝 9701/22 Feb/Mar 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-12">
<h3>9701/22 Feb/Mar 2024</h3>
<div class="q"><h4>Q4(c)(iii) <span class="mark">[2]</span></h4>
<p>T also has a single functional group. Use the information in (c)(i) and your answer to (c)(ii) to identify T and U. Draw the structures of T and U in the boxes.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b>（官方 MS 把这一问的题号误印成 "4(c)(iv)"， QP 上实际是 (c)(iii)，以 QP 为准）</p>
<ul><li>T = ethanol，HOCH₂CH₃</li><li>U = ethyl 2-methylbutanoate，CH₃CH₂CH(CH₃)COOCH₂CH₃（与 (c)(i) 的 7 个碳、S 的 2-methylbutanoic acid 结构一致）</li></ul>
<p>（结构图见 mark scheme）</p>
<p>（结构图见 mark scheme）</p>
<p><b>判分</b> — T、U 各 1 分</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-12"));</script>"""))
```

```python
# @title 📝 9701/23 May/June 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-13">
<h3>9701/23 May/June 2024</h3>
<div class="q"><h4>Q5(d)(iii) <span class="mark">[1]</span></h4>
<p>In step 3, X reacts with reducing agent Z to produce Y. Complete the equation for the reaction of X with Z. Use a molecular formula to represent the organic product. Use [H] to represent one atom of hydrogen from Z.</p>
<p>…… C₈H₁₂O₄ + ……[H] →</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — C₈H₁₂O₄ + 8[H] → C₈H₁₆O₂ + 2H₂O</p>
</div>
</div>
<div class="q"><h4>Q5(d)(iv) <span class="mark">[1]</span></h4>
<p>Identify Z.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — LiAlH₄</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-13"));</script>"""))
```

```python
# @title 📝 9701/22 Oct/Nov 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-14">
<h3>9701/22 Oct/Nov 2024</h3>
<p>Compound F reacts with reagent G to form compound H. The infrared spectrum of H is shown in Fig. 4.3. <img alt="Figure 4.3" src=""> Table 4.2</p>
<table><tr><th>bond</th><th>functional groups containing the bond</th><th>characteristic infrared absorption range (in wavenumbers) / cm⁻¹</th></tr><tr><td>C–O</td><td>hydroxy, ester</td><td>1040–1300</td></tr><tr><td>C=C</td><td>aromatic compound, alkene</td><td>1500–1680</td></tr><tr><td>C=O</td><td>amide</td><td>1640–1690</td></tr><tr><td>C=O</td><td>carbonyl, carboxyl</td><td>1670–1740</td></tr><tr><td>C=O</td><td>ester</td><td>1710–1750</td></tr><tr><td>C≡N</td><td>nitrile</td><td>2200–2250</td></tr><tr><td>C–H</td><td>alkane</td><td>2850–2950</td></tr><tr><td>N–H</td><td>amine, amide</td><td>3300–3500</td></tr><tr><td>O–H</td><td>carboxyl</td><td>2500–3000</td></tr><tr><td>O–H</td><td>hydroxy</td><td>3200–3600</td></tr></table>
<p>H also shows a molecular ion peak at <i>m</i>/<i>e</i> = 60 in its mass spectrum.</p>
<div class="q"><h4>Q4(e)(ii) <span class="mark">[1]</span></h4>
<p>Suggest the role of reagent G.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — oxidising agent（把 aldehyde F 氧化成 carboxylic acid H）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-14"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-15">
<h3>9701/22 Feb/Mar 2025</h3>
<div class="q"><h4>Q4(a) <span class="mark">[5]</span></h4>
<p>Table 4.1 gives details of tests on J to M. In each test, only two compounds give a positive result. Complete Table 4.1.</p>
<table><tr><th>reagent</th><th>observation of positive result</th><th>compounds that give a positive result</th></tr><tr><td>acidified K₂Cr₂O₇(aq)</td><td><i>(空)</i></td><td>J and L</td></tr><tr><td>alkaline I₂(aq)</td><td>yellow precipitate</td><td><i>(空)</i></td></tr><tr><td><i>(空)</i></td><td>orange precipitate</td><td>J and K</td></tr><tr><td>Na(s)</td><td><i>(空)</i></td><td><i>(空)</i></td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — 完整表格：</p>
<table><tr><th>reagent</th><th>observation of positive result</th><th>compounds that give a positive result</th></tr><tr><td>acidified K₂Cr₂O₇(aq)</td><td>(solution) turns (from orange to) green</td><td>J and L</td></tr><tr><td>alkaline I₂(aq)</td><td>yellow precipitate (forms)</td><td>K and L</td></tr><tr><td>2,4-dinitrophenylhydrazine / 2,4-DNPH</td><td>orange precipitate (forms)</td><td>J and K</td></tr><tr><td>Na(s)</td><td>effervescence</td><td>L and M</td></tr></table>
<p><b>判分</b> — 5 个空各 1 分：K₂Cr₂O₇ 一行的 observation、I₂ 一行的 compounds、reagent 一行填 2,4-DNPH、Na(s) 一行的 observation 与 compounds</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-15"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-16">
<h3>9701/22 May/June 2025</h3>
<div class="q"><h4>Q4(a)(i) <span class="mark">[1]</span></h4>
<p>Propanoic acid, CH₃CH₂COOH, reacts with reducing agent Q to produce propan-1-ol, CH₃CH₂CH₂OH. Suggest the formula of reducing agent Q.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — LiAlH₄</p>
</div>
</div>
<div class="q"><h4>Q4(a)(ii) <span class="mark">[1]</span></h4>
<p>Complete the equation to show the reduction of propanoic acid to propan-1-ol. Use [H] to represent one atom of hydrogen from Q.</p>
<p>……CH₃CH₂COOH + ……[H] →</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — CH₃CH₂COOH + 4[H] → CH₃CH₂CH₂OH + H₂O</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-16"));</script>"""))
```

```python
# @title 📝 9701/23 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-17">
<h3>9701/23 May/June 2025</h3>
<p>W is a colourless liquid. <img alt="Figure 5.1" src="data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAACCgAAAGkCAIAAAB2WhFoAAAACXBIWXMAAC4jAAAuIwF4pT92AAAoZ0lEQVR4nO3d93uUZb748fOfLJBCKImoWCACgoqgHkVRsaxY1i5WRARdBSyAZS0o4loQxbJgWXEtqDQFQaUoiroKRBABKQshAQIkOd9cX/eafXaGhCRzzzyT5PW6nh/OydzP7WeG3573zj3/838AAAAAAACB/E/cAwAAAAAAAK2H8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAMChrVi+/JQBA5KuZ/7614bvuuWmm1LvqruWLF7cwF3btm1LveWFqVNDvh8AACBjhAcAAODQdpWX57Vr1+EPf4he5517bgO37N69uzAvL+mW368J48c3cONHH36YekvdH0O/JwAAICOEBwAAoFEGnnxyUgzoUlRUXV1d3/p5c+cetDrUXWeecUYD/6GJEyak3rJjx44MvCcAACA84QEAAGiUP99xR2oP+P677+pbf8/YsfWFh4IOHSorK+u78fwhQ5LWn3zSSZl5TwAAQHjCAwAA0Ciz3n47NSG8+sor9a0f0L9/feGh7po/f/5B76quru5SVJS0+M7RozP2tgAAgMCEBwAAoFE2btyY2g9Gjhhx0MVbt2xpoDo08DMPq1atSl38zqxZmXxnAABASMIDAADQWKU9ejTyEKS3//73hsNDfT/z8PL06amLN23alMm3BQAAhCQ8AAAAjXXj9dcnJYG8du0qKipSV946fHjD4aG+n3kYfvPNSSv7HHdc5t8ZAAAQjPAAAAA01vSXXkpNCJ8tWpS6svTYY5OWHV5S0pifeeh3/PFJy2695ZbMvzMAACAY4QEAAGisH77/PjU8TH7yyaRla1avTlrTt3fv22+77ZA/87Bjx47U/Wf87W/Zen8AAEAAwgMAANBYtbW13YqLk8LAlZdfnrTsxWnTktbcMXr0O7NmHfJnHubNnZsaHsrKyrL09gAAgBCEBwAAoAkuvfjipDBwdPfuSWuuuuKKpDXvv/fe9u3bU3/mIen3IR5+6KGkNcekbA4AAOQ44QEAAGiCpyZPTv1Swq8bNiQWVFdXH/bf34rIa9du586ddS+dOnBg0o3z5s2Lbv7HCy5IWjDs2muz/AYBAIA0CQ8AAEATfPnFF6nh4R/vvJNYsGL58qRXTz/ttN9fuv/ee5NeGn///Ykba2pqSrp0SVow7YUXsv0OAQCA9AgPAABAE1RVVRUVFCTlgXvHjUsseGLSpKRXJ06Y8PtLC+bPT3pp0OmnJ2785w8/pCaN77/7LtvvEAAASI/wAAAANM3ZZ52VlAfq/pJ49bxzz016ddHChb+/tHfv3qRoEf2Zh9defTXpxsOKi2tqamJ4hwAAQBqEBwAAoGkmjB+fVAg6FRYeOHCg7qU9e/Z0zM+PvlRUWFhVVZW49/whQ+r7mYeRI0YkvfSnSy+N5Q0CAADpEB4AAICmmfPxx6lnIn3zzTd1L81POUzpwvPPj96behBT4mce+p9wQtJLT0+ZEsPbAwAA0iM8AAAATbNz587U8DD9pZfqXrrvnnuS/v7U5MnRe1N/evr3n3nYVV6e165d0kvLly+P5x0CAABpEB4AAIAmO/nEE5MiwfCbb677+ykDBiT9feXXX0dvrK6uPqy4OLogv337ioqKTxYsSLqxc8eO+/fvj+n9AQAAzSc8AAAATTZ61KikTnBC377btm1L+uPhJSWpvw591RVXpP7Mw+OPPpr0xwvOOy+WtwYAAKRJeAAAAJrsrTffTD1t6eXp05P+cs1VV6Xe++K0aak/83Dp0KFJf3zskUey/74AAID0CQ8AAECTbfjll9Tw0Lu0NDVFpN67ZvXqpGVn/O//Hl5SkvTHzxYtyv77AgAA0ic8AAAAzVF67LGp7SHpKisra969hXl5e/bsye4bAgAAwhAeAACA5rh+2LCG48FxPXvWd++tw4c3fO9ZgwZl870AAAABCQ8AAEBzpP5UQ9J126231nfv3996q+F7x99/fzbfCwAAEJDwAAAANMd3q1Y1HA9mvf12ffdu3bKl4XvnzpmTzfcCAAAEJDwAAADNUVNTU9K1awPxYNvWrQ3cfvJJJ9V3Y167duXl5Vl7IwAAQFjCAwAA0EyXDB1aXzwYePLJDd87buzY+u49ZcCA7MwPAABkgvAAAAA00xOTJtUXD+4dN67he+fOmVPfvXf/+c/ZmR8AAMgE4QEAAGimJYsX1xcP5s2b1/C9lZWVhXl5B7333X/8IyvjAwAAGSE8AAAAzVRVVdUxPz+1HNT9cffu3Ye8/ZzBgw8aHrZs2ZKF4QEAgAwRHgAAgOY7a9Cg1HIw5JxzGnPvY488knpv3z59MjwyAACQWcIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAABADDZv3vztt9/Onz//nVmz5s6Zs2zZsl3l5XEPBQAABCA8AAAAWXLgwIEPZ8++/bbbehx9dIc//CH16tu797ixY1csX96kbW+56aZTBgz4/Xp6ypSmTvXaq68mbr906NCm3g4AACQRHgAAgGx47913+x1//EF7Q+p1zuDBSxYvbuTOZ55xRuLGsWPGNHWwSY8/nrj9uJ49m3o7AACQRHgAAAAya/v27eede24jk0P0evzRR2traw+5v/AAAAA5RXgAAAAyaM3q1cf36pUaFQadfvq4sWOfmjz5xWnTJowff90115zUr1/qsmuvvnr37t0N/yeEBwAAyCnCAwAAkCllZWWHl5REQ0LXTp0eeuCBbVu3HnT950uWXPGnPyW1h9MGDtyzZ08D/xXhAQAAcorwAAAAZERVVdWpAwdGE8IFQ4Zs2bLlkDd+tmjRYcXF0RvH3HVXA+uFBwAAyCnCAwAAkBGjb789Gg/uGD26urq6kff+9NNPvUpLo7d/tmhRfYuFBwAAyCnCAwAAEN5PP/0UzQYXnn9+46vD79asXt0xPz+xQ2mPHhUVFQddKTwAAEBOER4AAIDwRgwfnniaX9y587Zt25qxycMPPRStF6++8spBlwkPAACQU4QHAAAgsC1btkS/rDDm7rubt09VVVX0wKULhgw56DLhAQAAcorwAAAABDZzxozoNxXKysqavdWDEycm9slr127z5s2pa4QHAADIKcIDAAAQ2K2Rc5b69u6dzlYrV66MNoznn3sudY3wAAAAOUV4AAAAAju+V6/Eo/ybbrghzd2O69kzsdt111yTukB4AACAnCI8AAAAIe3evfuQ31Fokssvuyyx29lnnZW6IBoervjTn+Z8/HGTrptvvFF4AACAgIQHAAAgpF83bIiGh3dmzUpzw9tvuy2xW6/S0tQF0fCQ5iU8AABA+oQHAAAgpFWrVkUf5c+fPz/NDR9+8MHEbkUFBakLhAcAAMgpwgMAABDSooULo4/yly9fnuaGLzz/fHTDnTt3Ji0QHgAAIKcIDwAAQEhLFi+OPspfkXZ4mDJ5cnTDysrKpAXR8HDqwIFjx4xp0jV40CDhAQAAAhIeAACAkNatWxftBPPmzUtzwwnjxyd2y2vXrra2NmlBNDyMHTOmqftPevxx4QEAAAISHgAAgJD27dsXDQ9/f+utNDccOWJEYrceRx+dukB4AACAnCI8AAAAgXU/4ojEo/wJ48enudtpAwcmdrvowgtTFwgPAACQU4QHAAAgsIsuvDDxKH/woEHpbLV79+789u0Tuz0wYULqGuEBAAByivAAAAAENnPGjMSj/MK8vO3btzd7q9mzZ0cPbvr+u+9S1wgPAACQU4QHAAAgsF27dhUVFiae5j/2yCPN3uqCIUMS+wzo3/+ga4QHAADIKcIDAAAQ3jVXXZV4mn9kt27l5eXN2GTF8uXRrzs8/dRTB10mPAAAQE4RHgAAgPCWLVuW165d4oH+VVdc0dQdKisr+/bundihqLBw06ZNB10pPAAAQE4RHgAAgIy4d9y46PcVmnTg0r59+66+8sro7S9MnVrfYuEBAAByivAAAABkRFVV1Ql9+0bjwZ2jR1dXVx/yxvLy8uhPO9RdF190UW1tbX3rhQcAAMgpwgMAAJApX61Y0bljx2hCOPnEE+fPn1/f+urq6ldefvmoI4+M3nJkt26//fZbA/8V4QEAAHKK8AAAAGTQqlWrSnv0iIaEuqtvnz73jhv37j/+8fmSJd9/993CTz998403Rgwf3v2II5JW1v1l5ddfN/yfEB4AACCnCA8AAEBmbdu27dyzz04qCo25zho0aN26dYfcX3gAAICcIjwAAAAZt3///kmPPVbSpUsjk0Pf3r1fevHFmpqaxmwuPAAAQE4RHgAAgCz517/+9fSUKacOHFhfbygqKLh+2LDPPvusgZ+STiU8AABAThEeAACAbNu2dev8+fNfmDp10mOPTZww4cknnnjl5ZeXLl26e/fuuEcDAADSJTwAAADhfbVixdKlS+OeAgAAiIHwAAAABLZ3796+ffrktWs35q67Kioq4h4HAADIKuEBAAAI7N5x4xK/mlDao0fZ2rVxTwQAAGSP8AAAAIS0bNmy/PbtE+Hh9FNPPXDgQNxDAQAA2SM8AAAAwVRVVZ3Ur1+iOnTMz//h++/jHgoAAMgq4QEAAAjmwYkTE9Wh7nr80UfjnggAAMg24QEAAAhj5cqVBR06JKrDqQMH7t+/P+6hAACAbBMeAACAAPbt2zegf/9EdSjMy1u1alXcQwEAADEQHgAAgAAe+ctfoocs/eXhh+OeCAAAiIfwAAAApOuH77/vmJ+fqA4D+vfft29f3EMBAADxEB4AAIC0HDhw4PRTT01Uh/z27b/+6qu4hwIAAGIjPAAAAGl58oknoocsjb///rgnAgAA4iQ8AAAAzffjjz8WFRQkqsOJ/fpVVVXFPRQAABAn4QEAAGim6urqwYMGRQ9ZWrZsWdxDAQAAMRMeAACAZnr2mWeihyzdO25c3BMBAADxEx4AAIDmKFu7tlNhYaI69O3TZ+/evXEPBQAAxE94AAAAmqympuacwYMT1SGvXbsvv/gi7qEAAICcIDwAAABN9uK0adFDlsbcdVfcEwEAALlCeAAAAJpm/fr1XTt1SlSH3qWllZWVcQ8FAADkCuEBAABogtra2gvPPz/6dYdFCxfGPRQAAJBDhAcAAKAJXpk+PVodRo8aFfdEAABAbhEeAABo/fbs2bN2zZovPv/8g/fff/+99xYtXFj3/9bU1MQ9V8uzcePGki5dEtWhtEePioqKuIcCAAByi/AAAECrtW7duqenTDnv3HMLOnSI/o/0f7+6duo09I9/fO3VV8vLy+OetMW4dOjQ6Gf4yYIFcU8EAADkHOEBAIBW6Lfffht9++0H7Q2pV0mXLg898ID/5f4hvT5zZvRzGzliRNwTAQAAuUh4AACgtXl95szOHTs2JjlErxP79Stbuzbu2XPX5s2buxUXJz6unsccs8s3RQAAgIMRHgAAaD1qa2sfeuCB1KhwZLduNwwb9vCDD74wdeqUyZNH3377Hy+4oEtRUdKyw4qL58+fH/ebyFFXXn559LOa8/HHcU8EAADkKOEBAIDWY8xddyW1hMGDBn36yScH/R3pnTt3PjV58rFHHRVdn9++/d/feiv7k+e4d2bNin5KN994Y9wTAQAAuUt4AACglZj19tvRh+OdCgtfnznzkHdVVVUNu/ba6I3FnTv/8ssvWRi4pdi2deuR3bolPp+ju3ffsWNH3EMBAAC5S3gAAKA1WLtmTddOnRIPx7sffviqVasaeW9tbe3ECROi7eH8IUPq/pjRgVuQa6++OvrhzP7gg7gnAgAAcprwAABAa3D9dddFj0tatHBhU3cYfvPN0cfrLzz/fCbmbHHef++96Mdy/bBhcU8EAADkOuEBAIAWb/369fnt2ycejk+cMKEZm2zfvv3wkpLEJj2POeagX3r4+eefp0ye3EauRx5+uPvhh//neyRHHFH3KaX9zwUAALRywgMAAC1e9DelO+bnb92ypXn7vDJ9evR/3f/5kiWpa+bOmZP0+9Vt53rv3XfT+4cCAADaBOEBAIAWr7RHj8TD8ZtvvLHZ+1RWVhYVFia2umP06NQ1bTY8XHPVVWn8EwEAAG2I8AAAQMv264YN0efjf3/rrXR2+9Ollya2OrJbtwMHDiQtaJvh4fCSki3N/R4JAADQ1ggPAAC0bG+9+Wb0EXlZWVk6u82cMSO6248//pi0YMXy5X+84IJWfx171FHRz6HuQ07nUwUAANoU4QEAgJbt/nvvTTwfP6y4OM3dvvnmm+gD94WffhpkyJblkwULoh/C5ZddFvdEAABASyI8AADQso0YPjzxiPzEfv3S3G3Tpk3RZ+6vz5wZZMgWpKKiIvqbGd2Kizdv3hz3UAAAQEsiPAAA0LJdefnliafkZ55xRpq7HThwIBoeJj/5ZJAhW5DRo0a18fQCAACkSXgAAKBlO2fw4MRT8osvuij9DbsVFyc2HHP33elv2IIsWrgwWh0uHTo07okAAICWR3gAAKBlO3/IkMSD8ktCPCgvKixMbHjfPfekv2EL8t6770bDw/XDhsU9EQAA0PIIDwAAtGy33HRT4kH5oNNPT3O3qqqq6JP3xx99NMiQLcjll10W/QQ+/uijuCcCAABaGOEBAICW7cGJExNPyfv27p3mbhs3bow+dn/t1VeDDNmCbNq0qaRr18Qn0POYY3aVl8c9FAAA0JIIDwAAtGzTX3op8ZS8oEOHysrKdHb76MMPo+Fh6dKlSQv27NmzrrV7YtKk6IcwetSodD5SAACgrREeAABo2ZYuXRp9Sv7JggXp7DZh/PjEVkUFBRUVFUkL5s6ZE/3PtZFryeLF6XyqAABAmyI8AADQstXW1h7Xs2fiEfk9Y8ems9spAwYktrrqiitSF7TN8NC3T5+9e/em88ECAABth/AAAECLN3HChMQj8pIuXXbt2tW8fT5fsiT6tP2D999PXdM2w0PdNf7++9P7VwIAANoK4QEAgBbvnz/8EH1E/sSkSc3b59KhQxObHFZcXFVVlbqmzYaHgg4dVq5cmd4/FAAA0CYIDwAAtAaXRJpBYV7eVytWNHWHl6dPjz5nv/vPfz7osv379+9sM7788su6DzPxmZw6cGDd20/73woAAGjlhAcAAFqDjRs3lnTtmnhEflzPnht++aXxt3/x+eedCgsTt/cqLW32eU2tzMMPPRTtMU8+8UTcEwEAALlOeAAAoJWYOWNG9BH50d27N/JooH+8805RQUHixvz27b/84otMT9tSVFVV9T/hhMSHU1RYuHr16riHAgAAcprwAABA63Ht1VdH20NhXt7YMWN27NhR3/qff/75qiuuSPoxgwcnTszmzLlv2bJlee3aJT6fcwYPrqmpiXsoAAAgdwkPAAC0HtXV1feMHZsUEooKCy+75JKXXnxx3ty533777fLly2fPnv3U5Mlnn3lm9Hn679eokSPrNon7feScsWPGRD+lF6dNi3siAAAgdwkPAAC0Nq/PnBk9OqmRV1Fh4bPPPBP37DmqsrKyd2lp4rMq7tz51w0b4h4KAADIUcIDAACt0MqVK4ecc04jk0PH/Pxh115bVlYW99Q57ZMFC6If2qVDh8Y9EQAAkKOEBwAAWq1FCxfectNNxZ0715ccTujb969PP71t27a4J20Zbh0+PPrpvfXmm3FPBAAA5CLhAQCAVu7AgQPfrVr1xuuvPz1lyoMTJz780EPPPvPMB++/v27durhHa2F27tx5dPfuifBwZLdu27ZujXsoAAAg5wgPAABAY73/3nvRLz1cP2xY3BMBAAA5R3gAAACa4Oorr4y2h48/+ijuiQAAgNwiPAAAAE3w22+/dSsuToSHnsccs6u8PO6hAACAHCI8AAAATTNzxozolx5GjxoV90QAAEAOER4AAICmqa2tvejCC6PtYcnixXEPBQAA5ArhAQAAaLL169d3KSpKhIe+vXvv3bs37qEAAICcIDwAAADNMfW556Jfehh/331xTwQAAOQE4QEAAGiO6urqwYMGJcJDfvv2K1eujHsoAAAgfsIDAADQTP/85z+LCgoS7eGUAQP2798f91AAAEDMhAcAAKD5Hn/00eiBS09MmhT3RAAAQMyEBwAAoPn27ds3oH//RHgoKihYvXp13EMBAABxEh4AAIC0fP3VV/nt2yfaw9lnnVVTUxP3UAAAQGyEBwAAIF333XNP9MClaS+8EPdEAABAbIQHAAAgXXv27Dm+V69EeOjaqdOvGzbEPRQAABAP4QEAAAjgs88+i37p4ZKhQ+OeCAAAiIfwAAAAhDFq5Mhoe3jzjTfinggAAIiB8AAAAISxq7z82KOOSoSHIw47bNvWrXEPBQAAZJvwAAAABPPh7NnRLz1cf911cU8EAABkm/AAAACEdP1110Xbw0cffhj3RAAAQFYJDwAAQEjbtm494rDDEuGhx9FH7yovj3soAAAge4QHAAAgsDffeCP6pYfRt98e90QAAED2CA8AAEB4lw4dGm0PixcvjnsiAAAgS4QHAAAgvF83bCju3DkRHvr27r137964hwIAALJBeAAAADLixWnTol96GH/ffXFPBAAAZIPwAAAAZERNTc05gwcnwkN++/YrV66MeygAACDjhAcAACBT1qxeXVRYmGgPpwwYsH///riHAgAAMkt4AAAAMmjyk09GD1x6YtKkuCcCAAAyS3gAAAAy6MCBA6cOHJgID0UFBT/99FPcQwEAABkkPAAAAJn1zTffFHTokGgPZ595Zk1NTdxDAQAAmSI8AAAAGTdxwoTogUsvTJ0a90QAAECmCA8AAEDGVVVV9Tv++ER46Nqp04Zffol7KAAAICOEBwAAIBu++Pzz6JceLr7oorgnAgAAMkJ4AAAAsuTPd9wRbQ9vvP563BMBAADhCQ8AAECWVFRUlB57bCI8HF5Ssm3r1riHAgAAAhMeAACA7Jk3d270Sw/Drr027okAAIDAhAcAACCrbrrhhmh7+HD27LgnAgAAQhIeAACArNq+fXv3I45IhIdjjzpqV3l53EMBAADBCA8AAEC2vTNrVvRLD6NGjox7IgAAIBjhAQAAiMHll10WbQ+fffZZ3BMBAABhCA8AAEAMNm7cWNK1ayI8HN+r1969e+MeCgAACEB4AAAA4vHKyy9Hv/Rw/733xj0RAAAQgPAAAADEo7a29vwhQxLhIb99+5Vffx33UAAAQLqEBwAAIDZlZWWdCgsT7WHgySfv378/7qEAAIC0CA8AAECcnp4yJXrg0qTHH497IgAAIC3CAwAAEKeysrL89u0T4aGooOCnn36KeygAAKD5hAcAACA2tbW1F55/fvQbD3XX2WeeWVNTE/doAABAMwkPAABAbF6cNi2pOvx+vTB1atyjAQAAzSQ8AAAA8Vi/fn3XTp0SsaEwLy/xf9f9/aMPP1z46aeNvFZ+/fXKRvvmm2/WNcUvv/yysykqKiri/mgBACBOwgMAABCD2traC4YMiX7L4flnn81r1+6gX4Bo3deN118f978GAACEJDwAAAAxSDpkaeSIEXV/HHP33bFnAOEBAADSJDwAAADZlnTIUmmPHr8fT1RZWdmrtDT2EiA8AABAOoQHAAAgq2pra88799zok/dPFixIvDp//vw+xx13eElJt+Liuuuw4uKk85dKunb9/aXoFXs8EB4AACBBeAAAALIq6ZCl0bff3sDiu+68M7r4lenT0x+gtra2ST8WvX379ib9GPXaNWsa/0vXddavX5/+mwIAgNwhPAAAANnz888/H/SQpYNatHBhtDoM/eMfa2trszktAADQDMIDAACQJamHLC389NP6FldWVvaO/N5DcefOv27YkM1pAQCA5hEeAACALEk+ZGnUqAYW3zl69H8dsvTyy1mbEwAASIfwAAAAZMPPP//cpagoERJ6lZY6ZAkAAFol4QEAAMi41EOWFi1cWN9ihywBAECLJjwAAAAZ98LUqdHqcMfo0Q0sHj1qVHTxq6+8krU5AQCA9AkPAABAZiUdstS7tLSysrK+xQ5ZAgCAlk54AAAAMqhJhyxVVFT0csgSAAC0cMIDAACQQVOfey5aHe66884GFicdsvTaq69mbU4AACAU4QEAAMiUJh2ytPDTTx2yBAAArYDwAAAAZERNTc2Qc86JtoTFixfXt7iioqK0Rw+HLAEAQCsgPAAAABmRziFLf3vttazNCQAAhCU8AAAA4SUdsnR8r14NHLL0yYIFDlkCAIBWQ3gAAAACq6mpOffss6MtYUmjD1kq6dLFIUsAANCiCQ8AAEBgSYcsjbn77gYWjxo50iFLAADQmggPAABASGVlZUmHLO3evbu+xQ5ZAgCA1kd4AAAAgqmpqTln8OBESMhr1+7zJUvqW5x8yFLXrg5ZAgCAVkB4AAAAgkk6ZGnsmDENLB45YoRDlgAAoPURHgAAgDDK1q7t3LFjIiT07d3bIUsAANAGCQ8AAEAAqYcsffnFF/Ut3lVennzI0q+/ZnNaAAAgc4QHAAAggOcdsgQAAPx/wgMAAJCusrVrOxUW/ueQpT599u7dW9/iBfPnO2QJAABaMeEBAABIS1MPWep5zDHRQ5Y2btyYzWkBAIBMEx4AAIC0PPfss9FvMNw7blwDix2yBAAArZ7wAAAANF+TDlmaN29etDpcfNFFDlkCAIDWR3gAAACaqbq6OnrIUn779kuXLq1vsUOWAACgjRAeAACAZnr2mWcaf8jSbbfe6pAlAABoC4QHAACgOZIOWTqhb9/GH7J0ydChDlkCAIDWSngAAACarLq6+uwzz4wesrRs2bL6FjtkCQAA2hThAQAAaLKkQ5buu+eeBhbfOny4Q5YAAKDtEB4AAICmWbtmTfSQpRP79auqqqpvcdIhS5c6ZAkAAFo74QEAAGiC6urqwYMGNfKQpZ07dzpkCQAA2hrhAQAAaIKkQ5bG339/A4sdsgQAAG2Q8AAAADTWmtWro4csndTwIUtz5yYdspTNUQEAgLgIDwAAQKOkHrL01YoV9S3euXPnsUcd5ZAlAABog4QHAACgUdI5ZGnG3/6WtTkBAIB4CQ8AAMChrV69uihyyNLJJ57YwCFLcz7+2CFLAADQZgkPAADAIThkCQAAaDzhAQAAOIS/Pv109BsMD0yY0MDiW2+5Jbp45owZWZsTAADIBcIDAADQkCYdsvTxRx85ZAkAANo44QEAAKhX0iFLBR06fP3VV/UtdsgSAADwf8IDAADQgKefeir6DYYHJ05sYPEtN90UXfz6zJlZmxMAAMgdwgMAAHBwyYcsnXSSQ5YAAIBDEh4AAICDqK6uPuu/D1lauXJlfYt37NgRPWTpsOJihywBAECbJTwAAAAHkXTI0kMPPNDA4ptvvNEhSwAAwO+EBwAAIFnSIUsD+vfft29ffYs/+vBDhywBAAAJwgMAAPBfqqurB51+euMPWTq6e/fE4m4OWQIAgDZPeAAAAP5L0iFLf3n44QYWO2QJAABIIjwAAAD/8eOPPzb+kKXZs2c7ZAkAAEgiPAAAAP+WdMhSYV7et99+W9/i1EOWNm3alM1pAQCA3CQ8AAAA/9akQ5ZuuuEGhywBAACphAcAAODfpj73XOKcpVMGDGjokKUPPnDIEgAAcFDCAwAA8B8//vjj6aedVpiXt2rVqvrWpB6ytHnz5mwOCQAA5DLhAQAA+C8HDhxYsnhxAwtuGDbMIUsAAEB9hAcAAKAJkg9ZuvjiuCcCAAByi/AAAAA01r/+9S+HLAEAAA0THgAAgMa63iFLAADAoQgPAABAoyQdsnTZJZfEPREAAJCLhAcAAKBRfvj++9MGDnTIEgAA0DDhAQAAaKz9+/c//uijhXl5DlkCAADqIzwAAABNU7Z2bdwjAAAAuUt4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgmP8HP6i5seMUZBIAAAAASUVORK5CYII="> W = CH₂=C(CH₃)COOH（2-methylprop-2-enoic acid / methacrylic acid）。</p>
<div class="q"><h4>Q5(a)(ii) <span class="mark">[4]</span></h4>
<p>Two different reagents are each added to separate samples of W as shown in Table 5.1. Complete Table 5.1.</p>
<p><b>Table 5.1</b></p>
<table><tr><th>reagent and conditions</th><th>observation when reagent is added to W</th><th>structural formula of organic product</th></tr><tr><td>solid Na</td><td>fizz / bubbles etc <b>OR</b> solid Na disappears / dissolves</td><td>CH₂=C(CH₃)CO₂Na <b>OR</b> CH₂=C(CH₃)CO₂⁻Na⁺</td></tr><tr><td>Br₂ in the absence of ultraviolet light</td><td>brown to colourless / decolourise</td><td>CH₂BrCBr(CH₃)CO₂H</td></tr></table>
<p>Na(s) 与 W 的 carboxyl 反应放出 H₂；Br₂ 在暗处（无 UV）对 C=C 做 electrophilic addition，不是 free-radical substitution。</p>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-17"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-18">
<h3>9701/22 Feb/Mar 2026</h3>
<div class="q"><h4>Q4(d)(i) <span class="mark">[1]</span></h4>
<p>Identify the type of reaction that occurs in step 1.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — condensation</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-18"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-19">
<h3>9701/22 May/June 2026</h3>
<p>An unlabelled sample contains one of compounds P, R or S. <img alt="Figure 6.4" src=""> The mass spectrum of the sample is shown in Figure 6.5. <img alt="Figure 6.5" src=""></p>
<div class="q"><h4>Q6(c)(iii) <span class="mark">[2]</span></h4>
<p>Suggest one reagent that is used to distinguish between samples of R and S. Include appropriate observations.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b>（任选一组）</p>
<ul><li>M1 Br₂(aq) / bromine water　　M2 orange / brown to colourless in S (not R)</li><li>或　M1 Na₂CO₃ / sodium carbonate / NaHCO₃ / sodium hydrogencarbonate　　M2 fizzing / effervescence in R (not S)</li></ul>
<p><b>判分</b></p>
<ul><li><b>R</b> 大小写只扣一次，如 KMNO₄</li><li>M2 的现象必须对应 M1 试剂的明确指认，例：M1 Br₂ → M2 decolourises S ✓； M1 K₂Cr₂O₇ / Cr₂O₇²⁻（不加酸）→ M2 (orange) to green ✓；M1 Na₂CO₃ → M2 bubbles in R ✓ 等</li><li>M2 <b>I</b> 直接说 alkene for S / carboxylic acid for R，但 A alcohol for S</li><li>M1 A bromine / Br₂；A aqueous iodine / aqueous I₂ / iodine water（不能只写 iodine）； <b>R</b> iodoform</li><li>M2 A decolourises；A yellow to colourless；<b>R</b> red；A bubbles；<b>I</b> limewater (turns milky)</li><li><b>同样给 2 分的其他试剂</b>（MS 续页 Alternatives）：</li><li>acidified potassium dichromate / H⁺ AND K₂Cr₂O₇（BOD H⁺ / Cr₂O₇²⁻） → (orange) to green with S (not R)</li><li>potassium permanganate / KMnO₄（BOD potassium manganate / MnO₄⁻） → decolorises / (purple) to colourless (solution) / brown ppt with S (not R)； A green solution formed if alkaline KMnO₄ used</li><li>加入 named alcohol + (conc) sulfuric acid（或 named alcohol AND catalyst） → sweet smell with R (not S)</li><li>加入 named carboxylic acid + (conc) sulfuric acid（或 named carboxylic acid AND catalyst） → sweet smell with S (and not R)</li><li>named acid-base indicator → appropriate colour change with R (and not S)</li><li>A Mg / magnesium → A faster bubbles with R（需要比较）</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-19"));</script>"""))
```

```python
# @title ✅ Learning outcomes 自查清单 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
</style>
<div class="k" id="k-ch18-20">
<h3>Learning outcomes 自查 <span class="faint count"></span></h3><p class="faint">勾选会存在这个浏览器里（localStorage）；Colab 不允许时只在当前页面有效。</p><div class="lolist" data-key="ch18"><div class="bar"><i></i></div><h4>18.1</h4>
<label class="chk"><input type="checkbox" data-id="18.1.1"><span><b>18.1.1</b> recall the reactions by which carboxylic acids can be produced:<ul><li>(a) oxidation of primary alcohols and aldehydes with acidified K₂Cr₂O₇ or acidified KMnO₄ and refluxing</li><li>(b) hydrolysis of nitriles with dilute acid or dilute alkali followed by acidification</li><li>(c) hydrolysis of esters with dilute acid or dilute alkali and heat followed by acidification</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="18.1.2"><span><b>18.1.2</b> describe:<ul><li>(a) the redox reaction with reactive metals to produce a salt and H₂(g)</li><li>(b) the neutralisation reaction with alkalis to produce a salt and H₂O(l)</li><li>(c) the acid–base reaction with carbonates to produce a salt and H₂O(l) and CO₂(g)</li><li>(d) esterification with alcohols with concentrated H₂SO₄ as catalyst</li><li>(e) reduction by LiAlH₄ to form a primary alcohol</li></ul></span></label>
<h4>18.2</h4>
<label class="chk"><input type="checkbox" data-id="18.2.1"><span><b>18.2.1</b> recall the reaction (reagents and conditions) by which esters can be produced:<ul><li>(a) the condensation reaction between an alcohol and a carboxylic acid with concentrated H₂SO₄ as catalyst</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="18.2.2"><span><b>18.2.2</b> describe the hydrolysis of esters by dilute acid and by dilute alkali and heat</span></label></div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        var ok = answers.some(function (a) {
          if (a === v) return true;
          var x = parseFloat(a), y = parseFloat(v);
          return !isNaN(x) && !isNaN(y) && /^[-+]?[\d.]+$/.test(v) && Math.abs(x - y) < 1e-9;
        });
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0;
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length;
      }
      if (eaten !== s.length) return null;
      return occ;
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch18-20"));</script>"""))
```
