## 16.1 Alcohols

### A. 怎样制得

> **LO 16.1.1** — recall the reactions (reagents and conditions) by which alcohols can be produced:
> - (a) electrophilic addition of steam to an alkene, H₂O(g) and H₃PO₄ catalyst
> - (b) reaction of alkenes with cold dilute acidified potassium manganate(VII) to form a diol
> - (c) substitution of a halogenoalkane using NaOH(aq) and heat
> - (d) reduction of an aldehyde or ketone using NaBH₄ or LiAlH₄
> - (e) reduction of a carboxylic acid using LiAlH₄
> - (f) hydrolysis of an ester using dilute acid or dilute alkali and heat

- (b) 得到的是 **diol**（两个 OH），例如 ethene → ethane-1,2-diol。

### B. primary / secondary / tertiary，以及物理性质

> **LO 16.1.3** — (a) classify alcohols as primary, secondary and tertiary alcohols, to include examples with more than one alcohol group
>
> (b) state characteristic distinguishing reactions, e.g. mild oxidation with acidified K₂Cr₂O₇, colour change from orange to green

按**连着 OH 的那个 C** 上直接连了几个 alkyl group 来分（与 [halogenoalkane](https://colab.research.google.com/drive/16jDQ0vRcndaJ9WwShvAMR9thtMUQgZjW#scrollTo=ch15-halogenoalkanes) 同一套标准）：

| 类型 | 例 | acidified K₂Cr₂O₇（warm） |
|---|---|---|
| primary（1°） | ethanol、propan-1-ol、2-methylpropan-1-ol | orange → **green** |
| secondary（2°） | propan-2-ol、butan-2-ol | orange → **green** |
| tertiary（3°） | 2-methylpropan-2-ol | **remains orange**（不被 oxidise） |

- 一个 molecule 有不止一个 OH 时**逐个分类**：例如 HOCH₂CH₂C(OH)(CH₂OH)₂（9701/22 May/June 2026 Q4(b)）有三个 1° OH 和一个 3° OH。
- (b) 的 **distinguishing reaction**：mild oxidation with acidified K₂Cr₂O₇，**orange → green**（Cr₂O₇<sup>2−</sup> 被 reduce 成 Cr<sup>3+</sup>）。1° 和 2° 变绿，3° 不变——所以它能把 3° 区分出来；再区分 1° 和 2° 要看 product（aldehyde 还是 ketone，chapter 17 的 Tollens' / Fehling's）。
- alcohol 之间有 **hydrogen bonding**，所以 boiling point 比相近 *M*ᵣ 的 alkane 高得多；短链 alcohol 和 water 能形成 hydrogen bond，完全 miscible；链越长，non-polar 的 alkyl 部分越大，越不溶。

### C. 反应一览（16.1.2）

> **LO 16.1.2** — describe:
> - (a) the reaction with oxygen (combustion)
> - (b) substitution to form halogenoalkanes, e.g. by reaction with HX(g); or with KCl and concentrated H₂SO₄ or concentrated H₃PO₄; or with PCl₃ and heat; or with PCl₅; or with SOCl₂
> - (c) the reaction with Na(s)
> - (d) oxidation with acidified K₂Cr₂O₇ or acidified KMnO₄ to:
>   - (i) carbonyl compounds by distillation
>   - (ii) carboxylic acids by refluxing
>
> > primary alcohols give aldehydes which can be further oxidised to carboxylic acids, secondary alcohols give ketones, tertiary alcohols cannot be oxidised
>
> - (e) dehydration to an alkene, by using a heated catalyst, e.g. Al₂O₃ or a concentrated acid
> - (f) formation of esters by reaction with carboxylic acids and concentrated H₂SO₄ as catalyst as exemplified by ethanol

**(a) combustion**：alcohol 完全燃烧 → CO₂ + H₂O，例：C₂H₅OH + 3O₂ → 2CO₂ + 3H₂O（清洁的 blue flame，可作 fuel）。

### D. (b) substitution → halogenoalkane

–OH 被 halogen 取代。syllabus 列出的几种 reagent：

| reagent | equation（以 ethanol 为例） | 说明 |
|---|---|---|
| HX(g)（如 HBr） | C₂H₅OH + HBr → C₂H₅Br + H₂O | — |
| KCl and concentrated H₂SO₄（或 conc. H₃PO₄） | KCl + H₂SO₄ → KHSO₄ + HCl，再同上 | 就地（in situ）产生 HX |
| PCl₃ **and heat** | 3C₂H₅OH + PCl₃ → 3C₂H₅Cl + H₃PO₃ | 要加热 |
| PCl₅ | C₂H₅OH + PCl₅ → C₂H₅Cl + POCl₃ + HCl | 室温；放出 steamy fumes of HCl（也用作 –OH 的测试） |
| SOCl₂ | C₂H₅OH + SOCl₂ → C₂H₅Cl + SO₂ + HCl | 两个 by-product 都是 gas，自己跑掉，product 好分离 |

判分（9701/22 May/June 2026 Q4(a)(i)）：PCl₅ / SOCl₂ / HCl / PCl₃ 都接受；**单写 hydrochloric acid / HCl(aq) / dilute HCl 不给**；PCl₅(aq) 判错。

### E. (c) 与 Na(s) 的反应，以及 alcohol 的 acidity

2C₂H₅OH + 2Na → 2C₂H₅O<sup>−</sup>Na<sup>+</sup> + H₂（sodium ethoxide，蒸干是白色固体，strongly basic）

- 和 Na + water 类似，但**温和得多**；链越长越温和。
- 观察要写 **effervescence / bubbles**（9701 多卷：只写「产生 gas」、写 gas 名称或检验方法都不给分）。
- carboxylic acid 也和 Na 放出 H₂（[18.1](https://colab.research.google.com/drive/1gxHJxFjn1RQSXR5y5n5CSeW5q82c6-6E#scrollTo=ch18-rx-acid-reactions)），所以这个 test 只说明有 –OH（alcohol 或 acid）。

> **LO 16.1.5** — explain the acidity of alcohols compared with water

**acidity：alcohol 比 water 更弱**。

ROH ⇌ RO<sup>−</sup> + H<sup>+</sup>；H₂O ⇌ OH<sup>−</sup> + H<sup>+</sup>。alcohol 解离得更少，因为：

- alkyl group 有 **positive inductive effect**，把电子推向 O；
- 于是 alkoxide ion RO<sup>−</sup> 的 O 上**负电荷更集中**（比 OH<sup>−</sup> 更不分散）→ RO<sup>−</sup> 更不稳定、更容易重新结合 H<sup>+</sup> → 平衡更偏左。

与 14.2 D（[carbocation](https://colab.research.google.com/drive/10uq0QFZnHK9NPDSSsCmw7w3-66mo9ltw#scrollTo=ch14-markovnikov)）是同一个 inductive effect，只是这里它让负离子**更不稳定**，那里让正离子**更稳定**。

```python
# @title 🔬 alcohol vs water：与 Na 的反应和 acidity { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-acidity">
<h3>🔬 alcohol vs water：与 Na 的反应和 acidity 对比</h3>
<p class="mut">醇（alcohol）与水（water）都含有羟基（–OH），但在酸性（acidity）以及与金属钠（Na）反应的剧烈程度上存在根本差异：① <b>acidity</b>：alkyl group 具有推电子的 <b>positive inductive effect</b>，向氧原子推电子，使 alkoxide ion（RO⁻）的氧原子上<b>负电荷更集中</b>、稳定性更低，更容易重新结合 H⁺，因此 alcohol 的酸性弱于水（equilibrium 偏左）；② <b>与 Na(s) 反应</b>：alcohol 与 Na 反应放出 H₂（观察现象必须写 <b>effervescence / bubbles</b>），但比水与 Na 的反应温和得多，碳链越长、分支越多反应越慢。</p>

<div class="row ac-banner">
  <div class="ac-banner-col">
    <b>Acidity 电离平衡：</b>
    <span>H₂O ⇌ OH⁻ + H⁺ (p<i>K</i><sub>a</sub> = 15.7) &nbsp;|&nbsp; ROH ⇌ RO⁻ + H⁺ (p<i>K</i><sub>a</sub> ≈ 16–18，更偏左)</span>
  </div>
  <div class="ac-banner-col">
    <b>与 Na(s) 反应方程式：</b>
    <span>2H₂O + 2Na → 2NaOH + H₂↑ &nbsp;|&nbsp; 2ROH + 2Na → 2RO⁻Na⁺ + H₂↑（产生 sodium alkoxide 白色固体）</span>
  </div>
</div>

<div class="ac-wrap">
  <div class="tabbar">
    <button type="button" class="on" data-tab="theory">① 电子效应与酸性原理 (Inductive Effect &amp; Acidity)</button>
    <button type="button" data-tab="sodium">② 与金属钠反应模拟 (Reaction with Na)</button>
    <button type="button" data-tab="quiz">③ 真题考点自测 (Practice Quiz)</button>
  </div>

  <!-- ================= Tab 1: 电子效应与酸性原理 ================= -->
  <div data-pane="theory" class="show">
    <!-- Molecule selector bar -->
    <div class="card ac-sec-card">
      <div class="row ac-ctrl-row">
        <span class="ac-lbl">选择分子（Molecule）：</span>
        <div class="row ac-mol-btns" role="group" aria-label="选择分子">
          <button type="button" class="ac-mol-btn" data-mol="water">water (H₂O)</button>
          <button type="button" class="ac-mol-btn" data-mol="methanol">methanol (CH₃OH)</button>
          <button type="button" class="ac-mol-btn on" data-mol="ethanol">ethanol (1°)</button>
          <button type="button" class="ac-mol-btn" data-mol="propan2ol">propan-2-ol (2°)</button>
          <button type="button" class="ac-mol-btn" data-mol="tbutanol">2-methylpropan-2-ol (3°)</button>
        </div>
        <button type="button" class="ac-btn-contrast">⚡ 对比模式：water vs ethanol 并排</button>
      </div>
    </div>

    <!-- Single Molecule View -->
    <div class="ac-single-view">
      <div class="grid2 ac-stage-grid">
        <!-- Left: Stage Card (SVG) -->
        <div class="card ac-stage-card">
          <div class="ac-stage-head">
            <span class="ac-mol-title">ethanol (CH₃CH₂OH)</span>
            <span class="ac-badge ac-type-badge">1° alcohol</span>
            <span class="ac-badge ac-pka-badge">p<i>K</i><sub>a</sub> ≈ 15.9</span>
          </div>
          <div class="ac-stage-box">
            <svg class="ac-svg ac-theory-svg" viewBox="0 0 540 260" role="img" aria-label="电子推拉效应与解离平衡示意图"></svg>
          </div>
          <div class="ac-stage-legend">
            <span class="ac-leg-item"><i class="ac-arr-icon">➔➔</i> positive inductive effect (推电子)</span>
            <span class="ac-leg-item"><i class="ac-halo-icon"></i> 氧原子负电荷密度云 (charge density)</span>
          </div>
        </div>

        <!-- Right: Analysis & Data Card -->
        <div class="card ac-data-card">
          <div class="ac-dh">💡 核心原理与 Cambridge 9701 得分点</div>
          <div class="ac-analysis-body">
            <div class="ac-step-point">
              <span class="ac-sp-num">1</span>
              <div>
                <b>Alkyl group 的 positive inductive effect：</b>
                <span class="ac-txt-ind">ethyl group（–C₂H₅）向氧原子推电子（electron-donating）。</span>
              </div>
            </div>
            <div class="ac-step-point">
              <span class="ac-sp-num">2</span>
              <div>
                <b>负电荷集中程度（charge density on oxygen）：</b>
                <span class="ac-txt-dens">由于 alkyl group 推电子，共轭碱 RO⁻ 氧原子上的负电荷<b>更集中（intensified / concentrated）</b>。</span>
              </div>
            </div>
            <div class="ac-step-point">
              <span class="ac-sp-num">3</span>
              <div>
                <b>阴离子稳定性与质子结合倾向：</b>
                <span class="ac-txt-stab">负电荷更集中导致 RO⁻ 离子<b>更不稳定（less stable than OH⁻）</b>，更容易重新结合 H⁺。</span>
              </div>
            </div>
            <div class="ac-step-point">
              <span class="ac-sp-num">4</span>
              <div>
                <b>平衡位置与酸性强弱：</b>
                <span class="ac-txt-eq">解离平衡 ROH ⇌ RO⁻ + H⁺ <b>更偏向左边（equilibrium lies further to the left）</b>，因此酸性更弱！</span>
              </div>
            </div>
          </div>

          <!-- Relative Acidity Meter -->
          <div class="ac-meter-box">
            <div class="ac-meter-title">相对酸性强弱排序（Acidity Order）：</div>
            <div class="ac-meter-bar">
              <div class="ac-meter-node" data-m="water"><span>water</span><small>15.7</small></div>
              <div class="ac-meter-arrow">&gt;</div>
              <div class="ac-meter-node" data-m="methanol"><span>methanol</span><small>15.5*</small></div>
              <div class="ac-meter-arrow">&gt;</div>
              <div class="ac-meter-node on" data-m="ethanol"><span>ethanol</span><small>15.9</small></div>
              <div class="ac-meter-arrow">&gt;</div>
              <div class="ac-meter-node" data-m="propan2ol"><span>propan-2-ol</span><small>16.5</small></div>
              <div class="ac-meter-arrow">&gt;</div>
              <div class="ac-meter-node" data-m="tbutanol"><span>2-methylpropan-2-ol</span><small>18.0</small></div>
            </div>
            <div class="faint" style="font-size:12px;margin-top:4px">*注：methanol 气相中因溶剂化效应略反常，但在 A-Level 体系中通常统一按 alkyl group 越大推电子越多、酸性越弱掌握：water &gt; 1° &gt; 2° &gt; 3° alcohol。</div>
          </div>
        </div>
      </div>
    </div>

    <!-- Contrast Mode View (hidden by default) -->
    <div class="card ac-contrast-view" hidden>
      <div class="ac-contrast-head">
        <b>⚡ 直观并排对照：Water (H₂O) vs Ethanol (C₂H₅OH)</b>
        <button type="button" class="ac-btn-close-contrast">返回单分子视角 ✕</button>
      </div>
      <div class="ac-stage-box">
        <svg class="ac-svg ac-contrast-svg" viewBox="0 0 740 280" role="img" aria-label="水与乙醇酸性对比图"></svg>
      </div>
      <div class="ac-contrast-takeaway">
        <div class="ac-ct-col">
          <b style="color:var(--ok)">Water (H₂O ⇌ OH⁻ + H⁺)：</b><br>
          H 原子没有显著的 positive inductive effect。OH⁻ 的负电荷相对分散（less concentrated），相对较稳定，解离平衡偏左程度较轻（p<i>K</i><sub>a</sub> = 15.7）。
        </div>
        <div class="ac-ct-col">
          <b style="color:var(--bad)">Ethanol (C₂H₅OH ⇌ C₂H₅O⁻ + H⁺)：</b><br>
          –C₂H₅ 基团推电子（positive inductive effect），使 ethoxide ion（C₂H₅O⁻）氧原子上的负电荷<b>更集中、更不稳定</b>，极易拉回 H⁺，平衡更偏左（p<i>K</i><sub>a</sub> = 15.9）。
        </div>
      </div>
    </div>
  </div>

  <!-- ================= Tab 2: 与金属钠反应模拟 ================= -->
  <div data-pane="sodium">
    <div class="card ac-sec-card">
      <div class="row ac-na-ctrls">
        <button type="button" class="ac-btn-play on">▶ 投入金属钠 (Add Sodium)</button>
        <button type="button" class="ac-btn-pause">⏸ 暂停</button>
        <button type="button" class="ac-btn-reset">↺ 重置</button>
        <div class="ac-timer-wrap">
          <span>模拟计时：<b class="ac-time">0.0</b> s</span>
          <input type="range" class="ac-time-slider" min="0" max="10" step="0.1" value="0" aria-label="反应时间滑块">
        </div>
      </div>
    </div>

    <!-- Tubes Stage -->
    <div class="card ac-stage-card">
      <div class="ac-tubes-stage-box">
        <svg class="ac-svg ac-tubes-svg" viewBox="0 0 720 300" role="img" aria-label="金属钠在水、乙醇与叔丁醇中反应对比"></svg>
      </div>
      <div class="row ac-tubes-legend">
        <div class="ac-tleg-col">
          <div class="ac-tleg-name"><b>试管 1：Water (H₂O)</b></div>
          <div class="ac-badge ac-badge-vig">💥 极其剧烈 (Extremely vigorous)</div>
          <div class="ac-tleg-desc">Na 熔成闪亮银色小球，在水面狂游，伴有嘶嘶声与大量气泡（vigorous effervescence），放热剧烈并自燃（yellow flame）。</div>
        </div>
        <div class="ac-tleg-col">
          <div class="ac-tleg-name"><b>试管 2：Ethanol (1° alcohol)</b></div>
          <div class="ac-badge ac-badge-mod">⚡ 温和放气 (Steady effervescence)</div>
          <div class="ac-tleg-desc">Na 沉在试管底部（密度大），稳定释放无色气泡（<b>effervescence / bubbles</b> of H₂），逐渐溶解，生成 sodium ethoxide 白色固体。</div>
        </div>
        <div class="ac-tleg-col">
          <div class="ac-tleg-name"><b>试管 3：2-methylpropan-2-ol (3°)</b></div>
          <div class="ac-badge ac-badge-slow">🐢 极为迟缓 (Very slow / sluggish)</div>
          <div class="ac-tleg-desc">Na 沉在底部，非常缓慢地冒出少量微弱气泡，反应极其温和，需温热才能加快反应。</div>
        </div>
      </div>
    </div>

    <!-- Cambridge Exam Note Callout -->
    <div class="card ac-obs-card">
      <div class="ac-dh">📝 Cambridge 9701 官方标答规范（Mark Scheme Key Points）</div>
      <ul class="ac-ms-list">
        <li><b>现象描述 Observation</b>：写 <b>effervescence</b> 或 <b>bubbles / fizzing</b>；或 <b>sodium dissolves / gets smaller</b>。仅写「产生气体 (gas is produced)」或写气体检验证明均<b>不给分</b>！</li>
        <li><b>反应产物 Product</b>：生成 hydrogen gas（H₂）以及对应的 <b>sodium alkoxide</b>（如 sodium ethoxide，蒸干后为白色固体，呈强碱性 strongly basic）。</li>
        <li><b>反应配平 Equation</b>：<code>2C₂H₅OH + 2Na → 2C₂H₅O⁻Na⁺ + H₂</code> 或 <code>C₂H₅OH + Na → C₂H₅ONa + ½H₂</code>。</li>
      </ul>
    </div>
  </div>

  <!-- ================= Tab 3: 真题考点自测 ================= -->
  <div data-pane="quiz">
    <div class="card ac-quiz-card">
      <div class="ac-qh">
        <h4>📝 Cambridge 9701 真题核心考点自测</h4>
        <span class="ac-qscore">已正确 <b class="ac-sc">0</b> / 4 题</span>
      </div>
      <div class="ac-qitems">
        <!-- Q1 -->
        <div class="ac-qitem" data-qi="0">
          <div class="ac-qp"><b>1.</b> 解释为什么 ethanol 的酸性比 water 更弱？（Cambridge mark scheme 核心满分踩分点）</div>
          <div class="ac-opts">
            <button type="button" class="ac-opt-btn" data-opt="A">Ethanol 含有分子间氢键，阻碍了 O–H 键断裂</button>
            <button type="button" class="ac-opt-btn" data-opt="B">Alkyl group（ethyl group）具有 positive inductive effect 向氧推电子，使 ethoxide ion（RO⁻）的氧上负电荷更集中、稳定性降低，平衡更偏左</button>
            <button type="button" class="ac-opt-btn" data-opt="C">Ethanol 在水中的溶解度低于纯水，因此无法电离出自由移动的离子</button>
            <button type="button" class="ac-opt-btn" data-opt="D">O–H 键的键能（bond energy）在 ethanol 中显著大于 water</button>
          </div>
          <div class="ac-fb" hidden></div>
        </div>
        <!-- Q2 -->
        <div class="ac-qitem" data-qi="1">
          <div class="ac-qp"><b>2.</b> 将一小块 sodium 加入干燥的 ethanol 中。下列哪项是考试中<b>允许给分</b>的实验现象描述（observation）？</div>
          <div class="ac-opts">
            <button type="button" class="ac-opt-btn" data-opt="A">生成了 hydrogen 气体</button>
            <button type="button" class="ac-opt-btn" data-opt="B">用点燃的木条检验有 pop sound</button>
            <button type="button" class="ac-opt-btn" data-opt="C">effervescence / bubbles（持续产生气泡），sodium 逐渐减小溶解</button>
            <button type="button" class="ac-opt-btn" data-opt="D">溶液颜色瞬间由橙色变绿色</button>
          </div>
          <div class="ac-fb" hidden></div>
        </div>
        <!-- Q3 -->
        <div class="ac-qitem" data-qi="2">
          <div class="ac-qp"><b>3.</b> 下列物质按酸性（acidity）<b>由强到弱</b>的排列顺序，正确的是？</div>
          <div class="ac-opts">
            <button type="button" class="ac-opt-btn" data-opt="A">water &gt; ethanol &gt; 2-methylpropan-2-ol</button>
            <button type="button" class="ac-opt-btn" data-opt="B">2-methylpropan-2-ol &gt; ethanol &gt; water</button>
            <button type="button" class="ac-opt-btn" data-opt="C">ethanol &gt; water &gt; 2-methylpropan-2-ol</button>
            <button type="button" class="ac-opt-btn" data-opt="D">2-methylpropan-2-ol &gt; water &gt; ethanol</button>
          </div>
          <div class="ac-fb" hidden></div>
        </div>
        <!-- Q4 -->
        <div class="ac-qitem" data-qi="3">
          <div class="ac-qp"><b>4.</b> 干燥的 propan-1-ol 与 sodium 反应生成的有机产物（organic product）名称是？</div>
          <div class="ac-opts">
            <button type="button" class="ac-opt-btn" data-opt="A">sodium propanoate</button>
            <button type="button" class="ac-opt-btn" data-opt="B">sodium propoxide (或 sodium propan-1-olate)</button>
            <button type="button" class="ac-opt-btn" data-opt="C">propanal</button>
            <button type="button" class="ac-opt-btn" data-opt="D">propene</button>
          </div>
          <div class="ac-fb" hidden></div>
        </div>
      </div>
      <div class="row ac-qtools">
        <button type="button" class="ac-qshow">显示标准答案</button>
        <button type="button" class="ac-qreset">重置练习</button>
      </div>
    </div>
  </div>
</div>

<style>
.k [hidden]{display:none!important}
.k .ac-wrap{margin-top:8px}
.k .ac-banner{justify-content:space-between;gap:8px 16px;background:var(--acc-soft);border:1px solid var(--line);border-radius:8px;padding:8px 12px;font-size:13px;margin:6px 0 10px}
.k .ac-banner-col{flex:1 1 280px;display:flex;flex-direction:column;gap:2px}
.k .ac-sec-card{padding:8px 12px;margin-bottom:10px}
.k .ac-ctrl-row{align-items:center;gap:6px 12px}
.k .ac-lbl{font-weight:600;font-size:13.5px}
.k .ac-mol-btns{gap:5px}
.k .ac-mol-btn{font-size:13px;padding:3px 9px}
.k .ac-btn-contrast{margin-left:auto;font-size:13px;padding:3px 10px;background:var(--card);border:1px dashed var(--acc);color:var(--acc)}
.k .ac-btn-contrast:hover{background:var(--acc-soft)}
.k .ac-stage-grid{grid-template-columns:minmax(320px,1.2fr) minmax(280px,1fr);align-items:stretch}
@media (max-width:760px){
  .k .ac-stage-grid{grid-template-columns:1fr}
  .k .ac-btn-contrast{margin-left:0;width:100%}
}
.k .ac-stage-card{display:flex;flex-direction:column;padding:10px 12px}
.k .ac-stage-head{display:flex;flex-wrap:wrap;align-items:center;gap:6px 10px;margin-bottom:6px}
.k .ac-mol-title{font-size:15px;font-weight:700}
.k .ac-badge{font-size:12px;font-weight:600;padding:2px 7px;border-radius:4px;border:1px solid var(--line);background:var(--card)}
.k .ac-type-badge{color:var(--acc);border-color:var(--acc)}
.k .ac-pka-badge{color:var(--mut)}
.k .ac-badge-vig{background:var(--bad-soft);color:var(--bad);border-color:var(--bad)}
.k .ac-badge-mod{background:var(--ok-soft);color:var(--ok);border-color:var(--ok)}
.k .ac-badge-slow{background:var(--acc-soft);color:var(--mut);border-color:var(--line)}
.k .ac-stage-box{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:4px;overflow:hidden;display:flex;align-items:center;justify-content:center}
.k .ac-theory-svg{width:100%;height:auto;display:block}
.k .ac-stage-legend{display:flex;flex-wrap:wrap;gap:12px;margin-top:8px;font-size:12px;color:var(--mut)}
.k .ac-leg-item{display:inline-flex;align-items:center;gap:5px}
.k .ac-arr-icon{color:var(--acc);font-weight:700;letter-spacing:-2px}
.k .ac-halo-icon{display:inline-block;width:10px;height:10px;border-radius:50%;background:rgba(207,34,46,0.35);border:1px solid var(--bad)}
.k .ac-data-card{display:flex;flex-direction:column;padding:10px 14px}
.k .ac-dh{font-weight:700;font-size:14px;color:var(--fg);margin-bottom:8px;border-bottom:1px solid var(--line);padding-bottom:5px}
.k .ac-analysis-body{display:flex;flex-direction:column;gap:8px;font-size:13px;line-height:1.45}
.k .ac-step-point{display:flex;gap:8px;align-items:flex-start}
.k .ac-sp-num{flex:none;width:18px;height:18px;border-radius:50%;background:var(--acc-soft);color:var(--acc);font-size:11.5px;font-weight:700;display:flex;align-items:center;justify-content:center}
.k .ac-meter-box{margin-top:auto;padding-top:10px;border-top:1px dashed var(--line)}
.k .ac-meter-title{font-size:12.5px;font-weight:600;margin-bottom:6px;color:var(--fg)}
.k .ac-meter-bar{display:flex;flex-wrap:wrap;align-items:center;gap:4px;font-size:12px}
.k .ac-meter-node{display:flex;flex-direction:column;align-items:center;padding:2px 6px;border:1px solid var(--line);border-radius:4px;background:var(--card)}
.k .ac-meter-node small{color:var(--mut);font-size:11px}
.k .ac-meter-node.on{border-color:var(--acc);background:var(--acc-soft);font-weight:650;color:var(--acc)}
.k .ac-meter-arrow{color:var(--mut);font-weight:700}
.k .ac-contrast-view{padding:12px;margin-top:6px}
.k .ac-contrast-head{display:flex;justify-content:space-between;align-items:center;margin-bottom:8px;font-size:14px}
.k .ac-btn-close-contrast{font-size:12.5px;padding:2px 8px}
.k .ac-contrast-svg{width:100%;height:auto;display:block}
.k .ac-contrast-takeaway{display:grid;grid-template-columns:1fr 1fr;gap:12px;margin-top:10px;font-size:13px;line-height:1.45}
@media (max-width:600px){
  .k .ac-contrast-takeaway{grid-template-columns:1fr}
}
.k .ac-ct-col{padding:8px 10px;background:var(--card);border:1px solid var(--line);border-radius:6px}

/* Tab 2: Sodium controls & tubes */
.k .ac-na-ctrls{align-items:center;gap:8px 12px}
.k .ac-na-ctrls button{font-size:13px;padding:3px 10px}
.k .ac-timer-wrap{margin-left:auto;display:inline-flex;align-items:center;gap:8px;font-size:13px}
.k .ac-time-slider{width:130px}
.k .ac-tubes-stage-box{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:4px;overflow:hidden}
.k .ac-tubes-svg{width:100%;height:auto;display:block}
.k .ac-tubes-legend{display:grid;grid-template-columns:repeat(3,1fr);gap:10px;margin-top:10px;padding-top:8px;border-top:1px dashed var(--line)}
@media (max-width:680px){
  .k .ac-tubes-legend{grid-template-columns:1fr}
  .k .ac-timer-wrap{margin-left:0;width:100%}
}
.k .ac-tleg-col{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:6px 8px;font-size:12.5px}
.k .ac-tleg-name{margin-bottom:4px}
.k .ac-tleg-desc{margin-top:4px;color:var(--mut);line-height:1.35}
.k .ac-obs-card{margin-top:10px;padding:10px 12px}
.k .ac-ms-list{margin:4px 0 0;padding-left:18px;font-size:13px;line-height:1.5}
.k .ac-ms-list code{font-size:12.5px}

/* Tab 3: Quiz */
.k .ac-quiz-card{padding:12px}
.k .ac-qh{display:flex;justify-content:space-between;align-items:center;margin-bottom:10px;border-bottom:1px solid var(--line);padding-bottom:6px}
.k .ac-qh h4{margin:0;font-size:15px}
.k .ac-qscore{font-size:13px;color:var(--mut)}
.k .ac-qscore b{color:var(--ok)}
.k .ac-qitems{display:flex;flex-direction:column;gap:12px}
.k .ac-qitem{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k .ac-qp{font-size:13.5px;margin-bottom:8px}
.k .ac-opts{display:flex;flex-direction:column;gap:5px}
.k .ac-opt-btn{text-align:left;font-size:13px;padding:5px 10px;border-radius:6px;line-height:1.35}
.k .ac-opt-btn.ok{background:var(--ok-soft);border-color:var(--ok);color:var(--fg);font-weight:600}
.k .ac-opt-btn.bad{background:var(--bad-soft);border-color:var(--bad);color:var(--fg)}
.k .ac-fb{margin-top:6px;padding:6px 10px;border-radius:6px;font-size:12.5px;line-height:1.4}
.k .ac-fb.ok{background:var(--ok-soft);color:var(--ok);border:1px solid var(--ok)}
.k .ac-fb.bad{background:var(--bad-soft);color:var(--bad);border:1px solid var(--bad)}
.k .ac-qtools{margin-top:12px;gap:8px}
.k .ac-qtools button{font-size:13px;padding:4px 12px}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var $ = function (sel) { return $root.querySelector(sel); };
var $$ = function (sel) { return Array.prototype.slice.call($root.querySelectorAll(sel)); };

/* Molecule details data */
var MOL_DATA = {
    water: {
      id: 'water',
      name: 'water (H₂O)',
      type: 'inorganic reference',
      pka: '15.7',
      ka: '1.8 × 10⁻¹⁶',
      formula: 'H–O–H',
      base: 'OH⁻',
      rName: 'H',
      indTxt: 'Hydrogen（H）没有显著的 positive inductive effect，作为酸性基准。',
      densTxt: '由于没有基团向 O 推电子，OH⁻ 上负电荷相对分散（less concentrated）。',
      stabTxt: 'OH⁻ 比 alkoxide ion（RO⁻）相对更稳定（more stable），重新拉回 H⁺ 趋势更温和。',
      eqTxt: 'H₂O ⇌ OH⁻ + H⁺ 解离常数 pKa = 15.7，在常温下酸性显著强于脂肪醇。',
      indArrows: 0,
      chargeHalo: 14,
      eqLeftRatio: 0.55
    },
    methanol: {
      id: 'methanol',
      name: 'methanol (CH₃OH)',
      type: '1° alcohol',
      pka: '15.5*',
      ka: '3.2 × 10⁻¹⁶',
      formula: 'CH₃–O–H',
      base: 'CH₃O⁻',
      rName: '–CH₃',
      indTxt: 'Methyl group（–CH₃）向氧推电子（positive inductive effect）。',
      densTxt: '推电子效应使 methoxide（CH₃O⁻）氧原子上的负电荷较集中。',
      stabTxt: 'CH₃O⁻ 略不稳定，易重新结合 H⁺。',
      eqTxt: 'CH₃OH ⇌ CH₃O⁻ + H⁺ 解离平衡明显偏向反应物。',
      indArrows: 1,
      chargeHalo: 18,
      eqLeftRatio: 0.65
    },
    ethanol: {
      id: 'ethanol',
      name: 'ethanol (CH₃CH₂OH)',
      type: '1° alcohol',
      pka: '15.9',
      ka: '1.3 × 10⁻¹⁶',
      formula: 'CH₃CH₂–O–H',
      base: 'CH₃CH₂O⁻',
      rName: '–C₂H₅',
      indTxt: 'Ethyl group（–CH₂CH₃）具有明显的 positive inductive effect（推电子）。',
      densTxt: '推电子效应使 ethoxide（C₂H₅O⁻）氧上的负电荷更集中（concentrated）。',
      stabTxt: 'C₂H₅O⁻ 比 OH⁻ 更不稳定（less stable），更容易重新捕获质子 H⁺。',
      eqTxt: 'C₂H₅OH ⇌ C₂H₅O⁻ + H⁺ 平衡显著更偏向左边，酸性弱于水。',
      indArrows: 2,
      chargeHalo: 22,
      eqLeftRatio: 0.75
    },
    propan2ol: {
      id: 'propan2ol',
      name: 'propan-2-ol ((CH₃)₂CHOH)',
      type: '2° alcohol',
      pka: '16.5',
      ka: '3.2 × 10⁻¹⁷',
      formula: '(CH₃)₂CH–O–H',
      base: '(CH₃)₂CHO⁻',
      rName: '–CH(CH₃)₂',
      indTxt: '两个 methyl groups 通过二级碳向氧强烈推电子（stronger positive inductive effect）。',
      densTxt: 'Prop-2-oxide 氧原子上的负电荷高度集中，电子云密度进一步增大。',
      stabTxt: 'Alkoxide 阴离子稳定性极低，与 H⁺ 结合能力极强。',
      eqTxt: '解离平衡极度偏向左边，酸性明显弱于 1° 醇和水。',
      indArrows: 3,
      chargeHalo: 27,
      eqLeftRatio: 0.85
    },
    tbutanol: {
      id: 'tbutanol',
      name: '2-methylpropan-2-ol ((CH₃)₃COH)',
      type: '3° alcohol',
      pka: '18.0',
      ka: '1.0 × 10⁻¹⁸',
      formula: '(CH₃)₃C–O–H',
      base: '(CH₃)₃CO⁻',
      rName: '–C(CH₃)₃',
      indTxt: '三个 methyl groups 同时通过三级碳向氧推电子（three electron-donating alkyl groups）。',
      densTxt: 'Tert-butoxide 氧上的负电荷最为集中（most concentrated negative charge）。',
      stabTxt: 'RO⁻ 最不稳定（least stable alkoxide），平衡最偏向分子状态。',
      eqTxt: 'pKa 高达 18.0，是常见醇中酸性最弱的之一，几乎极难解离。',
      indArrows: 4,
      chargeHalo: 32,
      eqLeftRatio: 0.95
    }
  };

  var curMol = 'ethanol';
  var contrastMode = false;
  var curTab = 'theory';

  /* Animation & Sodium Simulation state */
  var naTime = 0.0;
  var naRunning = false;
  var naRaf = null;
  var lastTs = null;

  /* Helper to escape/set SVG */
  function renderTheorySvg() {
    var svg = $('.ac-theory-svg');
    if (!svg) return;
    var d = MOL_DATA[curMol];

    var h = [];
    // Background card border & guides
    h.push('<defs>');
    h.push('  <marker id="arr-ind" markerWidth="6" markerHeight="6" refX="5" refY="3" orient="auto">');
    h.push('    <polygon points="0 0, 6 3, 0 6" fill="var(--acc)" />');
    h.push('  </marker>');
    h.push('  <marker id="arr-eq-r" markerWidth="7" markerHeight="5" refX="6" refY="2.5" orient="auto">');
    h.push('    <polygon points="0 0, 7 2.5, 0 5" fill="var(--fg)" />');
    h.push('  </marker>');
    h.push('  <marker id="arr-eq-l" markerWidth="7" markerHeight="5" refX="1" refY="2.5" orient="auto">');
    h.push('    <polygon points="7 0, 0 2.5, 7 5" fill="var(--fg)" />');
    h.push('  </marker>');
    h.push('  <radialGradient id="halo-glow" cx="50%" cy="50%" r="50%">');
    h.push('    <stop offset="0%" stop-color="var(--bad)" stop-opacity="0.45" />');
    h.push('    <stop offset="70%" stop-color="var(--bad)" stop-opacity="0.18" />');
    h.push('    <stop offset="100%" stop-color="var(--bad)" stop-opacity="0" />');
    h.push('  </radialGradient>');
    h.push('</defs>');

    // Title inside SVG
    h.push('<text x="20" y="28" font-size="13" font-weight="700" fill="var(--fg)">' + d.name + ' 解离平衡与电子推拉效应：</text>');

    // Left Box: Undissociated molecule (R–O–H)
    h.push('<g transform="translate(30, 60)">');
    h.push('  <rect x="0" y="0" width="160" height="150" rx="8" fill="var(--card)" stroke="var(--line)" />');
    h.push('  <text x="80" y="24" text-anchor="middle" font-size="12" fill="var(--mut)" font-weight="600">未解离分子 (Molecule)</text>');
    
    // Draw Molecule representation
    if (d.id === 'water') {
      h.push('  <text x="45" y="80" font-size="18" font-weight="700" fill="var(--fg)">H</text>');
      h.push('  <line x1="62" y1="75" x2="80" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="83" y="80" font-size="19" font-weight="700" fill="var(--bad)">O</text>');
      h.push('  <line x1="102" y1="75" x2="120" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="123" y="80" font-size="18" font-weight="700" fill="var(--fg)">H</text>');
      h.push('  <text x="80" y="125" text-anchor="middle" font-size="12" fill="var(--mut)">H–O bond 极性强</text>');
    } else {
      h.push('  <text x="35" y="80" font-size="16" font-weight="700" fill="var(--acc)">' + d.rName + '</text>');
      h.push('  <line x1="72" y1="75" x2="90" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="93" y="80" font-size="19" font-weight="700" fill="var(--bad)">O</text>');
      h.push('  <line x1="112" y1="75" x2="128" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="131" y="80" font-size="18" font-weight="700" fill="var(--fg)">H</text>');
      // Inductive arrow pushing from R to O
      h.push('  <line x1="45" y1="60" x2="88" y2="60" stroke="var(--acc)" stroke-width="2.5" marker-end="url(#arr-ind)" />');
      h.push('  <text x="80" y="125" text-anchor="middle" font-size="12" fill="var(--mut)">alkyl group 推电子</text>');
    }
    h.push('</g>');

    // Middle Equilibrium arrows
    var leftLen = 30 + d.eqLeftRatio * 25;
    var rightLen = 30 - d.eqLeftRatio * 15;
    h.push('<g transform="translate(205, 120)">');
    // Top arrow (forward ->)
    h.push('  <line x1="0" y1="-4" x2="' + rightLen + '" y2="-4" stroke="var(--fg)" stroke-width="1.8" marker-end="url(#arr-eq-r)" />');
    // Bottom arrow (backward <-, longer!)
    h.push('  <line x1="' + (leftLen - 15) + '" y1="8" x2="-8" y2="8" stroke="var(--fg)" stroke-width="2.6" marker-end="url(#arr-eq-l)" />');
    h.push('  <text x="12" y="32" text-anchor="middle" font-size="12" fill="var(--mut)">⇌ 平衡偏左</text>');
    h.push('</g>');

    // Right Box: Conjugate Base (RO⁻ / OH⁻) + H⁺
    h.push('<g transform="translate(265, 60)">');
    h.push('  <rect x="0" y="0" width="245" height="150" rx="8" fill="var(--card)" stroke="var(--line)" />');
    h.push('  <text x="122" y="24" text-anchor="middle" font-size="12" fill="var(--mut)" font-weight="600">共轭碱 (Conjugate Base) + H⁺</text>');

    // Draw anion & charge density
    var oxX = d.id === 'water' ? 70 : 85;
    var oxY = 75;

    // Glowing circle around O representing negative charge concentration
    h.push('  <circle cx="' + oxX + '" cy="' + oxY + '" r="' + d.chargeHalo + '" fill="url(#halo-glow)" />');

    if (d.id === 'water') {
      h.push('  <text x="35" y="80" font-size="18" font-weight="700" fill="var(--fg)">H</text>');
      h.push('  <line x1="52" y1="75" x2="65" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="68" y="80" font-size="19" font-weight="700" fill="var(--bad)">O</text>');
      h.push('  <text x="85" y="70" font-size="14" font-weight="700" fill="var(--bad)">⁻</text>');
    } else {
      h.push('  <text x="25" y="80" font-size="16" font-weight="700" fill="var(--acc)">' + d.rName + '</text>');
      h.push('  <line x1="62" y1="75" x2="78" y2="75" stroke="var(--fg)" stroke-width="2" />');
      h.push('  <text x="81" y="80" font-size="19" font-weight="700" fill="var(--bad)">O</text>');
      h.push('  <text x="98" y="70" font-size="14" font-weight="700" fill="var(--bad)">⁻</text>');

      // Multiple inductive arrows based on degree
      for (var a = 0; a < d.indArrows; a++) {
        var offY = 52 - a * 8;
        h.push('  <line x1="30" y1="' + offY + '" x2="75" y2="' + offY + '" stroke="var(--acc)" stroke-width="2" marker-end="url(#arr-ind)" />');
      }
    }

    // Plus sign and H+
    h.push('  <text x="135" y="80" font-size="16" fill="var(--fg)">+</text>');
    h.push('  <text x="155" y="80" font-size="18" font-weight="700" fill="var(--fg)">H⁺</text>');

    // Annotation
    var stCol = d.id === 'water' ? 'var(--ok)' : 'var(--bad)';
    h.push('  <text x="122" y="118" text-anchor="middle" font-size="12" font-weight="600" fill="' + stCol + '">');
    h.push(d.id === 'water' ? 'OH⁻ 相对稳定 (more stable)' : 'RO⁻ 负电荷集中、不稳定 (less stable)');
    h.push('  </text>');
    h.push('  <text x="122" y="136" text-anchor="middle" font-size="12" fill="var(--mut)">极易重新捕获 H⁺ 逆向反应</text>');
    h.push('</g>');

    // Bottom summary banner inside SVG
    h.push('<rect x="30" y="222" width="480" height="28" rx="4" fill="var(--acc-soft)" stroke="var(--line)" />');
    h.push('<text x="270" y="241" text-anchor="middle" font-size="12.5" font-weight="600" fill="var(--fg)">');
    h.push('结论：' + d.rName + ' 推电子 ➔ O 上负电荷更集中 ➔ RO⁻ 更不稳定 ➔ 酸性更弱');
    h.push('</text>');

    svg.innerHTML = h.join('\n');
  }

  function renderContrastSvg() {
    var svg = $('.ac-contrast-svg');
    if (!svg) return;
    var h = [];
    h.push('<defs>');
    h.push('  <marker id="carr-ind" markerWidth="6" markerHeight="6" refX="5" refY="3" orient="auto">');
    h.push('    <polygon points="0 0, 6 3, 0 6" fill="var(--acc)" />');
    h.push('  </marker>');
    h.push('  <marker id="carr-eq-r" markerWidth="7" markerHeight="5" refX="6" refY="2.5" orient="auto">');
    h.push('    <polygon points="0 0, 7 2.5, 0 5" fill="var(--fg)" />');
    h.push('  </marker>');
    h.push('  <marker id="carr-eq-l" markerWidth="7" markerHeight="5" refX="1" refY="2.5" orient="auto">');
    h.push('    <polygon points="7 0, 0 2.5, 7 5" fill="var(--fg)" />');
    h.push('  </marker>');
    h.push('  <radialGradient id="halo-low" cx="50%" cy="50%" r="50%">');
    h.push('    <stop offset="0%" stop-color="var(--ok)" stop-opacity="0.35" />');
    h.push('    <stop offset="100%" stop-color="var(--ok)" stop-opacity="0" />');
    h.push('  </radialGradient>');
    h.push('  <radialGradient id="halo-high" cx="50%" cy="50%" r="50%">');
    h.push('    <stop offset="0%" stop-color="var(--bad)" stop-opacity="0.55" />');
    h.push('    <stop offset="100%" stop-color="var(--bad)" stop-opacity="0" />');
    h.push('  </radialGradient>');
    h.push('</defs>');

    // Left half: Water
    h.push('<g transform="translate(15, 10)">');
    h.push('  <rect x="0" y="0" width="345" height="255" rx="8" fill="var(--card)" stroke="var(--ok)" stroke-width="1.4" />');
    h.push('  <text x="172" y="26" text-anchor="middle" font-size="14" font-weight="700" fill="var(--ok)">Water (H₂O) — 酸性相对较强 (pKa 15.7)</text>');
    
    // Water equation & inductive effect representation
    h.push('  <g transform="translate(20, 50)">');
    h.push('    <text x="20" y="45" font-size="17" font-weight="700" fill="var(--fg)">H–O–H</text>');
    h.push('    <line x1="88" y1="36" x2="125" y2="36" stroke="var(--fg)" stroke-width="1.8" marker-end="url(#carr-eq-r)" />');
    h.push('    <line x1="120" y1="46" x2="80" y2="46" stroke="var(--fg)" stroke-width="2.2" marker-end="url(#carr-eq-l)" />');
    h.push('    <circle cx="170" cy="40" r="14" fill="url(#halo-low)" />');
    h.push('    <text x="160" y="45" font-size="17" font-weight="700" fill="var(--bad)">OH⁻</text>');
    h.push('    <text x="202" y="45" font-size="16" fill="var(--fg)">+ H⁺</text>');
    h.push('  </g>');

    h.push('  <rect x="18" y="125" width="310" height="110" rx="6" fill="var(--ok-soft)" />');
    h.push('  <text x="30" y="148" font-size="12.5" font-weight="700" fill="var(--ok)">得分分析（Cambridge Mark Scheme）：</text>');
    h.push('  <text x="30" y="168" font-size="12" fill="var(--fg)">• H 原子没有推电子效应 (no positive inductive effect)</text>');
    h.push('  <text x="30" y="188" font-size="12" fill="var(--fg)">• OH⁻ 氧上的负电荷相对分散 (less concentrated)</text>');
    h.push('  <text x="30" y="208" font-size="12" fill="var(--fg)">• 稳定性较高，重新捕获 H⁺ 趋势更低</text>');
    h.push('  <text x="30" y="226" font-size="12" font-weight="600" fill="var(--ok)">➔ 解离程度较高，酸性更强 (water is a stronger acid)</text>');
    h.push('</g>');

    // Right half: Ethanol
    h.push('<g transform="translate(380, 10)">');
    h.push('  <rect x="0" y="0" width="345" height="255" rx="8" fill="var(--card)" stroke="var(--bad)" stroke-width="1.4" />');
    h.push('  <text x="172" y="26" text-anchor="middle" font-size="14" font-weight="700" fill="var(--bad)">Ethanol (C₂H₅OH) — 酸性更弱 (pKa 15.9)</text>');

    // Ethanol equation & inductive effect representation
    h.push('  <g transform="translate(15, 50)">');
    h.push('    <text x="10" y="45" font-size="15" font-weight="700" fill="var(--acc)">C₂H₅–OH</text>');
    h.push('    <line x1="88" y1="36" x2="115" y2="36" stroke="var(--fg)" stroke-width="1.4" marker-end="url(#carr-eq-r)" />');
    h.push('    <line x1="125" y1="46" x2="75" y2="46" stroke="var(--fg)" stroke-width="3" marker-end="url(#carr-eq-l)" />');
    h.push('    <circle cx="185" cy="40" r="23" fill="url(#halo-high)" />');
    h.push('    <text x="150" y="45" font-size="15" font-weight="700" fill="var(--bad)">C₂H₅O⁻</text>');
    h.push('    <text x="210" y="45" font-size="16" fill="var(--fg)">+ H⁺</text>');
    // Inductive arrows above C2H5O-
    h.push('    <line x1="145" y1="20" x2="182" y2="20" stroke="var(--acc)" stroke-width="2.2" marker-end="url(#carr-ind)" />');
    h.push('  </g>');

    h.push('  <rect x="18" y="125" width="310" height="110" rx="6" fill="var(--bad-soft)" />');
    h.push('  <text x="30" y="148" font-size="12.5" font-weight="700" fill="var(--bad)">得分分析（Cambridge Mark Scheme）：</text>');
    h.push('  <text x="30" y="168" font-size="12" fill="var(--fg)">• Ethyl group 有 <b>positive inductive effect</b> 向氧推电子</text>');
    h.push('  <text x="30" y="188" font-size="12" fill="var(--fg)">• C₂H₅O⁻ 氧上负电荷<b>更集中 (more concentrated)</b></text>');
    h.push('  <text x="30" y="208" font-size="12" fill="var(--fg)">• 稳定性更低 (less stable)，极易重新结合质子 H⁺</text>');
    h.push('  <text x="30" y="226" font-size="12" font-weight="600" fill="var(--bad)">➔ 平衡更偏向左边，酸性更弱 (ethanol is weaker)</text>');
    h.push('</g>');

    svg.innerHTML = h.join('\n');
  }

  /* Render Sodium Test Tubes SVG */
  function renderTubesSvg(time) {
    var svg = $('.ac-tubes-svg');
    if (!svg) return;
    var t = Math.max(0, Math.min(10, time));
    var h = [];

    h.push('<defs>');
    h.push('  <linearGradient id="na-metal" x1="0%" y1="0%" x2="100%" y2="100%">');
    h.push('    <stop offset="0%" stop-color="#f0f3f6" />');
    h.push('    <stop offset="50%" stop-color="#b5bec8" />');
    h.push('    <stop offset="100%" stop-color="#6e7781" />');
    h.push('  </linearGradient>');
    h.push('  <radialGradient id="na-flame" cx="50%" cy="50%" r="50%">');
    h.push('    <stop offset="0%" stop-color="#fffb00" stop-opacity="0.9" />');
    h.push('    <stop offset="40%" stop-color="#ff9900" stop-opacity="0.7" />');
    h.push('    <stop offset="100%" stop-color="#ff3300" stop-opacity="0" />');
    h.push('  </radialGradient>');
    h.push('</defs>');

    // Draw 3 test tubes
    var tubes = [
      { id: 'water', name: 'Water (H₂O)', x: 120, rate: 1.0 },
      { id: 'ethanol', name: 'Ethanol (1°)', x: 360, rate: 0.35 },
      { id: 'tbutanol', name: '2-methylpropan-2-ol (3°)', x: 600, rate: 0.08 }
    ];

    tubes.forEach(function (tb) {
      var tx = tb.x;
      // Tube outline
      h.push('<g>');
      // Test tube rack holder
      h.push('  <rect x="' + (tx - 50) + '" y="160" width="100" height="8" rx="2" fill="var(--line)" />');
      h.push('  <rect x="' + (tx - 50) + '" y="250" width="100" height="12" rx="2" fill="var(--line)" />');

      // Tube glass body
      h.push('  <path d="M ' + (tx - 35) + ' 50 L ' + (tx - 35) + ' 230 Q ' + (tx - 35) + ' 265 ' + tx + ' 265 Q ' + (tx + 35) + ' 265 ' + (tx + 35) + ' 230 L ' + (tx + 35) + ' 50 Z" fill="var(--card)" stroke="var(--line)" stroke-width="2.5" />');
      // Tube rim
      h.push('  <ellipse cx="' + tx + '" cy="50" rx="36" ry="6" fill="var(--card)" stroke="var(--line)" stroke-width="2" />');

      // Liquid level (y = 110 to 260)
      var liqFill = tb.id === 'water' ? 'rgba(74, 162, 255, 0.12)' : 'rgba(122, 162, 255, 0.08)';
      h.push('  <path d="M ' + (tx - 33) + ' 110 L ' + (tx - 33) + ' 230 Q ' + (tx - 33) + ' 263 ' + tx + ' 263 Q ' + (tx + 33) + ' 263 ' + (tx + 33) + ' 230 L ' + (tx + 33) + ' 110 Z" fill="' + liqFill + '" />');
      h.push('  <ellipse cx="' + tx + '" cy="110" rx="33" ry="5" fill="' + liqFill + '" stroke="var(--line)" stroke-width="1" />');

      // Sodium metal & Reaction dynamics
      var prog = Math.min(1.0, t * tb.rate);

      if (tb.id === 'water') {
        // Water: sodium floats on surface (y around 108), darts back and forth
        if (t > 0) {
          var remainingNa = Math.max(0, 1.0 - t * 0.45);
          if (remainingNa > 0.05) {
            var wiggle = Math.sin(t * 14) * 16;
            var naX = tx + wiggle;
            var naR = 9 * remainingNa;
            // Sodium bead
            h.push('  <circle cx="' + naX + '" cy="108" r="' + naR + '" fill="url(#na-metal)" stroke="var(--fg)" stroke-width="1" />');
            // Flame / hiss flare
            h.push('  <circle cx="' + naX + '" cy="104" r="' + (naR * 2.2) + '" fill="url(#na-flame)" />');
            // Smoke puffs above
            for (var p = 0; p < 4; p++) {
              var py = 95 - p * 12 - (t * 20 % 30);
              var px = naX + Math.sin(t * 8 + p) * 10;
              h.push('  <circle cx="' + px + '" cy="' + py + '" r="' + (4 + p * 2) + '" fill="var(--mut)" opacity="0.25" />');
            }
          } else {
            h.push('  <text x="' + tx + '" y="100" text-anchor="middle" font-size="12" font-weight="700" fill="var(--bad)">Na 全部反应完毕</text>');
          }
        } else {
          // Unreacted Na waiting
          h.push('  <circle cx="' + tx + '" cy="108" r="9" fill="url(#na-metal)" stroke="var(--fg)" stroke-width="1" />');
        }

        // Bubbles in water (vigorous!)
        if (t > 0 && t < 2.5) {
          for (var b = 0; b < 18; b++) {
            var bx = tx - 22 + (b * 19 % 44);
            var by = 110 + (b * 27 + t * 90) % 130;
            var br = 2.5 + (b % 3);
            h.push('  <circle cx="' + bx + '" cy="' + by + '" r="' + br + '" fill="none" stroke="var(--acc)" stroke-width="1.2" />');
          }
        }
      } else if (tb.id === 'ethanol') {
        // Ethanol: sodium sinks to bottom (density 0.97 > 0.79)
        var remainingNaE = Math.max(0.1, 1.0 - t * 0.12);
        var naW = 16 * remainingNaE;
        var naH = 12 * remainingNaE;
        var naY = 245 - naH / 2;

        h.push('  <rect x="' + (tx - naW / 2) + '" y="' + naY + '" width="' + naW + '" height="' + naH + '" rx="2" fill="url(#na-metal)" stroke="var(--fg)" stroke-width="1" />');

        // Steady bubble stream rising from bottom
        if (t > 0) {
          for (var eb = 0; eb < 12; eb++) {
            var ebx = tx - 12 + ((eb * 13) % 24);
            var eby = 240 - ((eb * 22 + t * 45) % 125);
            var ebr = 2 + (eb % 2);
            h.push('  <circle cx="' + ebx + '" cy="' + eby + '" r="' + ebr + '" fill="none" stroke="var(--ok)" stroke-width="1.4" />');
          }
        }
      } else {
        // 2-methylpropan-2-ol: sodium sinks, very sluggish bubbles
        var naW3 = 16;
        var naH3 = 12;
        var naY3 = 245 - naH3 / 2;
        h.push('  <rect x="' + (tx - naW3 / 2) + '" y="' + naY3 + '" width="' + naW3 + '" height="' + naH3 + '" rx="2" fill="url(#na-metal)" stroke="var(--fg)" stroke-width="1" />');

        if (t > 0) {
          for (var sb = 0; sb < 4; sb++) {
            var sbx = tx - 8 + ((sb * 15) % 16);
            var sby = 240 - ((sb * 35 + t * 15) % 125);
            h.push('  <circle cx="' + sbx + '" cy="' + sby + '" r="2" fill="none" stroke="var(--mut)" stroke-width="1.2" />');
          }
        }
      }

      // Tube label
      h.push('  <text x="' + tx + '" y="25" text-anchor="middle" font-size="13" font-weight="700" fill="var(--fg)">' + tb.name + '</text>');
      h.push('</g>');
    });

    svg.innerHTML = h.join('\n');
  }

  function updateMolecule(molId) {
    curMol = molId;
    var d = MOL_DATA[curMol];

    $$('.ac-mol-btn').forEach(function (b) {
      b.classList.toggle('on', b.dataset.mol === molId);
    });

    // Update meter nodes
    $$('.ac-meter-node').forEach(function (n) {
      n.classList.toggle('on', n.dataset.m === molId);
    });

    // Update text
    $('.ac-mol-title').textContent = d.name;
    $('.ac-type-badge').textContent = d.type;
    $('.ac-pka-badge').innerHTML = 'p<i>K</i><sub>a</sub> ≈ ' + d.pka;

    $('.ac-txt-ind').textContent = d.indTxt;
    $('.ac-txt-dens').textContent = d.densTxt;
    $('.ac-txt-stab').textContent = d.stabTxt;
    $('.ac-txt-eq').textContent = d.eqTxt;

    renderTheorySvg();
    K.fit();
  }

  function setContrast(show) {
    contrastMode = !!show;
    $('.ac-single-view').hidden = contrastMode;
    $('.ac-contrast-view').hidden = !contrastMode;
    $('.ac-btn-contrast').classList.toggle('on', contrastMode);
    if (contrastMode) {
      renderContrastSvg();
    } else {
      renderTheorySvg();
    }
    K.fit();
  }

  /* Sodium Simulation Timer Loop */
  function updateSodiumTime(t) {
    naTime = Math.max(0, Math.min(10, t));
    $('.ac-time').textContent = naTime.toFixed(1);
    $('.ac-time-slider').value = naTime;
    renderTubesSvg(naTime);
  }

  function naTick(ts) {
    if (!naRunning) return;
    if (lastTs == null) lastTs = ts;
    var dt = (ts - lastTs) / 1000;
    lastTs = ts;
    var nt = naTime + dt * 1.5;
    if (nt >= 10) {
      nt = 10;
      updateSodiumTime(nt);
      naRunning = false;
      $('.ac-btn-play').classList.remove('on');
      return;
    }
    updateSodiumTime(nt);
    naRaf = requestAnimationFrame(naTick);
  }

  function playSodium() {
    if (naTime >= 10) naTime = 0;
    naRunning = true;
    lastTs = null;
    $('.ac-btn-play').classList.add('on');
    $('.ac-btn-pause').classList.remove('on');
    naRaf = requestAnimationFrame(naTick);
  }

  function pauseSodium() {
    naRunning = false;
    $('.ac-btn-play').classList.remove('on');
    $('.ac-btn-pause').classList.add('on');
    if (naRaf) cancelAnimationFrame(naRaf);
  }

  function resetSodium() {
    pauseSodium();
    $('.ac-btn-pause').classList.remove('on');
    updateSodiumTime(0);
  }

  /* Quiz Logic */
  var QUIZ_ANS = {
    0: {
      correct: 'B',
      okFb: '✓ 正确！Alkyl group 具有推电子的 positive inductive effect，使 ethoxide ion（C₂H₅O⁻）氧上的负电荷更加集中，导致其更不稳定（less stable than OH⁻），极易重新结合质子 H⁺，解离平衡显著偏向左边，因此酸性比水更弱。',
      badFb: '✗ 错误。核心踩分点：alkyl group 的 positive inductive effect 使 RO⁻ 氧原子上的负电荷更集中、稳定性降低，平衡更偏左。'
    },
    1: {
      correct: 'C',
      okFb: '✓ 正确！Cambridge 9701 官方标答必须写 effervescence / bubbles（冒气泡）或 sodium dissolves / decreases in size。只写产生气体或写检验方法判 0 分！',
      badFb: '✗ 错误！剑桥官方阅卷标准中，必须直接描述可见现象「effervescence / bubbles」或「sodium 逐渐溶解变小」，写产生氢气或点燃爆鸣均不得分。'
    },
    2: {
      correct: 'A',
      okFb: '✓ 正确！Water (pKa 15.7) > ethanol (pKa 15.9) > 2-methylpropan-2-ol (pKa 18.0)。Alkyl group 数量越多（3° > 2° > 1°），推电子效应越强，酸性越弱。',
      badFb: '✗ 错误。推电子烷基越多，RO⁻ 氧上负电荷越集中越不稳定，酸性越弱。因此酸性顺序为：water > 1° alcohol > 3° alcohol。'
    },
    3: {
      correct: 'B',
      okFb: '✓ 正确！Propan-1-ol 与金属钠反应生成 sodium propoxide（或 sodium propan-1-olate）以及氢气 H₂。',
      badFb: '✗ 错误。醇与金属钠发生置换反应生成 alkoxide 盐（sodium propoxide）并放出 H₂。'
    }
  };

  var userAnswers = {};

  function updateQuizScore() {
    var sc = 0;
    for (var i = 0; i < 4; i++) {
      if (userAnswers[i] === QUIZ_ANS[i].correct) sc++;
    }
    $('.ac-sc').textContent = String(sc);
  }

  function answerQuestion(qi, opt) {
    userAnswers[qi] = opt;
    var item = $('.ac-qitem[data-qi="' + qi + '"]');
    var fb = item.querySelector('.ac-fb');
    var btns = item.querySelectorAll('.ac-opt-btn');

    btns.forEach(function (b) {
      b.classList.remove('ok', 'bad');
      if (b.dataset.opt === opt) {
        b.classList.add(opt === QUIZ_ANS[qi].correct ? 'ok' : 'bad');
      }
    });

    fb.hidden = false;
    var isOk = (opt === QUIZ_ANS[qi].correct);
    fb.className = 'ac-fb ' + (isOk ? 'ok' : 'bad');
    fb.innerHTML = isOk ? QUIZ_ANS[qi].okFb : QUIZ_ANS[qi].badFb;

    updateQuizScore();
    K.fit();
  }

  /* Event Listeners */
  $$('.ac-mol-btn').forEach(function (btn) {
    btn.addEventListener('click', function () {
      updateMolecule(btn.dataset.mol);
    });
  });

  $('.ac-btn-contrast').addEventListener('click', function () {
    setContrast(!contrastMode);
  });

  $('.ac-btn-close-contrast').addEventListener('click', function () {
    setContrast(false);
  });

  $('.ac-btn-play').addEventListener('click', playSodium);
  $('.ac-btn-pause').addEventListener('click', pauseSodium);
  $('.ac-btn-reset').addEventListener('click', resetSodium);

  $('.ac-time-slider').addEventListener('input', function (e) {
    pauseSodium();
    updateSodiumTime(parseFloat(e.target.value));
  });

  $$('.ac-opt-btn').forEach(function (btn) {
    btn.addEventListener('click', function () {
      var qi = parseInt(btn.closest('.ac-qitem').dataset.qi, 10);
      answerQuestion(qi, btn.dataset.opt);
    });
  });

  $('.ac-qshow').addEventListener('click', function () {
    for (var i = 0; i < 4; i++) {
      answerQuestion(i, QUIZ_ANS[i].correct);
    }
  });

  $('.ac-qreset').addEventListener('click', function () {
    userAnswers = {};
    $$('.ac-opt-btn').forEach(function (b) { b.classList.remove('ok', 'bad'); });
    $$('.ac-fb').forEach(function (fb) { fb.hidden = true; fb.innerHTML = ''; });
    updateQuizScore();
    K.fit();
  });

  // Tab switching custom listener
  $root.addEventListener('tab', function (e) {
    curTab = e.detail;
    if (curTab === 'sodium') {
      renderTubesSvg(naTime);
    } else if (curTab === 'theory') {
      if (contrastMode) renderContrastSvg();
      else renderTheorySvg();
    }
  });

  /* Test Hook attached to window and root */
  var hook = {
    setTab: function (tabId) {
      var btn = $('.tabbar button[data-tab="' + tabId + '"]');
      if (btn) btn.click();
    },
    getTab: function () { return curTab; },
    setMolecule: function (molId) { updateMolecule(molId); },
    getMolecule: function () { return curMol; },
    toggleContrast: function (on) { setContrast(on); },
    isContrast: function () { return contrastMode; },
    setT: function (t) { updateSodiumTime(t); },
    getT: function () { return naTime; },
    play: playSodium,
    pause: pauseSodium,
    reset: resetSodium,
    isPlaying: function () { return naRunning; },
    answerQuiz: function (qi, opt) { answerQuestion(qi, opt); },
    getScore: function () {
      var sc = 0;
      for (var i = 0; i < 4; i++) {
        if (userAnswers[i] === QUIZ_ANS[i].correct) sc++;
      }
      return sc;
    },
    showAnswers: function () { $('.ac-qshow').click(); },
    resetQuiz: function () { $('.ac-qreset').click(); },
    getMoleculesList: function () { return Object.keys(MOL_DATA); },
    getMinFontSize: function () {
      var minPx = 999;
      var svgs = $$('.ac-svg');
      svgs.forEach(function (svg) {
        if (svg.closest('[hidden]')) return;
        var r = svg.getBoundingClientRect();
        if (!r.width) return;
        var vb = (svg.getAttribute('viewBox') || '').trim().split(/\s+/).map(Number);
        var scale = vb[2] ? r.width / vb[2] : 1;
        var texts = svg.querySelectorAll('text');
        texts.forEach(function (t) {
          if (!t.textContent.trim()) return;
          var fs = parseFloat(getComputedStyle(t).fontSize) * scale;
          if (fs < minPx) minPx = fs;
        });
      });
      return minPx;
    }
  };

  window.T_acidity = hook;
  $root.acidityHook = hook;

  // Initialize
  K.init($root);
  renderTheorySvg();
  renderTubesSvg(0);

K.init($root);
})(document.getElementById("k-ch16-acidity"));</script>"""))
```

### F. (d) oxidation：distillation 还是 reflux

oxidising agent：**acidified K₂Cr₂O₇**（orange → green）或 **acidified KMnO₄**（purple → colourless）。

| alcohol | **distillation**（product 一生成就蒸出） | **reflux**（excess oxidising agent，长时间加热） |
|---|---|---|
| **primary** | **aldehyde**：CH₃CH₂OH + [O] → CH₃CHO + H₂O | **carboxylic acid**：CH₃CH₂OH + 2[O] → CH₃COOH + H₂O |
| **secondary** | **ketone**：CH₃CH(OH)CH₃ + [O] → CH₃COCH₃ + H₂O | 仍是 ketone（ketone 不再被 oxidise） |
| **tertiary** | 不反应 | 不反应 |

- 为什么 distillation 得到 aldehyde：aldehyde 的 bp 比 alcohol、acid 低得多（没有 O–H，不形成 hydrogen bond），一生成就蒸出，来不及被进一步 oxidise。
- reflux：冷凝管竖直，蒸气冷凝回流，反应物和 product 一直留在烧瓶里 → aldehyde 继续变成 acid。
- 多个 OH 的 molecule 被 excess K₂Cr₂O₇ reflux 时，**每个 1° OH → COOH，每个 2° OH → C=O，3° OH 不变**（9701/22 May/June 2026 Q4(b) 的 X → C₅H₆O₇）。
- 配平用 [O]：每个 1° → COOH 用 2[O]、放 1 H₂O；每个 → C=O 用 1[O]、放 1 H₂O。

aldehyde / ketone 之后的反应见 [chapter 17](https://colab.research.google.com/drive/10P0lQfoNh6lRy7HRbjJZxH7AnzI4V3vL#scrollTo=ch17-carbonyls)，carboxylic acid 见 [chapter 18](https://colab.research.google.com/drive/1gxHJxFjn1RQSXR5y5n5CSeW5q82c6-6E#scrollTo=ch18-acids)。

```python
# @title 🔬 K₂Cr₂O₇ 试管：1° / 2° / 3°，distillation 与 reflux { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-oxidation-tubes">
<h3>K₂Cr₂O₇ 氧化：distillation 与 reflux <span class="faint">选底物与条件 · 点加热观察变色与产物</span></h3>

<!-- 精简一行高亮提示卡 -->
<div class="row ox-summary-strip">
  <div class="ox-sum-item"><b>1° alcohol</b>：蒸馏（distillation）得 <b>aldehyde</b>；回流（reflux）完全氧化为 <b>carboxylic acid</b></div>
  <div class="ox-sum-item"><b>2° alcohol</b>：蒸馏或回流均只氧化为 <b>ketone</b>（无进一步反应）</div>
  <div class="ox-sum-item"><b>3° alcohol</b>：羟基碳上无氢原子（no α-H），<b>不发生反应 (remains orange)</b></div>
</div>

<div class="ox-wrap">
  <div class="tabbar">
    <button type="button" class="on" data-tab="explorer">① 反应装置与产物探索 (Distillation vs Reflux)</button>
    <button type="button" data-tab="parallel">② 三试管并排鉴别模式 (3-Tube Parallel Test)</button>
    <button type="button" data-tab="quiz">③ 真题考点自测 (Practice Quiz)</button>
  </div>

  <!-- ================= Tab 1: 反应装置与产物探索 ================= -->
  <div data-pane="explorer" class="show">
    <!-- 紧凑精致的控制栏 -->
    <div class="card ox-ctrl-card">
      <div class="row ox-ctrl-row">
        <span class="ox-ctrl-lbl">底物 Substrate：</span>
        <div class="row ox-sub-pills" role="group" aria-label="选择醇底物">
          <button type="button" class="ox-sub-btn on" data-sub="ethanol">ethanol (1°)</button>
          <button type="button" class="ox-sub-btn" data-sub="propan1ol">propan-1-ol (1°)</button>
          <button type="button" class="ox-sub-btn" data-sub="mpropan1ol">2-methylpropan-1-ol (1°)</button>
          <button type="button" class="ox-sub-btn" data-sub="propan2ol">propan-2-ol (2°)</button>
          <button type="button" class="ox-sub-btn" data-sub="butan2ol">butan-2-ol (2°)</button>
          <button type="button" class="ox-sub-btn" data-sub="tbutanol">2-methylpropan-2-ol (3°)</button>
        </div>
      </div>
      <div class="row ox-ctrl-row ox-cond-row">
        <div class="row ox-sub-group">
          <span class="ox-ctrl-lbl">装置 Condition：</span>
          <button type="button" class="ox-cond-btn" data-cond="distil">🧪 Distillation (蒸馏出产物)</button>
          <button type="button" class="ox-cond-btn on" data-cond="reflux">⚗️ Reflux (长时间充分回流)</button>
        </div>
        <div class="row ox-sub-group ox-ox-group">
          <span class="ox-ctrl-lbl">氧化剂 Oxidant：</span>
          <button type="button" class="ox-ox-btn on" data-ox="dichromate">K₂Cr₂O₇/H⁺ (orange→green)</button>
          <button type="button" class="ox-ox-btn" data-ox="manganate">KMnO₄/H⁺ (purple→colourless)</button>
        </div>
      </div>
    </div>

    <!-- 舞台双栏 -->
    <div class="grid2 ox-stage-grid">
      <!-- 左侧：仪器与动画模拟 -->
      <div class="card ox-stage-card">
        <div class="ox-stage-head">
          <span class="ox-cur-title">ethanol (1°) + Reflux</span>
          <span class="ox-badge ox-class-badge">1° alcohol</span>
          <span class="ox-badge ox-res-badge">orange → green</span>
        </div>

        <div class="row ox-play-bar">
          <button type="button" class="ox-btn-heat on">🔥 加热反应 (Heat)</button>
          <button type="button" class="ox-btn-reset">↺ 重置</button>
          <div class="ox-slider-wrap">
            <span class="faint">反应进度：</span>
            <input type="range" class="ox-progress-slider" min="0" max="100" value="0" aria-label="反应进度滑块">
            <b class="ox-prog-val">0%</b>
          </div>
        </div>

        <!-- 装置状态标头（外置，完全避免与 SVG 内部重叠） -->
        <div class="ox-apparatus-banner">
          <span class="ox-app-badge">⚗️ 装置特征</span>
          <span class="ox-app-desc">Reflux 竖直冷凝管：冷水下进上出，蒸气遇冷液化回流，确保反应物与过量氧化剂彻底反应</span>
        </div>

        <div class="ox-stage-box">
          <svg class="ox-svg ox-app-svg" viewBox="0 0 520 340" role="img" aria-label="反应装置与试管颜色变化图"></svg>
        </div>

        <div class="ox-apparatus-notes">
          <div class="ox-an-col">
            <b class="ox-an-title">冷凝水进出方向规范（Cambridge 9701 经典实验考点）：</b>
            <span class="ox-an-text">冷水必须从<b>底部接口流入 (water in)</b>、<b>顶部接口流出 (water out)</b>。这样可依靠水压与重力使冷却水完全充满套管夹层，避免留存气穴导致冷凝失效或受热不均。</span>
          </div>
        </div>
      </div>

      <!-- 右侧：化学方程式与得分点卡片 -->
      <div class="card ox-info-card">
        <div class="ox-ih">📋 反应产物与 Cambridge 9701 核心得分点</div>

        <!-- Reactant vs Product comparison -->
        <div class="ox-prod-box">
          <div class="ox-pb-item">
            <span class="faint">底物醇 (Reactant)：</span>
            <div class="ox-p-name ox-reac-name">ethanol</div>
            <div class="ox-p-form ox-reac-form">CH₃CH₂OH</div>
          </div>
          <div class="ox-pb-arr">➔</div>
          <div class="ox-pb-item">
            <span class="faint">有机产物 (Organic Product)：</span>
            <div class="ox-p-name ox-prod-name">ethanoic acid</div>
            <div class="ox-p-form ox-prod-form">CH₃COOH</div>
          </div>
        </div>

        <!-- Chemical Equation Box -->
        <div class="ox-eq-box">
          <span class="faint">配平反应方程式 (Balanced equation with [O])：</span>
          <div class="ox-eq-text">CH₃CH₂OH + 2[O] → CH₃COOH + H₂O</div>
        </div>

        <!-- Exam Analysis Bullet Points -->
        <div class="ox-points-box">
          <div class="ox-pt-item">
            <b class="ox-pt-h">1. 为什么该条件生成此产物？</b>
            <span class="ox-pt-desc ox-why-cond">回流条件下使用过量 acidified K₂Cr₂O₇，初生的 ethanal 蒸发后在竖直冷凝管中遇冷液化回流，无法逃逸，被继续完全氧化为 ethanoic acid。</span>
          </div>
          <div class="ox-pt-item">
            <b class="ox-pt-h">2. 沸点与分子间作用力差异（极常考）：</b>
            <span class="ox-pt-desc ox-bp-desc">Ethanal (bp 20°C，无氢键，仅有 dipole-dipole) &lt; Ethanol (bp 78°C，有氢键) &lt; Ethanoic acid (bp 118°C，强氢键二聚体)。蒸馏时 ethanal 沸点最低最先逸出。</span>
          </div>
          <div class="ox-pt-item">
            <b class="ox-pt-h">3. 考试观察现象与得分词（Observation）：</b>
            <span class="ox-pt-desc ox-obs-desc">必须完整写 <b>orange to green</b>（或 orange solution turns green）。单写 green 或写变成黄色不给分。</span>
          </div>
        </div>
      </div>
    </div>
  </div>

  <!-- ================= Tab 2: 三试管并排鉴别模式 ================= -->
  <div data-pane="parallel">
    <div class="card ox-sec-card">
      <div class="row ox-p-ctrls">
        <button type="button" class="ox-btn-warm on">🔥 水浴温热全部试管 (Warm in water bath)</button>
        <button type="button" class="ox-btn-p-reset">↺ 重置试管</button>
        <span class="faint" style="margin-left:auto">分别向 1°、2°、3° 醇各试管加入等量 acidified K₂Cr₂O₇</span>
      </div>
    </div>

    <!-- Parallel tubes Stage -->
    <div class="card ox-stage-card">
      <div class="ox-stage-box">
        <svg class="ox-svg ox-parallel-svg" viewBox="0 0 660 260" role="img" aria-label="三种醇并排氧化试管颜色对比"></svg>
      </div>

      <!-- Tubes status cards below -->
      <div class="row ox-ptubes-row">
        <div class="ox-ptube-card">
          <div class="ox-ptc-name"><b>试管 1：Primary (1°)</b></div>
          <div class="faint">ethanol (CH₃CH₂OH)</div>
          <div class="ox-badge ox-badge-p1">orange → green</div>
          <div class="ox-ptc-desc">发生氧化反应，Cr₂O₇²⁻ 被还原为 Cr³⁺（绿色溶液）。蒸馏得 aldehyde，回流得 carboxylic acid。</div>
        </div>

        <div class="ox-ptube-card">
          <div class="ox-ptc-name"><b>试管 2：Secondary (2°)</b></div>
          <div class="faint">propan-2-ol (CH₃CH(OH)CH₃)</div>
          <div class="ox-badge ox-badge-p2">orange → green</div>
          <div class="ox-ptc-desc">发生氧化反应，生成 ketone（丙酮 propanone）。溶液由橙色变为绿色。</div>
        </div>

        <div class="ox-ptube-card">
          <div class="ox-ptc-name"><b>试管 3：Tertiary (3°)</b></div>
          <div class="faint">2-methylpropan-2-ol ((CH₃)₃COH)</div>
          <div class="ox-badge ox-badge-p3">remains orange</div>
          <div class="ox-ptc-desc">连有 –OH 的 C 上无 H 原子（no α-H），不发生反应，溶液<b>保持橙色 (remains orange)</b>！</div>
        </div>
      </div>
    </div>

    <!-- Cambridge Distinguishing Decision Tree Card -->
    <div class="card ox-tree-card">
      <div class="ox-dh">💡 Cambridge 9701 醇分类鉴别实验逻辑树（Distinguishing Strategy）</div>
      <div class="ox-tree-grid">
        <div class="ox-tree-step">
          <div class="ox-ts-num">步骤 1</div>
          <b>加入 acidified K₂Cr₂O₇ 加热温热：</b>
          <ul>
            <li><b style="color:var(--bad)">保持橙色 (remains orange)</b> ➔ 检出为 <b>3° alcohol</b>！</li>
            <li><b style="color:var(--ok)">橙色变绿色 (orange → green)</b> ➔ 说明为 <b>1° 或 2° alcohol</b>（还需进一步区分）。</li>
          </ul>
        </div>
        <div class="ox-tree-step">
          <div class="ox-ts-num">步骤 2</div>
          <b>进一步区分 1° 与 2° 醇（检验蒸馏氧化产物）：</b>
          <ul>
            <li>收集氧化产物，加入 <b>Tollens' reagent</b> 温热：产生 <b>silver mirror</b> 为 <b>1° 醇产物 (aldehyde)</b>；无现象为 <b>2° 醇产物 (ketone)</b>。</li>
            <li>或加入 <b>Fehling's solution</b> 温热：产生 <b>red precipitate of Cu₂O</b> 为 aldehyde（来自 1° 醇）。</li>
          </ul>
        </div>
      </div>
    </div>
  </div>

  <!-- ================= Tab 3: 真题考点自测 ================= -->
  <div data-pane="quiz">
    <div class="card ox-quiz-card">
      <div class="ox-qh">
        <h4>📝 Cambridge 9701 醇氧化核心考点真题自测</h4>
        <span class="ox-qscore">已正确 <b class="ox-sc">0</b> / 4 题</span>
      </div>
      <div class="ox-qitems">
        <!-- Q1 -->
        <div class="ox-qitem" data-qi="0">
          <div class="ox-qp"><b>1.</b> 将 propan-2-ol 与过量 acidified K₂Cr₂O₇ 混合并加热回流（heated under reflux）。在试卷中描述实验现象（observation），正确的写法是？</div>
          <div class="ox-opts">
            <button type="button" class="ox-opt-btn" data-opt="A">溶液变成黄色</button>
            <button type="button" class="ox-opt-btn" data-opt="B">orange solution turns green (或 orange to green)</button>
            <button type="button" class="ox-opt-btn" data-opt="C">产生大量 colourless bubbles</button>
            <button type="button" class="ox-opt-btn" data-opt="D">溶液保持不变依然为 orange</button>
          </div>
          <div class="ox-fb" hidden></div>
        </div>

        <!-- Q2 -->
        <div class="ox-qitem" data-qi="1">
          <div class="ox-qp"><b>2.</b> 将 ethanol 与过量 acidified K₂Cr₂O₇ 在<b>回流（reflux）</b>条件下长时间充分加热。反应的主要有机产物是？</div>
          <div class="ox-opts">
            <button type="button" class="ox-opt-btn" data-opt="A">ethanal (CH₃CHO)</button>
            <button type="button" class="ox-opt-btn" data-opt="B">ethanoic acid (CH₃COOH)</button>
            <button type="button" class="ox-opt-btn" data-opt="C">ethyl ethanoate (CH₃COOCH₂CH₃)</button>
            <button type="button" class="ox-opt-btn" data-opt="D">ethene (CH₂=CH₂)</button>
          </div>
          <div class="ox-fb" hidden></div>
        </div>

        <!-- Q3 -->
        <div class="ox-qitem" data-qi="2">
          <div class="ox-qp"><b>3.</b> 解释为什么 2-methylpropan-2-ol 无法被酸化重铬酸钾氧化（remains orange）？（Cambridge mark scheme 得分点）</div>
          <div class="ox-opts">
            <button type="button" class="ox-opt-btn" data-opt="A">分子量过大，空间位阻阻碍了重铬酸根离子接近</button>
            <button type="button" class="ox-opt-btn" data-opt="B">连有 –OH 基团的碳原子上没有与之相连的氢原子（no hydrogen on the C–OH carbon）</button>
            <button type="button" class="ox-opt-btn" data-opt="C">叔丁醇在酸性溶液中溶解度为零，无法与水溶液相反应</button>
            <button type="button" class="ox-opt-btn" data-opt="D">叔丁醇分子中没有 C–C 单键可供断裂</button>
          </div>
          <div class="ox-fb" hidden></div>
        </div>

        <!-- Q4 -->
        <div class="ox-qitem" data-qi="3">
          <div class="ox-qp"><b>4.</b> 在 1° 醇氧化制备 aldehyde 的实验中，必须采用<b>蒸馏装置（distillation）</b>而不是回流装置。其主要原因是？</div>
          <div class="ox-opts">
            <button type="button" class="ox-opt-btn" data-opt="A">Aldehyde 分子间没有氢键，沸点显著低于反应物 alcohol 和进一步氧化的 acid，一生成便挥发逸出被冷凝收集，防止被继续氧化</button>
            <button type="button" class="ox-opt-btn" data-opt="B">回流冷凝管会催化 aldehyde 分解为二氧化碳和水</button>
            <button type="button" class="ox-opt-btn" data-opt="C">Distillation 装置中水分较少，能促进醇分子的脱氢反应</button>
            <button type="button" class="ox-opt-btn" data-opt="D">Aldehyde 的密度大于水，必须在冷凝管侧支处依靠重力沉淀析出</button>
          </div>
          <div class="ox-fb" hidden></div>
        </div>
      </div>

      <div class="row ox-qtools">
        <button type="button" class="ox-qshow">显示标准答案</button>
        <button type="button" class="ox-qreset">重置练习</button>
      </div>
    </div>
  </div>
</div>

<style>
.k [hidden]{display:none!important}
.k .ox-wrap{margin-top:10px}

/* 精简一行高亮提示卡 */
.k .ox-summary-strip{
  display:flex;
  flex-wrap:wrap;
  gap:8px 14px;
  background:var(--acc-soft);
  border:1px solid var(--line);
  border-radius:8px;
  padding:7px 12px;
  font-size:12.5px;
  margin:6px 0 10px;
  align-items:center;
}
.k .ox-sum-item{
  flex:1 1 200px;
  line-height:1.4;
  color:var(--fg);
}

/* 控制栏精致化 */
.k .ox-ctrl-card{
  padding:8px 12px;
  margin-bottom:10px;
  display:flex;
  flex-direction:column;
  gap:6px;
}
.k .ox-ctrl-row{
  display:flex;
  flex-wrap:wrap;
  align-items:center;
  gap:6px 10px;
}
.k .ox-ctrl-lbl{
  font-size:12.5px;
  font-weight:700;
  color:var(--fg);
  white-space:nowrap;
}
.k .ox-sub-pills{
  display:flex;
  flex-wrap:wrap;
  gap:4px 6px;
}
.k .ox-sub-btn{
  font-size:12px;
  padding:3px 9px;
  border-radius:14px;
  border:1px solid var(--line);
  background:var(--card);
  color:var(--fg);
  cursor:pointer;
  transition:all 0.15s ease;
}
.k .ox-sub-btn:hover{
  border-color:var(--acc);
}
.k .ox-sub-btn.on{
  background:var(--acc);
  color:#fff;
  border-color:var(--acc);
  font-weight:600;
}
.k .ox-cond-row{
  justify-content:space-between;
  border-top:1px dashed var(--line);
  padding-top:6px;
  margin-top:2px;
}
.k .ox-sub-group{
  display:flex;
  flex-wrap:wrap;
  align-items:center;
  gap:6px;
}
.k .ox-cond-btn,.k .ox-ox-btn{
  font-size:12px;
  padding:3px 10px;
  border-radius:6px;
  border:1px solid var(--line);
  background:var(--card);
  color:var(--fg);
  cursor:pointer;
  transition:all 0.15s ease;
}
.k .ox-cond-btn:hover,.k .ox-ox-btn:hover{
  border-color:var(--acc);
}
.k .ox-cond-btn.on,.k .ox-ox-btn.on{
  background:var(--acc);
  color:#fff;
  border-color:var(--acc);
  font-weight:600;
}
@media (max-width:760px){
  .k .ox-cond-row{
    flex-direction:column;
    align-items:flex-start;
    gap:8px;
  }
}

/* 舞台双栏 */
.k .ox-stage-grid{
  grid-template-columns:minmax(330px,1.2fr) minmax(290px,1fr);
  align-items:stretch;
  gap:12px;
}
@media (max-width:760px){
  .k .ox-stage-grid{
    grid-template-columns:1fr;
  }
}

/* 左侧舞台卡 */
.k .ox-stage-card{
  display:flex;
  flex-direction:column;
  padding:10px 12px;
}
.k .ox-stage-head{
  display:flex;
  flex-wrap:wrap;
  align-items:center;
  gap:6px 10px;
  margin-bottom:6px;
}
.k .ox-cur-title{
  font-size:14.5px;
  font-weight:700;
  color:var(--fg);
}
.k .ox-badge{
  font-size:11.5px;
  font-weight:600;
  padding:2px 8px;
  border-radius:4px;
  border:1px solid var(--line);
  background:var(--card);
}
.k .ox-class-badge{
  color:var(--acc);
  border-color:var(--acc);
  background:var(--acc-soft);
}
.k .ox-res-badge{
  color:var(--ok);
  border-color:var(--ok);
  background:var(--ok-soft);
}
.k .ox-play-bar{
  align-items:center;
  gap:8px;
  margin-bottom:8px;
}
.k .ox-play-bar button{
  font-size:12.5px;
  padding:3px 10px;
}
.k .ox-slider-wrap{
  margin-left:auto;
  display:inline-flex;
  align-items:center;
  gap:6px;
  font-size:12.5px;
}
.k .ox-progress-slider{
  width:105px;
}
.k .ox-prog-val{
  min-width:34px;
  font-size:12.5px;
  color:var(--acc);
}

/* 装置外置特征条 */
.k .ox-apparatus-banner{
  display:flex;
  align-items:center;
  gap:8px;
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  padding:5px 8px;
  margin-bottom:8px;
  font-size:12px;
  line-height:1.35;
}
.k .ox-app-badge{
  font-weight:700;
  color:var(--fg);
  white-space:nowrap;
  padding:1px 6px;
  border-radius:4px;
  background:var(--line);
}
.k .ox-app-desc{
  color:var(--mut);
}

.k .ox-stage-box{
  background:var(--card);
  border:1px solid var(--line);
  border-radius:8px;
  padding:4px;
  overflow:hidden;
  display:flex;
  align-items:center;
  justify-content:center;
}
.k .ox-app-svg{
  width:100%;
  height:auto;
  display:block;
}
.k .ox-apparatus-notes{
  margin-top:8px;
  padding:6px 10px;
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  font-size:12px;
  line-height:1.4;
}
.k .ox-an-title{
  color:var(--fg);
}
.k .ox-an-text{
  color:var(--mut);
}

/* 右侧信息卡 */
.k .ox-info-card{
  display:flex;
  flex-direction:column;
  padding:10px 14px;
}
.k .ox-ih{
  font-weight:700;
  font-size:13.5px;
  color:var(--fg);
  margin-bottom:8px;
  border-bottom:1px solid var(--line);
  padding-bottom:5px;
}
.k .ox-prod-box{
  display:flex;
  align-items:center;
  justify-content:space-between;
  gap:8px;
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  padding:8px 12px;
  margin-bottom:8px;
}
.k .ox-pb-item{
  display:flex;
  flex-direction:column;
  gap:2px;
}
.k .ox-p-name{
  font-size:14px;
  font-weight:700;
  color:var(--fg);
}
.k .ox-p-form{
  font-size:13px;
  color:var(--acc);
  font-family:monospace;
}
.k .ox-pb-arr{
  font-size:18px;
  font-weight:700;
  color:var(--mut);
}
.k .ox-eq-box{
  background:var(--acc-soft);
  border:1px solid var(--line);
  border-radius:6px;
  padding:7px 10px;
  margin-bottom:8px;
  font-size:12.5px;
}
.k .ox-eq-text{
  font-size:13px;
  font-weight:700;
  margin-top:2px;
  color:var(--fg);
  font-family:monospace;
}
.k .ox-points-box{
  display:flex;
  flex-direction:column;
  gap:7px;
  font-size:12.5px;
  line-height:1.45;
}
.k .ox-pt-item{
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  padding:6px 10px;
}
.k .ox-pt-h{
  font-size:12.5px;
  display:block;
  margin-bottom:2px;
  color:var(--fg);
}
.k .ox-pt-desc{
  color:var(--mut);
}

/* Tab 2: 并排试管模式 */
.k .ox-sec-card{
  padding:8px 12px;
  margin-bottom:10px;
}
.k .ox-p-ctrls{
  align-items:center;
  gap:8px;
}
.k .ox-p-ctrls button{
  font-size:12.5px;
  padding:3px 10px;
}
.k .ox-parallel-svg{
  width:100%;
  height:auto;
  display:block;
}
.k .ox-ptubes-row{
  display:grid;
  grid-template-columns:repeat(3,1fr);
  gap:10px;
  margin-top:10px;
}
@media (max-width:680px){
  .k .ox-ptubes-row{
    grid-template-columns:1fr;
  }
}
.k .ox-ptube-card{
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  padding:8px 10px;
  font-size:12px;
  display:flex;
  flex-direction:column;
  gap:3px;
}
.k .ox-ptc-name{
  font-size:13px;
  color:var(--fg);
}
.k .ox-badge-p1,.k .ox-badge-p2{
  background:var(--ok-soft);
  color:var(--ok);
  border-color:var(--ok);
  align-self:flex-start;
}
.k .ox-badge-p3{
  background:var(--acc-soft);
  color:var(--mut);
  border-color:var(--line);
  align-self:flex-start;
}
.k .ox-ptc-desc{
  color:var(--mut);
  margin-top:2px;
  line-height:1.35;
}
.k .ox-tree-card{
  margin-top:10px;
  padding:10px 12px;
}
.k .ox-dh{
  font-weight:700;
  font-size:13.5px;
  color:var(--fg);
  margin-bottom:6px;
}
.k .ox-tree-grid{
  display:grid;
  grid-template-columns:1fr 1fr;
  gap:10px;
  margin-top:6px;
}
@media (max-width:640px){
  .k .ox-tree-grid{
    grid-template-columns:1fr;
  }
}
.k .ox-tree-step{
  background:var(--card);
  border:1px solid var(--line);
  border-radius:6px;
  padding:8px 10px;
  font-size:12px;
  line-height:1.45;
}
.k .ox-ts-num{
  display:inline-block;
  padding:1px 6px;
  border-radius:4px;
  background:var(--acc-soft);
  color:var(--acc);
  font-weight:700;
  font-size:11px;
  margin-bottom:4px;
}
.k .ox-tree-step ul{
  margin:4px 0 0;
  padding-left:16px;
}
.k .ox-tree-step li{
  margin:2px 0;
}

/* Tab 3: 自测练习 */
.k .ox-quiz-card{
  padding:12px;
}
.k .ox-qh{
  display:flex;
  justify-content:space-between;
  align-items:center;
  margin-bottom:10px;
  border-bottom:1px solid var(--line);
  padding-bottom:6px;
}
.k .ox-qh h4{
  margin:0;
  font-size:14px;
}
.k .ox-qscore{
  font-size:12.5px;
  color:var(--mut);
}
.k .ox-qscore b{
  color:var(--ok);
}
.k .ox-qitems{
  display:flex;
  flex-direction:column;
  gap:10px;
}
.k .ox-qitem{
  background:var(--card);
  border:1px solid var(--line);
  border-radius:8px;
  padding:10px 12px;
}
.k .ox-qp{
  font-size:13px;
  margin-bottom:8px;
  line-height:1.4;
}
.k .ox-opts{
  display:flex;
  flex-direction:column;
  gap:5px;
}
.k .ox-opt-btn{
  text-align:left;
  font-size:12.5px;
  padding:5px 10px;
  border-radius:6px;
  border:1px solid var(--line);
  background:var(--card);
  color:var(--fg);
  line-height:1.35;
  cursor:pointer;
  transition:all 0.15s ease;
}
.k .ox-opt-btn:hover{
  border-color:var(--acc);
}
.k .ox-opt-btn.ok{
  background:var(--ok-soft);
  border-color:var(--ok);
  color:var(--fg);
  font-weight:600;
}
.k .ox-opt-btn.bad{
  background:var(--bad-soft);
  border-color:var(--bad);
  color:var(--fg);
}
.k .ox-fb{
  margin-top:6px;
  padding:6px 10px;
  border-radius:6px;
  font-size:12px;
  line-height:1.4;
}
.k .ox-fb.ok{
  background:var(--ok-soft);
  color:var(--ok);
  border:1px solid var(--ok);
}
.k .ox-fb.bad{
  background:var(--bad-soft);
  color:var(--bad);
  border:1px solid var(--bad);
}
.k .ox-qtools{
  margin-top:12px;
  gap:8px;
}
.k .ox-qtools button{
  font-size:12.5px;
  padding:4px 12px;
}
</style>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

var $ = function (sel) { return $root.querySelector(sel); };
var $$ = function (sel) { return Array.prototype.slice.call($root.querySelectorAll(sel)); };

/* Alcohol dataset */
var ALCOHOLS = {
  ethanol: {
    id: 'ethanol',
    name: 'ethanol',
    formula: 'CH₃CH₂OH',
    degree: '1°',
    distilProd: { name: 'ethanal', formula: 'CH₃CHO', type: 'aldehyde', eq: 'CH₃CH₂OH + [O] → CH₃CHO + H₂O' },
    refluxProd: { name: 'ethanoic acid', formula: 'CH₃COOH', type: 'carboxylic acid', eq: 'CH₃CH₂OH + 2[O] → CH₃COOH + H₂O' },
    bpTable: 'Ethanal (bp 20°C，无分子间氢键) 远低于 Ethanol (bp 78°C) 和 Ethanoic acid (bp 118°C)。'
  },
  propan1ol: {
    id: 'propan1ol',
    name: 'propan-1-ol',
    formula: 'CH₃CH₂CH₂OH',
    degree: '1°',
    distilProd: { name: 'propanal', formula: 'CH₃CH₂CHO', type: 'aldehyde', eq: 'CH₃CH₂CH₂OH + [O] → CH₃CH₂CHO + H₂O' },
    refluxProd: { name: 'propanoic acid', formula: 'CH₃CH₂COOH', type: 'carboxylic acid', eq: 'CH₃CH₂CH₂OH + 2[O] → CH₃CH₂COOH + H₂O' },
    bpTable: 'Propanal (bp 49°C) 远低于 Propan-1-ol (bp 97°C) 和 Propanoic acid (bp 141°C)。'
  },
  mpropan1ol: {
    id: 'mpropan1ol',
    name: '2-methylpropan-1-ol',
    formula: '(CH₃)₂CHCH₂OH',
    degree: '1°',
    distilProd: { name: '2-methylpropanal', formula: '(CH₃)₂CHCHO', type: 'aldehyde', eq: '(CH₃)₂CHCH₂OH + [O] → (CH₃)₂CHCHO + H₂O' },
    refluxProd: { name: '2-methylpropanoic acid', formula: '(CH₃)₂CHCOOH', type: 'carboxylic acid', eq: '(CH₃)₂CHCH₂OH + 2[O] → (CH₃)₂CHCOOH + H₂O' },
    bpTable: '2-methylpropanal (bp 64°C) 低于原醇 (bp 108°C) 和羧酸 (bp 155°C)。'
  },
  propan2ol: {
    id: 'propan2ol',
    name: 'propan-2-ol',
    formula: 'CH₃CH(OH)CH₃',
    degree: '2°',
    distilProd: { name: 'propanone', formula: 'CH₃COCH₃', type: 'ketone', eq: 'CH₃CH(OH)CH₃ + [O] → CH₃COCH₃ + H₂O' },
    refluxProd: { name: 'propanone', formula: 'CH₃COCH₃', type: 'ketone', eq: 'CH₃CH(OH)CH₃ + [O] → CH₃COCH₃ + H₂O' },
    bpTable: 'Propanone 沸点 56°C。作为 ketone，没有 α-H，在温和酸化重铬酸钾下无法被进一步氧化。'
  },
  butan2ol: {
    id: 'butan2ol',
    name: 'butan-2-ol',
    formula: 'CH₃CH₂CH(OH)CH₃',
    degree: '2°',
    distilProd: { name: 'butanone', formula: 'CH₃CH₂COCH₃', type: 'ketone', eq: 'CH₃CH₂CH(OH)CH₃ + [O] → CH₃CH₂COCH₃ + H₂O' },
    refluxProd: { name: 'butanone', formula: 'CH₃CH₂COCH₃', type: 'ketone', eq: 'CH₃CH₂CH(OH)CH₃ + [O] → CH₃CH₂COCH₃ + H₂O' },
    bpTable: 'Butanone 沸点 80°C。无论是蒸馏还是长时间回流，产物始终为丁酮 (butanone)。'
  },
  tbutanol: {
    id: 'tbutanol',
    name: '2-methylpropan-2-ol',
    formula: '(CH₃)₃COH',
    degree: '3°',
    distilProd: { name: 'No reaction', formula: '不反应', type: 'none', eq: '(CH₃)₃COH + [O] → No reaction (remains orange)' },
    refluxProd: { name: 'No reaction', formula: '不反应', type: 'none', eq: '(CH₃)₃COH + [O] → No reaction (remains orange)' },
    bpTable: '连有 –OH 的叔碳上没有氢原子（no hydrogen on C–OH carbon），在酸性重铬酸钾下不发生氧化反应。'
  }
};

/* State variables */
var curSub = 'ethanol';
var curCond = 'reflux'; // 'distil' or 'reflux'
var curOx = 'dichromate'; // 'dichromate' or 'manganate'
var curProg = 0; // 0 to 100
var curTab = 'explorer';

var isHeating = false;
var heatRaf = null;
var lastHeatTs = null;

/* Interpolate colors for oxidant */
function getOxidantColor(ox, degree, prog) {
  var p = Math.max(0, Math.min(100, prog)) / 100;
  if (degree === '3°') {
    return ox === 'dichromate' ? '#f97316' : '#9333ea';
  }
  if (ox === 'dichromate') {
    // Orange (#ea580c -> rgb(234, 88, 12)) to Emerald green (#15803d -> rgb(21, 128, 61))
    var r = Math.round(234 + (21 - 234) * p);
    var g = Math.round(88 + (128 - 88) * p);
    var b = Math.round(12 + (61 - 12) * p);
    return 'rgb(' + r + ',' + g + ',' + b + ')';
  } else {
    // Purple (#9333ea -> rgb(147, 51, 234)) to very faint pale pink/colourless (#f4f6fa)
    var r2 = Math.round(147 + (244 - 147) * p);
    var g2 = Math.round(51 + (246 - 51) * p);
    var b2 = Math.round(234 + (250 - 234) * p);
    return 'rgb(' + r2 + ',' + g2 + ',' + b2 + ')';
  }
}

/* Render Apparatus SVG in Tab 1 */
function renderApparatusSvg() {
  var svg = $('.ox-app-svg');
  if (!svg) return;
  var alc = ALCOHOLS[curSub];
  var col = getOxidantColor(curOx, alc.degree, curProg);
  var h = [];

  h.push('<defs>');
  h.push('  <linearGradient id="flask-grad" x1="0%" y1="0%" x2="0%" y2="100%">');
  h.push('    <stop offset="0%" stop-color="' + col + '" stop-opacity="0.85" />');
  h.push('    <stop offset="100%" stop-color="' + col + '" stop-opacity="0.98" />');
  h.push('  </linearGradient>');
  h.push('  <linearGradient id="jacket-grad" x1="0%" y1="0%" x2="100%" y2="100%">');
  h.push('    <stop offset="0%" stop-color="var(--acc)" stop-opacity="0.10" />');
  h.push('    <stop offset="100%" stop-color="var(--acc)" stop-opacity="0.22" />');
  h.push('  </linearGradient>');
  h.push('  <marker id="ox-arr" markerWidth="6" markerHeight="6" refX="4" refY="3" orient="auto">');
  h.push('    <path d="M 0 0 L 6 3 L 0 6 Z" fill="var(--acc)" />');
  h.push('  </marker>');
  h.push('</defs>');

  if (curCond === 'distil') {
    // ==========================================
    // Distillation Apparatus Drawing (蒸馏装置)
    // 烧瓶置于 cx=135，温度计标注在左侧有充足 130px 留白，接收瓶在 cx=400
    // ==========================================
    h.push('<g>');

    // 1. Heating mantle under flask (cx=135, cy=245, r=42)
    h.push('  <path d="M 98 252 A 44 44 0 0 0 172 252 L 176 275 A 6 6 0 0 1 170 282 L 100 282 A 6 6 0 0 1 94 275 Z" fill="var(--line)" opacity="0.88" />');
    h.push('  <rect x="104" y="282" width="62" height="6" rx="2" fill="var(--mut)" opacity="0.6" />');
    h.push('  <circle cx="158" cy="272" r="2.5" fill="var(--bad)" opacity="0.8" />');
    h.push('  <text x="135" y="302" text-anchor="middle" font-size="13" font-weight="600" fill="var(--mut)">加热套 (Electric Mantle)</text>');

    // 2. Round-bottom Flask body & neck
    // Neck x=124..146, y=105..185. Flask bulb cx=135, cy=235, r=42
    h.push('  <path d="M 124 105 L 124 185 A 42 42 0 1 0 146 185 L 146 105 Z" fill="var(--card)" stroke="var(--fg)" stroke-width="2" />');

    // 3. Liquid in distillation flask
    h.push('  <path d="M 96 235 A 39 39 0 0 0 174 235 Z" fill="url(#flask-grad)" />');

    // Anti-bumping granules (沸石)
    h.push('  <polygon points="120,268 125,263 129,268" fill="var(--fg)" opacity="0.75" />');
    h.push('  <polygon points="140,270 145,265 149,271" fill="var(--fg)" opacity="0.75" />');

    // Boiling bubbles if heating
    if (curProg > 0 && alc.degree !== '3°') {
      for (var bb = 0; bb < 6; bb++) {
        var bx = 112 + (bb * 10) % 45;
        var by = 265 - (bb * 7 + curProg * 3) % 36;
        h.push('  <circle cx="' + bx + '" cy="' + by + '" r="2.2" fill="none" stroke="#fff" stroke-width="1.2" />');
      }
    }

    // 4. Stopper at flask neck
    h.push('  <rect x="121" y="97" width="28" height="12" rx="2" fill="var(--line)" stroke="var(--fg)" stroke-width="1.5" />');

    // 5. Thermometer (水银球恰好位于侧口水平位置: y=145)
    h.push('  <line x1="135" y1="48" x2="135" y2="145" stroke="var(--fg)" stroke-width="2.5" />');
    h.push('  <line x1="135" y1="95" x2="135" y2="145" stroke="var(--bad)" stroke-width="1.8" />');
    h.push('  <circle cx="135" cy="145" r="3.5" fill="var(--bad)" stroke="var(--fg)" stroke-width="1" />');

    // Thermometer annotation (居中置于左上方空旷处，直角引线指向水银球，绝不越界)
    h.push('  <text x="75" y="70" text-anchor="middle" font-size="13" font-weight="600" fill="var(--fg)">温度计水银球</text>');
    h.push('  <text x="75" y="86" text-anchor="middle" font-size="12" fill="var(--mut)">与侧支管口齐平</text>');
    h.push('  <path d="M 75 92 L 75 145 L 128 145" fill="none" stroke="var(--mut)" stroke-width="1.2" stroke-dasharray="3 2" />');

    // 6. Angled Liebig Condenser (从 146, 145 向右下斜伸至 355, 230)
    // Inner tube
    h.push('  <line x1="146" y1="145" x2="355" y2="230" stroke="var(--fg)" stroke-width="4.5" stroke-linecap="round" />');
    // Outer water jacket (斜管外层水套)
    h.push('  <line x1="185" y1="161" x2="325" y2="218" stroke="url(#jacket-grad)" stroke-width="22" stroke-linecap="round" />');
    h.push('  <line x1="185" y1="161" x2="325" y2="218" stroke="var(--line)" stroke-width="22" fill="none" stroke-linecap="round" opacity="0.45" />');

    // Water out (顶部接口: 左上方，清晰不挤压)
    h.push('  <path d="M 205 155 L 195 130" stroke="var(--acc)" stroke-width="2.2" marker-end="url(#ox-arr)" />');
    h.push('  <text x="215" y="122" text-anchor="middle" font-size="13" font-weight="700" fill="var(--acc)">← water out (冷水出)</text>');

    // Water in (底部接口: 右下方，清晰向下指引)
    h.push('  <path d="M 315 225 L 325 248" stroke="var(--acc)" stroke-width="2.2" marker-end="url(#ox-arr)" />');
    h.push('  <text x="330" y="264" font-size="13" font-weight="700" fill="var(--acc)">water in (冷水进) →</text>');

    // 7. Receiver Adaptor & Receiving Flask (接收锥形瓶)
    // Curved adaptor downwards from 355, 230 to 385, 255
    h.push('  <path d="M 355 230 Q 380 240 385 258" fill="none" stroke="var(--fg)" stroke-width="3" />');
    // Conical flask (cx=390, cy=285)
    h.push('  <path d="M 378 258 L 365 300 A 5 5 0 0 0 370 306 L 410 306 A 5 5 0 0 0 415 300 L 402 258 Z" fill="var(--card)" stroke="var(--fg)" stroke-width="2" />');

    // Distillate collected in receiver
    if (alc.degree === '1°' && curProg > 20) {
      var dOpacity = 0.25 + (curProg / 100) * 0.55;
      h.push('  <path d="M 369 290 L 366 302 A 4 4 0 0 0 370 305 L 410 305 A 4 4 0 0 0 414 302 L 411 290 Z" fill="var(--acc)" opacity="' + dOpacity + '" />');
      // Falling condensate drop
      h.push('  <ellipse cx="387" cy="272" rx="1.8" ry="2.5" fill="var(--acc)" />');
    }
    h.push('  <text x="390" y="325" text-anchor="middle" font-size="13" font-weight="600" fill="var(--fg)">接收瓶 (Aldehyde 馏分)</text>');

    h.push('</g>');

  } else {
    // ==========================================
    // Reflux Apparatus Drawing (回流装置)
    // 中心对齐 cx=240，对称开朗
    // ==========================================
    h.push('<g>');

    // 1. Electric Heating Mantle (cx=240, cy=250, r=45)
    h.push('  <path d="M 198 258 A 47 47 0 0 0 282 258 L 287 282 A 6 6 0 0 1 281 289 L 199 289 A 6 6 0 0 1 193 282 Z" fill="var(--line)" opacity="0.88" />');
    h.push('  <rect x="205" y="289" width="70" height="6" rx="2" fill="var(--mut)" opacity="0.6" />');
    h.push('  <circle cx="266" cy="278" r="2.5" fill="var(--bad)" opacity="0.8" />');
    h.push('  <text x="240" y="310" text-anchor="middle" font-size="13" font-weight="600" fill="var(--mut)">电热套加热 (Heated under reflux)</text>');

    // 2. Round-bottom Flask body (cx=240, cy=240, r=44)
    h.push('  <path d="M 228 175 A 44 44 0 1 0 252 175 Z" fill="var(--card)" stroke="var(--fg)" stroke-width="2" />');

    // 3. Solution in flask
    h.push('  <path d="M 200 245 A 41 41 0 0 0 280 245 Z" fill="url(#flask-grad)" />');

    // Anti-bumping granules
    h.push('  <polygon points="222,274 227,269 231,274" fill="var(--fg)" opacity="0.75" />');
    h.push('  <polygon points="248,276 253,270 258,276" fill="var(--fg)" opacity="0.75" />');

    // Bubbles in boiling flask
    if (curProg > 0 && alc.degree !== '3°') {
      for (var rb = 0; rb < 7; rb++) {
        var rbx = 214 + (rb * 9) % 52;
        var rby = 275 - (rb * 6 + curProg * 4) % 36;
        h.push('  <circle cx="' + rbx + '" cy="' + rby + '" r="2.2" fill="none" stroke="#fff" stroke-width="1.2" />');
      }
    }

    // 4. Vertical Liebig Condenser (竖直冷凝管: 中心 x=240)
    // Inner tube: width 18 (x=231..249), from y=48 down to y=175
    h.push('  <rect x="231" y="48" width="18" height="127" fill="var(--card)" stroke="var(--fg)" stroke-width="2" />');
    // Ground glass joint indicator
    h.push('  <rect x="228" y="166" width="24" height="10" fill="var(--line)" opacity="0.6" stroke="var(--fg)" stroke-width="1" />');
    // Top rim of inner tube (顶部开口管口)
    h.push('  <ellipse cx="240" cy="48" rx="10" ry="3.5" fill="var(--line)" stroke="var(--fg)" stroke-width="1.5" />');

    // Top open annotation (右侧清晰引线，绝不重叠！)
    h.push('  <path d="M 252 48 L 295 48" stroke="var(--mut)" stroke-width="1.2" stroke-dasharray="3 2" />');
    h.push('  <text x="300" y="52" font-size="13" font-weight="600" fill="var(--mut)">顶部开口（防密闭憋压）</text>');

    // Outer water jacket: width 44 (x=218..262), from y=70 to y=155
    h.push('  <rect x="218" y="70" width="44" height="85" rx="5" fill="url(#jacket-grad)" stroke="var(--line)" stroke-width="1.8" />');

    // 5. Water in & Water out (下进上出)
    // Water in (底部接口: 右侧 y=140)
    h.push('  <path d="M 262 140 L 285 140" stroke="var(--acc)" stroke-width="2.2" marker-end="url(#ox-arr)" />');
    h.push('  <text x="292" y="144" font-size="13" font-weight="700" fill="var(--acc)">water in (冷水进) →</text>');

    // Water out (顶部接口: 左侧 y=80)
    h.push('  <path d="M 218 80 L 195 80" stroke="var(--acc)" stroke-width="2.2" marker-end="url(#ox-arr)" />');
    h.push('  <text x="188" y="84" text-anchor="end" font-size="13" font-weight="700" fill="var(--acc)">← water out (冷水出)</text>');

    // 6. Reflux condensate droplets inside tube (位置在 y=115，与 y=80 的 water out 保持安全间距)
    if (curProg > 10) {
      h.push('  <ellipse cx="240" cy="88" rx="2.5" ry="3.5" fill="var(--acc)" opacity="0.8" />');
      h.push('  <ellipse cx="237" cy="118" rx="2.2" ry="3" fill="var(--acc)" opacity="0.75" />');
      h.push('  <ellipse cx="243" cy="142" rx="2" ry="3" fill="var(--acc)" opacity="0.8" />');

      // Reflux ring annotation
      h.push('  <path d="M 231 118 L 180 118" stroke="var(--mut)" stroke-width="1.2" stroke-dasharray="3 2" />');
      h.push('  <text x="175" y="115" text-anchor="end" font-size="13" font-weight="600" fill="var(--fg)">蒸气冷凝回流</text>');
      h.push('  <text x="175" y="130" text-anchor="end" font-size="12" fill="var(--mut)">Reflux ring</text>');
    }

    h.push('</g>');
  }

  svg.innerHTML = h.join('\n');
}

/* Render Parallel 3-Tube SVG in Tab 2 */
function renderParallelSvg(isWarmed) {
  var svg = $('.ox-parallel-svg');
  if (!svg) return;
  var h = [];

  var pTubes = [
    { id: '1deg', name: '1°: ethanol', x: 125, col: isWarmed ? '#15803d' : '#ea580c', res: isWarmed ? 'orange → green' : 'orange' },
    { id: '2deg', name: '2°: propan-2-ol', x: 330, col: isWarmed ? '#15803d' : '#ea580c', res: isWarmed ? 'orange → green' : 'orange' },
    { id: '3deg', name: '3°: 2-methylpropan-2-ol', x: 535, col: '#ea580c', res: isWarmed ? 'remains orange' : 'orange' }
  ];

  h.push('<defs>');
  pTubes.forEach(function (tb, idx) {
    h.push('  <linearGradient id="ptube-grad-' + idx + '" x1="0%" y1="0%" x2="0%" y2="100%">');
    h.push('    <stop offset="0%" stop-color="' + tb.col + '" stop-opacity="0.82" />');
    h.push('    <stop offset="100%" stop-color="' + tb.col + '" stop-opacity="0.96" />');
    h.push('  </linearGradient>');
  });
  h.push('</defs>');

  // Water bath tub background (y=120..225, 高105)
  h.push('<rect x="35" y="120" width="590" height="105" rx="8" fill="var(--card)" stroke="var(--line)" stroke-width="2" />');
  h.push('<rect x="37" y="135" width="586" height="88" fill="var(--acc)" opacity="0.08" />');

  // 水浴文字放在试管 1 与试管 2 之间的空旷区 (x=228)，分两行紧凑居中，绝对不与试管重叠！
  h.push('<text x="228" y="140" text-anchor="middle" font-size="13" fill="var(--acc)" font-weight="700">恒温热水浴</text>');
  h.push('<text x="228" y="156" text-anchor="middle" font-size="12" fill="var(--acc)" opacity="0.85">Water bath (60°C)</text>');
  h.push('<path d="M 205 163 Q 228 158 251 163" stroke="var(--acc)" fill="none" opacity="0.35" stroke-width="1.5" />');

  pTubes.forEach(function (tb, idx) {
    var tx = tb.x;
    h.push('<g>');
    // Test tube glass (底部到 y=205)
    h.push('  <path d="M ' + (tx - 26) + ' 35 L ' + (tx - 26) + ' 185 Q ' + (tx - 26) + ' 210 ' + tx + ' 210 Q ' + (tx + 26) + ' 210 ' + (tx + 26) + ' 185 L ' + (tx + 26) + ' 35 Z" fill="var(--card)" stroke="var(--fg)" stroke-width="2" />');
    h.push('  <ellipse cx="' + tx + '" cy="35" rx="27" ry="5" fill="var(--line)" stroke="var(--fg)" stroke-width="1.6" />');

    // Colored solution in tube
    h.push('  <path d="M ' + (tx - 24) + ' 90 L ' + (tx - 24) + ' 185 Q ' + (tx - 24) + ' 208 ' + tx + ' 208 Q ' + (tx + 24) + ' 208 ' + (tx + 24) + ' 185 L ' + (tx + 24) + ' 90 Z" fill="url(#ptube-grad-' + idx + ')" />');
    h.push('  <ellipse cx="' + tx + '" cy="90" rx="24" ry="4" fill="' + tb.col + '" opacity="0.85" />');

    // Tube labels (字号 13.5 确保窄屏下不被判定为过小)
    h.push('  <text x="' + tx + '" y="24" text-anchor="middle" font-size="13.5" font-weight="700" fill="var(--fg)">' + tb.name + '</text>');
    var badgeCol = (isWarmed && tb.id === '3deg') ? 'var(--mut)' : (isWarmed ? 'var(--ok)' : 'var(--fg)');
    h.push('  <text x="' + tx + '" y="248" text-anchor="middle" font-size="13.5" font-weight="700" fill="' + badgeCol + '">' + tb.res + '</text>');
    h.push('</g>');
  });

  svg.innerHTML = h.join('\n');
}

/* Update Tab 1 UI Details */
function updateExplorerUI() {
  var alc = ALCOHOLS[curSub];
  var isReflux = (curCond === 'reflux');
  var prod = isReflux ? alc.refluxProd : alc.distilProd;

  // Title and badges
  $('.ox-cur-title').textContent = alc.name + ' (' + alc.degree + ') + ' + (isReflux ? 'Reflux' : 'Distillation');
  $('.ox-class-badge').textContent = alc.degree + ' alcohol';

  var resText = (alc.degree === '3°') ? 'remains ' + (curOx === 'dichromate' ? 'orange' : 'purple') : (curOx === 'dichromate' ? 'orange → green' : 'purple → colourless');
  $('.ox-res-badge').textContent = resText;
  $('.ox-res-badge').style.color = (alc.degree === '3°') ? 'var(--mut)' : 'var(--ok)';
  $('.ox-res-badge').style.borderColor = (alc.degree === '3°') ? 'var(--line)' : 'var(--ok)';
  $('.ox-res-badge').style.background = (alc.degree === '3°') ? 'var(--card)' : 'var(--ok-soft)';

  // Apparatus description banner
  if (isReflux) {
    $('.ox-app-badge').textContent = '⚗️ Reflux 回流装置';
    $('.ox-app-desc').textContent = '竖直冷凝管下进上出，蒸气液化回流反应瓶，过量氧化剂彻底氧化为 carboxylic acid';
  } else {
    $('.ox-app-badge').textContent = '🧪 Distillation 蒸馏装置';
    $('.ox-app-desc').textContent = '斜置冷凝管接接收瓶，低沸点 aldehyde（无分子间氢键）一生成即蒸出冷凝收集';
  }

  // Product cards
  $('.ox-reac-name').textContent = alc.name;
  $('.ox-reac-form').textContent = alc.formula;
  $('.ox-prod-name').textContent = prod.name;
  $('.ox-prod-form').textContent = prod.formula;
  $('.ox-eq-text').textContent = prod.eq;

  // Analysis Points
  if (alc.degree === '1°') {
    if (isReflux) {
      $('.ox-why-cond').textContent = 'Reflux 竖直冷凝管将蒸发的 aldehyde 持续回流至烧瓶中，在过量氧化剂和高温下被完全氧化为 carboxylic acid。';
    } else {
      $('.ox-why-cond').textContent = 'Distillation 蒸馏装置使沸点较低的 aldehyde 一生成便挥发逸出冷凝收集，防止与氧化剂进一步接触而被氧化为酸。';
    }
  } else if (alc.degree === '2°') {
    $('.ox-why-cond').textContent = 'Secondary alcohol 氧化生成 ketone（酮）。酮的羰基碳上没有氢原子，在酸性重铬酸钾下无法继续被氧化，因此蒸馏与回流产物相同。';
  } else {
    $('.ox-why-cond').textContent = 'Tertiary alcohol 连有 –OH 的碳原子上没有与之相连的氢原子（no hydrogen on C–OH carbon），无法脱氢发生氧化，溶液保持橙色。';
  }

  $('.ox-bp-desc').textContent = alc.bpTable;
  $('.ox-obs-desc').innerHTML = (alc.degree === '3°') ? '考试必须写 <b>remains orange</b>（或 no reaction / no visible change）。' : '考试必须完整写 <b>orange to green</b>（或 orange solution turns green）。单写 green 不给分。';

  $('.ox-progress-slider').value = curProg;
  $('.ox-prog-val').textContent = Math.round(curProg) + '%';

  renderApparatusSvg();
  K.fit();
}

/* Heat Animation Loop */
function setHeatProgress(prog) {
  curProg = Math.max(0, Math.min(100, prog));
  $('.ox-progress-slider').value = curProg;
  $('.ox-prog-val').textContent = Math.round(curProg) + '%';
  renderApparatusSvg();
}

function heatTick(ts) {
  if (!isHeating) return;
  if (lastHeatTs == null) lastHeatTs = ts;
  var dt = (ts - lastHeatTs) / 1000;
  lastHeatTs = ts;
  var nextP = curProg + dt * 35; // 100% in ~3 seconds
  if (nextP >= 100) {
    nextP = 100;
    setHeatProgress(nextP);
    isHeating = false;
    $('.ox-btn-heat').classList.remove('on');
    return;
  }
  setHeatProgress(nextP);
  heatRaf = requestAnimationFrame(heatTick);
}

function startHeat() {
  if (curProg >= 100) curProg = 0;
  isHeating = true;
  lastHeatTs = null;
  $('.ox-btn-heat').classList.add('on');
  heatRaf = requestAnimationFrame(heatTick);
}

function stopHeat() {
  isHeating = false;
  $('.ox-btn-heat').classList.remove('on');
  if (heatRaf) cancelAnimationFrame(heatRaf);
}

function resetHeat() {
  stopHeat();
  setHeatProgress(0);
}

/* Tab 2 Parallel State */
var parallelWarmed = false;
function setParallelWarmed(warmed) {
  parallelWarmed = !!warmed;
  renderParallelSvg(parallelWarmed);
}

/* Tab 3 Quiz Logic */
var QUIZ_ANS = {
  0: {
    correct: 'B',
    okFb: '✓ 正确！Cambridge 9701 官方标答必须完整描述初始与最终颜色：orange to green（或 orange solution turns green）。单写 green 不给分。',
    badFb: '✗ 错误。Cambridge 官方标准要求完整写出颜色转变过程「orange to green」或「orange solution turns green」。'
  },
  1: {
    correct: 'B',
    okFb: '✓ 正确！在过量酸化重铬酸钾和回流（reflux）条件下，初生 ethanal 无法逃逸，被完全氧化为 ethanoic acid (CH₃COOH)。',
    badFb: '✗ 错误。回流加热（reflux）将 primary alcohol 彻底氧化为 carboxylic acid（ethanoic acid）。若要得到 ethanal 必须使用蒸馏（distillation）。'
  },
  2: {
    correct: 'B',
    okFb: '✓ 正确！Tertiary alcohol 连有 –OH 基团的碳原子上没有连有任何氢原子（no hydrogen on C–OH carbon），无法被氧化，因此保持橙色。',
    badFb: '✗ 错误。核心得分点是：连有 –OH 基团的碳原子上没有直接相连的氢原子（no hydrogen on the C–OH carbon），无法脱氢。'
  },
  3: {
    correct: 'A',
    okFb: '✓ 正确！Aldehyde 分子之间没有氢键（仅有 dipole-dipole forces），沸点显著低于含有分子间氢键的 alcohol 和 carboxylic acid，因此在蒸馏时最先挥发逸出冷凝收集，防止被继续氧化。',
    badFb: '✗ 错误。核心原因是沸点差异：醛没有羟基（无分子间氢键），沸点远低于醇和羧酸，一生成立即蒸馏出来，避免被进一步氧化。'
  }
};

var userAnswers = {};

function updateQuizScore() {
  var sc = 0;
  for (var i = 0; i < 4; i++) {
    if (userAnswers[i] === QUIZ_ANS[i].correct) sc++;
  }
  $('.ox-sc').textContent = String(sc);
}

function answerQuestion(qi, opt) {
  userAnswers[qi] = opt;
  var item = $('.ox-qitem[data-qi="' + qi + '"]');
  var fb = item.querySelector('.ox-fb');
  var btns = item.querySelectorAll('.ox-opt-btn');

  btns.forEach(function (b) {
    b.classList.remove('ok', 'bad');
    if (b.dataset.opt === opt) {
      b.classList.add(opt === QUIZ_ANS[qi].correct ? 'ok' : 'bad');
    }
  });

  fb.hidden = false;
  var isOk = (opt === QUIZ_ANS[qi].correct);
  fb.className = 'ox-fb ' + (isOk ? 'ok' : 'bad');
  fb.innerHTML = isOk ? QUIZ_ANS[qi].okFb : QUIZ_ANS[qi].badFb;

  updateQuizScore();
  K.fit();
}

/* Event listeners */
$$('.ox-sub-btn').forEach(function (btn) {
  btn.addEventListener('click', function () {
    $$('.ox-sub-btn').forEach(function (b) { b.classList.remove('on'); });
    btn.classList.add('on');
    curSub = btn.dataset.sub;
    resetHeat();
    updateExplorerUI();
  });
});

$$('.ox-cond-btn').forEach(function (btn) {
  btn.addEventListener('click', function () {
    $$('.ox-cond-btn').forEach(function (b) { b.classList.remove('on'); });
    btn.classList.add('on');
    curCond = btn.dataset.cond;
    resetHeat();
    updateExplorerUI();
  });
});

$$('.ox-ox-btn').forEach(function (btn) {
  btn.addEventListener('click', function () {
    $$('.ox-ox-btn').forEach(function (b) { b.classList.remove('on'); });
    btn.classList.add('on');
    curOx = btn.dataset.ox;
    resetHeat();
    updateExplorerUI();
  });
});

$('.ox-btn-heat').addEventListener('click', function () {
  if (isHeating) stopHeat();
  else startHeat();
});

$('.ox-btn-reset').addEventListener('click', resetHeat);

$('.ox-progress-slider').addEventListener('input', function (e) {
  stopHeat();
  setHeatProgress(parseFloat(e.target.value));
});

// Parallel mode buttons
$('.ox-btn-warm').addEventListener('click', function () {
  setParallelWarmed(true);
});

$('.ox-btn-p-reset').addEventListener('click', function () {
  setParallelWarmed(false);
});

// Quiz buttons
$$('.ox-opt-btn').forEach(function (btn) {
  btn.addEventListener('click', function () {
    var qi = parseInt(btn.closest('.ox-qitem').dataset.qi, 10);
    answerQuestion(qi, btn.dataset.opt);
  });
});

$('.ox-qshow').addEventListener('click', function () {
  for (var i = 0; i < 4; i++) {
    answerQuestion(i, QUIZ_ANS[i].correct);
  }
});

$('.ox-qreset').addEventListener('click', function () {
  userAnswers = {};
  $$('.ox-opt-btn').forEach(function (b) { b.classList.remove('ok', 'bad'); });
  $$('.ox-fb').forEach(function (fb) { fb.hidden = true; fb.innerHTML = ''; });
  updateQuizScore();
  K.fit();
});

function onTabChange(tabId) {
  curTab = tabId;
  if (curTab === 'parallel') {
    renderParallelSvg(parallelWarmed);
  } else if (curTab === 'explorer') {
    renderApparatusSvg();
  }
  K.fit();
}

$root.addEventListener('tab', function (e) {
  onTabChange(e.detail);
});

var oxWrap = $('.ox-wrap');
if (oxWrap) {
  oxWrap.addEventListener('tab', function (e) {
    onTabChange(e.detail);
  });
}

// 代理 tabbar 按钮直接点击，确保无论 kit.js 是否冒泡都能精确切换渲染
$$('.tabbar button[data-tab]').forEach(function (btn) {
  btn.addEventListener('click', function () {
    onTabChange(btn.getAttribute('data-tab'));
  });
});

/* Test Hook */
var hook = {
  setTab: function (tabId) {
    var btn = $('.tabbar button[data-tab="' + tabId + '"]');
    if (btn) btn.click();
  },
  getTab: function () { return curTab; },
  setAlcohol: function (id) {
    var btn = $('.ox-sub-btn[data-sub="' + id + '"]');
    if (btn) btn.click();
  },
  getAlcohol: function () { return curSub; },
  setCondition: function (cond) {
    var btn = $('.ox-cond-btn[data-cond="' + cond + '"]');
    if (btn) btn.click();
  },
  getCondition: function () { return curCond; },
  setOxidant: function (ox) {
    var btn = $('.ox-ox-btn[data-ox="' + ox + '"]');
    if (btn) btn.click();
  },
  getOxidant: function () { return curOx; },
  setProgress: function (p) { setHeatProgress(p); },
  getProgress: function () { return curProg; },
  heat: startHeat,
  stopHeat: stopHeat,
  resetHeat: resetHeat,
  isHeating: function () { return isHeating; },
  setParallelWarmed: function (w) { setParallelWarmed(w); },
  isParallelWarmed: function () { return parallelWarmed; },
  getColor: function () { return getOxidantColor(curOx, ALCOHOLS[curSub].degree, curProg); },
  answerQuiz: function (qi, opt) { answerQuestion(qi, opt); },
  getScore: function () {
    var sc = 0;
    for (var i = 0; i < 4; i++) {
      if (userAnswers[i] === QUIZ_ANS[i].correct) sc++;
    }
    return sc;
  },
  showAnswers: function () { $('.ox-qshow').click(); },
  resetQuiz: function () { $('.ox-qreset').click(); }
};

window.T_oxidationTubes = hook;
$root.oxTubesHook = hook;

// Initialize
K.init($root);
updateExplorerUI();
renderParallelSvg(false);

K.init($root);
})(document.getElementById("k-ch16-oxidation-tubes"));</script>"""))
```

### G. (e) dehydration → alkene

alcohol 蒸气通过 **heated Al₂O₃**（或与 **concentrated H₂SO₄ / H₃PO₄** 加热）→ alkene + H₂O。这是 **elimination**（消去 H₂O）。

- C₂H₅OH → CH₂=CH₂ + H₂O
- Al₂O₃ 的作用：**catalyst** 或 **dehydrating agent**（9701/22 May/June 2026 Q4(a)(ii)；写 redox 判错）。
- 2-methylpropan-2-ol → 2-methylpropene；不对称的 alcohol（butan-2-ol）可得多种 alkene（but-1-ene、cis / trans but-2-ene）。

### H. (f) esterification

alcohol + carboxylic acid，**concentrated H₂SO₄ as catalyst**，加热 → ester + H₂O（condensation）。ester 名字的前一半来自 alcohol（ethyl …），后一半来自 acid（… ethanoate）。详见 [18.2 A](https://colab.research.google.com/drive/1gxHJxFjn1RQSXR5y5n5CSeW5q82c6-6E#scrollTo=ch18-rx-esterification)。

```python
# @title 🔬 alcohol 反应中心：选 alcohol × reagent { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-alcohol-hub">
<h3>alcohol 反应中心：选 alcohol × reagent</h3><p class="faint">待合并：<code>colab/src/ch16/alcohol-hub.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-alcohol-hub"));</script>"""))
```

### I. tri-iodomethane（iodoform）test

> **LO 16.1.4** — deduce the presence of a CH₃CH(OH)– group in an alcohol, CH₃CH(OH)–R, from its reaction with alkaline I₂(aq) to form a yellow precipitate of tri-iodomethane and an ion, RCO₂<sup>−</sup>

- reagent：**alkaline I₂(aq)**（I₂ + NaOH(aq)，warm）；
- positive：**pale yellow precipitate of tri-iodomethane, CHI₃**（有消毒水的气味）；
- 结构条件：alcohol 里有 **CH₃CH(OH)–** 原子团（OH 在第二个 C，旁边是 CH₃）——它先被 oxidise 成 CH₃CO–，再发生 iodoform 反应；
- product：CHI₃ **和 RCO₂<sup>−</sup>**（R 是 CH(OH) 另一边的基团）：CH₃CH(OH)R → CHI₃ + RCO₂<sup>−</sup>。

| alcohol | 有 CH₃CH(OH)–？ | 结果 |
|---|---|---|
| ethanol CH₃CH₂OH（R = H） | ✔（唯一的 1° alcohol） | yellow ppt + HCO₂<sup>−</sup>（methanoate） |
| propan-2-ol | ✔ | yellow ppt + CH₃CO₂<sup>−</sup> |
| butan-2-ol | ✔ | yellow ppt + CH₃CH₂CO₂<sup>−</sup>（propanoate；9701 卷：写成 sodium propanoate 也可以） |
| propan-1-ol、2-methylpropan-2-ol | ✘ | no visible change |

同一个 test 也用来检出 **CH₃CO–**（methyl ketone、ethanal），见 [17.1 D](https://colab.research.google.com/drive/10P0lQfoNh6lRy7HRbjJZxH7AnzI4V3vL#scrollTo=ch17-rx-tests)——数 product 的碳：CHI₃ 拿走一个 C，RCO₂<sup>−</sup> 比原 molecule **少一个 C**。

```python
# @title 🔬 iodoform test：找 CH₃CH(OH)– / CH₃CO– { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-iodoform">
<h3>iodoform test：找 CH₃CH(OH)– / CH₃CO–</h3><p class="faint">待合并：<code>colab/src/ch16/iodoform.html</code> 还没写好</p>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-iodoform"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2024 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-15">
<div class="k-ptable"></div>
<h3>9701/22 May/June 2024</h3>
<p>Y is a structural isomer of W. Both W and Y produce colourless bubbles when sodium is added to them. Y does not react when heated with acidified K₂Cr₂O₇. Y does not react when warmed with alkaline I₂(aq).</p>
<div class="q"><h4>Q5(c)(i) <span class="mark">[1]</span></h4>
<p>Name the functional group present in Y.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — hydroxyl</p>
</div>
</div>
<div class="q"><h4>Q5(c)(ii) <span class="mark">[1]</span></h4>
<p>Complete the equation to describe the reaction of W or Y with sodium.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — (1)C₄H₁₀O + (1)Na → C₄H₉ONa + ½H₂</p>
</div>
</div>
<div class="q"><h4>Q5(c)(iii) <span class="mark">[1]</span></h4>
<p>Draw the structure of Y.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — Y = (CH₃)₃COH（2-methylpropan-2-ol，tert-butanol）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-15"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-16">
<div class="k-ptable"></div>
<h3>9701/22 Feb/Mar 2025</h3>
<div class="q"><h4>Q4(a) <span class="mark">[5]</span></h4>
<p>Table 4.1 gives details of tests on J to M. In each test, only two compounds give a positive result. Complete Table 4.1.</p>
<table><tr><th>reagent</th><th>observation of positive result</th><th>compounds that give a positive result</th></tr><tr><td>acidified K₂Cr₂O₇(aq)</td><td><i>(空)</i></td><td>J and L</td></tr><tr><td>alkaline I₂(aq)</td><td>yellow precipitate</td><td><i>(空)</i></td></tr><tr><td><i>(空)</i></td><td>orange precipitate</td><td>J and K</td></tr><tr><td>Na(s)</td><td><i>(空)</i></td><td><i>(空)</i></td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — 完整表格：</p>
<table><tr><th>reagent</th><th>observation of positive result</th><th>compounds that give a positive result</th></tr><tr><td>acidified K₂Cr₂O₇(aq)</td><td>(solution) turns (from orange to) green</td><td>J and L</td></tr><tr><td>alkaline I₂(aq)</td><td>yellow precipitate (forms)</td><td>K and L</td></tr><tr><td>2,4-dinitrophenylhydrazine / 2,4-DNPH</td><td>orange precipitate (forms)</td><td>J and K</td></tr><tr><td>Na(s)</td><td>effervescence</td><td>L and M</td></tr></table>
<p><b>判分</b> — 5 个空各 1 分：K₂Cr₂O₇ 一行的 observation、I₂ 一行的 compounds、reagent 一行填 2,4-DNPH、Na(s) 一行的 observation 与 compounds</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-16"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-17">
<div class="k-ptable"></div>
<h3>9701/22 May/June 2025</h3>
<div class="q"><h4>Q4(c)(ii) <span class="mark">[1]</span></h4>
<p>Identify the gas produced when Na(s) is added to separate samples of each alcohol.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — hydrogen / H₂</p>
</div>
</div>
<div class="q"><h4>Q4(c)(iii) <span class="mark">[1]</span></h4>
<p>Describe the role of Na(s) when it reacts with alcohols.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — (It / Na) reduces A / B / C / the alcohol (group) OR (it / Na behaves as) reducing agent OR (it / Na) donates electrons to A / B / C / the alcohol (group)</p>
</div>
</div>
<div class="q"><h4>Q4(d)(i) <span class="mark">[1]</span></h4>
<p>Table 4.1 shows the results of two tests on separate samples of A, B and C. Deduce the structure of A.</p>
<p><b>Table 4.1</b></p>
<table><tr><th>compound</th><th>heat under reflux with acidified K₂Cr₂O₇</th><th>warm with alkaline I₂(aq)</th></tr><tr><td>A</td><td>remains orange</td><td>no visible change</td></tr><tr><td>B</td><td>orange to green</td><td>no visible change</td></tr><tr><td>C</td><td>orange to green</td><td>pale yellow precipitate</td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — A = (CH₃)₃COH（2-methylpropan-2-ol，tertiary alcohol，不被 K₂Cr₂O₇ 氧化）</p>
</div>
</div>
<div class="q"><h4>Q4(d)(ii) <span class="mark">[1]</span></h4>
<p>Deduce a possible name for B.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — butan-1-ol OR (2-)methylpropan-1-ol</p>
</div>
</div>
<div class="q"><h4>Q4(d)(iii) <span class="mark">[1]</span></h4>
<p>Deduce the identity of C.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — butan-2-ol OR CH₃CH₂CH(OH)CH₃</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-17"));</script>"""))
```

```python
# @title 📝 9701/23 May/June 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-18">
<div class="k-ptable"></div>
<h3>9701/23 May/June 2025</h3>
<p>Fig. 5.2 shows two reactions of W to produce organic compounds Y and Z. <img alt="Figure 5.2" src=""> W 经 LiAlH₄ 还原 COOH→CH₂OH 得 Y（C=C 不受影响）；W 与 methanol / 浓 H₂SO₄ 发生酯化得 Z = CH₂=C(CH₃)COOCH₃（methyl 2-methylprop-2-enoate，methyl methacrylate）。</p>
<div class="q"><h4>Q5(b)(ii) <span class="mark">[1]</span></h4>
<p>W reacts with LiAlH₄ to produce Y. Draw the structure of Y.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — Y = CH₂=C(CH₃)CH₂OH（2-methylprop-2-en-1-ol）</p>
</div>
</div>
<div class="q"><h4>Q5(b)(iii) <span class="mark">[1]</span></h4>
<p>Identify the role of LiAlH₄ when it reacts with W.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — reducing agent</p>
</div>
</div>
<div class="q"><h4>Q5(b)(iv) <span class="mark">[1]</span></h4>
<p>Suggest the reagent and conditions required when W is converted into Z.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — methanol / CH₃OH <b>AND</b> conc(entrated) (sulfuric) acid（catalyst）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-18"));</script>"""))
```

```python
# @title 📝 9701/22 Oct/Nov 2025 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-19">
<div class="k-ptable"></div>
<h3>9701/22 Oct/Nov 2025</h3>
<div class="q"><h4>Q4(c)(i) <span class="mark">[1]</span></h4>
<p>Reaction 3 takes place when CH₃CH₂CH₂OH is heated under reflux with acidified potassium dichromate(VI) solution. State the colour change that takes place in the reaction mixture.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — orange to green</p>
</div>
</div>
<div class="q"><h4>Q4(c)(ii) <span class="mark">[1]</span></h4>
<p>Construct an equation to represent reaction 3. Use [O] to represent an atom of oxygen from the oxidising agent.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — CH₃CH₂CH₂OH + 2[O] → CH₃CH₂COOH + H₂O</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-19"));</script>"""))
```

```python
# @title 📝 9701/22 Feb/Mar 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-20">
<div class="k-ptable"></div>
<h3>9701/22 Feb/Mar 2026</h3>
<div class="q"><h4>Q3(e) <span class="mark">[3]</span></h4>
<p>The organic product of hydrolysis of each of B, C and D is butan-2-ol. Complete Table 3.1 to give details of some typical chemical reactions of butan-2-ol.</p>
<p>Table 3.1</p>
<table><tr><th>reagent and conditions</th><th>observation with butan-2-ol</th><th>organic product(s)</th></tr><tr><td>acidified K₂Cr₂O₇</td><td>(orange solution) turns green</td><td>butanone：CH₃CH₂COCH₃</td></tr><tr><td>Na(s)</td><td>effervescence</td><td>sodium butan-2-olate：CH₃CH₂CH(ONa)CH₃</td></tr><tr><td>alkaline I₂(aq)</td><td>yellow precipitate</td><td>CH₃CH₂COO⁻（propanoate ion，以 sodium propanoate 形式存在） and CHI₃</td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>判分</b></p>
<ul><li>M1 acidified K₂Cr₂O₇ 一行：(orange solution) turns green</li><li>M2 Na(s) 一行：organic product = sodium butan-2-olate</li><li>M3 alkaline I₂(aq) 一行：reagent = alkaline I₂(aq)（iodoform test，butan-2-ol 含 CH₃–CH(OH)– 结构，碳链在羰基处断裂，缩短一个碳生成 propanoate）</li></ul>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-20"));</script>"""))
```

```python
# @title 📝 9701/21 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-21">
<div class="k-ptable"></div>
<h3>9701/21 May/June 2026</h3>
<p>Compound Z is oxidised to produce organic compound W, C₅H₆O₇, by heating under reflux with an excess of acidified Cr₂O₇²⁻(aq). <img alt="Figure 4.1" src="data:image/png;base64,iVBORw0KGgoAAAANSUhEUgAACCgAAAGYCAIAAAAF4ta+AAAACXBIWXMAAC4jAAAuIwF4pT92AAAonUlEQVR4nO3d+XsV9b3A8fufFAkEIrIIiBepVFAUqldxqVQRtFoqbpXWijG1yqIVBJ8WUdzqvtArLoitihXNvbIpESsVlwvK2iIIGCDhyQLpTR/vM2fuScg235xJznm9nnn6WDPz5TPnHPxh3nMy//ZPAAAAAACAQP4t7QEAAAAAAID8ITwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPQFs9uGjRD8eOTb4tfeWVtE8FAAAAAOgswgPQVrffdlvP730v+fb4Y4+lfSoAAAAAQGcRHoC2Eh4AAAAAgFYJD0BbCQ8AAAAAQKuEB6CthAcAAAAAoFXCA9BWTz355MRLLmnjdubppzdbHYp69Fi9enXapwIAAAAAdBbhAQjv0KFDxwoPix54IO3pAAAAAIBOJDwAgTU0NEy56qpmq8MN112X9nQAAAAAQOcSHoDA7p0/v9nqcPa4cYcPH057OgAAAACgcwkPQEhvvP56s9Vh6Ikn7tyxI+3pAAAAAIBOJzwAwXz+2Wf9+vZtWh169+y5ds2atKcDAAAAAHJBeADC2L9//8gRI5r9usPTTz2V9nQAAAAAQI4ID0AA9fX1Ey+5pNnqUDp9etrTAQAAAAC5IzwAAcycMaPZ6nDR+efX1tamPR0AAAAAkDvCA5DUi0uWNFsdhg8btmfPnrSnAwAAAABySngAEvlo/fq+vXs3rQ59i4s//utf054OAAAAAMg14QHouN27d//7SSc1+3WHl196Ke3pAAAAAIAUCA9AB9XU1Fwwfnyz1eHOWbPSng4AAAAASIfwAHTQzTfd1Gx1mDRx4pEjR9KeDgAAAABIh/AAdMSTjz/ebHU47dRTKysr054OAAAAAEiN8AC02+rVq3v37Nm0OpxQUvLFF1+kPR0AAAAAkCbhAWifHTt2DB44sNmvOyxfvjzt6QAAAACAlAkPQDtUV1ePO+usZqvDgt/9Lu3pAAAAAID0CQ9AO1w7dWqz1eHqKVMaGhrSng4AAAAASJ/wALTVwvvua7Y6nDVmTFVVVdrTAQAAAABdgvAAtMnbf/lLs9XhxAEDtm7dmvZ0AAAAAEBXITwArdu0adOAfv2aVodexx238r330p4OAAAAAOhChAegdb+9885mv+4w+rTT7po9uwNbZWVl2ucEAAAAAHQK4QFo3V2zZzcbHjq87dixI+1zAgAAAAA6hfAAtE54AAAAAADaSHgAWic8AAAAAABtJDwArRMeAAAAAIA2Eh6A1gkPAAAAAEAbCQ8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgMAAAAAABCM8AAAAAAAAAQjPAAAAAAAAMEIDwAAAAAAQDDCAwAAAAAAEIzwAAAAAAAABCM8AAAAAAAAwQgPAAAAAABAMMIDAAAAAAAQjPAAAAAAAAAEIzwAAAAAAADBCA8AAAAAAEAwwgPwL19//fUnn3xSXl7+2rJl76xY8eGHHx48cCDtoQAAAACA7kd4gMJVX1//1vLlpdOnDx82rOf3vtd0GzVy5KyZMz9av77ta1ZUVPxw7NhoO3ToUHunOvfss6PD//LWW+09HAAAAABIl/AABer1P/959GmnNdsbmm4/uvDCtWvWtGXZ8vLy+IEH2v+1iV7HHRcd/uKSJe0/MwAAAAAgTcIDFJx9+/b9+OKL25gc4tt9v/99Q0NDy4sLDwAAAABQ4IQHKCxfbt582qmnNo0K4889d9bMmQ8uWvT0U0/Nufvu6665Zszo0U13u3bq1Orq6hbWFx4AAAAAoMAJD1BAtmzZcuKAAfEwcEJJyfx77tn7zTfN7v/+2rVTrroqqz2cM27c4cOHj/VHCA8AAAAAUOCEBygUNTU1Z48bF68Cl06YsGfPnlYPXL1q1cD+/eMHzrj99mPtLDwAAAAAQIETHqBQlJWWxpPAr8vKjhw50sZjN23adOqIEfHDV69a1eyewgMAAAAAFDjhAQrCpk2b4j1g4iWXtL06fOfLzZv79OoVrTBi+PBDhw413U14AAAAAIACJzxAQbj5ppuiq/n9jz9+7969HVjk3vnz41Fh8fPPN91HeAAAAACAAic8QP7bs2dP/MsKM+64o2Pr1NTUxH/h0qUTJjTdR3gAAAAAgAInPED+W/LCC/EYsGXLlg4vNW/u3Gidoh49vv7666wdhAcAAAAAKHDCA+S/X8V+z9KokSOTLLVhw4Z4V3j8sceydhAeAAAAAKDACQ+Q/0479dToUv60n/884WrfP+WUaLXrrrkm66fCQ974+uuvP/nkk8Y39LVly95ZseLDDz882P53EwAAAIACJDxAnquurm75Owrt9dMrr4xWu+iCC7J+mhUeXlu2bMXbb7drK+rRQ3hIS319/VvLl5dOnz582LD4+xj/xsysmTM/Wr++Xcv+ctq0H44d+9328EMPtXeqPy5eHB3+k8mT23s4AAAAADkmPJCH3Kkd9/edO7NKQMIFS6dPj1Y7dcSIrJ9mhYeEm/CQS6//+c+jTzutjW/Njy68cO2aNW1c+fzzzosOnDljRnsHW3jffdHh3z/llPYeDgAAAECOCQ8EltatzZ10p3Ye2LhxY/x1KC8vT7jgvfPmRav17d0766fCQ3e0b9++H198cQfeoPt+//uGhoZW1xceAAAAAAqK8EBgqVxh7Lw7tfPAqpUr46e/PnF6efLxx+MLVlZWxn8qPHQ7X27eHH8KSLSNP/fcWTNnPrho0dNPPTXn7ruvu+aaMaNHN93t2qlTq6urW/4jhAcAAACAgiI8EFiOrzB29p3aeWDtmjXxE0/+nY+HFi2KL1hVVRX/aVZ4+HVZWePHoF2b8JBLW7ZsOXHAgPhrfkJJyfx77tn7zTfN7v/+2rVTrroq62/TOePGHT58uIU/RXgAAAAAKCjCA4Hl8gpjDu7UzgPbtm2Ln/W7776bcMHGlzRarahHj6x+kxUeDrT/6Rq9jjtOeMiNmpqas8eNi79fl06YsGfPnlYPXL1q1cD+/eMHzrj99hb2Fx4AAAAACorwQGA5u8KYmzu180BtbW38lJe+8krCBW+5+eZoteHDhmX9VHjoRspKS7O+nnLkyJE2Hrtp06ZTR4yIH7561apj7Sw8AAAAABQU4YHAcnOFMWd3aueHoYMHR+c75+67E652TuyVnzRxYtZPhYfuYtOmTfF3auIll7S9Onzny82b+/TqFa0wYvjwQ4cONbun8AAAAABQUIQHAsvNFcac3amdHyZNnBid7IXjxydZqrq6Oh4G7pkzJ2sH4aG7uPmmm6LXuf/xx+/du7cDi9w7f3787V78/PPN7iY8AAAAABQU4YHAcnCFMZd3aueHJS+8EJ1scVHRvn37OrzU8uXL4y/+Z59+mrWD8NAt7NmzJ/5XYMYdd3RsnZqamnjGu3TChGZ3Ex4AAAAACorwQGA5uMKYyzu188PBgwf7FhdHJ7vgd7/r8FKXTpgQrTP2zDOb7iA8dAvxFtW4bdmypcNLzZs7N1qnqEePr7/+uuk+wgMAAABAQREeCKyzrzDm+E7tvHHN1VdHJztk0KAO9IBGH61fH79a/fCDDzbdR3joFn4Vq3ejRo5MstSGDRvi7/jjjz3WdB/hAQAAAKCgCA8E1tlXGHN8p3be+PDDDxvPMTrfq6dMae8KVVVVo0aOjFboW1y8a9euprsJD93CaaeeGr3I037+84SrNf5VjVa77pprmu4gPAAAAAAUFOGBwDr7CmOO79TOJ3fOmhU/33b9wqXa2tqpP/tZ/PAnn3ii2T2Fh66vuro67Cf/p1deGa120QUXNN0h/p+FKVddteLtt9u1/eLGG4UHAAAAgG5EeCCwzg4POb5TO5/U1NScPmpU/IrzbWVlbXku94EDB+KPdmjcLp80qaGhodmdhYeu7+87d8bfo9eWLUu4YOn06dFqp44Y0XSH+H8WEm7CAwAAAEDXJzwQWKfe2pz7O7XzzF8/+uj4Pn3ir+FZZ5xRXl5+rP2PHDny/HPPnTRkSPyQIYMG7d69+1iHFHJ4aDzZDV3Mt99+23TOjRs3xt+jFj4AbXTvvHnRan179266g/AAAAAAUFCEBwLr1CuMub9TO/9s3LhxxPDhWS/1qB/84M5Zs/78pz+9v3btZ59+uvK9915+6aWbb7pp6ODBWXs2/psNH3/cwvqFHB6unTo11Ic/1Dawf/+nn3oqa85VK1fG91m/fn3CE3/y8cfjC1ZWVmbtIDwAAAAAFBThgcA69Qpj7u/Uzkt79+69+KKLOvCOXDB+/LZt21pevGDDw+rVq1MMDC1vWaOuXbMm/tOPEoeHhxYtii9YVVWVtUP8Pwtnjxs3c8aMdm0Xjh8vPAAAAAB0I8IDgXVqeMj9ndr5qq6ubuGCBQP69WvjezFq5Mhnnn766NGjra5cmOGhvr7+rDFjUg8MbQwP27Zti//03XffTXj6c+6+O1qtqEePps//6OxHvwAAAADQpQgPBNaptzbn/k7t/LZ///6HH3qo8W061gXrvr1733D99atXrz7Wo6SbKszw8NSTT8bPekC/foP69+8K23n/8R9r1qzJmra2tjY+7dJXXkl4+rfcfHO02vBhw5ruIDwAAAAAFBThgcA69Qpj7u/ULhB7v/mmvLz8ySeeWLhgwdw5cx64//7nn3uuoqKiuro67dG6gf379w8eODD6IJ01Zkx9fX3aQ7Ui/vSOxr8FCVc7J9auJk2c2HQH4QEAAACgoAgPBNapVxhzf6c2tOq2srL4x3LVypVpT9S6SRMnRgNfOH58kqWqq6vjX1K5Z86cpvsIDwAAAAAFRXggsM6+wpjjO7WhZV98/nnvnj2jT9HVU6akPVGbLHnhhWjm4qKiffv2dXip5cuXx7vLZ59+2nQf4QEAAACgoAgPBNbZVxhzfKc2tOzSCRMyj8QoLt6+fXvaE7XJwYMHG6eNJl/wu991eKn4KzD2zDOb3Ud4AAAAACgowgOBdfYVxhzfqQ0teOP11+MfoXlz56Y9UTtcc/XV0eRDBg3qwGPAG320fn38FXj4wQeb3U14AAAAACgowgOBdfYVxhzfqQ3HUlNTM3LEiPgzQqqqqtIeqh0+/PDDoh49kvySqMbzHTVyZPwLH7t27Wp2T+EBAAAAoKAIDwSWgyuMubxTG45l4YIF8Y/Qq0uXpj1Ru905a1b8FNqV8Wpra6f+7Gfxw5984olj7Sw8AAAAABQU4YHAcnCFMZd3akOzGj8w/fr2jT5CF4wf39DQkPZQ7VZTU3P6qFHxeHBbWdmRI0daPfDAgQPxLww1bpdPmtTCKyA8AAAAABQU4YHAcnOFMWd3akOzbrzhhujzU9Sjx4aPP057og7660cfHd+nT/yvw1lnnFFeXn6s/Y8cOfL8c8+dNGRI/JAhgwbt3r27hT9FeAAAAAAoKMIDgeXmCmPO7tSGpirWrYt/hKb/6ldpT5TIxo0bRwwfHj+jxm3UD35w56xZf/7Tn95fu/azTz9d+d57L7/00s033TR08OCsPRv/TavdRXgAAAAAKCjCA4Hl7Apjbu7UhixHjx4995xzoo/QgH799uzZk/ZQSe3du/fiiy7KKgpt2S4YP37btm2tri88AAAAABQU4YHAcnmFMQd3akOWPy5eHP8UPfrII2lPFEZdXd3CBQsG9OvXxuQwauTIZ55++ujRo21ZXHgAAAAAKCjCA4Hl+ApjZ9+pXYB27tixYcOGv+/cmfYgXdHBgwfjBWvUD35QV1eX9lAh7d+//+GHHjp73Lhj/cXp27v3Dddfv3r16nb9gjLhAQAAAKCgCA8ElvsrjJ16p3ahOXz48Cknn9z4KjX+b+M/pz1Ol3PX7Nnxz9I7K1akPVFn2fvNN+Xl5U8+8UTjX665c+Y8cP/9zz/3XEVFRXV1ddqjAQAAANDVCQ8EltatzZ10p3ahaXwNo1fsicceS3ucrmXz5s3FRUXR63PlFVekPREAAAAAdEXCA/nGndpJ3FZWFl1Yb3z10h6na7nyiiuiF6e4qGjz5s1pTwQAAAAAXZHwAGQID8fyzooV8S/Q3DV7dtoTAQAAAEAXJTwAGcJDs+rq6kafdlr0ypw0ZMjBgwfTHgoAAAAAuijhAcgQHpr16COPxL/u8MfFi9OeCAAAAAC6LuEByBAemtqzZ8+Afv2il+Xcc845evRo2kMBAAAAQNclPFC4Ghoalr/55t937kx7kC5EeGiqdPr0+NcdKtatS3siAAAAAOjShAcKVENDw48vvrjn977Xp1ev+xcuTHucrkJ4yLLh44+LevSIXpMbb7gh7YkAAAAAoKsTHihc8dvY165Zk/Y4XYLwENfQ0HDR+edHL8gJJSW7du1KeygAAAAA6OqEBwpXSXFxdE15zOjRNTU1aU+UPuEh7tWlS+N1auGCBWlPBAAAAADdgPBA4Xpw0SKXlbMID5Gqqqrhw4ZFr8bIESOkKQAAAABoC+GBwlVXV3fWmDHRleW+xcVbvvoq7aFSJjxE7p0/P96llr/5ZtoTAQAAAED3IDxQ0CoqKuIXlydecklDQ0PaQ6VJePjO9u3b+8Z+E9elEyakPREAAAAAdBvCA4Wu7NZb4+3hlZdfTnuiNAkP37l6ypTodejds+cXn3+e9kQAAAAA0G0IDxS6AwcOnDRkSHSVeeiJJ+7fvz/toVIjPDRavXp1vEXd8ZvfpD0RAAAAAHQnwgP887Vly+IXmm+5+ea0J0qN8FBfXx9/8sfggQMLOUQBAAAAQAcID/Avl0+aFG8PH7z/ftoTpUN4ePqpp+KfhGefeSbtiQAAAACgmxEe4F+2bdtWEnuY8JjRo2tra9MeKgUFHh72798/eODA6BU4a8yY+vr6tIcCAAAAgG5GeID/8+CiRfFb3e9fuDDtiVJQ4OHhN7/+dfwzsHr16rQnAgAAAIDuR3iA/1NXVxf/5f59i4u3bNmS9lC5Vsjh4YvPP+/ds2d0+tdOnZr2RAAAAADQLQkPkFFRURG/4f2ySy9Ne6JcK+TwcOmECfHstH379rQnAgAAAIBuSXiA/6fs1lvj7eGVl19Oe6KcKtjw8OYbb8Tf93vnz097IgAAAADoroQH+H8qKytPGjIkugA9dPDgb7/9Nu2hcqcww0NNTc3IESOiEz/l5JMPHz6c9lAAAAAA0F0JD5Bt2auvxm9+v/WWW9KeKHcKMzwsXLAg/o6/unRp2hMBAAAAQDcmPEAzLp80KX4let0HH6Q9UY4UYHjYtWvXCSUl0VlffNFFDQ0NaQ8FAAAAAN2Y8ADN2Lp1a0lxcXQx+szTT6+rq0t7qFwowPBw4w03RKdc1KPHhg0b0p4IAAAAALo34QGat+iBB+Jfenjg/vvTnigXCi08VFRUxN/l0unT054IAAAAALo94QGaV1dXd9aYMdEl6b7FxVu3bk17qE5XUOHh6NGj555zTnS+A044Ye/evWkPBQAAAADdnvAAx1Sxbl38dvhJEyemPVGnK6jw8MfFi+Pv76OPPJL2RAAAAACQD4QHaEnZrbfGr02/unRp2hN1rsIJDwcPHjxpyJDoZM8YPbpAHuMBAAAAAJ1NeICWVFZWxi9PN/5z479Je6hOVDjh4a7Zs+NJqby8PO2JAAAAACBPCA/QimWvvhq/Ql1WWpr2RJ2oQMLDl5s3FxcVRWd65RVXpD0RAAAAAOQP4QFad/mkSfH2ULFuXdoTdZYCCQ9XXnFF5rHhvXtv+eqrtCcCAAAAgPwhPEDrtm7dWlJcHF2qPuuMM/L1eQCFEB7eWbEinpF+e+edaU8EAAAAAHlFeIA2eeD+++NXqxc98EDaE3WKvA8PdXV1p48aFZ3jsKFDDx48mPZQAAAAAJBXhAdok7q6urPOOCO6YF1SXLx169a0hwov78PDo488Eg9ILy5ZkvZEAAAAAJBvhAdoq4p16+LXrC+fNCnticLL7/DwzZ49A/r1i07w3HPOOXr0aNpDAQAAAEC+ER6gHcpKS+Pt4bVly9KeKLD8Dg+l06fH377169enPREAAAAA5CHhAdqhsrLypCFDoivXjf984MCBtIcKKY/Dw4aPPy7q0SM6u19Om5b2RAAAAACQn4QHaJ9Xly6N3zVfduutaU8UUr6Gh4aGhosuuCA6tRNKSnbt2pX2UAAAAACQn4QHaLfJl10Wbw8VFRVpTxRMvoaHrFz04KJFaU8EAAAAAHlLeIB227p1a0lxcXQVe+yZZ9bV1aU9VBh5GR6qq6uHDxsWndfIESNqamrSHgoAAAAA8pbwAB3xwP335+Ud9HkZHu6dPz/+Zi1fvjztiQAAAAAgnwkP0BF1dXVnnXFGdC37+D597v7tb+fcfXd3384eNy46qfPPOy/1eZJvje9L39jXUyZfdlnanx0AAAAAyHPCA3RQxbp18fvobV1/692z5xeff572BwcAAAAA8pzwAB1XVlqa+sV0W9u3GXfckfZHBgAAAADyn/AAHVdZWTl08ODUr6fb2rINPfHEAwcOpP2RAQAAAID8JzxAIu+++25ZaWnebGeefnp0pf7ssWNTnyfg9pe33kr7wwIAAAAABUF4ADJuKyuLwsPcOXPSHgcAAAAA6H6EByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIEB4AAAAAgISEByBDeAAAAAAAEhIegAzhAQAAAABISHgAMoQHAAAAACAh4QHIiIeHX5eVpT0OAAAAAND9CA9ARjw8NG5XTJ782aefpj0UAAAAANCdCA9ARlZ4aNyKevSY/qtf7d69O+3RAAAAAIDuQXgAMv72t79ddMEFWe2hcTuhpOTeefOqqqrSHhAAAAAA6OqEByDb8jffHDVyZNP8cPLQoS/853/W19enPSAAAAAA0HUJD0Azamtrn3jssSGDBjXND2eefnp5eXnaAwIAAAAAXZTwABzTgQMH7po9u0+vXk3zw+TLLvvkk0/SHhAAAAAA6HKEB6AVO3bs+Pn11zdtD547DQAAAAA0JTwAbbJ+/fpmnztdUlw8/557PHcaAAAAAPiO8AC0QwvPnX7+2Wc9dxoAAAAAEB6A9vHcaQAAAACgBcID0BEtPHf60h//2HOnAQAAAKBgCQ9Ax23fvr2F507/fefOtAcEAAAAAHJNeACS+tdzp88/33OnAQAAAIB/Cg9AKMd67vRJQ4Y88/TTnjsNAAAAAAVCeACCqa2tffzYz51+Z8WKtAcEAAAAADqd8AAE1sJzpy+ZMMFzpwEAAAAgvwkPQKdo4bnTv5w2zXOnAQAAACBfCQ9AJ1q/fv2F48cf67nThw4dSrj+4cOHv/ryyw/ef//NN9544/XXV61c2fh/jx49GmR4AAAAAKADhAeg0x3rudNDBg3q2HOnt23b9vBDD/344ot79+zZdNkTSkomX3bZHxcvPnDgQNvX/OW0aT8cO/a7rXHx9o7U+MdFh/9k8uT2Hg4AAAAAeUN4AHKhtrb2sT/84VjPnV6+fHkb19m9e3dZaWmzvaHpNqBfv7Z/r+L8886LDpw5Y0Z7T3DhffdFh3//lFPaezgAAAAA5A3hAcidhM+dfnHJkuP79GlLcohvZ4weveWrr1qdTXgAAAAAgCCEByDXtm3b1uxzp3sdd9zGjRubPaShoWH+Pfc0+8uaGpe6d968J5944qFFi8pKSy+79NJ+fftm7Tawf//y8vKWpxIeAAAAACAI4QFIR7PPnb5kwoRmd55x++1ZezYe+95//3ezz5GurKx8cNGifz/ppKyqsfSVV1qYR3gAAAAAgCCEByBNWc+dListbbrPsldfjSeEkuLiF5csaXXlmpqa66+9Nn5g/+OP37Fjx7H2Fx4AAAAAIAjhAUhZbW3tE489VlZa2rjt2bMn66dfffnlCSUl0TX9oSeeeKxfx9RUQ0PD3Dlzsr5R0fgvm91ZeAAAAACAIIQHoEu74brr4r8uadXKle1d4aZf/CLeHp58/PFmdxMeAAAAACAI4QHourZv397ruOOiC/pz58zpwCL79u07ccCAaJFTTj652S89CA8AAAAAEITwAHRd8WdK9+nV65smv4ipjZ5/9tn4lx7eX7u26T7CAwAAAAAEITwAXdeI4cOjq/m/uPHGDq9TVVXVt7g4WurXZWVN9xEeAAAAACAI4QHoov6+c2f8awpLX3klyWpX/eQn0VJDBg2qr6/P2kF4AAAAAIAghAegi3rl5Zfj4WHLli1JVlvywgvx1f7nf/4nawfhAQAAAACCEB6ALuq3d94ZXcof2L9/wtX+9re/xcPDyvfey9ohHh6mXHXVirffbtf2ixtvFB4AAAAA4J/CA9Bl3XzTTdGl/DNGj0642q5du+Lh4cUlS7J2iIeHhJvwAAAAAEAhEx6ALupnP/1pdCn//PPOS7hafX19vA0seuCBrB2EBwAAAAAIQngAuqgfXXhhdCn/8kmTki84qH//aMEZd9yR9VPhAQAAAACCEB6ALuqSCROiS/lXTJ6cfMG+xcXRgnfNnp3103h4OHvcuJkzZrRru3D8eOEBAAAAAP4pPABd1i+nTYsu5Y8/99yEq9XU1MS/lHDf73+ftUM8PMycMaO96y+87z7hAQAAAAD+KTwAXda8uXOjS/mjRo5MuNo//vGPeHj44+LFWTsIDwAAAAAQhPAAdFHPPvNMdCm/d8+eVVVVSVb7y1tvxcNDRUVF1g7CAwAAAAAEITwAXVRFRUU8Ffz3f/1XktXm3H13tFTf3r0PHTqUtYPwAAAAAABBCA9AF9XQ0PD9U06JrubPnjkzyWo/HDs2WurqKVOa7iA8AAAAAEAQwgPQdc2dMye6mj+gX7+DBw92bJ33166Nf3nizTfeaLqP8AAAAAAAQQgPQNf1xeefx4PB/QsXdmydn0yeHC0ysH//mpqapvsIDwAAAAAQhPAAdGlXxJpBcVHRXz/6qL0rPPfss/F6ccdvftPsbsIDAAAAAAQhPABd2j/+8Y8BJ5wQv6a/c8eOth/+wfvvlxQXR4efOmLEsX5fk/AAAAAAAEEID0BXt+SFF+JfWRg2dOiGDRvacuCfXnutb+/e0YG9jjtu3QcfHGtn4QEAAAAAghAegG7g2qlT4+2huKho5owZ33777bH237p169VTpsQPadzmzZ3bwh8hPAAAAABAEMID0A0cOXJk9syZWSGhb3HxlVdc8czTT7/7zjuffPLJ+vXrly9f/uCiRRedf35Rjx5ZO996yy2Ni7TwRwgPAAAAABCE8AB0Gy8uWRL/1Ult3PoWF//h0UdbXVx4AAAAAIAghAegO9mwYcOEH/2ojcmhT69e11977ZYtW9qysvAAAAAAAEEID0D3s2rlyl9Om9b/+OOPlRxOHzXqkYcf3rt3b9vXFB4AAAAAIAjhAeiu6uvrP9248aUXX3z4oYfmzZ177/z5f3j00TffeGPbtm1pjwYAAAAAhUt4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYIQHAAAAAAAgGOEBAAAAAAAIRngAAAAAAACCER4AAAAAAIBghAcAAAAAACAY4QEAAAAAAAhGeAAAAAAAAIIRHgAAAAAAgGCEBwAAAAAAIBjhAQAAAAAACEZ4AAAAAAAAghEeAAAAAACAYP4XnWKGZtQ20EUAAAAASUVORK5CYII="> Z 的结构：中心碳直接连一个 OH，另接三条支链 —CH₂OH、—CH₂OH、—CH₂CH₂OH（经渲染页逐键核对，右侧支链是 2 个碳，不是 1 个碳，否则凑不出 C₅H₁₂O₄）。</p>
<div class="q"><h4>Q4(a)(i) <span class="mark">[1]</span></h4>
<p>Describe the colour change in this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — colour of mixture at start of the reaction: orange；colour of mixture after the reaction: green</p>
<p><b>判分</b> — A 橙/绿的任何色号；BOD 绿写成 blue/grey；BOD 绿写成 brown（因为 dichromate 过量）</p>
</div>
<p>Compound A is a structural isomer of Z. <img alt="Figure 4.2" src=""> A 的结构（渲染页逐键核对后确认为 5 个碳）：HOCH₂CH₂–CH(OH)–CH(OH)–CH₂OH（pentane-1,2,3,5-tetraol）。</p>
</div>
<div class="q"><h4>Q4(b)(i) <span class="mark">[2]</span></h4>
<p>Complete Table 4.1 to show what is observed when separate samples of A are added to each reagent. State 'none' if no reaction occurs.</p>
<p><b>Table 4.1</b></p>
<table><tr><th>reagent</th><th>observation with A</th></tr><tr><td>Na(s)</td><td>effervescence / fizzing / bubbling</td></tr><tr><td>2,4-dinitrophenylhydrazine (2,4-DNPH)</td><td>none</td></tr><tr><td>Fehling's reagent</td><td>none</td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>判分</b> — Na(s) 一行：I 只写「产生气体」不写现象；I 气体名称/检验气体的方法；I 固体消失/变无色溶液。Fehling's 一行：A（仍为）orange 溶液/无沉淀/无反应/无现象；I 写成"变橙色"</p>
</div>
</div>
<div class="q"><h4>Q4(b)(iii) <span class="mark">[2]</span></h4>
<p>Compound A is heated under reflux with an excess of acidified Cr₂O₇²⁻(aq) to produce organic compound Y. Draw the structure of Y.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — A 的 2 个 primary alcohol（两端 CH₂OH）被氧化为 COOH，2 个 secondary alcohol（中间两个 CHOH）被氧化为酮 C=O，中间的 CH₂（无 OH）不受影响：</p>
<p>Y = HOOC–CO–CO–CH₂–COOH</p>
<p><b>判分</b> — A 任何无歧义的结构；I connectivity 表达方式</p>
</div>
</div>
<div class="q"><h4>Q4(c) <span class="mark">[2]</span></h4>
<p>Compound B reacts with an excess of LiAlH₄ to produce organic compound C.</p>
<p><img alt="Figure 4.3" src=""></p>
<p>Complete the equation in Figure 4.3 to show the reaction of B with an excess of LiAlH₄ to produce C. [H] represents one atom of hydrogen from the reducing agent, LiAlH₄.</p>
<p>B 的结构（渲染页逐键核对）：CH₃–CO–CH₂–CO–C(OH)(CH₃)–COOH（7 个碳：2 个 ketone + 1 个 tertiary alcohol + 1 个 carboxylic acid）。</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — 过量 LiAlH₄ 把 COOH 还原成 CH₂OH（耗 4[H]，副产物 H₂O），把 2 个 ketone 分别还原成 CH(OH)（各耗 2[H]），tertiary alcohol 不受影响：</p>
<p>C = HOCH₂–C(OH)(CH₃)–CH(OH)–CH₂–CH(OH)–CH₃（共 8 个 [H]）</p>
<p>B + 8[H] → C + H₂O</p>
<p><b>判分</b> — M1 A C 的任何无歧义结构；M2 按该结构配平的化学计量数；A ecf：若结构错但仍正确配平，且结构含 7 个碳，同样给分</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-21"));</script>"""))
```

```python
# @title 📝 9701/22 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-22">
<div class="k-ptable"></div>
<h3>9701/22 May/June 2026</h3>
<p>Figure 4.1 shows two reactions of compound D. <img alt="Figure 4.1" src=""> D = (CH₃)₃C–OH（2-methylpropan-2-ol / tert-butanol，图中以十字骨架 + OH 标注表示）。</p>
<div class="q"><h4>Q4(a)(i) <span class="mark">[1]</span></h4>
<p>D reacts with a single reagent, G, to produce compound E. Identify reagent G.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — PCl₅ <b>OR</b> SOCl₂ <b>OR</b> HCl <b>OR</b> PCl₃（或等效名称）</p>
<p><b>判分</b> — A thionyl chloride / sulfur oxide dichloride（SOCl₂）；A phosphorus pentachloride / phosphorus trichloride；A conc. HCl / concentrated hydrochloric acid；A hydrogen chloride / HCl(g)；I 单独 hydrochloric acid / HCl(aq) / dilute HCl；I NaCl／KCl + conc. H₂SO₄／H₃PO₄；R PCl₅(aq)</p>
</div>
</div>
<div class="q"><h4>Q4(a)(ii) <span class="mark">[1]</span></h4>
<p>D is heated with Al₂O₃ to produce organic compound F. Identify the role of Al₂O₃ in this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — catalyst <b>OR</b> dehydrating agent</p>
<p><b>判分</b> — A causes dehydration；A 加快反应速率且本身不变；A 降低 E꜀；I heterogeneous/homogeneous；R redox</p>
</div>
</div>
<div class="q"><h4>Q4(a)(iii) <span class="mark">[1]</span></h4>
<p>Draw the structure of F in Figure 4.1.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — F = 2-methylpropene，H₂C=C(CH₃)₂</p>
<p><b>判分</b> — A 任何无歧义结构；I 只写 C₄H₈</p>
</div>
<p>The structure of compound X is shown in Figure 4.2. <img alt="Figure 4.2" src=""> > 读图提示：X 的左支链在折点处<b>有一个 CH₂</b>（骨架式的拐点就是一个碳），所以 X = HOCH₂CH₂–C(OH)(CH₂OH)₂，共 5 个碳，正是题干给的 C₅H₁₂O₄（<i>M</i>ᵣ = 136）。注意 X <b>不是</b> pentaerythritol C(CH₂OH)₄，两者是结构异构体。 The molecular formula of X is C₅H₁₂O₄. Complete oxidation of 0.255 g of X occurs when it reacts with exactly 10.0 cm³ of 0.500 mol dm⁻³ acidified K₂Cr₂O₇(aq).</p>
</div>
<div class="q"><h4>Q4(b)(i) <span class="mark">[1]</span></h4>
<p>Describe the colour change in this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — colour of mixture at start of the reaction: orange；colour of mixture after the reaction: green</p>
<p><b>判分</b> — A 橙/绿的任何色号；BOD 绿写成 blue/grey 或 brown（dichromate 过量）</p>
</div>
</div>
<div class="q"><h4>Q4(c)(ii) <span class="mark">[2]</span></h4>
<p>Each molecule of W contains four hydroxyl groups. Only two of these are primary hydroxyl groups. Table 4.1 shows the observations when separate samples of Z are added to each reagent. Use the information in 4(c) to suggest the structure of W.</p>
<p><b>Table 4.1</b>（题目给出，非填空）</p>
<table><tr><th>reagent</th><th>observations with Z</th></tr><tr><td>2,4-dinitrophenylhydrazine (2,4-DNPH)</td><td>orange precipitate</td></tr><tr><td>Fehling's reagent</td><td>none</td></tr></table>
<p>Z 对 2,4-DNPH 有反应（含 C=O）但对 Fehling's 无反应（不是 aldehyde，是 ketone）——说明 W 的 4 个 hydroxyl 里，2 个 primary 被完全氧化成 carboxylic acid（COOH），1 个 secondary 被氧化成 ketone（C=O），1 个 tertiary 氧化不了、原样保留。按此推出 W = HOCH₂–CH(OH)–C(OH)(CH₃)–CH₂OH 时，Z = HOOC–CO–C(OH)(CH₃)–COOH，分子式 C₅H₆O₆，与 (c)(i) 配平方程里的 C₅H₆O₆ 完全对上，交叉验证成立。</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — W = HOCH₂CH(OH)C(OH)(CH₃)CH₂OH （<b>OR</b> (HOCH₂)₂C(OH)(CHOHCH₃)）：2 个 primary alcohol（两端 CH₂OH）、1 个 secondary alcohol（CH(OH)）、1 个 tertiary alcohol（C(OH)(CH₃)）</p>
<p><b>判分</b> — A 任何无歧义结构；I 尝试命名；Max 1：全部基团正确但有 1 处或以上 connectivity 错误（如写成 C−H−O）</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-22"));</script>"""))
```

```python
# @title 📝 9701/23 May/June 2026 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-23">
<div class="k-ptable"></div>
<h3>9701/23 May/June 2026</h3>
<div class="q"><h4>Q4(a)(i) <span class="mark">[2]</span></h4>
<p>Complete Table 4.1.</p>
<p><b>Table 4.1</b></p>
<table><tr><th></th><th>primary hydroxyl groups</th><th>secondary hydroxyl groups</th><th>tertiary hydroxyl groups</th></tr><tr><td>number of groups in one molecule of V</td><td>2</td><td>1</td><td>1</td></tr></table>
</div>
<div class="q"><h4>Q4(a)(ii) <span class="mark">[2]</span></h4>
<p>Complete Table 4.2 to show what is observed when separate samples of V are added to each reagent. State 'none' if no reaction occurs.</p>
<p><b>Table 4.2</b></p>
<table><tr><th>reagent</th><th>observation with V</th></tr><tr><td>Na(s)</td><td>effervescence / fizzing / bubbling</td></tr><tr><td>2,4-dinitrophenylhydrazine (2,4-DNPH)</td><td>none</td></tr><tr><td>Fehling's reagent</td><td>none</td></tr></table>
<button class="reveal">答案</button><div class="ans">
<p><b>判分</b> — Na(s) 一行：I 只写「产生气体」不写现象／写气体名称／写检验方法／固体消失或变无色溶液；2,4-DNPH 一行：A（仍为）orange 溶液/无沉淀/无反应/无现象；I 写成"变橙色"</p>
</div>
<p>Figure 4.1 is repeated for use in 4(b). V is completely oxidised by heating under reflux with an excess of acidified K₂Cr₂O₇(aq) to produce organic compound Z.</p>
</div>
<div class="q"><h4>Q4(b)(i) <span class="mark">[1]</span></h4>
<p>Describe the colour change in this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — colour of mixture at start of the reaction: orange；colour of mixture after the reaction: green</p>
<p><b>判分</b> — A 橙/绿的任何色号；BOD 绿写成 blue/grey 或 brown（dichromate 过量）</p>
</div>
</div>
<div class="q"><h4>Q4(b)(ii) <span class="mark">[1]</span></h4>
<p>Draw the structure of Z.</p>
<p>V 的 2 个 primary alcohol（两端 CH₂OH）氧化成 COOH，1 个 secondary alcohol（CH(OH)）氧化成 ketone（C=O），1 个 tertiary alcohol（C(OH)(CH₃)）氧化不了、原样保留：</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — Z = HOOC–C(OH)(CH₃)–CO–COOH （<b>OR</b> (HOOC)C=OC(COOH)(OH)CH₃，即同一结构的另一种写法）</p>
<p><b>判分</b> — A 任何无歧义结构；I connectivity 表达方式</p>
</div>
<p>Compounds W, X and Y are structural isomers of V. <img alt="Figure 4.2" src=""> > 读图提示：骨架式的每个拐点都是一个碳。Figure 4.2 的四个结构<b>全是 C₅H₁₂O₄</b>——V = CH₃–C(OH)(CH₂OH)–CH(OH)–CH₂OH；W = HOCH₂–CH(OH)–CH(OH)–CH₂–CH₂OH（左侧竖线下端到 CH₂OH 之间那个折点是 CH₂）；X = HOCH₂CH₂–C(OH)(CH₂OH)₂；Y = C(CH₂OH)₄（pentaerythritol）。W 的 C2、C3 各连 4 个不同基团，所以 (c)(ii) 说「both have 2 chiral centres」成立。</p>
</div>
<div class="q"><h4>Q4(c)(iii) <span class="mark">[1]</span></h4>
<p><img alt="Figure 4.3" src=""></p>
<p>Y reacts with reagent R to produce compound E（图中 E 是 Y 的四个 –CH₂OH 都与 reagent R 发生酯化后的产物）。Name reagent R.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — methanoic acid</p>
<p><b>判分</b> — A formic acid</p>
</div>
<p>Exactly 0.03 mol of one of the isomers, V, W, X or Y, is completely oxidised by 0.08 mol of acidified K₂Cr₂O₇(aq) to produce only one organic compound with molecular formula C₅H₄O₈.</p>
</div>
<div class="q"><h4>Q4(d)(i) <span class="mark">[1]</span></h4>
<p>Identify which isomer, V, W, X or Y, is used in this reaction.</p>
<button class="reveal">答案</button><div class="ans">
<p><b>答案</b> — (isomer) Y</p>
<p><b>判分</b> — A the structure of isomer Y</p>
</div>
</div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-23"));</script>"""))
```

```python
# @title ✅ Learning outcomes 自查清单 { display-mode: "form" }
from IPython.display import HTML, display
display(HTML(r"""<style>/* che_note Colab kit — shared styles, inlined into every html cell by tools/build_colab.py.
   Everything is scoped under .k so a preview page can stack many cells without clashes. */
/* Noto Sans draws superscript ⁺ ⁻ (U+207A/B) only a few px wide — I⁻ reads as "I'" at 15px.
   Take just those two glyphs from a font whose ⁺⁻ are full-size (unicode-range: nothing else changes). */
@font-face{font-family:"KitCharge";src:local("DejaVu Sans"),local("DejaVuSans"),local("Segoe UI"),local("SegoeUI"),local("Arial");unicode-range:U+207A-207B}
.k{
  --bg:#fbfaf7; --card:#ffffff; --fg:#1f2328; --mut:#5c6370; --faint:#8b929c; --line:#d9d6cf;
  --acc:#2f5fd0; --acc-soft:#e7edfb; --pos:#c0392b; --neg:#2f5fd0; --neu:#6b7280;
  --ok:#1a7f37; --ok-soft:#dcf3e2; --bad:#cf222e; --bad-soft:#fbe3e4; --hl:#fff3c4;
  --s1:#2f5fd0; --s2:#c0392b; --s3:#1a7f37; --s4:#b26b00; --s5:#7a3fb0;
  /* Latin fonts first: they carry every superscript digit (⁴⁵⁶…); CJK fonts lack U+2074–2079,
     so with a CJK font first ² and ⁶ come from different fonts and look mismatched */
  font:15px/1.55 "KitCharge","Segoe UI","Noto Sans","DejaVu Sans","Helvetica Neue",-apple-system,"Noto Sans SC","PingFang SC","Microsoft YaHei",sans-serif;
  color:var(--fg); background:var(--bg); border:1px solid var(--line); border-radius:10px;
  padding:14px 16px; margin:4px 0; max-width:920px; box-sizing:border-box;
}
@media (prefers-color-scheme:dark){
  .k{
    color-scheme:dark;   /* native checkbox / radio / scrollbars follow the theme (were white blocks) */
    --bg:#1b1e24; --card:#23272f; --fg:#e6e8eb; --mut:#aab1bb; --faint:#7d8590; --line:#3a3f48;
    --acc:#7aa2ff; --acc-soft:#2a3550; --pos:#ff7b6b; --neg:#7aa2ff; --neu:#9aa3ad;
    --ok:#4ac26b; --ok-soft:#1d3a26; --bad:#ff7b72; --bad-soft:#4a2224; --hl:#4a4020;
    --s1:#7aa2ff; --s2:#ff7b6b; --s3:#4ac26b; --s4:#e3a64a; --s5:#c39bf0;
  }
}
.k *{box-sizing:border-box}
.k h3{font-size:17px;margin:0 0 8px;font-weight:650}
.k h4{font-size:15px;margin:14px 0 6px;font-weight:650}
.k p{margin:6px 0}
.k .mut{color:var(--mut)} .k .faint{color:var(--faint);font-size:13px}
.k .row{display:flex;flex-wrap:wrap;gap:8px;align-items:center;margin:6px 0}
.k .grid2{display:grid;grid-template-columns:repeat(auto-fit,minmax(260px,1fr));gap:12px}
.k .card{background:var(--card);border:1px solid var(--line);border-radius:8px;padding:10px 12px}
.k button,.k select,.k input{font:inherit;color:var(--fg)}
.k button{background:var(--card);border:1px solid var(--line);border-radius:6px;padding:4px 11px;cursor:pointer}
.k button:hover{border-color:var(--acc)}
.k button.on{background:var(--acc);border-color:var(--acc);color:#fff}
@media (prefers-color-scheme:dark){.k button.on{color:#0d1220}}
.k select,.k input[type=number],.k input[type=text],.k input.fill:not([type]){background:var(--card);color:var(--fg);border:1px solid var(--line);border-radius:6px;padding:3px 7px}
.k input[type=number]{width:5.2em}
.k input[type=range],.k input[type=checkbox],.k input[type=radio]{accent-color:var(--acc)}
.k label{cursor:pointer}
.k table{border-collapse:collapse;margin:8px 0;font-size:14px}
.k th,.k td{border:1px solid var(--line);padding:4px 9px;text-align:left;vertical-align:top}
.k th{background:var(--acc-soft);font-weight:600}
/* mono stack: Latin fonts first, the CJK mono default has no ⁴–⁹ */
.k code{font-family:"Cascadia Mono","DejaVu Sans Mono","Noto Sans Mono","SF Mono",Consolas,ui-monospace,monospace;font-size:13.5px;background:var(--acc-soft);padding:0 4px;border-radius:4px}
.k ul,.k ol{margin:6px 0;padding-left:22px}
.k svg{max-width:100%;height:auto;display:block}
.k svg text{fill:var(--fg)}
.k img{max-width:100%;height:auto;background:#fff;border-radius:4px}
.k .q img{display:block;margin:6px 0}
/* chart labels: a halo in the card colour so grid lines never cut through the glyphs */
.k svg text.halo{paint-order:stroke;stroke:var(--card);stroke-width:4px;stroke-linejoin:round}
/* nuclide notation: <span class="nuc"><sup>A</sup><sub>Z</sub></span>X */
.k .nuc{display:inline-flex;flex-direction:column;vertical-align:.42em;line-height:1.05;font-size:.72em;text-align:right;margin-right:1px}
/* an inline-flex box sits on the baseline of its FIRST line (the mass number), so it is raised,
   not lowered: mass number ends up top-left, proton number bottom-left */
.k .nuc sup,.k .nuc sub{position:static;font-size:1em;line-height:1.05}
/* reveal */
.k .reveal{margin-top:6px}
.k .ans{display:none;margin-top:6px;padding:8px 11px;border-left:3px solid var(--ok);background:var(--ok-soft);border-radius:0 6px 6px 0}
.k .ans.show{display:block}
/* self-check inputs */
.k input.fill{width:4.6em;text-align:center}
.k input.fill.good:focus{outline:2px solid var(--ok)}
.k input.fill.badv:focus{outline:2px solid var(--bad)}
.k input.fill.good{border-color:var(--ok);background:var(--ok-soft)}
.k input.fill.badv{border-color:var(--bad);background:var(--bad-soft)}
.k .mcq button.good{background:var(--ok-soft);border-color:var(--ok)}
.k .mcq button.badv{background:var(--bad-soft);border-color:var(--bad)}
/* tabs */
.k .tabbar{display:flex;flex-wrap:wrap;gap:6px;margin-bottom:8px}
.k [data-pane]{display:none} .k [data-pane].show{display:block}
/* tooltip */
.k .tip{position:fixed;z-index:9;pointer-events:none;background:var(--fg);color:var(--bg);font-size:13px;padding:4px 8px;border-radius:5px;max-width:300px;opacity:0;transition:opacity .1s}
.k .tip.show{opacity:.95}
/* questions */
.k .q{border-top:1px dashed var(--line);padding-top:4px;margin-top:10px}
.k .q:first-of-type{border-top:0;margin-top:0}
.k .mark{color:var(--faint);font-weight:400}
.k .lo{border-left:3px solid var(--acc);padding:2px 10px;margin:6px 0;color:var(--mut);font-size:14px}
/* LO checklist */
.k .bar{height:8px;background:var(--line);border-radius:4px;overflow:hidden;margin:6px 0 10px}
.k .bar>i{display:block;height:100%;width:0;background:var(--ok);transition:width .2s}
.k .chk{display:flex;gap:8px;align-items:flex-start;padding:3px 0}
.k .chk input{margin-top:5px}
.k .chk.done span{color:var(--mut);text-decoration:line-through;text-decoration-color:var(--faint)}
/* 3D */
.k .stage{position:relative;width:100%;height:380px;border:1px solid var(--line);border-radius:8px;overflow:hidden;background:var(--card);touch-action:none}
.k .stage canvas{display:block;width:100%;height:100%;cursor:grab}
.k .stage .hud{position:absolute;left:0;top:0;right:0;font-size:12px;color:var(--mut);pointer-events:none;background:var(--card);padding:4px 8px;border-bottom:1px solid var(--line);z-index:1}
.k .fallback{display:none}
.k .fallback.show{display:block}
/* ChemFig structures from {{chemfig}} (tools/chemfig2svg.py) */
.k svg.cf{display:inline-block;vertical-align:middle;overflow:visible}
.k svg.cf .cf-b line{stroke:var(--fg);stroke-width:1.4;stroke-linecap:round}
.k svg.cf .cf-a{fill:var(--fg);font-size:13px}
.k a.xref{color:var(--acc);text-decoration:none;border-bottom:1px dotted var(--acc)}
.k a.xref:hover{border-bottom-style:solid}
/* Periodic Table behind a button: <div class="k-ptable"></div> (kit.js builds it on first open).
   Fixed-size cells; a narrow output scrolls INSIDE .pt-scroll, never the page. Every text ≥ 12px. */
.k .k-ptable{margin:2px 0 8px;max-width:100%;min-width:0}
.k .k-ptable .pt-btn{font-size:14px;padding:3px 11px}
.k .k-ptable.open .pt-btn{background:var(--acc-soft);border-color:var(--acc)}
.k .k-ptable .pt-wrap{display:none;margin-top:6px}
.k .k-ptable.open .pt-wrap{display:block}
.k .pt-scroll{overflow-x:auto;max-width:100%;padding-bottom:2px}
.k .pt-grid{display:grid;grid-template-columns:20px repeat(18,minmax(44px,1fr));grid-template-rows:auto repeat(7,auto) 8px auto auto;
  gap:2px;min-width:max-content;line-height:1.15;font-size:12px}
.k .pt-gh,.k .pt-ph{color:var(--mut);font-size:12px;display:flex;align-items:center;justify-content:center}
.k .pt-gh{padding-bottom:1px}
.k .pt-el{display:flex;flex-direction:column;align-items:center;justify-content:center;padding:2px 1px 3px;
  border:1px solid var(--line);border-radius:4px;background:var(--card);text-align:center;white-space:nowrap}
.k .pt-el .pt-z,.k .pt-el .pt-a{font-size:12px;color:var(--mut)}
.k .pt-el .pt-s{font-size:17px;font-weight:700;line-height:1.2;color:var(--fg);text-transform:none}
.k .pt-el .pt-sr{font-size:13px;line-height:1.45}
.k .pt-el:hover{border-color:var(--acc);box-shadow:0 0 0 1px var(--acc)}
.k .pt-el.pt-bs{background:color-mix(in srgb,var(--s1) 9%,var(--card))}
.k .pt-el.pt-bp{background:color-mix(in srgb,var(--s3) 9%,var(--card))}
.k .pt-el.pt-bd{background:color-mix(in srgb,var(--s4) 10%,var(--card))}
.k .pt-el.pt-bf{background:color-mix(in srgb,var(--s5) 10%,var(--card))}
.k .pt-key{display:flex;align-items:center;gap:8px;color:var(--mut);font-size:12px;line-height:1.25;padding-left:4px}
.k .pt-keyc{display:inline-flex;flex-direction:column;align-items:center;min-width:44px;padding:2px 4px 3px;border:1px dashed var(--faint);border-radius:4px}
.k .pt-keyc .pt-z,.k .pt-keyc .pt-a{font-size:12px} .k .pt-keyc .pt-s{font-size:17px;color:var(--fg);line-height:1.2}
.k .pt-fl{display:flex;align-items:center;justify-content:flex-end;padding-right:6px;color:var(--mut);font-size:12px;font-style:italic}
.k .pt-note{font-size:12px;color:var(--faint);margin:4px 0 0}
</style>
<div class="k" id="k-ch16-24">
<h3>Learning outcomes 自查 <span class="faint count"></span></h3><p class="faint">勾选会存在这个浏览器里（localStorage）；Colab 不允许时只在当前页面有效。</p><div class="lolist" data-key="ch16"><div class="bar"><i></i></div><h4>16.1</h4>
<label class="chk"><input type="checkbox" data-id="16.1.1"><span><b>16.1.1</b> recall the reactions (reagents and conditions) by which alcohols can be produced:<ul><li>(a) electrophilic addition of steam to an alkene, H₂O(g) and H₃PO₄ catalyst</li><li>(b) reaction of alkenes with cold dilute acidified potassium manganate(VII) to form a diol</li><li>(c) substitution of a halogenoalkane using NaOH(aq) and heat</li><li>(d) reduction of an aldehyde or ketone using NaBH₄ or LiAlH₄</li><li>(e) reduction of a carboxylic acid using LiAlH₄</li><li>(f) hydrolysis of an ester using dilute acid or dilute alkali and heat</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="16.1.2"><span><b>16.1.2</b> describe:<ul><li>(a) the reaction with oxygen (combustion)</li><li>(b) substitution to form halogenoalkanes, e.g. by reaction with HX(g); or with KCl and concentrated H₂SO₄ or concentrated H₃PO₄; or with PCl₃ and heat; or with PCl₅; or with SOCl₂</li><li>(c) the reaction with Na(s)</li><li>(d) oxidation with acidified K₂Cr₂O₇ or acidified KMnO₄ to:<ul><li>(i) carbonyl compounds by distillation</li><li>(ii) carboxylic acids by refluxing</li></ul></li></ul><br><span class="faint"><i>primary alcohols give aldehydes which can be further oxidised to carboxylic acids, secondary alcohols give ketones, tertiary alcohols cannot be oxidised</i></span><ul><li>(e) dehydration to an alkene, by using a heated catalyst, e.g. Al₂O₃ or a concentrated acid</li><li>(f) formation of esters by reaction with carboxylic acids and concentrated H₂SO₄ as catalyst as exemplified by ethanol</li></ul></span></label>
<label class="chk"><input type="checkbox" data-id="16.1.3"><span><b>16.1.3</b> (a) classify alcohols as primary, secondary and tertiary alcohols, to include examples with more than one alcohol group<br>(b) state characteristic distinguishing reactions, e.g. mild oxidation with acidified K₂Cr₂O₇, colour change from orange to green</span></label>
<label class="chk"><input type="checkbox" data-id="16.1.4"><span><b>16.1.4</b> deduce the presence of a CH₃CH(OH)– group in an alcohol, CH₃CH(OH)–R, from its reaction with alkaline I₂(aq) to form a yellow precipitate of tri-iodomethane and an ion, RCO₂⁻</span></label>
<label class="chk"><input type="checkbox" data-id="16.1.5"><span><b>16.1.5</b> explain the acidity of alcohols compared with water</span></label></div>
</div>
<script>/* che_note Colab kit — shared interactions, inlined into every html cell by tools/build_colab.py.
   Defines window.K once; each cell then calls K.init(root) on its own .k root. */
(function () {
  if (window.K) return;
  var K = {};
  K.$ = function (root, sel) { return root.querySelector(sel); };
  K.$$ = function (root, sel) { return Array.prototype.slice.call(root.querySelectorAll(sel)); };
  K.el = function (tag, attrs, html) {
    var e = document.createElement(tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    if (html != null) e.innerHTML = html;
    return e;
  };
  K.svg = function (tag, attrs) {
    var e = document.createElementNS('http://www.w3.org/2000/svg', tag);
    for (var a in attrs || {}) e.setAttribute(a, attrs[a]);
    return e;
  };
  /* superscript / subscript digits for building formula strings in JS */
  var SUP = { '0': '⁰', '1': '¹', '2': '²', '3': '³', '4': '⁴', '5': '⁵', '6': '⁶', '7': '⁷', '8': '⁸', '9': '⁹', '+': '⁺', '-': '⁻' };
  K.sup = function (s) { return String(s).split('').map(function (c) { return SUP[c] || c; }).join(''); };
  K.nuc = function (A, Z, sym, q) {
    return '<span class="nuc"><sup>' + A + '</sup><sub>' + Z + '</sub></span>' + sym + (q ? K.sup(q) : '');
  };

  /* ask Colab to grow the output iframe after content changes; silent everywhere else */
  K.fit = function () {
    try {
      var r = document.querySelector('.k');
      var h = r ? Math.ceil(r.getBoundingClientRect().height) + 20 : document.documentElement.scrollHeight;
      if (window.google && google.colab && google.colab.output && google.colab.output.setIframeHeight)
        google.colab.output.setIframeHeight(h, true, { maxHeight: 5000 });
    } catch (e) { /* not in Colab */ }
  };

  /* nice round axis ticks from 0 to ≥ max */
  K.ticks = function (max, n) {
    var raw = max / (n || 4), mag = Math.pow(10, Math.floor(Math.log10(raw))), step = mag;
    [1, 2, 2.5, 5, 10].some(function (m) { step = m * mag; return step >= raw; });
    var out = [];
    for (var v = 0; v <= max + step * 0.999; v += step) out.push(+v.toPrecision(12));
    return out;
  };

  /* place an SVG text label next to (x, y) without touching any line segment in `avoid`
     ([[x1,y1,x2,y2],…]) or any rect already in `placed`; tries the candidate offsets in order.
     Every opaque thing on a chart must be in these lists — a label that hides a line looks like
     a broken line. Returns the text element. */
  function segHitsRect(s, r) {
    function inside(x, y) { return x >= r.x && x <= r.x + r.w && y >= r.y && y <= r.y + r.h; }
    if (inside(s[0], s[1]) || inside(s[2], s[3])) return true;
    var edges = [[r.x, r.y, r.x + r.w, r.y], [r.x, r.y + r.h, r.x + r.w, r.y + r.h], [r.x, r.y, r.x, r.y + r.h], [r.x + r.w, r.y, r.x + r.w, r.y + r.h]];
    return edges.some(function (e) {
      var d = (s[2] - s[0]) * (e[3] - e[1]) - (s[3] - s[1]) * (e[2] - e[0]);
      if (!d) return false;
      var t = ((e[0] - s[0]) * (e[3] - e[1]) - (e[1] - s[1]) * (e[2] - e[0])) / d;
      var u = ((e[0] - s[0]) * (s[3] - s[1]) - (e[1] - s[1]) * (s[2] - s[0])) / d;
      return t >= 0 && t <= 1 && u >= 0 && u <= 1;
    });
  }
  function rectsHit(a, b) { return a.x < b.x + b.w && b.x < a.x + a.w && a.y < b.y + b.h && b.y < a.y + a.h; }
  K.label = function (svg, x, y, text, o) {
    o = o || {};
    var cands = o.cands || [[10, 22, 'start'], [10, -11, 'start'], [-10, 22, 'end'], [-10, -11, 'end'], [12, 5, 'start'], [-12, 5, 'end'],
                              [0, -14, 'middle'], [0, 26, 'middle'], [16, 34, 'start'], [-16, 34, 'end'], [16, -22, 'start'], [-16, -22, 'end']];
    var t = K.svg('text', { 'class': 'halo ' + (o.cls || '') });
    t.textContent = text; svg.appendChild(t);
    var vb = (svg.getAttribute('viewBox') || '0 0 9999 9999').split(/\s+/).map(Number);
    var best = null, bestBad = 1e9, pad = o.pad == null ? 2 : o.pad;
    for (var i = 0; i < cands.length; i++) {
      t.setAttribute('x', x + cands[i][0]); t.setAttribute('y', y + cands[i][1]); t.setAttribute('text-anchor', cands[i][2]);
      var b = t.getBBox(), r = { x: b.x - pad, y: b.y - pad, w: b.width + 2 * pad, h: b.height + 2 * pad };
      var bad = (o.avoid || []).filter(function (s) { return segHitsRect(s, r); }).length * 2 +
                (o.placed || []).filter(function (q) { return rectsHit(q, r); }).length * 2 +
                (r.x < vb[0] || r.y < vb[1] || r.x + r.w > vb[0] + vb[2] || r.y + r.h > vb[1] + vb[3] ? 1 : 0);
      if (bad < bestBad) { bestBad = bad; best = { c: cands[i], r: r }; }
      if (!bad) break;
    }
    t.setAttribute('x', x + best.c[0]); t.setAttribute('y', y + best.c[1]); t.setAttribute('text-anchor', best.c[2]);
    if (o.placed) o.placed.push(best.r);
    if (bestBad) t.setAttribute('data-clash', bestBad);   /* tests assert there is none */
    return t;
  };
  /* polyline points → segments for K.label's avoid list */
  K.segs = function (pts) {
    var out = [];
    for (var i = 1; i < pts.length; i++) out.push([pts[i - 1][0], pts[i - 1][1], pts[i][0], pts[i][1]]);
    return out;
  };

  K.store = {
    get: function (k) { try { return window.localStorage.getItem(k); } catch (e) { return null; } },
    set: function (k, v) { try { window.localStorage.setItem(k, v); } catch (e) { /* sandboxed */ } }
  };

  function norm(s) { return String(s).trim().toLowerCase().replace(/\s+/g, ' ').replace(/[−–]/g, '-'); }

  /* a typed number → { v, sf } or null. Accepts 0.00500, 5.00e-3, 5.00×10⁻³, 5.00 x 10^-3;
     s.f. counts trailing zeros after a decimal point (0.100 → 3) but not in a bare integer (50 → 1) */
  var SUPN = '⁰¹²³⁴⁵⁶⁷⁸⁹';
  function numOf(s) {
    var t = String(s).replace(/\s+/g, '').replace(/[−–]/g, '-').replace(/⁻/g, '-')
      .replace(/[⁰-⁹]/g, function (c) { return SUPN.indexOf(c); }).replace(/[¹²³]/g, function (c) { return '¹²³'.indexOf(c) + 1; })
      .replace(/(?:[×x*]10\^?)(-?\d+)$/i, 'e$1');
    var m = t.match(/^([-+]?)(\d*)(?:\.(\d*))?(?:e([-+]?\d+))?$/i);
    if (!m || !(m[2] || m[3])) return null;
    var dg = ((m[2] || '') + (m[3] || '')).replace(/^0+/, ''), sf = m[3] != null || m[4] != null ? dg.length : dg.replace(/0+$/, '').length;
    return { v: parseFloat(t), sf: Math.max(sf, 1) };
  }
  function roundSf(x, n) { return x === 0 ? 0 : +x.toPrecision(n); }

  /* bind every widget inside root that is not bound yet; safe to call again after a cell
     rebuilds part of its DOM (e.g. a new quiz question) */
  function fresh(root, sel) {
    return K.$$(root, sel).filter(function (e) { if (e.__kBound) return false; e.__kBound = true; return true; });
  }
  K.init = function (root) {
    if (!root) return;

    /* reveal: <button class="reveal">答案</button><div class="ans">…</div> */
    fresh(root, 'button.reveal').forEach(function (b) {
      var label = b.textContent;
      b.addEventListener('click', function () {
        var a = b.nextElementSibling;
        while (a && !a.classList.contains('ans')) a = a.nextElementSibling;
        if (!a) return;
        var on = a.classList.toggle('show');
        b.textContent = on ? '收起' : label;
        K.fit();
      });
    });

    /* self-check fill-in: <input class="fill" data-ans="12|twelve"> */
    fresh(root, 'input.fill').forEach(function (inp) {
      var answers = inp.getAttribute('data-ans').split('|').map(norm);
      inp.setAttribute('autocomplete', 'off');
      var longest = Math.max.apply(null, answers.map(function (a) { return a.length; }));
      if (longest > 4) inp.style.width = (longest * 0.62 + 1.6) + 'em';
      inp.addEventListener('input', function () {
        var v = norm(inp.value);
        inp.classList.remove('good', 'badv');
        if (!v) return;
        /* exact text, or the same number; or the most precise listed value correctly ROUNDED (not
           truncated) to the s.f. the student wrote, with at least 2 s.f.: 0.099968 → 0.100 ✓, 0.0999 ✗ */
        var u = numOf(v), nums = answers.map(numOf).filter(Boolean);
        var best = nums.reduce(function (b, a) { return !b || a.sf > b.sf ? a : b; }, null);
        var ok = answers.indexOf(v) >= 0 || !!(u && nums.some(function (a) { return Math.abs(a.v - u.v) <= 1e-9 * Math.max(1, Math.abs(a.v)); })) ||
          !!(u && best && u.sf >= Math.min(2, best.sf) && Math.abs(roundSf(best.v, u.sf) - u.v) <= 1e-12 * Math.max(1, Math.abs(u.v))) ||
          !!(u && best && u.sf > best.sf && Math.abs(roundSf(u.v, best.sf) - best.v) <= 1e-12 * Math.max(1, Math.abs(best.v)));   /* more digits that round to it */
        inp.classList.add(ok ? 'good' : 'badv');
      });
    });

    /* single choice: <div class="mcq" data-ans="b"><button data-v="a">…</button>…<div class="why">…</div></div> */
    fresh(root, '.mcq').forEach(function (box) {
      var ans = box.getAttribute('data-ans');
      K.$$(box, 'button[data-v]').forEach(function (b) {
        b.addEventListener('click', function () {
          K.$$(box, 'button[data-v]').forEach(function (o) { o.classList.remove('good', 'badv'); });
          b.classList.add(b.getAttribute('data-v') === ans ? 'good' : 'badv');
          var why = K.$(box, '.why');
          if (why) why.classList.add('show');
          K.fit();
        });
      });
    });

    /* tabs: <div class="tabbar"><button data-tab="x">…</button></div> … <div data-pane="x"> */
    fresh(root, '.tabbar').forEach(function (bar) {
      var scope = bar.parentNode;
      var btns = K.$$(bar, 'button[data-tab]');
      function show(id) {
        btns.forEach(function (b) { b.classList.toggle('on', b.getAttribute('data-tab') === id); });
        K.$$(scope, ':scope > [data-pane]').forEach(function (p) { p.classList.toggle('show', p.getAttribute('data-pane') === id); });
        scope.dispatchEvent(new CustomEvent('tab', { detail: id }));
        K.fit();
      }
      btns.forEach(function (b) { b.addEventListener('click', function () { show(b.getAttribute('data-tab')); }); });
      if (btns.length) show(btns[0].getAttribute('data-tab'));
    });

    /* LO checklist: <div class="lolist" data-key="ch01"> with <label class="chk"><input data-id="1.1.1">… */
    fresh(root, '.lolist').forEach(function (list) {
      var key = 'che9701:' + list.getAttribute('data-key');
      var saved = {};
      try { saved = JSON.parse(K.store.get(key) || '{}'); } catch (e) { saved = {}; }
      var boxes = K.$$(list, 'input[data-id]');
      var bar = K.$(list, '.bar > i'), count = K.$(list, '.count') || K.$(root, '.count');
      function refresh() {
        var n = 0;
        boxes.forEach(function (b) { b.closest('.chk').classList.toggle('done', b.checked); if (b.checked) n++; });
        if (bar) bar.style.width = (100 * n / boxes.length) + '%';
        if (count) count.textContent = n + ' / ' + boxes.length;
      }
      boxes.forEach(function (b) {
        b.checked = !!saved[b.getAttribute('data-id')];
        b.addEventListener('change', function () {
          saved[b.getAttribute('data-id')] = b.checked;
          K.store.set(key, JSON.stringify(saved));
          refresh();
        });
      });
      refresh();
    });

    /* Periodic Table, collapsed behind a button: <div class="k-ptable"></div> (built on first open) */
    fresh(root, '.k-ptable').forEach(function (box) {
      var btn = K.el('button', { type: 'button', 'class': 'pt-btn', 'aria-expanded': 'false' }, '▸ Periodic Table');
      var wrap = K.el('div', { 'class': 'pt-wrap' });
      box.appendChild(btn); box.appendChild(wrap);
      btn.addEventListener('click', function () {
        if (!wrap.firstChild) wrap.innerHTML = K.ptableHTML();
        var on = box.classList.toggle('open');
        btn.setAttribute('aria-expanded', on ? 'true' : 'false');
        btn.textContent = (on ? '▾' : '▸') + ' Periodic Table';
        K.fit();
      });
    });

    if (root.__kInit) return;
    root.__kInit = true;

    /* tooltip for anything with data-tip (HTML or SVG) */
    var tip = K.el('div', { 'class': 'tip' });
    root.appendChild(tip);
    root.addEventListener('mousemove', function (ev) {
      var t = ev.target.closest ? ev.target.closest('[data-tip]') : null;
      if (!t || !root.contains(t)) { tip.classList.remove('show'); return; }
      tip.innerHTML = t.getAttribute('data-tip');
      tip.classList.add('show');
      var x = ev.clientX + 14, y = ev.clientY + 12;
      var w = tip.offsetWidth, h = tip.offsetHeight;
      if (x + w > window.innerWidth - 4) x = ev.clientX - w - 10;
      if (y + h > window.innerHeight - 4) y = ev.clientY - h - 10;
      tip.style.left = x + 'px'; tip.style.top = y + 'px';
    });
    root.addEventListener('mouseleave', function () { tip.classList.remove('show'); });

    /* keyboard between answer boxes: Enter → next box in reading order (right on the same line,
       else the first box of the next line); ↑ ↓ → nearest box above / below; ← → only once the
       caret is at that end of the text. A cell that handles a key itself calls preventDefault. */
    root.addEventListener('keydown', function (ev) {
      var cur = ev.target;
      if (ev.defaultPrevented || ev.isComposing || ev.altKey || ev.ctrlKey || ev.metaKey || ev.shiftKey) return;
      if (!cur.matches || !cur.matches('input:not([type]), input[type=text], input[type=number]')) return;
      var num = cur.type === 'number', k = ev.key;
      if (k !== 'Enter' && !(!num && (k === 'ArrowUp' || k === 'ArrowDown')) && k !== 'ArrowLeft' && k !== 'ArrowRight') return;
      if (k === 'ArrowLeft' || k === 'ArrowRight') {
        var at = cur.selectionStart, to = cur.selectionEnd;
        if (num || at == null || at !== to || at !== (k === 'ArrowLeft' ? 0 : cur.value.length)) return;
      }
      var r0 = cur.getBoundingClientRect(), cy0 = r0.top + r0.height / 2;
      var all = K.$$(root, 'input:not([type]), input[type=text], input[type=number]').filter(function (e) {
        return e !== cur && !e.disabled && !e.readOnly && e.offsetParent !== null;
      }).map(function (e) { var r = e.getBoundingClientRect(); return { e: e, r: r, cy: r.top + r.height / 2 }; });
      var line = function (c) { return Math.abs(c.cy - cy0) < Math.max(r0.height, c.r.height) / 2; };
      var best = null, score = function (c, s) { if (best === null || s < best.s) best = { e: c.e, s: s }; };
      all.forEach(function (c) {
        var same = line(c);
        if (k === 'ArrowRight' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
        if (k === 'ArrowLeft' && same && c.r.right <= r0.left + 1) score(c, r0.left - c.r.right);
        if (k === 'ArrowDown' && !same && c.cy > cy0) score(c, (c.cy - cy0) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'ArrowUp' && !same && c.cy < cy0) score(c, (cy0 - c.cy) * 4 + Math.abs(c.r.left - r0.left));
        if (k === 'Enter' && same && c.r.left >= r0.right - 1) score(c, c.r.left - r0.right);
      });
      if (k === 'Enter' && !best) {   /* nothing to the right: the nearest line below, its leftmost box */
        var below = all.filter(function (c) { return !line(c) && c.cy > cy0; });
        var top = Math.min.apply(null, below.map(function (c) { return c.cy; }));
        below.forEach(function (c) { if (Math.abs(c.cy - top) < c.r.height / 2) score(c, c.r.left); });
      }
      if (!best) return;
      ev.preventDefault();
      best.e.focus();
      if (best.e.type !== 'number') { var n = best.e.value.length; best.e.setSelectionRange(n, n); }
    });

    if (window.ResizeObserver) new ResizeObserver(function () { K.fit(); }).observe(root);
    setTimeout(K.fit, 50);
  };

  /* ---------- chemistry helpers ---------- */
  var SYM = ('H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr ' +
    'Rb Sr Y Zr Nb Mo Tc Ru Rh Pd Ag Cd In Sn Sb Te I Xe Cs Ba La Ce Pr Nd Pm Sm Eu Gd Tb Dy Ho Er Tm Yb Lu ' +
    'Hf Ta W Re Os Ir Pt Au Hg Tl Pb Bi Po At Rn').split(' ');
  /* Aufbau order up to 4p (the syllabus stops at krypton) */
  var ORDER = [['1s', 2], ['2s', 2], ['2p', 6], ['3s', 2], ['3p', 6], ['4s', 2], ['3d', 10], ['4p', 6]];
  var LORD = { s: 0, p: 1, d: 2 };
  var CORES = [[36, 'Kr'], [18, 'Ar'], [10, 'Ne'], [2, 'He']];
  function writeOrder(a, b) { return a.n - b.n || LORD[a.l] - LORD[b.l]; }

  K.chem = {
    SYM: SYM,
    ORDER: ORDER,
    /* ground-state configuration of element Z with charge q (H–Kr, ≤ 36 electrons).
       Cr and Cu use 3d⁵4s¹ / 3d¹⁰4s¹; cations lose electrons from the highest n first (4s before 3d). */
    config: function (Z, q) {
      q = q || 0;
      var occ = {}, i, left = Z;
      ORDER.forEach(function (o) { occ[o[0]] = 0; });
      for (i = 0; i < ORDER.length && left > 0; i++) {
        var take = Math.min(ORDER[i][1], left); occ[ORDER[i][0]] = take; left -= take;
      }
      if (Z === 24) { occ['3d'] = 5; occ['4s'] = 1; }
      if (Z === 29) { occ['3d'] = 10; occ['4s'] = 1; }
      var subs = ORDER.map(function (o) { return { name: o[0], n: +o[0][0], l: o[0][1], cap: o[1] }; });
      if (q > 0) {
        for (var k = 0; k < q; k++) {
          var cand = subs.filter(function (s) { return occ[s.name] > 0; })
            .sort(function (a, b) { return b.n - a.n || LORD[b.l] - LORD[a.l]; })[0];
          if (!cand) return null;
          occ[cand.name]--;
        }
      } else if (q < 0) {
        for (k = 0; k < -q; k++) {
          var room = subs.filter(function (s) { return occ[s.name] < s.cap; })[0];
          if (!room) return null;
          occ[room.name]++;
        }
      }
      subs.forEach(function (s) { s.e = occ[s.name]; });
      var used = subs.filter(function (s) { return s.e > 0; }).sort(writeOrder);
      var total = used.reduce(function (t, s) { return t + s.e; }, 0);
      function str(list) { return list.map(function (s) { return s.name + K.sup(s.e); }).join(''); }
      var full = str(used), short = full;
      for (i = 0; i < CORES.length; i++) {
        if (CORES[i][0] >= total) continue;
        var core = K.chem.config(CORES[i][0], 0);
        if (core.used.every(function (c) { return occ[c.name] === c.e; })) {
          short = '[' + CORES[i][1] + ']' + str(used.filter(function (s) { return !core.used.some(function (c) { return c.name === s.name; }); }));
          break;
        }
      }
      var unpaired = 0, shells = {};
      subs.forEach(function (s) {
        var orbs = s.cap / 2; unpaired += s.e <= orbs ? s.e : s.cap - s.e;
        if (s.e) shells[s.n] = (shells[s.n] || 0) + s.e;
      });
      return { subs: subs, used: used, occ: occ, total: total, full: full, short: short, unpaired: unpaired, shells: shells };
    },
    /* arrows in each orbital box: Hund's rule, then pair up */
    boxes: function (e, cap) {
      var orbs = cap / 2, out = [], i;
      for (i = 0; i < orbs; i++) out.push(0);
      for (i = 0; i < e; i++) out[i % orbs]++;
      return out;
    },
    /* '1s2 2s2…' / '[Ar]3d6 4s2' / with superscripts → {name: count}, cores expanded; null if unreadable */
    parse: function (txt) {
      var s = String(txt).replace(/[⁰¹²³⁴⁵⁶⁷⁸⁹]/g, function (c) { return '⁰¹²³⁴⁵⁶⁷⁸⁹'.indexOf(c); })
        .replace(/\s+/g, '').toLowerCase();
      var occ = {}, m = s.match(/^\[(he|ne|ar|kr)\]/);
      if (m) {
        var Zc = { he: 2, ne: 10, ar: 18, kr: 36 }[m[1]];
        K.chem.config(Zc, 0).used.forEach(function (u) { occ[u.name] = u.e; });
        s = s.slice(m[0].length);
      }
      /* written without spaces '3d64s2' is ambiguous, so s/p take one digit and d takes '10' or one digit */
      var re = /([1-4])(?:([sp])(\d)|(d)(10|\d))/g, mm, eaten = 0, seq = [];
      while ((mm = re.exec(s))) {
        var name = mm[1] + (mm[2] || mm[4]);
        occ[name] = (occ[name] || 0) + (+(mm[3] || mm[5])); eaten += mm[0].length; seq.push(name);
      }
      if (eaten !== s.length) return null;
      /* the written order, kept off the enumerable keys so same() still compares counts only */
      Object.defineProperty(occ, 'seq', { value: seq });
      return occ;
    },
    /* written in the syllabus order (by n, then s p d: [Ar]3d⁶4s², not [Ar]4s²3d⁶), each sub-shell once */
    ordered: function (occ) {
      var seq = (occ && occ.seq) || [];
      return seq.every(function (nm, i) {
        if (i === 0) return true;
        var a = { n: +seq[i - 1][0], l: seq[i - 1][1] }, b = { n: +nm[0], l: nm[1] };
        return writeOrder(a, b) < 0;
      });
    },
    same: function (a, b) {
      if (!a || !b) return false;
      var keys = {}; Object.keys(a).concat(Object.keys(b)).forEach(function (k) { keys[k] = 1; });
      return Object.keys(keys).every(function (k) { return (a[k] || 0) === (b[k] || 0); });
    }
  };

  /* The Periodic Table of Elements, syllabus p.91 (reference.md, last section): symbol + Aᵣ for
     Z = 1…118 in order, '-' where the table gives none. colab/tests/kit_ptable_data.js checks it. */
  var PTD = 'H1.0 He4.0 Li6.9 Be9.0 B10.8 C12.0 N14.0 O16.0 F19.0 Ne20.2 Na23.0 Mg24.3 Al27.0 Si28.1 P31.0 S32.1 Cl35.5 Ar39.9 ' +
    'K39.1 Ca40.1 Sc45.0 Ti47.9 V50.9 Cr52.0 Mn54.9 Fe55.8 Co58.9 Ni58.7 Cu63.5 Zn65.4 Ga69.7 Ge72.6 As74.9 Se79.0 Br79.9 Kr83.8 ' +
    'Rb85.5 Sr87.6 Y88.9 Zr91.2 Nb92.9 Mo95.9 Tc- Ru101.1 Rh102.9 Pd106.4 Ag107.9 Cd112.4 In114.8 Sn118.7 Sb121.8 Te127.6 I126.9 Xe131.3 ' +
    'Cs132.9 Ba137.3 La138.9 Ce140.1 Pr140.9 Nd144.4 Pm- Sm150.4 Eu152.0 Gd157.3 Tb158.9 Dy162.5 Ho164.9 Er167.3 Tm168.9 Yb173.1 Lu175.0 ' +
    'Hf178.5 Ta180.9 W183.8 Re186.2 Os190.2 Ir192.2 Pt195.1 Au197.0 Hg200.6 Tl204.4 Pb207.2 Bi209.0 Po- At- Rn- ' +
    'Fr- Ra- Ac- Th232.0 Pa231.0 U238.0 Np- Pu- Am- Cm- Bk- Cf- Es- Fm- Md- No- Lr- ' +
    'Rf- Db- Sg- Bh- Hs- Mt- Ds- Rg- Cn- Nh- Fl- Mc- Lv- Ts- Og-';
  /* period / group from Z; the 57–71 and 89–103 run goes in the f rows below (f = index 0…14) */
  function ptPos(z) {
    if (z <= 2) return { p: 1, g: z === 1 ? 1 : 18 };
    var starts = [3, 11, 19, 37, 55, 87], p = 2;
    while (p < 7 && z >= starts[p - 1]) p++;
    var i = z - starts[p - 2];           /* 0-based index within the period */
    if (p <= 3) return { p: p, g: i < 2 ? i + 1 : i + 11 };
    if (p <= 5) return { p: p, g: i + 1 };
    if (i < 2) return { p: p, g: i + 1 };
    if (i < 17) return { p: p, g: 3, f: i - 2 };
    return { p: p, g: i - 13 };
  }
  K.chem.PT = PTD.split(' ').map(function (t, i) {
    var m = t.match(/^([A-Z][a-z]?)(.*)$/), pos = ptPos(i + 1);
    return { z: i + 1, sym: m[1], ar: m[2] === '-' ? null : m[2], p: pos.p, g: pos.g, f: pos.f };
  });
  function ptBlock(e) { return e.f != null ? 'f' : e.g <= 2 || e.z === 2 ? 's' : e.g >= 13 ? 'p' : 'd'; }
  K.ptableHTML = function () {
    var h = ['<div class="pt-scroll"><div class="pt-grid" role="table" aria-label="The Periodic Table of Elements">'];
    function at(col, row, cls, inner, attrs) {
      h.push('<div class="' + cls + '" style="grid-column:' + col + ';grid-row:' + row + '"' + (attrs || '') + '>' + inner + '</div>');
    }
    for (var g = 1; g <= 18; g++) at(g + 1, 1, 'pt-gh', g);
    for (var p = 1; p <= 7; p++) at(1, p + 1, 'pt-ph', p);
    at('5 / 14', 2, 'pt-key', '<span class="pt-keyc"><span class="pt-z">Z</span><b class="pt-s">X</b><span class="pt-a">Aᵣ</span></span>' +
      '<span>Key：atomic number<br>atomic symbol<br>relative atomic mass（– = not given）</span>');
    K.chem.PT.forEach(function (e) {
      var col = e.f != null ? e.f + 4 : e.g + 1, row = e.f != null ? (e.p === 6 ? 10 : 11) : e.p + 1;
      at(col, row, 'pt-el pt-b' + ptBlock(e), '<span class="pt-z">' + e.z + '</span><b class="pt-s">' + e.sym + '</b><span class="pt-a">' + (e.ar || '–') + '</span>',
        ' role="cell" data-z="' + e.z + '" data-sym="' + e.sym + '" data-p="' + e.p + '"' + (e.f != null ? ' data-f="' + e.f + '"' : ' data-g="' + e.g + '"'));
    });
    at(4, 7, 'pt-el pt-bf pt-ph6', '<span class="pt-z">57–71</span><b class="pt-s pt-sr">La–Lu</b><span class="pt-a">↓</span>', ' data-range="57-71"');
    at(4, 8, 'pt-el pt-bf pt-ph7', '<span class="pt-z">89–103</span><b class="pt-s pt-sr">Ac–Lr</b><span class="pt-a">↓</span>', ' data-range="89-103"');
    at('1 / 4', 10, 'pt-fl', 'lanthanoids');
    at('1 / 4', 11, 'pt-fl', 'actinoids');
    h.push('</div></div><p class="pt-note">上方数字 = Group 1–18，左侧数字 = Period 1–7；底色只是 s / p / d / f-block 的提示；窄屏时表格可左右滚动。数据：syllabus p.91 The Periodic Table of Elements。</p>');
    return h.join('');
  };

  window.K = K;
})();
</script>
<script>(function($root){

K.init($root);
})(document.getElementById("k-ch16-24"));</script>"""))
```
