- 命名：**C≡N 的 C 算进主链**：CH₃CH₂CN 是 propanenitrile（3 个 C），C₄H₉CN（直链）是 pentanenitrile；
- hydroxynitrile：OH 和 CN 连在同一个 C 上，名字是「2-hydroxy…nitrile」：ethanal → 2-hydroxypropanenitrile，propanone → 2-hydroxy-2-methylpropanenitrile，pentan-2-one → 2-hydroxy-2-methylpentanenitrile；
- 两条路线都让碳链 **+1 C**。

### B. nitrile 的 hydrolysis

> **LO 19.2.3** — describe the hydrolysis of nitriles with dilute acid or dilute alkali followed by acidification to produce a carboxylic acid

C≡N 整个变成 –COOH（碳数不变），N 以 NH₄<sup>+</sup> 或 NH₃ 离开：

| 条件 | equation（butanenitrile） | product |
|---|---|---|
| **dilute acid**（HCl(aq)），reflux | CH₃CH₂CH₂CN + 2H₂O + HCl → CH₃CH₂CH₂COOH + NH₄Cl | carboxylic acid + **NH₄<sup>+</sup>**（NH₄Cl） |
| **dilute alkali**（NaOH(aq)），reflux | CH₃CH₂CH₂CN + NaOH + H₂O → CH₃CH₂CH₂COONa + NH₃ | **carboxylate salt** + NH₃（alkaline gas）；**followed by acidification** 才得到 acid |

- 离子式也可以：RCN + 2H₂O + H<sup>+</sup> → RCOOH + NH₄<sup>+</sup>（9701/22 Feb/Mar 2024 Q4(b)(iii)）；
- alkali 时判分（9701/22 May/June 2025 Q4(b)(iii)）：M1 正确的 organic product（RCO₂Na 或 RCO₂<sup>−</sup>）；M2 其余部分正确（NH₃，或 2H₂O → NH₄OH 的写法）；
- 一个 molecule 里有两个可水解的基团（CN 和 COOH）时，NaOH(aq) 把两个都变成 –COONa：CH₂(CN)COOH → CH₂(COONa)₂（9701/22 Feb/Mar 2026 Q4(a)(iv)）。

**hydroxynitrile** 的 hydrolysis 得到 **hydroxy carboxylic acid**（2-hydroxy acid）：CH₃CH(OH)CN → CH₃CH(OH)COOH（2-hydroxypropanoic acid，lactic acid）。

**数碳**（chapter 21 的路线题最常用）：

- halogenoalkane C<sub>n</sub> → nitrile C<sub>n+1</sub> → carboxylic acid C<sub>n+1</sub>；
- aldehyde / ketone C<sub>n</sub> → hydroxynitrile C<sub>n+1</sub> → hydroxy acid C<sub>n+1</sub>。
